## 1. 동체 대조의 기준선이 잘못 연결됐다

깊이를 고정한 기준선과 각 동체 배율을 따로 비교한다. 동체의 영향 배제는 현재 대조에서 미확정이다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^4]

## 2. 자카드 옆에 기대 교집합 개수를 같은 척도로 배치했다

관측 교집합 개수와 기대 교집합 개수를 비교하고 자카드는 별도 열로 표시한다. 기대 자카드를 쓸 때는 귀무모형을 함께 명시한다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^5]

## 3. 광선 예산 민감도 대조를 독립 재표본화 널 구간으로 읽는다

광선 예산 변경에 따른 사건 집합의 민감도를 확인했다. 동일 예산에서의 독립 반복 분포와 동등성 판정은 후속 실험 대상이다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^6]

## 4. 프레임 배율은 허브 위치만 바꾸는 대조가 아니다

프레임과 허브 위치를 함께 바꾼 대조에서 사건 집합이 변했다. 허브 위치만의 기여는 미분리다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^7]

## 5. 날개 박자보다 낮은 노치라는 이유로 신호 불변을 주장한다

저주파 대역을 제거했으며 자유공간 짝에도 같은 처리를 적용해 비교한다. 박자 주파수 보존과 전체 복소 신호 보존을 구분한다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^8]

## 6. 같은 자세를 보간해도 노치와 보간의 순서가 결과를 바꾼다

원본 진폭에서 이상치를 검출해 복소 전계를 보간한 뒤 노치를 적용했다. 그림은 처리 순서와 비교 결과를 함께 설명한다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^9]

## 7. 표본 수가 가중치에 들어간다는 사실만으로 예산 대조를 부정하면 과하다

표본 수 변경은 추정량의 정규화와 표본 배치를 함께 바꾼다. 후보 누락·상한·편향·분산을 따로 확인해 예산 대조를 해석한다.

[원문·재계산·한정](PRECISION_REVIEW_0910.md) [^10]
