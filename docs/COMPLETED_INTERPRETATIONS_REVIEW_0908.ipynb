**유지하는 결과:** 선언된 송신 전력·기하·완전한 기준채널·단일 헤딩·가드 해제 조건의 거리 계산과 국소 보정값은 남는다.

**수정 반영:** 확인 · [10_results](../reports/10_results.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^4]

## 2. 취소한 눈감기 검증이 절 제목에 남아 있다

**유지하는 결과:** 해당 메쉬와 주파수 창에서 계산한 문헌 대조값은 남는다. 사전 비열람이라는 실험 지위만 제외한다.

**수정 반영:** 확인 · [03_anchor](../reports/03_anchor.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^5]

## 3. 예산 보정의 예측을 예산을 맞춘 관측으로 서술한다

**유지하는 결과:** 기하를 고정한 예산 사다리의 통계 변화와 그 관계를 외삽한 예측은 남는다.

**수정 반영:** 확인 · [06_6_microdoppler-limits](../reports/06_6_microdoppler-limits.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^6]

## 4. RCS 크기 일치를 복소 위상 검증으로 확대한다

**유지하는 결과:** 시험한 PEC 이면각의 크기·입사·주파수 조건에서 이중반사 RCS가 기준식과 가까웠다는 관측.

**수정 반영:** 확인 · [02_kernel](../reports/02_kernel.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^7]

## 5. 얇은 띠의 편파별 오차로 드론 탐지가 보수적이라고 읽는다

**유지하는 결과:** 시험한 얇은 띠에서 PO와 해당 기준 계산의 편파별 차이가 컸다는 관측과 격자 수렴 결과.

**수정 반영:** 확인 · [02_kernel](../reports/02_kernel.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^8]

**유지하는 결과:** 그 원장 세대에서 사전 지정한 그룹 사이의 레벨오차 차이와 NOT_VALIDATED 판정.

**수정 반영:** 확인 · [03_anchor](../reports/03_anchor.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^9]

**수정 반영:** 확인 · [04_elevation-coverage](../reports/04_elevation-coverage.ipynb) · [수정 전 인용·근거·권장 문장](COMPLETED_INTERPRETATIONS_REVIEW_0908.md) [^10]
