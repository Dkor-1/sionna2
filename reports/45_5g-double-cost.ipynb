## 세대가 최신일수록 조명원으로 유리하다는 통념

이 통념을 세 항목이 뒤집는다. 5G 는 채널이 넓지만 **상시로 켜는 부분**은 좁고, 그 부분이 다시 오는 간격은 길다. 패시브가 빌려 쓰는 것은 채널이 아니라 그 좁은 상시 부분이다.

**PRS 는 측위 세션이 설정될 때 켜지는 옵션**이고, 남의 셀을 빌리는 패시브 수신기의 기본선은 상시 신호인 **SSB** 다.
