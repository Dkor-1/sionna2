# 리포트 3 — 선행연구는 어디까지 왔고 우리는 어디 서는가

> 공개 문헌과 오픈소스를 전수로 세어, 우리가 **새로 하는 것과 빌려 쓰는 것**을 갈라 적는다.

이 권은 아래 절로 이루어진다. 각 절은 **한 일 · 결과 · 방법 · 재현**을 자기 앞에 달고 있어, 필요한 절만 따로 읽어도 된다.

| 절 | 무엇을 말하나 |
|---|---|
| 1 | 게재본 중 드론 메쉬에서 산란을 계산한 것은 0편이다 |
| 2 | 프리프린트는 따로 세고, Ziganshin 은 두 판을 구별해 인용한다 |
| 3 | 표적 서명을 어디서 조달했는지가 그 논문이 낼 수 있는 주장의 크기를 정한다 |
| 4 | 조달처 일곱 갈래 R1~R7 을 전수로 적었다 |

⭐ 숫자는 전부 계산 결과 JSON(원장)에서 주입된다 — 각주가 그 출처다. 원장이 다시 계산되면 빌더를 돌리는 것만으로 본문 숫자가 따라 바뀐다.

## 절 1. 게재본 중 드론 메쉬에서 산란을 계산한 것은 0편이다



> ### 한 일
> **아카이브 문서의 전문을 추출해 관문 넷에 통과시키고, 게재본을 «광선엔진 밖에서 서명을 얻은 쪽» 과 «엔진을 직접 돌리는 쪽» 으로 갈라 표로 적었다.**

### 결과
1. 문서 21 건 [^1] (저작 19 편 [^2])을 전문 판정했다 — 게재 11 편 [^3] · 프리프린트 10 편 [^4], 축자 인용 46 건 [^5] 은 PDF 안 그 쪽에서 기계가 다시 찾는다.
2. 광선엔진을 직접 돌리는 편은 10 편 [^6], 그 엔진 안 표적이 드론 메쉬인 편은 3 편 [^7], 그 메쉬에서 산란 진폭을 계산한 편은 0 편 [^8] 이다.
3. census 16편에 1차 사료 정산이 5 편 [^9] 을 더했다 — 그중 Rzewuski 는 최종물이 우리와 가장 가깝고 아카이브 밖에 있었다.
4. 전문 쪽수 합계는 240 쪽 [^10] 이고, 읽고 범위 밖으로 뺀 문서가 11 건 [^11] 이다 — 뺀 사유를 원장에 함께 남겼다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 범위 | 아카이브 2곳 + 패시브레이더 코퍼스에서 178 편 [^12] 을 낱말 대조하고, 21 건 [^1] 을 전문 판정했다 |
| 귀속 | 판정마다 축자 인용을 PDF 그 쪽에서 기계가 찾는다 — 못 찾으면 빌드가 죽는다 (`prior_work/src/build_prior_survey.py:85`) |
| 게재상태 | PDF 메타데이터 `subject`(IEEE 조판이 박는 "게재처;연도;권;호;DOI") · 지면 스탬프 · arXiv 스탬프로 문서마다 확정했다 |
| 관문 | G1 엔진 · G2 드론 메쉬 · G3 그 메쉬에서 산란 진폭 · G4 절대 dBsm 보고 |

## 절 2. 프리프린트는 따로 세고, Ziganshin 은 두 판을 구별해 인용한다



> ### 한 일
> **프리프린트를 게재본과 다른 표에 담고, 같은 저작의 회의판과 저널판이 보고량까지 다르다는 것을 판별해 판 단위로 셌다.**

### 결과
1. 프리프린트는 10 편 [^15] 이고 게재본은 11 편 [^16] 이다 — 같은 표에 섞지 않는다.
2. 문서 21 건 [^17] 은 저작 19 편 [^18] 이다. 차이는 Costa 와 Ziganshin 의 판 쌍이고, 판마다 게재상태와 보고량이 갈린다.
3. Ziganshin 학회판(EuCAP 2025)은 게재본이고 패싯 차량 RCS 를 dBsm 으로 인쇄한다. 저널판(arXiv 프리프린트)은 근거리 산란전계와 패싯 이산화 품질을 보고한다.
4. 본문에 지면 채택을 적은 arXiv 원고는 `일부` 로 세고 프리프린트로 남긴다 — 이 편의 모든 계수가 그 규칙 아래에서 계산됐다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 판 단위 계수 | 판마다 게재상태와 보고량이 달라 저작이 아니라 판으로 센다 — 저작 수를 같은 화면에 함께 적는다 |
| 게재상태 확정 | PDF 메타데이터 · 지면 스탬프 · arXiv 스탬프 셋으로 문서마다 확정한다. 본문의 채택 문장은 `일부` 로 센다 |
| 드론 문장 | 학회판 결론의 축자 인용을 그대로 싣는다 — 요약하지 않는다 |

## Ziganshin — 두 판을 구별해 인용한다

어느 판을 인용하는지가 사실을 가른다. 판마다 게재상태와 보고량이 다르다.

| 판 | 게재상태 | 보고한 양 | 표적 |
|---|---|---|---|
| 학회판 (EuCAP 2025, DOI 10.23919/EuCAP63536.2025.10999367) | 게재 | 패싯 차량 RCS 를 dBsm 으로 (Fig.6) | 단순화 차량(PEC) |
| 저널판 (`arXiv:2604.05991v2`) | IEEE OJAP 투고 프리프린트 | 근거리 산란전계 \|E\| 와 패싯 이산화 품질 | 구·원기둥·PEC 차량 |

드론을 향후 과제로 그 단어 그대로 적는 문장은 **학회판 p.5 결론**에 있다: *"can also be applied to compute scattering from other objects, such as drones, humans, and micro-Doppler effects"*. 우리가 서 있는 자리는 그 문장 바로 뒤이고, 그 문장은 동료심사를 통과한 문장이다.

## 절 3. 표적 서명을 어디서 조달했는지가 그 논문이 낼 수 있는 주장의 크기를 정한다



> ### 한 일
> **편마다 표적 서명의 조달처를 판정해 갈래로 묶고, 갈래가 사 주는 주장의 크기를 게재 편수와 함께 적었다.**

### 결과
1. 갈래는 7개다 — 측정은 절대 σ 를, 스톡 엔진은 운동학 서명을, 엔진 안 산란 구현은 메쉬에서 낸 RCS 의 **각도 구조**를 산다.
2. 절대 σ 를 dBsm(1 m² 를 0 으로 잡은 dB 눈금)으로 찍는 편은 6 편 [^20] 이고, 그중 광선엔진을 돌리는 것은 1 편 [^21] 이며 그 표적은 차량이다.
3. 전문에 `CFAR` 도 `false alarm` 도 0회인 논문이 14 편 [^22] 이다 — 오경보 바닥을 명목값에 고정하는 것은 잡음·클러터 분포를 통제하는 시뮬이 하는 일이다.
4. 두 극단의 표본이 갈래의 뜻을 보여준다 — md-rt 는 자기 폐형식 운동학 식을 대조 기준으로 삼고, LAMBDA 는 σ 를 CADFEKO 에서 풀어 광선경로에 조회로 싣는다.

## 고른 조달처가 주장의 크기를 정한다

편마다 표적 서명의 조달처를 골랐고, 고른 조달처가 그 편이 낼 수 있는 주장의 크기를 정한다. 무향실 측정은 기체 한 대의 절대 RCS 를 사고, 외부 완전파 솔버는 바이스태틱 RCS 와 그것을 먹는 커버리지 예산을 사며, 미리 계산한 RCS 의 주입은 엔진을 건드리지 않고 자세별 진폭을 산다.

해석 블레이드 모형은 마이크로도플러 서명의 **모양**을 사고, 추상 계수는 폐형식 신호 모형을 사며, 광선엔진의 스톡 상호작용은 운동학 구조를 산다. 갈래 전수와 편수는 [편 11 «조달처 일곱 갈래 R1~R7 을 전수로 적었다»](11_procurement-catalog.ipynb) 에 있다.

**그림 1.** 각 갈래에 몇 건이 서 있는가?

## 절대 σ 를 dBsm 으로 찍는 편은 어디인가

![report01_p3_dbsm](../outputs/figures/report01_p3_dbsm.png)

**그림 2.** 절대 σ 를 dBsm 으로 찍는 편은 어디인가?

엔진 밖 조달의 다른 갈래는 표적 진폭을 식 (108) CN(0, σ²) 에서 뽑는다. 그 클러터 서베이(Proc. IEEE)의 41쪽 전문을 세면 `physical optics` 0 회 [^39] · `dBsm` 0 회 [^40] · `max_depth` 0 회 [^41] · `mesh` 2 회 [^42] · `Blender` 1 회 [^43] 다.

클러터 서베이가 표적 축을 비운 것은 그 편의 설계다 — 표적 진폭은 통계 추출이고 광선추적은 건물 클러터에 쓴다. 그 빈 칸이 우리 자리이고, 메쉬 해상도·광선예산·기종을 인쇄하는 것이 우리 재현성 규약이다.


## 입장료는 검증이다
