# 리포트 79 — −75° 에서 추적 대역 몫은 고정 대역보다 38.55 dB 크고, 그 차이를 만든 것은 대역을 어디에 놓았는가 하나다

> ### 한 일
> **앙각 7 점의 같은 시계열에서 프로펠러 대역을 두 가지로 잡아 서로 견주고, 같은 양을 세 정규화(전체 전력·반송파·대역외 기준띠)로 함께 냈다.**

### 결과
1. −75° 에서 추적 대역 몫은 -10.51 dB [^1], 고정 대역 몫은 -49.06 dB [^2] — 차이 +38.55 dB [^3] 다. 두 몫의 분모가 같은 시계열이라 이 차는 대역 선택 하나만 담는다.
2. 고정 대역이 비는 것은 대역 산수가 정한다 — 그 하단이 그 앙각의 f_tip 보다 위라 겹치는 폭이 0 이다.
3. 같은 폭 대역외 기준띠로 견주면 추적 대역은 그 띠 위 +46.84 dB [^4], 고정 대역은 +0.16 dB [^5] 로 띠 높이에 앉는다 — 이 두 여유는 띠를 2.6 kHz 한 자리에 놓아 얻은 값이다.
4. −75° 추적 대역이 담은 것은 f_flash 의 1·2 차 빗살이고, 그 선의 SNR 은 64.8 dB [^6] 와 50.9 dB [^7] 다.
5. −60° 에서는 반송파 몫이 -8.31 dB [^8] 로 내려앉아 «전체 대비» 몫이 혼자 뛴다 — 그래서 반송파 기준 +2.10 dB [^9] 를 나란히 싣는다.

## 프로펠러 대역은 앙각을 따라 내려간다

블레이드 끝이 만드는 도플러는 **날개끝 주파수 f_tip** 에서 멈춘다 — 기체에서 가장 빠른 산란체가 날개끝이기 때문이다. 그 f_tip 은 관측 앙각을 내리면 함께 내려간다.

우리 팔의 표를 그대로 읽으면 f_tip 은 0° 의 1272.9 Hz [^17] 에서 −75° 의 329.5 Hz [^18] 로 내려가고, −90° 에서 0.0 Hz [^19] 가 된다 — cos(el) 을 곱한 값 그대로다.

그러니 **대역을 어디에 놓느냐**가 판정을 정한다. 여기서는 두 가지로 놓고 잰다.

## 두 대역, 그리고 바닥을 읽는 자

| 대역 | 어디에 놓나 | −75° 에서의 폭 |
|---|---|---|
| 추적 대역 | 0.35~1.0 × f_tip(el) — 앙각마다 다시 잡는다 [^10] | 214 Hz [^20] |
| 고정 대역 | 8/11 덱이 −15° 의 f_tip 으로 못 박은 자리에 그대로 둔다 | 799 Hz [^21] |

두 대역 모두 **자기와 같은 폭의 대역외 기준띠**와 견준다[^13]. 그 띠는 2.6 kHz 위에 있어 블레이드가 원리적으로 못 오는 자리다 — 대역 몫이 그 띠와 같으면 그것은 신호가 아니라 **바닥**이다.

## −75° 한 자리에서 두 대역이 갈린다

| 무엇을 | 대역 몫 | 같은 폭 기준띠 | 띠 위 여유 |
|---|---|---|---|
| 추적 대역, −75° | -10.51 dB [^1] | -57.35 dB [^22] | +46.84 dB [^4] |
| 고정 대역, −75° | -49.06 dB [^2] | -49.22 dB [^23] | +0.16 dB [^5] |
| 고정 대역, −90° | -52.50 dB [^24] | -51.72 dB [^25] | -0.78 dB [^26] |

추적 대역은 −75° 에서 고정 대역보다 좁다. 좁은 대역이 전체 전력의 +38.55 dB [^3] 만큼 더 큰 몫을 가져갔다는 것은 그 에너지가 거기에 **몰려 있다**는 뜻이다. 두 몫의 분모가 같은 시계열이라 이 차는 대역 선택 하나만 담는다.

고정 대역이 비는 것은 대역 산수가 정한다 — 그 하단 430.1 Hz 가 −75° 의 f_tip 329.5 Hz [^18] 보다 위라 겹치는 폭이 0.0000 [^27] 다.

## −75° 의 추적 대역을 채운 것은 저차 빗살이다

f_flash 의 정수배에 선 빗살 선 SNR 을 원장이 30 차까지 싣는다[^28]. −75° 에서 추적 대역(115.3~329.5 Hz)에 드는 차수는 1·2 차이고, 그 두 선은 64.8 dB [^6] 와 50.9 dB [^7] 로 서 있다. 덱 대역이 담는 4~9 차는 같은 칸에서 10.7 dB [^29] ~ 21.3 dB [^30] 다.

f_flash 는 앙각과 무관하고 f_tip 만 cos(el) 로 줄기 때문에, 창이 담는 차수는 앙각이 내려갈수록 낮아진다. −75° 에서 추적 대역이 얻는 이득은 날개끝 신호를 지킨 대가라기보다 창이 가장 센 저차 빗살 위로 내려앉은 결과다.

−90° 는 f_tip 이 0 이라 추적 대역의 폭도 0 이다 — 그 칸은 «측정 불가» 로 비워 뒀다[^31]. 값이 비는 이유는 신호가 약해서가 아니라 **잴 대역이 없어서**다.

같은 −90° 에서 고정 대역 -52.50 dB [^24] 와 그 기준띠 -51.72 dB [^25] 는 -0.78 dB [^26] 안에서 같다. 같다는 것은 두 띠에 **같은 f_flash 빗살이 지나간다**는 뜻이다 — 고정 대역 안 차수 4~9 는 국소 바닥 위 3.8 dB [^32] ~ 25.7 dB [^33], 기준띠 안 차수 21~26 은 21.5 dB [^34] ~ 16.6 dB [^35] 로 서 있다. 이 잣대가 −90° 에서 재는 것은 «대역이 비었나» 라기보다 «블레이드만의 초과분이 있나» 이고, 그 초과분은 -0.78 dB [^26] 안에서 0 이다.

**그림 1.** 앙각을 내리면 프로펠러 대역 에너지가 정말 주는가?

점선이 각 팔의 대역외 기준띠다. **자기 점선 위에 앉은 표식은 바닥에 앉은 것**이고, 오른쪽 패널의 −75°·−90° 에서 붉은 선이 정확히 그렇게 된다.

추적 대역이 자기 대역외 기준띠 위로 서는 여유는 f_tip 이 살아 있는 여섯 앙각에서 −15° 의 +16.05 dB [^37] 가 가장 좁고 −75° 의 +46.84 dB [^4] 가 가장 넓다. 이 여유는 한 칸 안에서 두 대역의 비라 팔마다 다른 절대 눈금과 무관하다.

같은 여섯 점을 전체 전력 대비 «몫» 으로 보면 −15° 의 -20.59 dB [^38] 와 −60° 의 -6.21 dB [^39] 사이, 폭 14.4 dB [^40] 안에 흩어진다. 그 폭은 잣대를 바꾸면 달라진다 — 기준띠 위 여유로는 30.79 dB [^41], 반송파 기준으로는 20.87 dB [^42] 다. 앙각이 내려가는 동안 몫은 단조롭지 않다: −15°, −30°, −45°, −60° 가 각각 -6.71 dB [^43] 와 -10.75 dB [^44] 를 사이에 두고 오르내린다.

고정 대역도 −60° 까지는 자기 기준띠 위로 +19.12 dB [^45] 남아 있다가 −75° 에서 그 띠에 닿는다. 그 지점부터 고정 대역은 f_tip 위에 통째로 놓인다.

**그림 2.** 전체 전력으로 나눈 몫과 반송파로 나눈 몫이 어디서 갈라지는가?

| 칸 | 반송파 몫 [dB] | 전체 대비 추적 몫 [dB] | 반송파 대비 추적 몫 [dB] |

| ours/el+0 | -1.89 | -16.11 | -14.22 |
| ours/el-15 | -1.82 | -20.59 | -18.77 |

| ours/el-90 | -1.76 | 측정 불가 | 측정 불가 |

출처 [^36]

몫의 분모는 **전체 전력**이고, 그 전력의 대부분은 동체선(반송파)이다 — 우리 팔의 반송파 몫은 −60° 를 뺀 여섯 점에서 -1.82 dB [^46] ~ -3.35 dB [^47] 안에 있다. 그래서 반송파가 내려앉는 칸에서는 분자가 그대로여도 몫이 올라간다. −60° 가 그 자리다 — 반송파 몫이 -8.31 dB [^8] 로 내려앉고, 전체 대비 추적 몫 -6.21 dB [^39] 가 우리 팔 여섯 점 중 가장 커진다.

이 내려앉음은 **우리 팔 한 칸의 성질이다.** 같은 −60° 에서 물리를 끈 PathSolver 는 -2.27 dB [^48], 켠 판은 -1.81 dB [^49] 로 다른 앙각과 같다.

정규화를 반송파로 바꾸는 것은 이 칸을 되돌려 놓지 못한다 — 전체 전력의 대부분이 반송파라 두 분모가 함께 내려앉기 때문이다. 반송파 기준으로 재면 −60° 는 +2.10 dB [^9] 로 우리 팔 여섯 점 중 유일한 양수가 되어 오히려 더 튄다.

분모가 약분되는 잣대는 **같은 폭의 대역외 기준띠와의 여유**다. 그 잣대에서 −60° 는 +27.01 dB [^50] 로 0° 의 +24.06 dB [^51] · −45° 의 +25.01 dB [^52] 와 나란하고, −75° 가 +46.84 dB [^4] 로 가장 높다. 헤드라인 자리인 −75° 는 반송파 몫이 -2.17 dB [^53] 로 통상값 안에 있어, 전체 대비와 반송파 대비가 여섯 점 중 같은 순위를 준다[^54].

출처 [^36]

−75° 에서 고정 대역은 f_tip = 329.5 Hz [^18] 위에 통째로 놓인다 — 겹치는 폭이 0.0000 [^27] 다. 우리 팔은 그 자리를 기준띠 높이로 비워 둔다. PathSolver 두 팔은 갈린다 — 물리를 끈 팔은 +11.27 dB [^55] 를 채워 넣고, 켠 팔은 +0.25 dB [^56] 로 우리 팔과 마찬가지로 띠 높이에 앉는다.

⭐그 세 숫자가 재는 범위는 여기까지다. 기준띠에도 빗살 21~26 차가 들어 있어서, 이 여유는 «빗살 든 칸 대 빈 칸» 이 아니라 **빗살 대 빗살**의 대비다 — 같은 −75° 에서 물리를 끈 팔의 고정 대역 차수 4~9 는 국소 바닥 위 29.4 dB [^57] ~ 37.1 dB [^58] 이고, 기준띠 차수 21~26 은 22.4 dB [^59] ~ 26.8 dB [^60] 다. 빗살의 간격은 회전 입력 f_flash 가 정하므로, 빗살이 섰다는 사실은 산란 커널의 점수와 다른 물건이다.

가르는 것은 대역을 f_tip 따라 옮겼을 때다 — 같은 −75° 에서 우리 팔의 추적 대역은 자기 기준띠보다 +46.84 dB [^4] 높다. 빗살이 f_tip 위에서도 살아남는다는 것은 관측 사실이고, 그 빗살이 블레이드 도플러인지 자세별 경로 집합의 깜빡임인지는 [편 87 «광선 예산»](87_budget-not-physics.ipynb) 이 예산 축에서 가른다.

⭐그 축의 한쪽은 값이 이미 나와 있다 — 물리를 끈 10 m 사다리에서 광선을 360 배 부어도 상한 위 빗살 선은 el −15 에서 -1.21 [^61] · el −60 에서 +2.43 dB [^62] 로 제자리다. 광선을 더 부어도 상한 위 빗살은 그 자리에 남으므로, 이 절이 기다리는 판정은 «광선을 늘리면 사라지나» 밖의 축에서 온다 — 그 사다리는 10 m·물리 끔·앙각 0~−60 자리이고 여기 −75° 는 그 밖이다.
