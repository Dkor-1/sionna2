# 리포트 38 — 동체가 날개를 가리면 변조 깊이와 레벨이 함께 바뀐다

> ### 한 일
> **광선 엔진·재질·기하·운동학·광선 격자를 전부 같게 두고 «동체가 막느냐» 만 다르게 한 두 열을 **SBR 팔(B)** 에서 돌렸다.**

### 결과
1. ⭐ 이 축은 **SBR 팔(B)에서만 선다** — 순수 PO 팔(P)은 모든 면이 항상 기여하는 설계라 «막느냐» 라는 스위치가 그 팔의 밖에 있다.
2. 동체가 막으면 변조 깊이와 레벨이 **함께 움직인다** — 막는 열(F)의 깊이 표준편차는 5.93 dB [^1], 안 막는 열(G)은 6.56 dB [^2] 다(DJI Matrice 4E [^3] · 배 쪽 한 칸 · 6,000 표본 [^4]).
3. ⛔ 그 움직임의 **dB 크기는 판 선택 위에 있다** — 격자 판을 반 칸 옮기면 레벨 차가 4.16 dB [^5], 깊이 **p-p** 차가 26.52 dB [^6] 흔들려 같은 p-p 자로 잰 원장 값(`findings.occlusion_ptp_db` = -4.79 dB [^7]) 을 넘는다 — ⚠ 판 축은 p-p 로만 재어 두었다. 두 열의 절대 레벨도 판 셋 사이 3.45 dB [^8] p-p 라 같은 규칙 아래 둔다 — 이 편은 크기 대신 **존재**를 세운다.
4. 두 열의 광선 격자를 같게 유지했다 — 막는 쪽은 동체를 완전흡수로 두고, 안 막는 쪽은 동체 면만 빼되 정점은 남겼다.
5. ⚠ 부호를 물리로 단정하지 않는다 — 합이 코히런트라 항이 줄어도 레벨이 올라갈 수 있고, 실제로 그런 칸이 있다.

## 가림만 남기고 다른 것을 전부 묶었다

![report07_f4](../outputs/figures/report07_f4.png)

**그림 1.** 같은 광선·같은 메쉬·같은 운동에서 동체가 막으면 무엇이 달라지는가?
한쪽은 동체를 완전흡수로 두어 **광선은 막되 산란은 안 하게** 하고, 다른 쪽은 동체 면만 빼되 정점 배열은 남겨 두었다.

정점을 남기면 경계상자가 같아서 두 팔의 광선 수와 간격이 같아진다. 정점을 빼면 «가림» 과 «표본화» 가 섞여 축이 둘이 된다.

## 두 열의 값 — 둘 다 SBR 팔이다

둘 다 **B** = Ours (SBR+PO, 기본) 이고, 여기서 «두 팔» 은 엔진이 아니라 **동체를 막는 열과 막지 않는 열**을 가리킨다. 오른쪽 두 열이 이 표를 읽는 법을 정한다 — 판 셋 흩어짐보다 작은 칸은 본문 밖에 둔다.

| 무엇을 | 막는 열 (F) | 안 막는 열 (G) | 차이 (F−G) | 판 셋 p-p |
|---|---|---|---|---|
| 변조 깊이 std ⭐ | 5.93 dB [^1] | 6.56 dB [^2] | ⛔ 본문 밖 | ⚠ 판 원장은 `ptp_db`·`level_db` 만 든다 |
| 변조 깊이 p-p ⛔ | 50.32 dB [^9] | 55.12 dB [^10] | ⛔ 본문 밖 | 26.52 dB [^6] |
| 레벨 | ⛔ 본문 밖 | ⛔ 본문 밖 | ⛔ 본문 밖 | 4.16 dB [^5] (팔별 절대 3.45 dB [^8]) |
| 동체:날개 비 | 0.761 [^11] | 0.644 [^12] | — | — |
| 날개끝 안쪽 에너지 비 | 0.8859 [^13] | 0.8066 [^14] | — | — |

⛔ «본문 밖» 칸의 값은 원장 `outputs/report15b_microdoppler.json` 에 그대로 있다. 크기를 뺀 이유가 둘이다 — 레벨 차(F−G)는 판 셋 흩어짐 4.16 dB [^5] 안에 들고, 팔별 절대 레벨은 판 한 장의 서브셀 오프셋에 3.45 dB [^8] p-p 로 걸린다. 얼린 복소장은 **무늬의 모양**을 내고, 절대 σ 는 디더를 켠 정적 경로가 낸다.

⛔ **p-p 두 칸은 표본 수 위에도 있다** — 같은 창을 등간격으로 솎아 N 을 6,000 [^4] 에서 60 으로 줄이면 F 의 p-p 가 50.3 → 27.5 dB, G 가 55.1 → 29.3 dB 로 내려가는데, 같은 솎기의 std 는 5.93 → 5.84 dB · 6.56 → 6.42 dB 로 붙어 있다. ⭐ 그래서 이 절이 깊이를 인용할 때 쓰는 자는 std 다. ⚠ 그 std 가 판 한 장에 얼마나 걸리는지는 이 원장이 답하지 않는다 — 판 원장(`outputs/freeze_plate_sensitivity.json`)은 팔마다 `ptp_db` 와 `level_db` 만 든다.

## 여섯 칸의 부호는 갈린다

위 표는 한 칸이다. 같은 원장의 여섯 칸을 전부 늘어놓으면 레벨 차의 **부호가 칸마다 갈린다** — 크기를 물리로 읽는 대신 «두 양이 함께 움직인다» 까지만 읽는 이유가 여기 있다.

| 칸 | 레벨 차 (F−G) | 변조 깊이 차 (F−G) |
|---|---|---|
| matrice4e/nose | -0.042 dB [^15] | +4.44 dB [^16] |
| matrice4e/belly | +1.315 dB [^17] | -4.79 dB [^7] |
| matrice4e/belly_side | -0.847 dB [^18] | +23.19 dB [^19] |
| mini5pro/nose | -0.000 dB [^20] | +0.00 dB [^21] |
| mini5pro/belly | -1.136 dB [^22] | -10.30 dB [^23] |
| mini5pro/belly_side | -1.057 dB [^24] | -5.74 dB [^25] |

⭐ `mini5pro/nose` 는 F 열과 G 열의 변조 깊이가 배정밀도 자릿수까지 같은 값이고 레벨 차도 그 자릿수 안에서 0 이다 — 그 자세에서 동체가 블레이드를 덮는 몫이 0 이다. 이 축이 자세에 얼마나 걸리는지가 그 칸에 있다.

⚠ 여섯 칸의 레벨 차는 전부 판 셋 흩어짐 4.16 dB [^5] 안에 든다. 그래서 이 표가 세우는 것은 **«칸마다 다르다»** 까지이고, 칸 하나의 dB 는 판 선택 위에 있다.

## ⭐이 표의 dB 는 «판 한 장» 에도 걸린다

두 팔은 같은 얼린 광선 격자를 쓴다. 그러면 격자를 어디에 놓았느냐는 두 팔에 똑같이 실리므로 차이에서 빠질 것 같다 — 그것이 «단일축» 이라는 말의 뒷받침이었다. 그 가정을 재 봤다: 판의 중심만 **반 칸** 옮긴 같은 크기의 판으로 같은 자세열을 다시 태우면 (광선 수·간격이 같고 서브셀 오프셋만 다르다) 이렇게 움직인다.

| 무엇이 | 판 셋 사이 p-p | 원장에 있는 값 | 이 편이 쓰는 자리 |
|---|---|---|---|
| 전체 팔의 절대 레벨 | 3.45 dB [^8] | `arms.*.level_db` | ⛔ 본문 밖 — 절대 σ 는 디더를 켠 정적 경로가 낸다 |
| 가림 · 레벨 (F−G) | 4.16 dB [^5] | `findings.occlusion_level_db` | ⛔ 본문 밖 — 판 흩어짐이 그 값을 덮는다 |
| 가림 · 변조 깊이 (F−G) | 26.52 dB [^6] | `findings.occlusion_ptp_db` | ⛔ 본문 밖 — 깊이 차의 **부호도** 판에 따라 바뀐다 |

⭐ 셋째 열은 **키 이름**이다. 값은 원장에 그대로 있고, 이 편의 본문·표는 그 크기를 빼고 «두 양이 함께 움직인다» 만 세운다 — 규칙과 본문이 한 자리에서 맞는다.

⭐ 그러니 이 편이 서는 자리를 좁힌다. **«동체가 막으면 두 양이 함께 움직인다» 는 그대로 서고, 그 움직임의 dB 크기는 판 선택 위에 있다.** 크기가 필요한 자리에서는 판 앙상블 평균(오프셋 여러 판의 평균)이 먼저다 [^26].

두 팔의 기하가 서로 달라서(막는 팔은 동체가 있고 안 막는 팔은 동체 면이 없다) 판 편향이 공통모드로 빠지지 않는다 — 그것이 이 표가 재서 확인한 내용이다 [^27].

이 비교가 같은 물리를 잰 것인지의 게이트도 함께 둔다 — 생산 판(P0)으로 낸 세 팔의 레벨이 생산 원장과 최대 0.53 dB [^28] 안에서 붙는다 [^29]. 그러니 위 흩어짐은 배선 차이가 아니라 판 오프셋 그 자체다.

## 부호를 물리로 단정하지 않는 이유

합이 코히런트라 항이 줄어도 남은 항끼리 상쇄가 덜 되면 레벨이 **올라갈 수 있다**. 실제로 그런 칸이 있다.

근거는 `outputs/report15b_microdoppler.json : cells.*.findings.occlusion_sign_note_ko` 에 있다. 그래서 이 편이 말하는 것은 «가림이 레벨을 내린다» 가 아니라 **«가림이 두 양을 함께 움직인다»** 다.

가림이 무는 자세가 어디인지는 [편 40 «자세 축»](40_md-attitude.ipynb) 이 잰다.
