# 리포트 32 — 모양의 유무는 부호가 확실하고(자릿수는 우리 격자가 정한다), 모양의 정밀도는 한 자릿수 dB 안에서 논다

> ### 한 일
> **교정 사다리에서 진짜 CAD 대비 오차를 방위별로 짝지어 빼고 «단순화의 값» 을 네 구간으로 갈랐다.**

### 결과
1. **낭떠러지** — 날개를 회전대칭 원판으로 바꾸면 검출 단면적 오차가 -56.48 dB [^1] · -54.44 dB [^2] 다. 회전대칭체의 참 변조는 정확히 0 이므로 이 두 수가 재는 것은 낭떠러지의 «깊이» 가 아니라 **메쉬가 우리 점구름의 수치 바닥보다 얼마나 위인가**이고, 그래서 부호까지만 인용한다.
2. **공짜 구간** — 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 +0.01 dB [^3] · -0.01 dB [^4], 템플릿 손실이 0.00 dB [^5] 움직인다.
3. **싼 구간** — 날개를 평판 두 장으로 바꾸면 검출 단면적이 -1.63 dB [^6] · -6.23 dB [^7] 틀리고 템플릿 손실은 7.15 dB [^8] 다.
4. «디테일 0 인 평판» 이 «메쉬 − 구» 간격의 94% [^9]~99% [^10] 를 이미 산다 — 남는 «CAD 정밀도 단독 몫»(메쉬 − 평판)은 0.51 [^11]~4.97 dB [^12] 다. ⛔ 간격 자체의 크기는 구 널의 격자 바닥이 정하므로 하한으로만 쓴다.
5. ⚠ «공짜» 는 **도는 부품에 한정된 진술**이다 — 동체까지 깎으면 AC 상대 변화가 9.7e-15 [^13] (기계 정밀도)로, 이 커널에서 동체는 AC 에 정확히 0 을 기여한다.

**그림 1.** 여섯 표적 모델의 마이크로도플러는 눈으로 얼마나 갈리는가?
열두 칸 모두 같은 색눈금과 같은 축이다. 세로축은 도플러(Hz), 가로축은 느린 시간(ms), 색은 그 도플러 칸에서 보이는 RCS 다. 점선은 운동학이 예측하는 날개 끝 도플러다. 윗줄이 있는 그대로의 사다리, 아랫줄이 교정 사다리다.

## 눈으로 읽히는 것 셋

1. 어느 칸에서나 가장 밝은 것은 0 Hz 가로줄 — 몸통이다. 날개 신호는 그보다 한참 어둡다. 이 낙차(호버 벌금)가 10 기체 평균 25.4 [^14]~38.7 dB [^15] 다.
2. 회전대칭 표적(구·원판)은 0 Hz 한 줄만 있고 위아래가 비어 있다. 돌려도 모양이 안 바뀌니 변조를 만들 방법이 막힌다 — 이것은 계산 결과가 아니라 기하학이다.
3. 아랫줄에서 배음 풍부도가 가장 높은 칸은 C(감싸는 상자)·D(평판)이고 E·F(진짜 CAD 쪽)에서 내려간다 — Matrice 4E 의 n_eff 는 C 12.02 [^16] · D 11.85 [^17] · F 7.94 [^18] 다. 평판이 진짜 CAD 보다 배음이 풍부한 이 역전이 이 편의 핵심 대조이고, 가림을 켜면 그 부호가 뒤집힌다([편 33 «이르다고 부르는 이유»](33_ladder-premature.ipynb) P2).

## 읽는 법 — 네 구간

- **공짜 — 해상도.** 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 -0.01 dB [^4], 템플릿 손실이 0.00 dB [^5] 움직인다. 잣대는 **같은 양**의 방위 산포다 — 검출 단면적 자체가 방위 24 점 [^41] 사이에서 Mini 2 4.05 dB [^42] · Matrice 4E 2.91 dB [^43] 흔들리는데, 방위를 짝지어 뺀 이 차이의 표준편차는 그 24 점 [^44] 에서 0.012 dB [^45] · 0.069 dB [^46] 다. ⚠ 잣대로 7.99 dB [^47] 를 쓰던 자리다 — 그것은 다른 양(방위별 총 RCS)의 방위 사이 표준편차를 12 칸 [^48] 짜리 풀에서 최대로 잡은 값이고, 그 최댓값이 나온 행은 mavic4pro 인데 이 절은 mini2 · matrice4e 를 다룬다.
- **싼 구간 — 날개를 평판으로.** 스팬·두께·부피가 같은 평판 두 장이면 검출 단면적이 한 자릿수 dB 안에 든다. «변조가 있나 없나» 만 보는 검출기라면 충분하고, 파형을 본뜨는 검출기라면 이미 비싸다.
- **비싼 구간 — 날개를 덩어리로.** 감싸는 상자는 속을 꽉 채웠으니 검출 단면적을 과대하게 낸다. 회전반경 위의 공은 Matrice 4E 에서 레벨만 보면 «잘 맞는» 것처럼 보이지만 파형 상관이 0.224 [^23] 뿐이라 템플릿 손실이 13.00 dB [^22] 다 — **레벨이 맞았다고 신호가 맞은 것이 아니다**.
- **낭떠러지 — 회전대칭.** 원판으로 바꾸면 검출 단면적 오차가 -54.44 dB [^2] 다. 어떤 커널 결함에도 안 흔들리는 것은 **부호**다 — 회전대칭이면 변조가 원리적으로 0 이라 오차는 음수다. **자릿수는 우리 격자가 정한다**: 원판 팔의 AC 중 운동학이 허용하는 대역에 남는 몫은 Mini 2 0.0090 [^49] · Matrice 4E 0.0007 [^50] 이고(첨두 차수 중앙값 49 [^51] · 111 [^52]), 같은 팔에서 점 간격을 1/4 로 줄이면(점 4116 [^53]→67550 [^54] · 21534 [^55]→384524 [^56]) 동체:날개 비가 +17.74 dB [^57] · +35.30 dB [^58] 움직인다. ⛔ 그래서 이 값은 부호까지만 인용한다.

## 사다리의 순서는 삼각형 수가 아니라 보존량이 정한다

감싸는 상자(C)가 평판(D)보다 오차가 큰 것은 «형상 정보가 더 적어서» 가 아니라 부피를 안 지켜서다. 즉 «단순화의 값» 을 정하는 것은 삼각형 수가 아니라 **무엇을 보존했는가**(회전대칭 깨짐 → 스팬 → 부피 → 두께 순)다.

⭐ 한 줄로: **«모양이 있느냐 없느냐» 는 우리가 잴 수 있는 바닥 끝까지 가르고, «모양이 얼마나 정밀하냐» 는 한 자릿수 dB 안에서 논다.** 지도교수의 지적은 앞쪽이 아니라 뒤쪽을 겨눈 것이고, 뒤쪽에서는 지적이 맞는다.

⛔ 앞쪽에 자릿수를 붙이는 것은 우리 격자다 — 회전대칭 팔(구·원판)의 참 변조가 정확히 0 이라 그 자리에 남는 값은 점 간격이 정한 바닥이다. 앞쪽에서 확실한 것은 부호와 «뒤쪽과 자릿수가 다르다» 까지다.

## 지도교수 지적에 답한다 — 네 축

| 축 | 팔 | 기체 | 무엇을 쟀나 | 결과 |
|---|---|---|---|---|
| ① 절대 세기 | B — SBR+PO (가림 O) | Phantom 3 | 실측 앵커 대비 rms | 자유 모수 0 개 [^59] 짜리 등가부피 구가 우리 Phantom 3 메쉬를 rms 1.73 dB [^60] 이긴다 |
| ② 방위 산포 | B — SBR+PO (가림 O) | Phantom 3 | 실측 대비 오차 | 우리 Phantom 3 메쉬 +1.42 dB [^61] · 구 -5.46 dB [^62] · 상자 +4.48 dB [^63] |
| ③ 시간 변조 | P — 순수 PO (가림 X) | Mini 2 · Matrice 4E | 메쉬 − 구 간격 — **하한만** | 간격의 크기는 구 널의 격자 바닥이 정한다(바로 아래 ⛔ 절). 그 간격에서 CAD 정밀도 단독 몫은 0.51 [^11]~4.97 dB [^12] 다 |
| ④ 단순화의 값 | P — 순수 PO (가림 X) | Mini 2 · Matrice 4E | 교정 사다리 | 위 두 표가 그 답이다 |

①② 는 Phantom 3 를 B 팔로 잰 축이고 그 엔진은 SBR (Mitsuba/OptiX first-hit) + PO surface integral — src/rcs_sbr.rcs_sbr_batch [^64] 이며 실측 앵커는 outputs/p3_validation_v2.json (Yuan θ90 복원 실측곡선 기준) [^65] 다. ③④ 는 Mini 2 · Matrice 4E 를 P 팔로 잰 축이고 그 엔진은 pure PO on point clouds (no occlusion, no edge diffraction, scalar |Gamma|) [^66] 다.

① 에서는 지적이 맞는다. ② 는 구가 원리적으로 못 내는 축이다 — 구의 방위 산포는 «작다» 가 아니라 정확히 0 이고, 회전대칭이라 만들 방법이 막혀 있다. ③ 의 큰 간격은 «메쉬가 정밀해서» 가 아니라 «회전대칭이 아니라서» 번 것이다.

⚠ ① 의 «자유 모수 0 개» 는 **구의 부피를 논문표기 상자로 잡았을 때**의 값이다 — 그 구는 레벨오차 +0.96 dB [^67] 이고, 같은 기체에서 **메쉬 부피**로 잡은 구는 -4.47 dB [^68] 로 우리 Phantom 3 메쉬 -3.30 dB [^69] 보다 멀다. 즉 이 축의 자유 매개변수는 구의 모수가 아니라 **어느 부피에 맞출지의 선택**이고, 그 대조군 표는 [편 27 «상자·구 대조군»](27_box-sphere-control.ipynb) 에 있다.

⭐ 그래서 다시 잡아야 할 방향은 «정밀도» 가 아니라 «무엇이 도는가» 다([편 30 «사다리가 셋이었다»](30_ladder-three.ipynb)).

## ⛔ ③ 의 분모가 무엇인지 — 구 팔의 값은 격자가 정한다

이 편의 규약(반송파 3.50 GHz [^70] · 구면파 · 방위 24 점 [^71] 평균)에서 «메쉬 − 구» 는 Mini 2 62.0 [^72] · Matrice 4E 76.8 dB [^73] 다. 분모에 서는 구 팔의 값은 경도 분할 수(seg)를 따라간다. 운동학이 허용하는 대역에 남는 AC 몫은 Mini 2 0.135 [^74] · Matrice 4E 0.036 [^75] 인데 같은 표에서 메쉬는 1.000 [^76] 이다.

seg 를 흔들면 그 바닥이 어디로 가는지 원장이 네 판을 적어 둔다 — Mini 2 는 seg 21 [^77] 에서 -95.94 [^78] · seg 39 [^79] 에서 -84.96 [^80] · seg 79 [^81] 에서 -118.33 [^82] · seg 157 [^83] 에서 -125.75 dB [^84] 다. 네 판을 그대로 싣는다 — 이 편이 여기서 쓰는 것은 «구 팔의 값이 seg 와 함께 움직인다» 까지다.

⛔ 그래서 ③ 은 **하한**으로만 쓴다. 회전대칭체의 참 변조가 정확히 0 이라 구 팔의 참값이 −∞ 쪽이고, 잰 값은 우리 점구름이 멈춘 자리다. 평판이 산 몫(%)도 같은 이유로 하한이다 — 분자·분모가 같은 바닥을 빼고 있어서 바닥이 내려갈수록 그 몫은 커진다.

## 자기반증 — 이 «공짜» 가 어디까지 참인가

이 커널에서 동체는 AC 에 정확히 0 을 기여한다. 동체를 아무리 깎아도 마이크로도플러 지표가 눈금 하나 움직이지 않는 것은 형상의 성질이 아니라 **가림 없는 커널의 성질**이다.

| 검사 | Mini 2 | Matrice 4E |
|---|---|---|
| 동체까지 깎았을 때 AC 의 상대 변화 | 9.7e-15 [^13] | 5.1e-15 [^85] |
| 같은 팔에서 동체:날개 비의 이동 | -1.88 dB [^86] | +0.34 dB [^87] |

첫 줄이 기계 정밀도라는 것은 AC 가 동체를 아예 안 본다는 뜻이다. 둘째 줄이 0 에서 떨어져 있는 것은 동체 간략화가 «몸통 대 날개 비» 는 움직인다는 뜻이다 — 면적·부피를 잃기 때문이다. 가림을 켜면 날개가 동체 뒤로 사라졌다 나타나면서 동체 형상이 AC 에 들어오기 시작한다.
