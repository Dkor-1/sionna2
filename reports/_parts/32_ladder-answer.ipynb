# 리포트 32 — 모양의 유무는 수십 dB 를 가르고, 모양의 정밀도는 한 자릿수 dB 안에서 논다

> ### 한 일
> **교정 사다리에서 진짜 CAD 대비 오차를 방위별로 짝지어 빼고 «단순화의 값» 을 네 구간으로 갈랐다.**

### 결과
1. **낭떠러지** — 날개를 회전대칭 원판으로 바꾸면 검출 단면적이 -56.48 dB [^1] · -54.44 dB [^2] 무너진다. 모양이 거칠어서가 아니라 «돌려도 안 바뀌는 물체» 라서다.
2. **공짜 구간** — 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 +0.01 dB [^3] · -0.01 dB [^4], 템플릿 손실이 0.00 dB [^5] 움직인다.
3. **싼 구간** — 날개를 평판 두 장으로 바꾸면 검출 단면적이 -1.63 dB [^6] · -6.23 dB [^7] 틀리고 템플릿 손실은 7.15 dB [^8] 다.
4. 메쉬와 구의 시간 변조 간격 62.0 [^9]~100.8 dB [^10] 중 94% [^11]~99% [^12] 를 «디테일 0 인 평판» 이 이미 산다.
5. ⚠ «공짜» 는 **도는 부품에 한정된 진술**이다 — 동체까지 깎으면 AC 상대 변화가 9.7e-15 [^13] (기계 정밀도)로, 이 커널에서 동체는 AC 에 정확히 0 을 기여한다.

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 오차 정의 | 교정 사다리의 각 단을 진짜 CAD 와 **같은 방위끼리** 빼고 방위평균했다 |
| 검출 단면적 오차 | 그 형상을 쓰면 검출 성능이 얼마나 틀리는가 — 가장 센 배음 한 줄의 RCS 차 |
| 템플릿 손실 | 그 형상으로 정합필터 본을 뜨면 잃는 SNR = −20 log10(방위평균 파형 상관) |
| 동체 감도 | 동체까지 삼각형을 깎아 AC 가 움직이는지를 따로 쟀다 — 커널이 동체를 보는지 여부가 «공짜» 의 성립 범위를 정한다 |

**그림 1.** 여섯 표적 모델의 마이크로도플러는 눈으로 얼마나 갈리는가?
열두 칸 모두 같은 색눈금과 같은 축이다. 세로축은 도플러(Hz), 가로축은 느린 시간(ms), 색은 그 도플러 칸에서 보이는 RCS 다. 점선은 운동학이 예측하는 날개 끝 도플러다. 윗줄이 있는 그대로의 사다리, 아랫줄이 교정 사다리다.

## 눈으로 읽히는 것 셋

1. 어느 칸에서나 가장 밝은 것은 0 Hz 가로줄 — 몸통이다. 날개 신호는 그보다 한참 어둡다. 이 낙차(호버 벌금)가 10 기체 평균 25.4 [^14]~38.7 dB [^15] 다.
2. 회전대칭 표적(구·원판)은 0 Hz 한 줄만 있고 위아래가 비어 있다. 돌려도 모양이 안 바뀌니 변조를 만들 방법이 막힌다 — 이것은 계산 결과가 아니라 기하학이다.
3. 아랫줄에서 배음 풍부도가 가장 높은 칸은 C(감싸는 상자)·D(평판)이고 E·F(진짜 CAD 쪽)에서 내려간다 — Matrice 4E 의 n_eff 는 C 12.02 [^16] · D 11.85 [^17] · F 7.94 [^18] 다. 평판이 진짜 CAD 보다 배음이 풍부한 이 역전이 이 편의 핵심 대조이고, 가림을 켜면 그 부호가 뒤집힌다([편 33 «이르다고 부르는 이유»](33_ladder-premature.ipynb) P2).

## 읽는 법 — 네 구간

- **공짜 — 해상도.** 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 -0.01 dB [^4], 템플릿 손실이 0.00 dB [^5] 움직인다. 방위에 따른 흔들림 7.99 dB [^41] 에 묻히는 크기라 실측이 가르는 범위 밖이다.
- **싼 구간 — 날개를 평판으로.** 스팬·두께·부피가 같은 평판 두 장이면 검출 단면적이 한 자릿수 dB 안에 든다. «변조가 있나 없나» 만 보는 검출기라면 충분하고, 파형을 본뜨는 검출기라면 이미 비싸다.
- **비싼 구간 — 날개를 덩어리로.** 감싸는 상자는 속을 꽉 채웠으니 검출 단면적을 과대하게 낸다. 회전반경 위의 공은 Matrice 4E 에서 레벨만 보면 «잘 맞는» 것처럼 보이지만 파형 상관이 0.224 [^23] 뿐이라 템플릿 손실이 13.00 dB [^22] 다 — **레벨이 맞았다고 신호가 맞은 것이 아니다**.
- **낭떠러지 — 회전대칭.** 원판으로 바꾸면 검출 단면적이 -54.44 dB [^2] 무너진다. 이것만은 어떤 커널 결함에도 안 흔들리는 기하학이다.

## 사다리의 순서는 삼각형 수가 아니라 보존량이 정한다

감싸는 상자(C)가 평판(D)보다 오차가 큰 것은 «형상 정보가 더 적어서» 가 아니라 부피를 안 지켜서다. 즉 «단순화의 값» 을 정하는 것은 삼각형 수가 아니라 **무엇을 보존했는가**(회전대칭 깨짐 → 스팬 → 부피 → 두께 순)다.

⭐ 한 줄로: **«모양이 있느냐 없느냐» 는 50 dB 대를 가르지만, «모양이 얼마나 정밀하냐» 는 한 자릿수 dB 안에서 논다.** 지도교수의 지적은 앞쪽이 아니라 뒤쪽을 겨눈 것이고, 뒤쪽에서는 지적이 맞는다.

## 지도교수 지적에 답한다 — 네 축

| 축 | 팔 | 기체 | 무엇을 쟀나 | 결과 |
|---|---|---|---|---|
| ① 절대 세기 | B — SBR+PO (가림 O) | Phantom 3 | 실측 앵커 대비 rms | 자유 모수 0 개 [^42] 짜리 등가부피 구가 우리 Phantom 3 메쉬를 rms 1.73 dB [^43] 이긴다 |
| ② 방위 산포 | B — SBR+PO (가림 O) | Phantom 3 | 실측 대비 오차 | 우리 Phantom 3 메쉬 +1.42 dB [^44] · 구 -5.46 dB [^45] · 상자 +4.48 dB [^46] |
| ③ 시간 변조 | P — 순수 PO (가림 X) | Mini 2 · Matrice 4E | 메쉬 − 구 간격 | 62.0 [^9]~100.8 dB [^10] 중 CAD 정밀도 단독 몫은 0.51 [^47]~4.97 dB [^48] |
| ④ 단순화의 값 | P — 순수 PO (가림 X) | Mini 2 · Matrice 4E | 교정 사다리 | 위 두 표가 그 답이다 |

①② 는 Phantom 3 를 B 팔로 잰 축이고 그 엔진은 SBR (Mitsuba/OptiX first-hit) + PO surface integral — src/rcs_sbr.rcs_sbr_batch [^49] 이며 실측 앵커는 outputs/p3_validation_v2.json (Yuan θ90 복원 실측곡선 기준) [^50] 다. ③④ 는 Mini 2 · Matrice 4E 를 P 팔로 잰 축이고 그 엔진은 pure PO on point clouds (no occlusion, no edge diffraction, scalar |Gamma|) [^51] 다.

① 에서는 지적이 맞는다. ② 는 구가 원리적으로 못 내는 축이다 — 구의 방위 산포는 «작다» 가 아니라 정확히 0 이고, 회전대칭이라 만들 방법이 막혀 있다. ③ 의 큰 간격은 «메쉬가 정밀해서» 가 아니라 «회전대칭이 아니라서» 번 것이다.

⚠ ① 의 «자유 모수 0 개» 는 **구의 부피를 논문표기 상자로 잡았을 때**의 값이다 — 그 구는 레벨오차 +0.96 dB [^52] 이고, 같은 기체에서 **메쉬 부피**로 잡은 구는 -4.47 dB [^53] 로 우리 Phantom 3 메쉬 -3.30 dB [^54] 보다 멀다. 즉 이 축의 자유 매개변수는 구의 모수가 아니라 **어느 부피에 맞출지의 선택**이고, 그 대조군 표는 [편 27 «상자·구 대조군»](27_box-sphere-control.ipynb) 에 있다.

⭐ 그래서 다시 잡아야 할 방향은 «정밀도» 가 아니라 «무엇이 도는가» 다([편 30 «사다리가 셋이었다»](30_ladder-three.ipynb)).

## 자기반증 — 이 «공짜» 가 어디까지 참인가

이 커널에서 동체는 AC 에 정확히 0 을 기여한다. 동체를 아무리 깎아도 마이크로도플러 지표가 눈금 하나 움직이지 않는 것은 형상의 성질이 아니라 **가림 없는 커널의 성질**이다.

| 검사 | Mini 2 | Matrice 4E |
|---|---|---|
| 동체까지 깎았을 때 AC 의 상대 변화 | 9.7e-15 [^13] | 5.1e-15 [^55] |
| 같은 팔에서 동체:날개 비의 이동 | -1.88 dB [^56] | +0.34 dB [^57] |

첫 줄이 기계 정밀도라는 것은 AC 가 동체를 아예 안 본다는 뜻이다. 둘째 줄이 0 에서 떨어져 있는 것은 동체 간략화가 «몸통 대 날개 비» 는 움직인다는 뜻이다 — 면적·부피를 잃기 때문이다. 가림을 켜면 날개가 동체 뒤로 사라졌다 나타나면서 동체 형상이 AC 에 들어오기 시작한다.
