# 리포트 19 — 스톡 솔버와 맞대면 «면이 많아서 에코가 커진다» 가설은 반증되고, 런타임의 96.9% 는 호스트가 쓴다

> ### 한 일
> **같은 메쉬를 스톡 경로 솔버에 그대로 넣고 경로 수·진폭·런타임을 우리 커널과 같은 카드에서 나란히 쟀다.**

### 결과
1. 삼각형 29,932 개 [^1] (mavic4pro)에서 광선 예산 2.6e+08 spp [^2] 일 때 스톡 경로는 자세당 213.8 개 [^3] 다.
2. 이미지법 정반사 경로는 36 자세 [^4] 를 통틀어 2 개 [^5] (자세 1 개 [^6])다 — 나머지 진폭은 확산 항이 낸다.
3. «면이 많아서 에코가 커진다» 가설은 스톡 기울기 0.093 dB/decade [^7] 로 REFUTED [^8] 다.
4. 평판 대조군에서 정반사 진폭은 무한거울 값 -75.37 dB [^9] 에 고정되어, 판 크기를 40 배 [^10] 로 키워도 7.4e-07 dB [^11] 안에 머문다.
5. 런타임은 같은 통계로 나란히 적는다 — S(스톡 PathSolver) 72.8 [^12]~128.0 ms [^13] (중앙값 106.3 ms [^14] · 12 설정 [^15]) 대 B(우리 per-pose) 9.2 [^16]~356.2 ms [^17] (중앙값 38.1 ms [^18]) 다. 이 대조가 통제하는 것은 하드웨어이고, 두 팔이 재는 양은 경로 해와 σ 로 갈린다 [^19].

## 같은 메쉬를 스톡에 그대로 넣으면

삼각형 29,932 개 [^1] (mavic4pro)에서 광선 예산 2.6e+08 spp [^2] 일 때 경로는 자세당 213.8 개 [^3] 다. 그중 **이미지법 정반사** 경로는 36 자세 [^4] 를 통틀어 2 개 [^5] (자세 1 개 [^6])뿐이다.

나머지 진폭은 확산 항이 낸다 — 그 항은 정규화 산란패턴을 쓰는 경험 모델이라 표면전류 적분이 아니다. 그래서 σ 를 얹을 자리가 생긴다.

## «면이 많아서 에코가 커진다» 를 반증했다

평판 대조군에서 정반사 진폭은 무한거울 값 -75.37 dB [^9] 에 고정되어, 판 크기를 40 배 [^10] 로 키워도 7.4e-07 dB [^11] 안에 머문다. 같은 축을 드론 메쉬에서 재면 스톡 기울기가 0.093 dB/decade [^7] 이고, 판정은 **REFUTED [^8]** 다.

⭐ S(스톡) 쪽에서 면 수가 값을 옮기는 자리는 **이미지법 중복 경로** 하나다 — 한 변 1 m [^20] 금속 평판을 2 [^21] → 512 삼각형 [^22] 으로 쪼개면 코히런트 전력이 9.54 dB [^23] 오르고, 늘어난 경로는 진폭·위상·지연이 같은 복사본이라 합이 20·log10(N) 을 0.0017 dB [^24] 안에서 따른다 [^25].

⭐ 이 반증이 부 1 의 결론과 같은 물건이다 — 경로 진폭은 면적·곡률·치수를 인자로 받지 않는다. 그 인자 목록은 [편 04 «필드 갱신 인자 여덟 개에 면적·곡률·치수·λ…»](04_eight-factors.ipynb) 에 있다.

## 비용은 어디에 있나

같은 카드에서 같은 통계로 나란히 쟀다 — S(스톡 PathSolver) 72.8 [^12]~128.0 ms [^13] (12 설정 [^15]) 대 B(우리 per-pose) 9.2 [^16]~356.2 ms [^17], 중앙값끼리는 106.3 [^14] 대 38.1 ms [^18] 다.

⚠ 두 팔은 다른 양을 잰다 — S 는 씬의 경로 해, B 는 표적 σ 이고 씬 내용도 갈린다. 이 대조가 통제하는 것은 하드웨어다 [^19]. 사다리는 자릿수를 놓는 것이고 같은 양을 잰 속도비는 그 뒤의 일이다 [^26].

그 비용의 96.9% [^27] 가 호스트에 있고 GPU 광선추적은 3.1% [^28] 다. PO 단계가 48.5% [^29] 이므로 디바이스로 옮길 자리는 거기다 — 전격자 재생성 추정은 0.91 h [^30] 다.
