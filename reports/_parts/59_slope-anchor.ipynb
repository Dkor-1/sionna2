# 리포트 59 — 레벨을 맞추려면 크기전이 법칙을 골라야 하므로 기울기만 받는다

> ### 한 일
> **σ = A(f)·B₁(φ,θ) 에서 A(f) 의 **기울기만** 측정에 맞추고, 레벨을 함께 맞추려면 무엇을 더 골라야 하는지와 그 대가의 크기를 원장에 적었다.**

### 결과
1. 재보정은 밴드별 스칼라 Δσ 하나씩이고, 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다.
2. 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB [^2] 다 — 절대 레벨은 우리 SBR+PO 커널(팔 B — sbr (rcs_sbr_batch, 1-bounce, penetrate=True) [^3]) 이 낸 값이다.
3. 레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB [^4] 갈린다 — 측정이 그 대가 없이 제약하는 것은 기울기뿐이다.
4. 앵커가 통제한 것 밖의 항은 4 개 [^5] 다 — 편파(UNRESOLVED) · 크기전이(UNRESOLVED) · 단일 실험실(UNRESOLVED) · 앙각 정합(PARTIAL). 규약 불확도 0.93 dB [^6] 는 원장에서 RESOLVED_EMPIRICALLY 이고, 채택 변환과 탈락 변환의 차로 앵커에 딸린다.
5. PO 오차가 1 dB 아래로 내려가려면 부품 폭이 파장의 0.729 [^7]배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

## 밴드 비교의 바닥 — 기울기 앵커 σ

σ = A(f)·B₁(φ,θ) 에서 **A(f) 의 기울기**를 Das 측정(IEEE WCL 2026 15:3731)에 맞춘다. **절대 레벨은 우리 SBR+PO 커널(팔 B — sbr (rcs_sbr_batch, 1-bounce, penetrate=True) [^3]) 이 낸 값**이고, 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB [^2] 다.

레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB [^4] 갈리므로, 측정이 그 대가 없이 제약하는 기울기만 받는다.

## 재보정은 밴드별 스칼라 하나씩이다

그래서 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다 — 앵커가 옮기는 것은 레벨이지 모양이 아니다.

앵커가 통제한 것 밖의 항은 **4 개 [^5]** 다 — 편파(UNRESOLVED) · 크기전이(UNRESOLVED) · 단일 실험실(UNRESOLVED) · 앙각 정합(PARTIAL). 개수는 원장 `uncontrolled`[^10] 의 status 로 센다.

규약 불확도 0.93 dB [^6] 는 같은 원장에서 RESOLVED_EMPIRICALLY 이고, 채택 변환과 탈락 변환의 차로 앵커에 딸려 다닌다. 앙각 정합 항의 크기는 0.73 dB [^11], 크기전이 항은 9.50 dB [^12] 다.

## PO 유효성 항이 이번에 크기를 얻었다

PO 오차가 1 dB 아래로 내려가려면 부품의 폭이 파장의 0.729 [^7]배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

그 항의 **부호는 아직 정하지 못한다** — 우리 적분이 편파(전파의 전기장이 흔들리는 방향)를 가르지 않기 때문이다([^13]). 참값은 두 편파에서 서로 다른 값을 주는데 우리 커널은 그 둘을 구분 없이 하나로 내므로, 한쪽 편파를 기준으로 보면 낮고 다른 쪽을 기준으로 보면 높다.

이 사실은 원장에 「VV 측정 대 무편파 커널」 로 기록되어 있다. 이 부는 그 개연성을 결과에 넣지 않고 순위만 든다.
