# 리포트 16 — 메쉬 외형이 실제 기체와 얼마나 맞는지를 사진 IoU·CAD 치수·실물 스캔 세 자로 쟀다

> ### 한 일
> **사진 실루엣 IoU · 제조사 CAD 치수 · 실물 유래 메쉬 σ 세 가지 검사를 각각 자기 바닥과 함께 재서 한 표에 올렸다.**

### 결과
1. 사진 21 장 [^1] 을 정합해 실루엣 IoU 를 쟀다 — 기체마다 최고 1장을 골라 세운 7 개 [^2] 값이 0.478 [^3]~0.875 [^4] (자기복제 상한 대비 50 [^5]~91% [^6]) 이고, 사진 전수로는 0.363 [^7]~0.875 [^8] 다.
2. 그 상한 자체가 0.864 [^9]~0.957 [^10] 다 — 같은 메쉬로 만든 가짜 사진을 같은 파이프라인에 넣었을 때의 값이고, 암·블레이드가 몇 px 이라 1.0 아래에 있다.
3. 제조사 CAD 어셈블리 3 종 [^11] 중 우리 기체에 대응하는 것은 1종(maps_to 열의 `x500v2`) [^12] 이고, 맞댄 치수 20 개 [^13] 는 전부 그 기체다 — 최대 편차는 CAD 대비 0.56% [^14] (치수 19 개 [^15], 설계 의도 실현 검사) · 발행 제원 대비 0.56% [^16] (치수 10 개 [^17], 독립) 다.
4. 실물 유래 메쉬와의 Δ 방위평균 σ 는 -2.0 [^18] · -0.4 [^19] · +1.8 dB [^20] 이고, 자세별 RMS 는 세 대조 전체에서 5.2 [^21]~10.0 dB [^22] 다 — 양쪽 메쉬를 전부 PEC 로 덮고 껍데기 형상만 맞댄 값이고, 방위 48 [^23]·72 [^24]·48 [^25]칸을 하나씩 빼는 잭나이프 표준오차가 1.0·0.5·1.8 dB 다. 인용은 소수 첫째 자리까지로 정한다.
5. ⭐ 이 세 자가 σ 의 **인용 단위**를 정한다 — 인용 단위는 **Δ 방위평균 σ** 하나이고 소수 첫째 자리까지다. 로브 위치와 널 깊이는 인용 단위에서 뺐다: 방위 격자가 7.5·5.0·7.5° 이고, 두 메쉬의 방위 칸을 첨자 순서로 짝짓는다. 그 둘을 다시 재는 일은 다음 단계 표에 세웠다.

## 사진과 맞댔다

사진 21 장 [^1] 을 각각 정합해 실루엣 **IoU**(두 실루엣이 겹친 넓이를 둘을 합친 넓이로 나눈 값 — 1 에 가까울수록 잘 맞는다)를 쟀다. 카메라 자세·원근·배율·위치와 로터별 프로펠러 위상을 맞춘 뒤 겹친다(`src/viz_mesh_photo.py`). 자료 규칙에 걸린 사진 12 장 [^29] 은 사유와 함께 원장에 남겼다.

표와 결과가 인용하는 IoU 범위는 **기체마다 최고 1장**을 골라 세운 7 개 [^2] 값이다. 사진 전수의 범위는 0.363 [^7]~0.875 [^8] 이고, 두 모집단을 함께 적어야 하한이 어느 쪽 값인지가 갈린다.

⚠ 2행의 지위는 원장이 그대로 적는다. x500v2 메쉬는 이 CAD 에서 뽑은 치수표로 지었다 — CAD 열과의 일치는 설계 의도가 메쉬로 실현됐는지의 검사이고, 독립 검증은 발행 제원 열과 제조사 CAD↔제원 간극(floor_pct)이다. [^26]

⚠ 3행 Typhoon 의 -0.4 dB [^19] 는 **외형을 참조에 맞춘 뒤**의 값이다 — `ours_body_only()` 가 참조 STL 의 bbox 를 우리 spec 에 주입하고 축별 배율을 한 번 더 건다. 그 자국이 원장에 남아 있다: 우리 bbox 와 참조 bbox 의 최대 상대차가 Phantom 4 17.0% · Typhoon 1e-14% · M600 56.5% 다. 정합을 끈 채로 다시 돌리는 일은 다음 단계 표에 세웠다.

⛔ 3행 M600 의 +1.8 dB [^20] 는 방위 격자와 함께 움직인다 — 같은 스윕을 한 칸씩 걸러 둘로 나누면 Δ 가 +5.6 와 -10.1 dB 로 갈린다(Phantom 4 -3.1/-1.0 · Typhoon -1.1/+0.8 dB). az 가 60° 의 배수인 칸은 격자의 12.5·8.3·12.5% 인데, σ 합에서 그 칸이 차지하는 몫은 우리 쪽 10.5·7.9·91.9% · 실물 쪽 19.4·7.9·8.9% 다.

⚠ 3행은 재질을 지운 정규화다 — 양쪽 메쉬를 전부 PEC 로 덮고 껍데기 형상만 맞댔으므로, 절 3 의 재질 가중 |Γ| 표·Σ|Γ|A 와는 다른 눈금 위에 선다.

## 이 표가 σ 의 인용 단위를 정한다

인용 단위는 **Δ 방위평균 σ** 하나이고, 소수 첫째 자리까지다. **로브**(방위를 돌릴 때 σ 가 솟는 봉우리) **위치**와 **널 깊이**(그 봉우리 사이에서 σ 가 꺼지는 골의 깊이)는 인용 단위에서 뺐다 — 아래 두 줄이 그 근거를 숫자로 적는다.

⛔ **로브 위치** — 방위 격자는 7.5·5.0·7.5° 이고, 로브 폭의 눈금 λ/2D 는 7.0·4.7·1.5° 다(D = 실물 메쉬 bbox 의 수평 최대 변, 3.5 GHz [^28]). 이웃 칸 lag-1 자기상관은 여섯 패턴 전체에서 -0.19~+0.11 이고, 최대 σ 방위는 실물 120·50·127.5° 대 우리 180·70·180° 다.

⛔ **널 깊이** — 두 메쉬의 방위 칸을 첨자 순서로 짝짓는다. 우리 쪽 방위열을 한 칸씩 돌려 가며 자세별 RMS 를 다시 내면 Phantom 4 5.0~6.9 · Typhoon 4.7~8.2 · M600 8.2~10.3 dB 사이를 오가고, 회전 0 인 우리 짝짓기는 그 회전 널에서 41/48 · 2/72 · 37/48 번째다. 요각 정합은 다음 단계 표에 있다.

IoU 에는 눈금이 붙어 있다 — 표 오른쪽 끝의 **자기복제 상한**은 같은 메쉬로 만든 가짜 사진을 같은 파이프라인에 넣었을 때 나오는 값, 즉 이 검사가 낼 수 있는 최고점이다. 그 값이 0.864 [^9]~0.957 [^10] 다(암·블레이드가 몇 px 이라 최고점이 1.0 아래다). 같은 메쉬끼리도 자세가 1° 어긋나면 0.906 [^33], 2° 면 0.822 [^44] 로 내려간다.

⭐ 그래서 이 편이 파는 것은 «메쉬가 맞다» 가 아니라 **«어느 인용 단위까지 맞다»** 다. 그 단위 안에서 σ 를 쓰는 자리가 [부 5 «앵커와 검증»](../README.md#부-5-앵커와-검증) 이고, 단위 밖의 양을 인용하지 않는다.

⚠ 사진 IoU·CAD 표는 2026-08-04 [^31] 형상 정정 **전** 메쉬의 산출이고, σ 대조 세 값은 그 메쉬 위에서 2026-08-07 10:58:22 [^45] Γ(θ) 각도 모양(기본 켬) **이전** 커널로 돌린 것이다 — 재측정은 다음 단계 표에 있다.
