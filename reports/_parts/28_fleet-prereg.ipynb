### 결과
1. 판정은 **NOT_VALIDATED (P3 산포) [^1]** 다 — 결정권을 준 항목은 레벨오차 산포가 6.0 dB 이하일 것이고, 실제 산포는 16.27 dB [^2] 다.
2. 봉인이 예측한 산포는 1.0 dB [^3] 였다 — 예측이 반증됐다.
3. ⭐ 갈린 축은 대역도 전기적 크기도 기체 크기도 아니고 **그 기체 자체의 형상 증거가 있는가** 다 — 증거가 있는 Mini 2 -0.51 [^4] · Phantom 3 -1.47 [^5] 와 얇은 Phantom 2 -7.19 [^6] · M350 RTK +9.08 dB [^7] 사이에 5.72 dB [^8] 의 빈 구간이 있다.
4. 예측이 반증됐고 일치도가 메쉬 구속도를 따라간다는 두 사실이 함께 서 있다 — 그것은 결과에 맞춰 맞춘 흔적의 정반대 서명이다 [^9].

## 봉인하고 채점했다

Das Table III 는 네 기체를 바이스태틱각 일곱 개로 준다 — 그 칸을 전부 채점했다([^10]). 합격규칙은 **계산 전에** 봉인했고(`outputs/das_fleet_prereg.json`) 그대로 썼다.

결정권을 준 항목은 「⭐ max(DL(0)) − min(DL(0)) ≤ 6.0 dB — 오차가 기체마다 흩어진 게 아니라 **공통 오프셋**일 것 [^11]」 이다.

## 판정

판정은 **NOT_VALIDATED (P3 산포) [^1]** 다 — 실제 산포가 16.27 dB [^2] 이고 봉인이 예측한 산포는 1.0 dB [^3] 였다.

| 봉인 항목 | 무엇을 요구했나 | 통과 |
|---|---|---|
| P1 | 4 기체 전부 \|DL(0)\| <= 6.0 dB [^12] | 아니오 [^13] |
| P2 | 4 중 **3 이상**이 \|DL(0)\| <= 4.0 dB [^14] | 아니오 [^15] (실제 2 기체 [^16]) |
| P3 ⭐ | ⭐ max(DL(0)) - min(DL(0)) <= 6.0 dB — 오차가 기체마다 흩어진 게 아니라 **공통 오프셋**일 것 [^11] | 아니오 [^17] |
| P4 | 4 중 3 이상이 같은 부호 [^18] | 예 [^19] |
| 기울기 | phantom3 · phantom2 둘 다 기울기차가 봉인 문턱 안 [^20] | 아니오 [^21] |
| 바이스태틱 | phantom2 의 바이스태틱 추세가 봉인 범위 안 [^22] | 예 [^23] |

네 항목 전부 통과가 요구조건이고([^24]), 판정 문자열이 이름으로 든 것은 산포(P3) 다 — 봉인이 «판정을 정하는 항목» 으로 미리 지목한 것이 그것이기 때문이다 [^25].

⚠ 용어 두 개를 푼다. 판정 문자열의 **P3 은 봉인한 합격조건 네 개 중 세 번째**(레벨오차 산포)를 가리키는 이름이고 기체 Phantom 3 를 뜻하지 않는다. 그리고 **레벨오차 DL(0)** 는 바이스태틱각 0°(송신기와 수신기가 같은 자리에 있는 배치)에서 측정 대비 우리 σ 가 몇 dB 높거나 낮은가다 — 양수면 우리가 더 밝게 냈다는 뜻이다.

## 갈린 축은 형상 증거의 유무다

⭐ 갈린 축은 대역도 전기적 크기도 기체 크기도 아니고 **그 기체 자체의 형상 증거가 있는가** 다 — 형상 증거가 있는 Mini 2 -0.51 [^4] · Phantom 3 -1.47 [^5] 와 증거가 얇은 Phantom 2 -7.19 [^6] · M350 RTK +9.08 dB [^7] 사이에 5.72 dB [^8] 의 빈 구간이 있다.

예측이 반증됐고 일치도가 메쉬 구속도를 따라간다는 두 사실이 함께 서 있고, 그것은 결과에 맞춰 맞춘 흔적의 정반대 서명이다 [^9].

⚠ 이 2:2 분할은 봉인이 계산 전에 지목한 것이고, 지목된 두 기체가 |DL(0)| 하위 두 자리를 다 차지할 확률은 무작위 배정에서 0.167 [^26] 다 — 기체가 넷뿐이라 유의수준이 아니라 **가설 생성**으로 읽는다 [^27]. 구속도와 |DL(0)| 의 상관도 같은 이유로 서술용이다 [^28].

⚠ M350 RTK 값은 계산이 다 끝나기 전의 스냅샷에서 집계됐고, Mini 2 행은 2026-08-04 [^29] 형상 정정 전 메쉬에서 나온 값이다 — 둘 다 재집계를 기다린다 [^30] · [^31].

⚠ **이 표의 σ 는 다중정적 경로(`rcs_sbr_multistatic`)로 냈고, 그 경로에 Γ(θ) 각도 모양이 들어온 것은 2026-08-10T15:13:58+00:00 [^32] 다** — 그 뒤로 복소장 경로와 σ 경로가 1.4e-14 dB [^33] 안에서 같은 값을 낸다([^34]). 이 표는 그 배선 이전 판이라 **재계산 대기**이고, 이동의 크기는 이 격자에서 아직 안 쟀다. 판정이 기댄 것은 산포 16.27 dB [^2] 와 위 표 P3 문턱 사이의 거리이고, 형상 증거 구간 5.72 dB [^8] 는 그 거리보다 좁다.
