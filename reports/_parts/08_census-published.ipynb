# 리포트 08 — 전문 판정한 게재본 중 드론 메쉬 산란을 엔진 안에서 «검증» 한 것은 0편이다

> ### 한 일
> **아카이브 문서의 전문을 추출해 관문 넷에 통과시키고, 게재본을 «광선엔진 밖에서 서명을 얻은 쪽» 과 «엔진을 직접 돌리는 쪽» 으로 갈라 표로 적었다.**

### 결과
1. 문서 21 건 [^1] (저작 19 편 [^2])을 전문 판정했다 — 게재 11 편 [^3] · 프리프린트 10 편 [^4], 축자 인용 46 건 [^5] 은 PDF 안 그 쪽에서 기계가 다시 찾는다.
2. 게재본 11 편 [^6] 중 광선엔진을 직접 돌리는 편은 4 편 [^7], 그 엔진 안 표적이 드론 메쉬인 편은 2 편 [^8], 그 메쉬에서 산란 진폭을 계산한 편은 0 편 [^9] 이다 — 프리프린트까지 합친 문서 21 건 [^10] 기준으로는 10 편 [^11] → 3 편 [^12] → 0 편 [^13] 이다.
3. 1차 사료 정산이 5 편 [^14] 을 더해 판정 문서가 21 건 [^15] 이 됐다 — 더한 것 중 Rzewuski 는 최종물이 우리와 가장 가깝고 아카이브 밖에 있었다.
4. 전문 쪽수 합계는 240 쪽 [^16] 이고, 읽고 범위 밖으로 뺀 문서가 11 건 [^17] 이다 — 뺀 사유를 원장에 함께 남겼다.

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 범위 | 아카이브 2곳 + 패시브레이더 코퍼스에서 178 편 [^18] 을 낱말 대조하고, 21 건 [^1] 을 전문 판정했다 |
| 귀속 | 판정마다 축자 인용을 PDF 그 쪽에서 기계가 찾는다 — 못 찾으면 빌드가 죽는다 (`prior_work/src/build_prior_survey.py:85`) |
| 게재상태 | PDF 메타데이터 `subject`(IEEE 조판이 박는 "게재처;연도;권;호;DOI") · 지면 스탬프 · arXiv 스탬프로 문서마다 확정했다 |
| 관문 | G1 엔진 · G2 드론 메쉬 · G3 그 메쉬에서 산란 진폭 · G4 절대 dBsm 보고 |

## 관문 넷과 그것이 세는 것

21 건 [^1] 을 전문 추출해 관문 네 개에 통과시켰다. 문서 21 건 [^1] 은 저작 19 편 [^2] 이다 — Costa 와 Ziganshin 은 회의판·저널판을 따로 세고, 그 이유는 [편 09 «프리프린트는 따로 세고, Ziganshin 은 두 판을 구별해 인용한다»](09_census-preprint.ipynb) 에 있다.

| 관문 | 통과 조건 |
|---|---|
| G1 | 이 논문이 Sionna 계열 광선엔진을 직접 돌린다 |
| G2 | 그 엔진 안의 표적이 드론 기하(메쉬)다 — 큐브·점산란체와 구분한다 |
| G3 | 그 메쉬에서 산란 **진폭**을 계산한다 |
| G4 | 결과를 절대 RCS(dBsm)로 보고한다 |

## 관문을 하나씩 걸면 몇 건이 남는가

![report01_p1_funnel](../outputs/figures/report01_p1_funnel.png)

**그림 1.** 관문을 하나씩 걸면 몇 건이 남는가?

게재본 열은 G1 4 편 [^7] → G2 2 편 [^8] → G3 0 편 [^9] 이고, 이 편의 아래 두 표가 그 열이다. 그림의 깔때기는 프리프린트를 합친 문서 21 건 [^10] 기준이라 G1 10 편 [^11] → G2 3 편 [^12] → G3 0 편 [^13] 로 읽는다 — 막대는 게재본과 프리프린트를 쌓고, 괄호 안 숫자가 게재본 몫이다.

프리프린트 열은 [편 09 «프리프린트는 따로 세고, Ziganshin 은 두 판을 구별해 인용한다»](09_census-preprint.ipynb) 가 따로 든다. ⭐ 이 깔때기가 이 편의 제목이고, 그 마지막 칸이 우리가 서는 자리다.
