# 리포트 84 — AC/DC 가 +0.89 dB 에서 −68.13 dB 로 내려간 것은 분모가 66.55 dB 이상 커진 결과다

> ### 한 일
> **스위치를 하나씩 켠 여섯 판의 경로 수·레벨·AC/DC 를 한 표에 놓고, 비가 내려간 것이 분자 쪽 사건인지 분모 쪽 사건인지를 레벨 열로 갈랐다.**

### 결과
1. 회절을 켜면 레벨이 -130.78 dB [^1] 에서 -64.23 dB [^2] 로 +66.55 dB 오르고, AC/DC 는 +0.89 dB [^3] 에서 -68.13 dB [^4] 로 -69.02 dB 내려간다.
2. 두 열의 평균 방식이 달라 정적 성분의 상승은 66.55 ~ 70.03 dB 구간으로, 변조 성분의 이동은 -2.47 ~ +1.01 dB 구간으로 묶인다 — 비를 끌어내린 것은 분모다.
3. 경로 중앙값을 11 개 [^5] 에서 2 개 [^6] 로 줄인 스위치는 굴절이고, 굴절 판의 레벨은 -132.48 dB [^7] 로 기준 자리에서 1.70 dB 안에 머문다.
4. 모서리회절 행은 스위치의 크기가 아니라 설정의 항등식이다 — 경로 11 개 [^8] · 레벨 -130.78 dB [^9] · AC/DC +0.89 dB [^10] 로 기준판의 네 값과 같다.
5. 자세 4,096 개 [^11] 로 돌린 스윕 본판의 물리 팔은 -64.23 dB [^12] · 경로 6 개 [^13] 로 진단판의 «전부 켬» 칸과 같고, 물리를 끈 행은 자세 수를 따라 0.49 dB 옮겨 앉는다.

## 비가 내려가는 길은 둘이다

AC/DC 는 **변조 전력 ÷ 정적 전력**이다. 비가 내려가는 길은 둘뿐이다 — 분자인 변조 성분이 줄거나, 분모인 정적 성분이 커지거나.

둘을 가르는 잣대는 같은 표의 **레벨 열**이다. 레벨은 자세평균 크기라서 정적 성분이 커지면 함께 오르고, 변조만 사라지면 제자리에 머문다. 그래서 «비는 내려갔는데 레벨은 올랐다» 는 곧 분모 쪽 사건이라는 뜻이다.

## 경로를 무너뜨린 것은 굴절이다

굴절을 켜면 경로 중앙값이 11 개 [^5] 에서 2 개 [^6] 로, 평균이 11.50 개 [^15] 에서 1.65 개 [^16] 로 내려간다.

굴절 스위치는 표면에 닿은 광선이 **매질 속으로 들어가는 길**을 연다. Sionna 는 교점마다 반사·확산·투과 중 하나를 |R|²·|T|² 에 비례한 확률로 뽑으므로(`sionna/rt/radio_materials/radio_material.py:764`~`776`), 투과 계수가 큰 플라스틱 면에서는 대부분의 광선이 투과 가지로 간다. `max_depth=1` 이라 그 광선은 그 자리에서 예산을 다 쓴다.

⇒ 줄어드는 것은 **발견된 사슬의 수**다. 한 번이라도 발견된 사슬은 이미지법으로 정확히 다시 풀리므로 진폭은 표본 수와 무관하고, 그래서 경로가 5.5 배 줄어도 레벨은 -132.48 dB [^7] 로 기준판 -130.78 dB [^1] 에서 1.70 dB 안에 머문다.

그 1.70 dB 는 나딧의 값이다 — 같은 스위치가 el −45° 에서는 레벨을 -125.96 [^17] 에서 -130.30 dB [^18] 로 내리고, el 0° 에서는 -59.65 [^19] 에서 -59.66 dB [^20] 다. 그 두 판은 자세 16 개 [^21] 로 돌린 표본이라 나딧 판(자세 20 개 [^22])과 표본 수가 다르다.

나딧 판의 AC/DC 는 +4.60 dB [^23] 로 올라가는데, 남은 경로 2 개 [^6] 가 자세마다 크게 흔들린 결과다.

굴절 하나는 `--physics` 의 경로 감소를 분해하지 않는다 — 넷을 다 켠 판의 중앙값은 6 개 [^24] 이고, 회절만 켠 판은 14 개 [^25] 로 오히려 늘어난다.

## 레벨을 올린 것은 회절이고, 그 크기는 판 대 판의 차다

회절만 켠 판의 레벨은 -64.23 dB [^2] 이고 넷을 다 켠 판도 -64.23 dB [^26] 다 — 귀속은 [편 83 «스위치 단일축»](83_physics-single-axis.ipynb) 가 세웠다.

⭐ 그 66.55 dB 는 «회절 항 대 정반사 항» 의 비가 아니라 **판 대 판의 레벨 차**다. 여섯 판 모두 정반사와 확산 반사를 함께 켜고 굴절을 끈 깊이 1 로 돌았으므로(`benchmark/diag_physics_paths.py:61`), 기준판의 이름은 «반사만» 이다.

이 차에는 등급이 있다 — el −45° 에서 -125.96 [^17] → -97.02 dB [^27] 로 28.94 dB, el 0° 에서 -59.65 [^19] → -59.65 dB [^28] 로 0.00 dB 다 — 두 판 다 자세 16 개 [^21] 표본이다.

⭐ 나딧은 표적 바닥이 파면과 정면으로 마주 보는 자리다. 회절을 켠 판은 el 0° 에서 나딧까지 4.58 dB 내려가는 동안 반사만 켠 판은 71.13 dB 내려간다. 그래서 이 레벨이 묻는 것은 «회절이 그만큼 세다» 가 아니라 «스톡 솔버의 반사 경로가 나딧에서 무엇을 잃고 회절 경로가 그 자리를 채우는가» 다.

## 그래서 −68 dB 는 분모 쪽 사건이다

회절 판의 AC/DC 는 -68.13 dB [^4] 라 그 판의 간극 δ 는 0.00 dB 이고, 레벨이 곧 정적 성분이다. 기준판은 AC/DC 가 +0.89 dB [^3] 라 δ 의 상한이 3.48 dB 다.

그래서 정적 성분의 상승은 66.55 ~ 70.03 dB 구간이고, 비가 -69.02 dB 내려간 것과 합치면 변조 성분의 이동은 -2.47 ~ +1.01 dB 구간이다 — 분자는 몇 dB 안에 갇혀 있고, 수십 dB 움직인 것은 분모다.

그러니 「물리를 켜니 나딧 도플러가 이론대로 0 이 됐다」는 문장은 이 분모 상승의 다른 이름이다. 회절이 정적 성분을 수십 dB 올렸고, 변조는 그 자리에 남았다.

⚠ 레벨차 66.55 dB 와 합산값 -2.47 dB 는 각각 구간의 **끝**으로 읽는다. 점값을 내려면 자세별 복소 합을 저장해 |mean E|² 를 직접 재야 하고, 그것이 다음 단계 표의 둘째 줄이다.

## 모서리회절 행은 항등식이고, 깊이 축은 둘째 자리다

모서리회절만 켠 판은 경로 11 개 [^8] · 레벨 -130.78 dB [^9] · AC/DC +0.89 dB [^10] 로 기준판과 **네 값이 모두 같다.** 그 스위치가 `diffraction` 의 하위 옵션이라 `diffraction=False` 인 판에서는 계산에 들어가지 않기 때문이다 — 세 앙각(0°·−45°·−90°) 전부에서 네 값이 똑같이 겹치는 것이 그 표시다. 모서리회절의 크기는 다음 단계 표의 판이 잰다.

다중반사(최대 깊이 3 [^29])는 이 앙각에서 레벨 -130.76 dB [^30] · AC/DC +0.88 dB [^31] 로 소수점 둘째 자리에서 갈리고, 경로 중앙값만 11 개 [^5] → 12 개 [^32] 로 하나 는다. 세 앙각을 다 보면 깊이 축이 레벨을 0.29 dB(el −45°), AC/DC 를 0.56 dB(el 0°) 안에서 움직인다[^33].

선행 스위치 연구는 깊이 축에서 갈린다 — 도시 규모 경로손실을 관측량으로 삼은 판은 `max_depth` 무감을 보고하고, 방위 추정과 라디오맵을 관측량으로 삼은 두 판은 깊이가 결과를 옮긴다고 적는다(`prior_work/sionna_solver_settings_survey.md` 의 도시 판·방위 판·라디오맵 판). 우리 소형 표적 후방산란은 앞쪽과 같은 방향이고, 갈리는 것은 불리언 스위치 쪽이다.

## 자세 수에 걸리는 칸과 걸리지 않는 칸

이 절이 여는 el −90 진단판은 자세 20 개 [^22] 로 돌린 표본이고, el −45·el 0 판은 자세 16 개 [^21] 다. 같은 스위치를 켜고 자세 4,096 개 [^11] 로 돌린 스윕 본판의 나딧 행은 레벨 -64.23 dB [^12] · 경로 6 개 [^13] 로 진단판의 «전부 켬» 칸과 **같은 값**이다. 그 판은 AC/DC 가 −68 dB 라 자세축 변동이 평균의 0.04 % 이고, 자세를 몇 개 뽑아도 자세평균 크기가 같은 자리에 온다.

물리를 끈 같은 엔진의 나딧 행은 **자세 수에 걸린다** — 본판은 레벨 -130.29 dB [^34] · 경로 12 개 [^35] 이고, 진단판의 기준 칸은 -130.78 dB [^1] · 11 개 [^5] 다. 그 판의 AC/DC 가 +0.89 dB [^3] 라 자세마다 크게 흔들리기 때문이다. 두 행 모두 자세가 하나도 비지 않은 완결 행이다[^36].

⇒ 이 절의 레벨 상승 66.55 dB 는 **el −90 진단판의 자세 20 개 기준**으로 닫아 계산한 값이다. 본판 행을 피감수로 쓰면 같은 뺄셈이 66.06 dB 가 되므로, 두 판의 행은 따로 읽는다.

⚠ 이 대조는 **PathSolver 계열 안에서만** 성립한다. 우리 커널은 정규화가 달라 레벨 축을 따로 쓴다.

## 이 판이 서는 범위

여섯 줄은 el -90° [^37] 한 점에서, 표적 `matrice4e` 하나로, 거리 10 m [^38] · 반송주파수 3.5 GHz[^39] · 광선 11,111,111 개 [^40] · 자세 20 개 [^22] 로 잰 값이다.

분자·분모 가르기는 부등식이 주는 **구간**이다. 회절 판은 AC/DC 가 -68.13 dB [^4] 라 간극이 0.00 dB 이고, 기준판은 +0.89 dB [^3] 라 간극의 상한이 3.48 dB 다. 그 폭이 -2.47 ~ +1.01 dB 라는 구간을 만든다.

같은 여섯 판을 돌린 원장은 앙각 3 점에 있다 — el −90° · el −45° · el 0° 이고, el −45·el 0 판은 자세 16 개 [^21] 표본이다. 남은 앙각으로 넓히는 일이 아래 표의 첫 줄이다.
