# 리포트 27 — 메쉬가 사는 축은 절대 크기가 아니라 각도 구조다

> ### 한 일
> **Phantom 3 를 상자·정육면체·구로 바꿔 넣고 같은 눈금에서 다시 채점해 메쉬가 값어치를 내는 축을 갈랐다.**

### 결과
1. 레벨 축에서 우리 메쉬는 상자 계열을 전부 이긴다 — 가장 가까운 정육면체가 +6.22 [^1] 인데 우리는 -3.30 dB [^2] 다.
2. ⚠ 레벨 하나만 보면 부피를 맞게 고른 구가 우리보다 낫다 — 논문표기 상자와 부피가 같은 구가 +0.96 dB [^3] · rms 1.98 dB [^4] 다.
3. ⭐ 그 구의 부피를 무엇으로 잡는가가 자유 매개변수다 — 메쉬 부피와 같은 구는 -4.47 dB [^5] 로 우리보다 나쁘다.
4. 각도 축에서는 갈린다 — Das 잣대에서 우리는 +1.42 [^6], 상자 계열은 +3.98 [^7]~+4.52 dB [^8] 다.
5. v1→v2 메쉬는 레벨오차 -4.91 [^9] → -3.30 dB [^10], 기울기 0.420 [^11] → 0.519 dB/GHz [^12] 로 움직인다.

## 무엇을 갈아끼웠나

**레벨을** 같은 눈금(Yuan θ90 복원 실측곡선 기준 밴드평균 잔차 [^13])으로 놓고 Phantom 3 를 상자·정육면체·구로 바꿔 넣어 다시 채점했다. 우리 메쉬는 **상자 계열을 전부 이긴다** — 가장 가까운 정육면체가 +6.22 [^1] 인데 우리는 -3.30 dB [^2] 다.

⚠ 그런데 **레벨 하나만 보면 부피를 맞게 고른 구가 우리보다 낫다** — 논문표기 상자와 부피가 같은 구가 +0.96 dB [^3] · 주파수 전체에 걸친 평균 오차 크기(rms) 1.98 dB [^4] 다. ⭐ 다만 **그 구의 부피를 무엇으로 잡는가가 자유 매개변수**(결과를 보고 골라 넣을 수 있는 값)다 — 메쉬 부피와 같은 구는 -4.47 dB [^5] 로 우리보다 나쁘다.

## v1 과 v2 는 다른 메쉬다

⭐ **이 편의 «우리» 는 v2 메쉬**(사진 실측으로 다시 지은 판)이고, [편 26 «Phantom 3 를 문헌값을 보지 않고 내고…»](26_blind-p3.ipynb) 표의 «우리» 는 v1 메쉬다. 같은 Yuan θ90 곡선 잣대에서 v1→v2 는 레벨오차 -4.91 [^9] → -3.30 dB [^10], 기울기 0.420 [^11] → 0.519 dB/GHz [^12] (촘촘하게 되살린 실측곡선 0.316 dB/GHz [^14])로 움직인다 — 레벨은 가까워지고 기울기는 멀어졌다.

형상은 삼각형 28,160 [^15]→28,366 [^16] · 실루엣 IoU 가 자기복제 상한 대비 77.9 [^17]→86.9% [^18] 다.

## 각도 축에서는 갈린다

구는 방위에 따라 σ 가 그대로라 방위 산포 ε(같은 기체를 여러 방위에서 볼 때 σ 가 얼마나 흩어지는지, dB)를 0.00 dB [^19] 로 낸다.

⚠ **ε 만은 잣대가 다르다** — 비교 상대가 위 레벨이 쓴 Yuan 곡선이 아니라 Das Table III 가 준 5.46 dB [^20] 다. 그 Das 잣대에서 우리는 +1.42 [^6], 상자 계열은 +3.98 [^7]~+4.52 dB [^8] 다.

**그래서 메쉬가 값어치를 내는 축은 σ 의 절대 크기가 아니라 각도에 따른 구조다** — 절대 레벨은 [편 70 «구가 σ 를 절대량으로 만들고»](70_calibration-sphere.ipynb) 의 교정구가 측정으로 앵커한다.

## 레벨에서 구가 이긴 것은 잣대가 아니라 어느 구를 넣었나가 갈랐다

⭐ 메쉬 부피로 잡은 같은 반지름의 구를 레벨까지 Das Table III 잣대로 다시 채점해도 같은 Phantom 3 에서 우리(-1.47 [^21])가 구(-2.59 dB [^22])를 이긴다. 그쪽 표는 구의 σ 를 단면적 πa² 로 굳힌 근사로, 이 표는 정확해(Mie)로 낸다 — 반지름은 같다.

즉 **두 잣대 모두에서 지는 것은 메쉬 부피로 잡은 구**이고, 레벨에서 우리를 앞선 것은 반지름을 그보다 키운 논문표기 상자부피 구 하나다. 두 결과는 «어느 부피를 골랐나» 와 함께 읽는다.
