# 리포트 86 — 우리 커널은 덱 3~40 m 판과 0.9141~0.9877 로 겹치고, 엔진이 바뀐 3 칸은 0.6291~0.6961 다

> ### 한 일
> **앙각 스윕의 el −15° 시계열을 8/11 덱이 쓴 같은 기하의 시계열과 같은 STFT 규약으로 나란히 놓고 dB 맵 상관·빗살 간격·빗살 높낮이로 대조했다.**

### 결과
1. 우리 팔은 덱 15 m 판과 0.9877 [^1], 40 m 판과 0.9774 [^2], 3 m 판과 0.9141 [^3] 로 겹친다.
2. 덱 세 판은 **같은 커널을 거리만 바꿔** 다시 돌린 것이라, 그 세 값은 우리 커널의 거리 무관성을 잰다[^4].
3. 엔진이 바뀐 칸은 3 개다 — 물리 끔 0.6961 [^5] · 물리 켬 0.6903 [^6] · 광선 250M 에 물리 켬 0.6291 [^7] 다.
4. 빗살 간격은 7 판이 모두 125.05 Hz [^8] 이고 예측 대비 오차도 -1.62 Hz [^9] 로 같다 — 그 칸의 눈금은 4.81 Hz 라 반 칸보다 작은 차이를 접는다.
5. 물리를 켠 두 팔의 el −15° 행은 자세 4,096 개 [^10] 가 다 차 있다 — `n_missing` 이 0 [^11] · 0 [^12] 다.

## 닮음을 눈이 아니라 수로 잰다

우리 팔의 dB 맵을 기준으로 나머지 6 판과의 상관계수를 잰다. 덱 15 m 판이 가장 가깝고, 광선 250M 에 물리를 켠 PathSolver 가 가장 멀다.

⭐덱 세 판은 **우리 커널을 거리만 바꿔 다시 돌린 것**이다(`benchmark/deck_ours_by_range.py`, 조명은 «위상만 구면파» [^19]). 그러니 그 세 값은 엔진 사이의 교차검증이라기보다 **한 커널이 3~40 m 에서 같은 무늬를 내는가**를 잰다. 엔진이 바뀐 칸은 아래 표의 `sionna` 계열 3 줄이다.

| 판 | 우리 팔과의 상관 |
|---|---|
| deck:R15/E | 0.9877 |
| deck:R40/E | 0.9774 |
| deck:R3/E | 0.9141 |
| sionna/el-15 | 0.6961 |
| sionna_phys/el-15 | 0.6903 |
| sionna_p250000000_phys/el-15 | 0.6291 |

출처 [^20]

⚠ 이 상관은 **패널마다 자기 최대값으로 정규화한 dB 맵**끼리 잰 값이다[^21]. 두 팔의 절대 레벨은 이 표에 들어 있지 않다 — 레벨 축은 [편 87 «예산은 물리가 아니다»](87_budget-not-physics.ipynb) 가 같은 엔진 안에서 다룬다.

**그림 1.** 우리 팔의 el −15° 무늬는 이미 발표한 덱의 판과 같은 물건인가?

위 줄은 우리 커널 · 물리를 끈 PathSolver · 물리를 켠 PathSolver(11.1M) 이고, 가운데 줄은 광선 250M 에 물리를 켠 판과 덱 3 · 15 m 판, 아래 줄이 덱 40 m 판이다. 흰 점선은 날개끝 속도가 정하는 예측 상한이다. 우리 팔과 덱 세 판은 그 선 아래에 같은 간격의 빗살을 세우고, 물리를 끈 PathSolver 는 그 위로 번진 얼룩을 얹는다. 물리를 켠 두 판은 0 Hz 띠와 자세 몇 개를 가로지르는 세로선으로 선다.

## 빗살 간격은 7 판이 한 값이다

7 판의 빗살 간격과 예측 대비 오차, 그리고 정규화와 무관한 양인 AC/DC(0 Hz 정지 성분 대비 변조 성분의 비)를 한 표에 놓는다.

| 판 | 빗살 간격 [Hz] | 예측 대비 오차 [Hz] | AC/DC [dB] (el −15° 판) |
|---|---|---|---|
| ours/el-15 | 125.05 | -1.62 | -19.63 |
| deck:R15/E | 125.05 | -1.62 | -19.67 |
| deck:R40/E | 125.05 | -1.62 | -19.72 |
| deck:R3/E | 125.05 | -1.62 | -19.54 |
| sionna/el-15 | 125.05 | -1.62 | -19.99 |
| sionna_phys/el-15 | 125.05 | -1.62 | -26.38 |
| sionna_p250000000_phys/el-15 | 125.05 | -1.62 | -30.27 |

출처 [^20]

간격 칸이 모든 판에서 한 값인 것은 **눈금이 한 칸뿐이기 때문**이다. 첨두는 주파수 격자 위에만 놓이고, PRF 19,700 Hz [^15] 를 4,096 자세로 잰 격자의 한 칸은 4.81 Hz 다. 125.05 Hz 는 그 26 칸이고, 입력한 f_flash 126.67 Hz [^17] 는 26.34 칸 자리라 26 칸으로 내려앉는다 — 예측 대비 오차 −1.62 Hz 는 커널이 낸 값이라기보다 그 내림 나머지다.

⇒ 이 칸이 가르는 것은 **반 칸(±2.4 Hz)보다 큰 박자 차이**뿐이고, 모든 판이 그 안에 들어 있다. 박자 자체는 우리가 입력한 회전수가 정하므로 이 칸은 산란 커널을 재지 않는다 — 팔을 가르는 잣대는 [편 80 «상한 위 누설»](80_el-above-tip-limit.ipynb) 가 앙각 여섯 점에서 다룬다. AC/DC 는 우리 팔·덱·물리 끔 팔이 -19.54 dB [^22] 에서 -19.99 dB [^23] 사이에 모이고, 물리를 켠 두 팔은 -26.38 dB [^24] · -30.27 dB [^25] 로 그 아래에 앉는다 — 이 값은 판 안에서의 비라 정규화가 달라도 나란히 놓을 수 있는 칸이다.

## 팔을 가르는 것은 빗살의 높낮이다

빗살 간격이 같으므로, 무늬를 가르는 것은 **몇 차 빗살이 얼마나 높은가**다. 1~8 차 빗살의 잡음바닥 대비 높이를 판마다 적는다.

| 판 | 1x | 2x | 3x | 4x | 5x | 6x | 7x | 8x |
|---|---|---|---|---|---|---|---|---|
| ours/el-15 | 38.44 | 34.15 | 24.30 | 27.95 | 37.06 | 42.26 | 42.90 | 37.43 |
| deck:R15/E | 38.17 | 34.09 | 24.25 | 26.92 | 36.23 | 41.67 | 42.42 | 37.07 |
| deck:R40/E | 37.86 | 34.11 | 24.14 | 26.54 | 35.09 | 40.89 | 41.80 | 36.56 |
| deck:R3/E | 39.84 | 33.70 | 24.41 | 31.58 | 39.80 | 44.21 | 44.31 | 37.50 |
| sionna/el-15 | 35.36 | 34.92 | 29.55 | 37.70 | 29.77 | 34.15 | 31.88 | 34.00 |
| sionna_phys/el-15 | 13.05 | 9.99 | 11.77 | 11.74 | 13.26 | 12.03 | 8.64 | 10.95 |
| sionna_p250000000_phys/el-15 | 11.18 | 12.64 | 12.99 | 7.23 | 12.64 | 7.66 | 7.64 | 8.48 |

출처 [^20]

우리 팔과 덱 판은 같은 모양의 사다리를 그린다 — 3 차에서 골이 지고 6~7 차에서 마루가 선다. 우리 팔의 3 차는 24.30 dB [^26], 7 차는 42.90 dB [^27] 이고, 덱 15 m 판의 같은 두 칸은 24.25 dB [^28] 과 42.42 dB [^29] 다.

물리를 끈 PathSolver 판에도 3 차 골은 있다 — 29.55 dB [^30] 로 그 행 여덟 칸 중 가장 낮다. 갈리는 것은 **골의 깊이와 마루의 자리**다. 그 판의 마루는 7 차(31.88 dB [^31])가 아니라 4 차(37.70 dB [^32])에 서고, 최고−최저가 8.15 dB 로 우리 팔 18.60 dB 의 절반보다 작다. 상관 0.6961 [^5] 의 정체가 이 **덜 굴곡짐**이다.

물리를 켠 두 팔은 사다리 자체가 낮다 — 여덟 칸이 8.64 [^33]~13.26 dB [^34] (물리 켬 11.1M) 과 7.23 [^35]~12.99 dB [^36] (물리 켬 250M) 안에 다 들어간다. 그 폭은 우리 팔 18.60 dB 의 3 분의 1 아래이고, 상관 0.6903 [^6] · 0.6291 [^7] 이 그 평평함의 점수다.

`line_snr_db` 는 팔마다 **자기 대역외 바닥**에 대한 비다. 바닥이 다르면 여덟 칸이 통째로 평행이동하므로, 비교되는 것은 한 팔 안의 **모양**이다.

## 이 대조에서 빠진 팔은 하나다

el −15° 의 채점표에 자리가 빈 팔은 모서리 회절을 켠 우리 커널 하나다[^37] — 그 팔의 시계열이 아직 계산되지 않았고, 그 계산이 다음 단계 표에 있다.

물리를 켠 두 팔은 el −15° 에서 자세가 다 차 있어 채점됐다 — `n_missing` 이 0 [^11] · 0 [^12] 다. 광선 250M 에 물리를 끈 팔(`sionna_p250000000/el-15`)도 완결 행이고[^38] 이 대조표에는 그 팔의 시계열이 아직 들어가 있지 않다.

완결성 문턱은 대조표를 만드는 스크립트 밖에 있다[^39] — 스크립트는 npz 에 키가 있으면 패널로 넣고, 자세 완결성은 `outputs/elevation_sweep_md.json` 의 `n_missing` 이 따로 들고 있다.

⚠ 그 점검 원장은 `gate.sionna_phys_el-15_n_missing` 을 1,536 [^40] 로 들고 있다 — 병합판이 그보다 뒤 판이고, 같은 팔의 지금 값은 위의 0 이다.

## 이 표는 명령 한 줄로 다시 나온다

같은 명령을 출력 경로만 바꿔 다시 돌려 원장과 맞대 보면, 공통 5 칸 [^41] 의 숫자 59 개 [^42] 가 최대 차이 0.000000 [^43] 로 같은 값을 낸다[^44]. 간격 · 오차 · AC/DC · 상관 · 1~8 차 빗살 높이가 전부 그 안에 든다.

재현에 걸린 시간은 35.5 초 [^45] 이고, 돌린 장치는 CPU 하나다[^46].
