# 리포트 86 — 우리 커널은 덱 3~40 m 판과 0.9141~0.9877 로 겹치고, 엔진이 바뀐 한 칸은 0.6961 이다

> ### 한 일
> **앙각 스윕의 el −15° 시계열을 8/11 덱이 쓴 같은 기하의 시계열과 같은 STFT 규약으로 나란히 놓고 dB 맵 상관·빗살 간격·빗살 높낮이로 대조했다.**

### 결과
1. 우리 팔은 덱 15 m 판과 0.9877 [^1], 40 m 판과 0.9774 [^2], 3 m 판과 0.9141 [^3] 로 겹친다.
2. 덱 세 판은 **같은 커널을 거리만 바꿔** 다시 돌린 것이라, 그 세 값은 우리 커널의 거리 무관성을 잰다[^4].
3. 물리를 끈 PathSolver 는 같은 잣대에서 0.6961 [^5] 다 — 엔진이 바뀐 유일한 칸이고, 광선 한 예산에서만 채점됐다.
4. 빗살 간격은 다섯 판이 모두 125.05 Hz [^6] 이고 예측 대비 오차도 -1.62 Hz [^7] 로 같다 — 그 칸의 눈금은 4.81 Hz 라 반 칸보다 작은 차이를 접는다.
5. 물리를 켠 팔의 el −15° 는 자세 1,536 [^8] / 4,096 [^9] 이 비어 있어 이 대조에서 뺀다.

## 덱과 다른 것은 거리 하나다

이 대조가 성립하는 이유는 두 자료가 **같은 물건을 같은 설정으로** 낸 것이기 때문이다[^10]. 아래 일곱 줄 중 여섯 줄이 같고, 다른 것은 거리 한 줄이다.

| 무엇 | 8/11 덱 | 앙각 스윕 |
|---|---|---|
| 표적 | `matrice4e` | `matrice4e` |
| 방위 · 앙각 | az 0° · el −15° | az 0° · el −15° |
| 반송주파수 | 3.5 GHz[^11] | 같음 |
| PRF | 19,700 Hz [^12] | 같음 |
| 자세 수 | 4,096 [^13] | 같음 |
| 플래시 예측 | 126.67 Hz [^14] | 같음 |
| 거리 | 3 · 15 · 40 m | 10 m [^15] |

## 닮음을 눈이 아니라 수로 잰다

우리 팔의 dB 맵을 기준으로 나머지 판과의 상관계수를 잰다. 덱 15 m 판이 가장 가깝고, 물리를 끈 PathSolver 가 가장 멀다.

⭐덱 세 판은 **우리 커널을 거리만 바꿔 다시 돌린 것**이다(`benchmark/deck_ours_by_range.py`, 조명은 «위상만 구면파» [^16]). 그러니 위 세 값은 엔진 사이의 교차검증이라기보다 **한 커널이 3~40 m 에서 같은 무늬를 내는가**를 잰다. 엔진이 바뀐 칸은 마지막 한 줄뿐이다.

| 판 | 우리 팔과의 상관 |
|---|---|
| deck:R15/E | 0.9877 |
| deck:R40/E | 0.9774 |
| deck:R3/E | 0.9141 |
| sionna/el-15 | 0.6961 |

출처 [^17]

⚠ 이 상관은 **패널마다 자기 최대값으로 정규화한 dB 맵**끼리 잰 값이다[^18]. 두 팔의 절대 레벨은 이 표에 들어 있지 않다 — 레벨 축은 [편 87 «예산은 물리가 아니다»](87_budget-not-physics.ipynb) 가 같은 엔진 안에서 다룬다.

![physics vs deck](../outputs/figures/physics_vs_deck_el-15.png)

**그림 1.** 우리 팔의 el −15° 무늬는 이미 발표한 덱의 판과 같은 물건인가?

왼쪽 위가 우리 커널, 가운데 위가 물리를 끈 PathSolver, 나머지 셋이 덱의 3 · 15 · 40 m 판이다. 흰 점선은 날개끝 속도가 정하는 예측 상한이다. 다섯 판 모두 그 선 아래에 같은 간격의 빗살이 서 있고, PathSolver 판에는 그 선 위로 번진 얼룩이 얹혀 있다.

## 빗살 간격은 다섯 판이 한 값이다

다섯 판의 빗살 간격과 예측 대비 오차, 그리고 정규화와 무관한 양인 AC/DC(0 Hz 정지 성분 대비 변조 성분의 비)를 한 표에 놓는다.

| 판 | 빗살 간격 [Hz] | 예측 대비 오차 [Hz] | AC/DC [dB] (el −15° 판) |
|---|---|---|---|
| ours/el-15 | 125.05 | -1.62 | -19.63 |
| sionna/el-15 | 125.05 | -1.62 | -19.99 |
| deck:R3/E | 125.05 | -1.62 | -19.54 |
| deck:R15/E | 125.05 | -1.62 | -19.67 |
| deck:R40/E | 125.05 | -1.62 | -19.72 |

출처 [^17]

간격 칸이 다섯 판에서 한 값인 것은 **눈금이 한 칸뿐이기 때문**이다. 첨두는 주파수 격자 위에만 놓이고, PRF 19,700 Hz [^12] 를 4,096 자세로 잰 격자의 한 칸은 4.81 Hz 다. 125.05 Hz 는 그 26 칸이고, 입력한 f_flash 126.67 Hz [^14] 는 26.34 칸 자리라 26 칸으로 내려앉는다 — 예측 대비 오차 −1.62 Hz 는 커널이 낸 값이라기보다 그 내림 나머지다.

⇒ 이 칸이 가르는 것은 **반 칸(±2.4 Hz)보다 큰 박자 차이**뿐이고, 다섯 판은 그 안에 들어 있다. 박자 자체는 우리가 입력한 회전수가 정하므로 이 칸은 산란 커널을 재지 않는다 — 팔을 가르는 잣대는 [편 80 «상한 위 누설»](80_el-above-tip-limit.ipynb) 가 앙각 여섯 점에서 다룬다. AC/DC 도 다섯 판이 -19.54 dB [^19] 에서 -19.99 dB [^20] 사이에 모인다 — 이 값은 판 안에서의 비라 정규화가 달라도 나란히 놓을 수 있는 칸이다.

## 팔을 가르는 것은 빗살의 높낮이다

빗살 간격이 같으므로, 무늬를 가르는 것은 **몇 차 빗살이 얼마나 높은가**다. 1~8 차 빗살의 잡음바닥 대비 높이를 판마다 적는다.

| 판 | 1x | 2x | 3x | 4x | 5x | 6x | 7x | 8x |
|---|---|---|---|---|---|---|---|---|
| ours/el-15 | 38.44 | 34.15 | 24.30 | 27.95 | 37.06 | 42.26 | 42.90 | 37.43 |
| deck:R15/E | 38.17 | 34.09 | 24.25 | 26.92 | 36.23 | 41.67 | 42.42 | 37.07 |
| deck:R40/E | 37.86 | 34.11 | 24.14 | 26.54 | 35.09 | 40.89 | 41.80 | 36.56 |
| deck:R3/E | 39.84 | 33.70 | 24.41 | 31.58 | 39.80 | 44.21 | 44.31 | 37.50 |
| sionna/el-15 | 35.36 | 34.92 | 29.55 | 37.70 | 29.77 | 34.15 | 31.88 | 34.00 |

출처 [^17]

우리 팔과 덱 판은 같은 모양의 사다리를 그린다 — 3 차에서 골이 지고 6~7 차에서 마루가 선다. 우리 팔의 3 차는 24.30 dB [^21], 7 차는 42.90 dB [^22] 이고, 덱 15 m 판의 같은 두 칸은 24.25 dB [^23] 과 42.42 dB [^24] 다.

PathSolver 판에도 3 차 골은 있다 — 29.55 dB [^25] 로 그 행 여덟 칸 중 가장 낮다. 갈리는 것은 **골의 깊이와 마루의 자리**다. 그 판의 마루는 7 차(31.88 dB [^26])가 아니라 4 차(37.70 dB [^27])에 서고, 최고−최저가 8.15 dB 로 우리 팔 18.60 dB 의 절반이 안 된다. 상관 0.6961 [^5] 의 정체가 이 **덜 굴곡짐**이다.

이 표를 읽는 두 가지 주의.

1. `line_snr_db` 는 팔마다 **자기 대역외 바닥**에 대한 비다. 바닥이 다르면 여덟 칸이 통째로 평행이동하므로, 비교되는 것은 한 팔 안의 **모양**이다.
2. 이 평평함은 광선 한 예산의 성질이다. 같은 기체·같은 앙각·같은 거리에서 예산을 250 M 으로 올린 팔은 완결 행으로 원장에 있고[^28] 이 대조표에는 아직 들어가 있지 않다 — 그 팔을 넣는 일이 다음 단계 표에 있다.

## 물리를 켠 팔이 이 대조에서 빠지는 이유

el −15° 에서 빠져 있는 팔은 둘이다 — 모서리 회절을 켠 우리 커널[^29] 과 물리를 켠 PathSolver[^30] 이다.

물리 팔의 el −15° 행은 자세 1,536 [^31] 개가 비어 있다 — 부분 병합이라 시계열에 0 이 박혀 있고, 그 0 이 STFT 를 가로질러 상관을 끌어내린다. 그래서 이 절은 그 팔을 판정에서 뺀다.

완결성 문턱은 대조표를 만드는 스크립트 밖에 있다[^32] — 스크립트는 npz 에 키가 있으면 패널로 넣고, 자세 완결성은 `outputs/elevation_sweep_md.json` 의 `n_missing` 이 따로 들고 있다.

## 이 표는 명령 한 줄로 다시 나온다

같은 명령을 출력 경로만 바꿔 다시 돌려 원장과 맞대 보면, 공통 5 칸 [^33] 의 숫자 59 개 [^34] 가 최대 차이 0.000000 [^35] 로 같은 값을 낸다[^36]. 간격 · 오차 · AC/DC · 상관 · 1~8 차 빗살 높이가 전부 그 안에 든다.

재현에 걸린 시간은 35.5 초 [^37] 이고, 돌린 장치는 CPU 하나다[^38].
