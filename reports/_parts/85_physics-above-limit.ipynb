# 리포트 85 — 광선을 250M 으로 맞추면 물리 스위치가 상한 위 몫에 남기는 것은 앙각 0° 에서 0.00084 이고, 기울인 세 앙각에서 0.55~0.77 다

> ### 한 일
> **물리 스위치를 켠 팔의 도플러 에너지를 날개끝 상한 위까지 넓혀 재고, 광선 예산을 맞춘 짝에서 스위치 축과 예산 축을 앙각 네 점에서 갈랐다.**

### 결과
1. 규칙 예산 11.1M · 앙각 0° 에서 상한 위 몫은 물리 끔 0.290 [^1] · 물리 켬 0.831 [^2] 다.
2. 그 자리에서는 예산 축 혼자서도 같은 상승을 만든다 — 물리를 끈 채 광선만 22.5 배 올린 팔이 0.870 [^3] 다.
3. 광선을 250M 으로 맞춘 짝에서 앙각 0° 의 뺄셈은 0.00084 이고, 그 자리는 두 팔이 이미 포화점에 닿아 스위치가 더할 자리가 없는 앙각이다.
4. 기울인 세 앙각에서는 같은 짝의 스위치가 몫을 0.55~0.77 올린다 — 예산을 맞춰도 스위치가 상한 위 누설을 만든다.
5. 앙각 0° 의 자세당 경로 중앙값은 250M 에서 127 [^4] → 33 [^5] 로 줄고, 우리 커널의 같은 칸은 0.02285 [^6] 다.

## 상한 위 몫이 무엇을 재는가

날개끝 속도가 정하는 도플러 상한은 f_tip = 1272.9 Hz [^9] × cos(앙각) 이다. 시선 방향으로 가장 빠른 점이 날개끝이므로 그보다 높은 주파수는 블레이드가 닿는 자리 밖이고, 거기 담긴 전력은 전부 엔진이 만든 인공물이다.

관찰 상한은 나이퀴스트 9850 Hz [^10] 라, 앙각 0° 에서는 f_tip 위로 7.7 배(= 9850 ÷ 1272.9) 폭의 자리를 들여다본다. 그 자리에 무엇이 얼마나 앉는지가 이 절의 잣대다.

몫은 팔마다 자기 전체 전력으로 나눈 값이라 눈금과 무관하다. 팔 사이 레벨 비교는 같은 엔진 안에서만 하고, 그 자리는 [편 87 «예산 사다리»](87_budget-not-physics.ipynb) 다.

## 앙각 0° 에서 스위치 축과 예산 축을 따로 움직인다

| 팔 | 광선 예산 | 물리 스위치 | 경로/자세(중앙값) | 상한 위 몫 |
|---|---|---|---|---|
| 우리 커널 (SBR+PO) | — | — | — | 0.02285 [^6] |
| PathSolver | 11.1M | 끔 | 9 개 [^11] | 0.29027 [^1] |
| PathSolver | 11.1M | 켬 | 5 개 [^12] | 0.83078 [^2] |
| PathSolver | 250M | 끔 | 127 개 [^4] | 0.87023 [^3] |
| PathSolver | 250M | 켬 | 33 개 [^5] | 0.87107 [^13] |

규칙 예산에서 스위치를 켜면 몫이 0.290 [^1] 에서 0.831 [^2] 로 올라간다. 그 상승은 예산 축 혼자서도 나온다 — 물리를 끈 채 광선만 22.5 배 올린 팔이 0.870 [^3] 다.

두 축을 가르는 자리는 광선을 맞춘 짝이다. 250M 에서 물리 끔 0.87023 대 물리 켬 0.87107 이고, 뺄셈이 0.00084 다.

앙각 0° 가 그렇게 되는 것은 그 자리가 이 잣대의 **포화점**이기 때문이다 — 상한 위 대역이 관찰 대역의 87.08% 를 덮으므로, 시간 구조가 남지 않은 시계열이 받는 점수가 그 값이다. 두 팔 다 이미 0.87 에 앉아 있어 스위치가 더할 자리가 남아 있다.

## 같은 광선 수는 팔마다 다른 경로 수로 갈린다

물리를 켜면 광선 하나가 굴절 · 회절 · 모서리회절 · 3 회 반사로 갈라진다. 앙각 0° 에서 자세당 경로 중앙값은 11.1M 에서 9 [^11] → 5 [^12], 250M 에서 127 [^4] → 33 [^5] 로 줄어든다.

부호는 앙각이 정한다 — 같은 11.1M 짝이 앙각 −30° 에서는 6 [^21] → 7 [^22] 로 오히려 는다. 굶기는 스위치는 굴절 하나이고, 회절만 켜면 나딧에서 경로가 늘어난다[^23].

이 열이 세는 것은 **자세당 되돌아온 경로의 개수**이고 표면 표본 밀도는 이 원장 밖이다. 그리고 굶음은 같은 예산 짝 안의 상대량이다 — 250M 물리 팔의 33 [^5] 개는 규칙 예산 물리 끔 팔의 9 [^11] 개보다 3.67 배 많다.

그래서 11.1M 짝의 0.290 [^1] 대 0.831 [^2] 는 스위치와 경로 수가 함께 움직인 값이고, 같은 예산에서 스위치만 남긴 크기는 위 표의 뺄셈 열이다.

## 비용 — 같은 예산 짝에서 벽시계가 2 배다

| 짝 | 물리 끔 | 물리 켬 | 배율 |
|---|---|---|---|
| 앙각 0° · 11.1M | 1549.9 s [^24] | 3361.5 s [^25] | 2.17 배 |
| 앙각 −90° · 11.1M | 1818.0 s [^26] | 3621.6 s [^27] | 1.99 배 |
| 앙각 0° · 250M | 1221.3 s [^28] | 1530.0 s [^29] | 1.25 배 |

«배율» 은 켬 ÷ 끔 이다. 초를 4096 자세 [^8] 로 나누면 자세당 0.38 → 0.82 초(앙각 0°), 0.44 → 0.88 초(앙각 −90°) 다.

이 초는 여러 실행이 같은 GPU 를 나눠 쓴 조건에서 잰 벽시계다. 광선을 22.5 배 더 쏜 팔의 벽시계가 오히려 짧은 칸이 있고(1221.3 ÷ 1549.9 = 0.79 배), 같은 스위치의 배율도 2.17 · 1.25 로 갈린다. 비용 축은 카드 한 장을 비워 놓고 다시 재야 확정된다 — 이 절이 비용에 대해 세우는 주장은 «같은 예산 짝에서 켜면 더 걸린다» 까지다.

## 병합판 원장은 이 표보다 뒤에 서 있다

공정 예산 팔(250M + 물리)의 행은 병합판 `outputs/elevation_sweep_md.json` 밖에 있다. 샤드는 앙각 0° 에서 8/8 로 서 있고, 이 절은 그 샤드를 병합기와 같은 규칙(`idx` 로 제자리 꽂기)으로 복원해 읽었다. 스냅숏 시각은 2026-08-12T04:47 [^35] 다.

자기시험이 그 복원을 보증한다 — 이미 병합된 19 [^31] 칸을 같은 코드로 다시 내 차이 0 [^32] 이고, 행 23 [^33] 개의 초 · 경로 · 레벨도 전부 같다.

### 그 게이트가 덮는 범위

| 칸 | 무엇이 대조됐나 |
|---|---|
| 칸 19 개 | 우리 커널 · 11.1M · 250M 이 각각 앙각 여섯 점, 그리고 물리 팔의 앙각 0° 한 칸 |
| 앙각 −90° | f_tip = 0 이라 «상한 위 몫» 이 정의되지 않아 어느 팔도 채점에서 빠진다 |
| 물리 경로 | 병합판에 채점 가능한 물리 칸이 앙각 0° 하나뿐이라, 물리 축의 대조는 그 한 점에서 성립한다 |
| «최대 차이 0.0» | 양쪽 값이 5 자리로 반올림된 상태의 일치다(허용치 1e−5 = 반올림 눈금) |
| 행 23 개 | 병합판 28 행에서 샤드가 없는 3 행과 부분 조립이라 미완결 수가 달라진 2 행(`stale_merge_rows`)을 뺀 나머지 |
| 경로 중앙값 | 우리 커널 7 행은 양쪽 다 값이 비어, 실제로는 Sionna 계열 16 행에서 비교된다 |

⇒ 이 게이트가 자격을 주는 범위는 **샤드 → 시계열 복원**이다. 대역 규약은 그림 스크립트와 같은 코드라 같은 오류를 같이 낸다. 원장 한 판으로 좁히는 일은 `--merge` 한 번이고, 그 명령은 기존 원장을 덮어쓴다.
