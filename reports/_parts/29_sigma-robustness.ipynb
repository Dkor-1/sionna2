# 리포트 29 — 공통모드 σ 오차는 파형 순위를 안 건드리고, 차분 오차가 뒤집는다

> ### 한 일
> **σ 오차를 공통모드와 차분 두 갈래로 나눠 검출거리 R90 순위에 넣고 각각이 순위를 언제 뒤집는지 쟀다.**

### 결과
1. 기체 5 [^1] × 밴드 3 [^2] = 15 셀 [^3] 에서 쟀다.
2. 공통모드에서 순위는 -10 [^4]~+10 dB [^5] 전 구간 15 셀 [^3] 에서 그대로다 — 움직이는 것은 절대거리뿐이다.
3. 그 기울기가 σ 1 dB 당 0.246 dB [^6] (0.223 [^7]~0.256 [^8]) 다 — 거리의 dB 는 10log10 규약이라 R90 ∝ σ^(1/4) 이고, σ ±10 dB 에서 거리 -43 [^9] %~+76% [^10] 다.
4. 자세평균 σ 로 인용하면 다섯 기체가 한 순위(LTE > 5G > WiFi [^11])에 합의하고, 단일자세에서는 순위 3 종 [^12] 이 나온다.
5. 거기에 측정 기울기를 얹으면 최악 뒤집힘 문턱이 0.09 [^13] → 3.72 dB [^14] 로 올라간다(+3.63 dB [^15]).

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 두 갈래 | **공통모드**는 한 기체의 세 밴드를 같은 dB 로 옮기고(절대 레벨 오차의 모양), **차분**은 밴드마다 다르게 옮긴다(기울기 오차의 모양) |
| 무엇에 넣나 | 검출거리 R90 — 검출확률 90 % 문턱을 마지막으로 넘는 거리다 (`benchmark/sigma_sensitivity.py`) |
| 무엇을 보나 | 절대거리가 아니라 **세 파형의 순위**가 언제 뒤집히는지를 본다 |
| 왜 이 축인가 | 차분오차가 순위를 정하는 축이고, 그래서 앵커가 잡는 축이 정확히 이것이다 |

## 오차를 두 갈래로 나눈다

σ 오차를 두 갈래로 나눠 검출거리 R90(검출확률 90 % 문턱을 마지막으로 넘는 거리) 순위에 넣었다(`benchmark/sigma_sensitivity.py`, 기체 5 [^1] × 밴드 3 [^2] = 15 셀 [^3]).

**공통모드**는 한 기체의 세 밴드를 같은 dB 로 옮긴다 — 절대 레벨 오차의 모양이다. **차분**은 밴드마다 다르게 옮긴다 — 기울기 오차의 모양이다.

## 공통모드 — 절대거리만 움직인다

공통모드에서 순위는 -10 [^4]~+10 dB [^5] 전 구간 15 셀 [^3] 에서 그대로다 — 움직이는 것은 절대거리뿐이고 그 기울기가 σ 1 dB 당 0.246 dB [^6] (0.223 [^7]~0.256 [^8]), 즉 σ ±10 dB 에서 거리 -43 [^9] %~+76% [^10] 다.

⚠ **거리의 dB 는 10log10 규약이다** — `benchmark/sigma_sensitivity.py` 가 10·log10(R90) 을 σ 오프셋에 적합해 이 기울기를 낸다. 그래서 σ 1 dB 당 약 1/4 dB 는 R90 ∝ σ^(1/4) 와 같은 말이고, 위 ±10 dB 의 거리 변화율이 그 지수와 맞는다.

**논문이 절대 σ 에 기대는 곳은 여기서 끝난다.**

## 차분 — 순위를 정하는 축

차분오차가 순위를 정하는 축이고, 그래서 [편 24 «σ = A(f)·B₁·B₂ 에서 A(f) 의…»](24_anchor-mode.ipynb) 의 앵커가 잡는 축이 정확히 이것이다.

자세평균 σ 로 인용하면 다섯 기체가 한 순위(LTE > 5G > WiFi [^11])에 합의하고(단일자세에서는 순위 3 종 [^12]), 거기에 측정 기울기를 얹으면 최악 뒤집힘 문턱이 0.09 [^13] → 3.72 dB [^14] 로 올라간다(+3.63 dB [^15]).

## 뒤집힘 문턱은 어느 열을 따라가나

뒤집힘 문턱은 최대 치수 열과 밴드간 σ 산포 열 **둘 다에 약하게** 걸린다 — 단일자세 문턱과의 상관이 크기 쪽 -0.62 [^17] · 산포 쪽 -0.31 [^18] 로, 크기 쪽이 더 강하다. 두 열 사이의 상관은 0.09 [^19] 라서 크기 열과 산포 열은 이 표에서 서로 다른 축이다.

⚠ 기체가 5 대 [^1] 뿐이라 이 세 수는 유의수준이 아니라 **서술용**이다. 원장의 같은 칸에 붙은 산문은 산포 쪽을 단독 원인으로 든다 [^20] — 본문은 위 세 상관계수를 그대로 읽는다.

⚠ Matrice 4E 행은 2026-08-04 [^21] 형상 정정 전 값이고, 표 전체가 2026-08-07 10:58:22 [^22] Γ(θ) 각도 모양(기본 켬) 이전 커널의 산출이기도 하다 — 전체 드론 σ 이동 +0.08 [^23] ~ +0.10 dB [^24] 다 — 이 표를 낸 `outputs/sigma_sensitivity.json` 가 그 기체의 형상을 먹는다 [^25]. 재생성 시 0.1 dB 급 문턱 값은 그 이동만으로도 움직이고, 위 상관 세 수도 함께 다시 읽는다.
