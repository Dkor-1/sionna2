### 결과

## 상한은 날개끝 속도가 긋는다

회전자 날개끝이 시선 방향으로 내는 속도가 **순시 도플러**의 상한을 정한다. 그 상한이 f_tip 이고, 앙각이 내려가면 날개 속도가 시선에 수직해지면서 cos(el) 로 줄어든다 — el 0° 에서 1272.9 Hz [^15], 직하방에서 0.0 Hz [^16] 다.

관찰 상한은 그와 무관하게 표본율이 정한다 — PRF 19700 Hz [^17] 의 절반인 나이퀴스트 9850 Hz [^2] 다.

⇒ el 0° 에서 순시 도플러가 닿는 폭 밖에 남는 자리는 관찰 대역의 87.08% [^3] 이고, 직하방에서는 100.00% [^18] 다. 원장은 그 자리를 인공물로 적는다[^19].

그 «인공물» 을 0 이 참값인 자리로 읽는 것은 이 잣대가 서는 범위 **밖**이다. f_tip 은 순시 주파수의 상한이지 스펙트럼이 끊기는 자리가 아니다 — 회전 날개의 위상변조 측대역은 그 위로 이어진다. 그래서 이상적인 날개도 상한 위에 작은 양수를 남긴다.

상한선의 정밀도도 여기까지다. f_tip 은 호버 회전수 한 값으로 긋는데 이 스윕의 네 로터는 «덱과 같은 결정론 패턴(OU 프리셋 아님) — 축을 하나만 바꾼다 [^20]» 로 돌아, 가장 빠른 로터의 순시 상한은 el 0° 에서 f_tip 보다 몇 Hz 위에 있다.

⇒ 이 절은 상한 위 몫을 **같은 앙각에서 팔끼리 견주는 상대 잣대**로 쓴다. 바닥이 0 인 절대 눈금으로 올리려면 진폭변조를 포함한 참조 신호로 물리 꼬리를 먼저 빼야 하고, 그 빼기는 다음 단계 표의 첫 줄이다.

## 이 잣대가 대역 안 에너지와 다른 점

대역 안(`500 Hz` 부터 f_tip 까지) 몫은 el 0° 에서 우리 -0.61 dB [^42], 물리 끔 -10.66 dB [^43], 물리 켬 -10.88 dB [^44] 로 갈린다. 셋 중 무엇이 참인지는 표적의 산란과 조명 기하가 정하고, 그 참값은 이 원장 밖에 있다.

상한 위의 참값도 같은 것이 정한다 — 날개 위 진폭 분포가 위상변조 꼬리의 크기를 정하기 때문이다. 두 잣대의 참값은 같은 이유로 원장 밖에 있다.

상한 위가 더 강한 것은 **그 참값이 작다고 알려져 있다**는 데 있다. 대역 안 몫은 0.2~0.9 사이 어디든 될 수 있지만, 상한 위 몫은 이상적 날개에서 한 자릿수 퍼센트 아래로 묶인다. 그래서 이 잣대는 큰 값을 인공물로 읽는 데 쓰고, 작은 값들 사이의 순위는 이 절의 범위 밖이다.

### 이 잣대가 서는 범위

| 조건 | 이 절이 재는 자리 |
|---|---|
| 관찰 상한 | 9850 Hz [^2] 아래만 읽는다 — 그 위 성분은 접혀 들어와 대역 안에 더해진다 |
| 표적의 운동 | 기체 고정 · 회전자만 회전. 기체가 움직이면 상한이 동체 도플러만큼 옮겨간다 |
| 앙각 | f_tip > 0 인 여섯 앙각. 직하방은 f_tip = 0 이라 몫이 정의되지 않는다 |
| 포화점 | 평평한 스펙트럼이 받는 점수 — el 0° 에서 87.08% [^3], el −75° 에서 96.65% [^45]. 그 값에 닿은 팔들 사이의 순위는 읽지 않는다 |
| 비교 축 | 같은 앙각에서 팔끼리 몫을 견준다. 앙각을 가로지르는 비교는 분모가 달라 이 표 밖이다 |
| 격자 | ⛔우리 팔의 값은 **λ/12 한 판**이고, 원장이 이 인용에 λ/12 한정 [^46] 꼬리표를 강제해 뒀다. 같은 팔·같은 자세 8192 개 [^47] 로 격자만 λ/24 로 조이면 el −15° 의 17.18% [^5] 가 4.63% [^7] 로 내려간다. 이 잣대의 격자 산포 밴드는 12.55 [^48] %p 이고, 두 팔의 차이가 그 밴드 안이면 판정 불가다 |
| 광선 예산 | ⭐물리를 끈 10 m 사다리에서 광선을 360 배 부어도 상한 위 바닥은 el −15 에서 -1.68 [^49] · el −45 에서 +0.33 dB [^50] 로 제자리다 — 광선을 더 부어도 이 몫은 그 자리에 남는다. ⚠물리를 켠 두 계단(22.5 배·깊이 3)은 같은 자리가 el −15 에서 +9.25 dB [^51] 올라간다 |
| 합격선 | 미정 — 임계는 다음 단계 표가 정한다 |

## λ/12 판에서 우리 팔이 물리 끔 PathSolver 보다 높은 자리는 세 곳이다

el −15° 는 17.18% [^5] 대 6.53% [^28], el −30° 는 10.62% [^31] 대 1.06% [^32], el −45° 는 1.81% [^35] 대 0.94% [^10] 다. 나머지 세 자리(0°·−60°·−75°)에서는 우리 팔이 아래에 있다.

⛔같은 커널·같은 자세 8192 개 [^47] 로 격자만 λ/24 로 조이면 el −15° 는 4.63% [^7], el −45° 는 0.27% [^9] 로 내려가 두 자리 다 물리 끔 **아래로** 간다 — 원장의 사전등록 판정이 그 두 칸을 ORDER_FLIP [^52] · ORDER_FLIP [^53] 로 찍는다. el −30° 는 1.36% [^8] 로 순서가 남고, 판정은 PRESERVED_BUT_WITHIN_BAND [^54] 다.

λ/12 판의 여유는 el −15° 10.65 [^55] · el −30° 9.56 [^56] · el −45° 0.88 [^57] %p 이고, 이 잣대의 격자 산포 밴드는 12.55 [^48] %p 다 — 셋 다 밴드 안이다.

그 밴드를 어떻게 읽는지는 원장이 적어 뒀다 — ⚠밴드는 바로 이 네 앙각의 |λ/24 − λ/12| 최대값이라, «뒤집힌 칸이 밴드 안» 이라는 진술은 부분적으로 동어반복이다. 그래서 판정의 무게는 «밴드 안/밖» 이 아니라 **순서가 실제로 뒤집혔다**는 사실에 둔다. [^58]

⛔세 자리가 높은 기전은 이 절의 범위 밖이다. 후보는 λ/12 격자의 표본화와 자세 시계열 창의 누설 둘이고, 둘을 가르는 실험은 다음 단계 표 넷째 줄이다. 거리 곡률은 후보에서 빠진다: 이 스윕은 원거리장 경계 밖에서 쟀다([편 78 «스윕 규약»](78_el-sweep-design.ipynb)).

⚠λ/8 → λ/32 사다리는 **다른 추정기**의 판이라 방향만 읽는다 — 같은 종류의 대역밖 몫이 1.37% [^59] 에서 0.085% [^60] 로 내려간다(얼린 격자 팔, 블레이드 대역을 분모로 쓴다).

⇒ 이 절이 확정한 것은 **λ/12 판 여섯 칸의 값**까지다. 우리 팔과 물리 끔 PathSolver 사이의 순서는 격자를 조이면 el −15°·−45° 에서 뒤집히므로 **판정 불가**로 남는다.

## ⭐물리 스위치를 켜면 여섯 앙각이 전부 상한 위로 간다

굴절·회절·모서리회절을 켠 팔은 같은 광선 예산·같은 자리에서 잰다. 갈리는 축은 스위치 하나뿐이다.

| 앙각 | 물리 끔 | 물리 켬 (깊이 1) | 물리 켬 (깊이 2) |
|---|---|---|---|
| +0° | 86.22% [^11] | 86.41% [^14] | 86.64% [^61] |
| -15° | 6.53% [^28] | 87.76% [^29] | 87.73% [^62] |
| -30° | 1.06% [^32] | 85.79% [^33] | 86.00% [^63] |
| -45° | 0.94% [^10] | 78.09% [^12] | 78.43% [^64] |
| -60° | 3.31% [^38] | 86.72% [^39] | 86.67% [^65] |
| -75° | 0.81% [^41] | 96.17% [^13] | 96.15% [^66] |

물리를 끈 팔은 el 0° 한 자리에서만 86.22% [^11] 로 뛰고 나머지 다섯 자리는 0.94% [^10] ~ 3.31% [^38] 사이다. 물리를 켠 팔은 여섯 자리가 전부 78.09% [^12] 위이고 el −75° 에서 96.17% [^13] 다.

⇒ 물리를 켜면 **움직이는 에너지의 대부분이 날개가 낼 수 없는 대역**에 놓인다. 그 자리의 참값은 0 에 가깝다고 알려져 있으므로, 그 몫은 표적의 운동이 아니라 엔진이 만든 것으로 읽는다.

반사 깊이는 이 잣대를 움직이지 않는다 — 여섯 자리 모두 깊이 1 과 깊이 2 의 차이가 소수점 아래다(el 0° 에서 86.41% [^14] 대 86.64% [^61]). 깊이를 1 에서 2 로 올리는 것은 이 표적·이 자리에서 사실상 같은 계산이다.

이 몫이 무엇에서 오는지는 이 절이 갈라 두지 않는다. 스위치를 하나씩만 켜서 축을 가른 실험은 [편 02 «경로가 무엇을 세나»](02_engine-paths.ipynb) 에 있고, 이 절이 확정한 것은 «켜면 여섯 자리가 전부 상한 위로 간다» 까지다.
