# 리포트 20 — 수신 방향 그림자 광선을 켜면 상반성 위반이 9.69 → 8.24 dB 로 내려간다

> ### 한 일
> **각 충돌점에서 수신기 방향으로 그림자 광선을 한 번 더 쏘아 출사 쪽 가림을 판정하고, 그 효과를 상반성으로 쟀다.**

### 결과
1. 상반성 위반 최대치(전 β 최악)가 9.69 [^1] → 8.24 dB [^2] 로 내려간다.
2. 모노스태틱에서 이 검사는 무연산이라 생산 σ 는 0.000e+00 dB [^3] 그대로다 — 켜도 옛 결과가 안 움직인다.
3. 바이스태틱 자세 패턴은 β ≤ 45° [^4] 에서 성립한다 — 그 범위의 상반성 RMS 가 2.57 dB [^5] 다.
4. 이 단계는 Sagitta(preprint, arXiv:2604.09243 각주 1)가 바이스태틱 SBR 에서 빠져 있다고 지목한 바로 그 단계다 — 선행이 이름 붙인 구멍을 메운 자리다.

## 무엇을 한 번 더 쏘나

각 충돌점에서 수신기 방향으로 그림자 광선을 한 번 더 쏘아 **출사 쪽 가림**을 판정한다(`src/rcs_sbr.py` `rcs_sbr_multistatic()`). 조명 쪽만 보면 «빛이 닿는 면» 까지는 맞지만, 그 면이 수신기에서 보이는지는 따로 물어야 한다.

Sagitta(preprint, arXiv:2604.09243 각주 1)가 바이스태틱 SBR 에서 빠져 있다고 지목한 바로 그 단계다.

## 켠 효과 — 상반성으로 잰다

**상반성**은 보내는 자리와 받는 자리를 맞바꿔도 σ 가 같아야 한다는 성질이다. 정리이므로 위반량이 그대로 모형오차의 크기다. 위반 최대치가 9.69 [^1] → 8.24 dB [^2] 로 내려간다.

모노스태틱에서 이 검사는 무연산이라 생산 σ 는 0.000e+00 dB [^3] 그대로다 — 새 단계를 켜도 옛 모노스태틱 결과가 한 눈금도 안 움직인다는 뜻이다.

## 두 계열을 나란히 — β 별 RMS

헤드라인 9.69 [^1] → 8.24 dB [^2] 는 **전 β 최악값** 하나의 개선이다. β 별 RMS 로 내려가면 두 방향이 함께 있다.

| β [°] | 출사 가시성 끔 [dB] | 출사 가시성 켬 [dB] |
|---|---|---|
| 15 [^6] | 1.36 [^7] | 0.79 [^8] |
| 30 [^9] | 2.72 [^10] | 1.57 [^11] |
| 45 [^4] | 2.36 [^12] | 2.57 [^5] |
| 60 [^13] | 2.22 [^14] | 1.67 [^15] |
| 75 [^16] | 1.83 [^17] | 2.84 [^18] |
| 90 [^19] | 6.11 [^20] | 4.02 [^21] |

β 45° [^4] (주장 창의 경계)와 75° [^16] 두 칸에서는 켠 계열이 더 크다 — 창 규칙의 문턱 RMS 는 그 경계 칸의 값으로 세워져 있다. 나머지 네 칸과 전 β 최악값에서는 켠 쪽이 더 작다.

## 자세 패턴을 주장할 범위

**바이스태틱 자세 패턴은 β ≤ 45° [^4] 에서 성립한다** — 창 규칙은 «경계 안 상반성 RMS ≤ 2.57 dB [^5]» 이고, 경계는 선언값이다. RMS 는 β 에 단조가 아니라 β 60° 에서 1.67 [^15] 로 내려왔다가 β 90° 에서 4.02 dB [^21] 까지 오른다. 그 위의 β 는 창 밖이고, 검출 기하가 이 창 안에 드는지는 [편 56 «TX·RX·표적 배치와 β·앙각·원거리장이 유…»](56_geometry.ipynb) 가 확인한다.

전방산란 쪽 끝(β→180°)은 조명 게이트와 수신 게이트가 상호배타라 σ ≡ 0 이 된다 — 주장 창을 후방~중간 바이스태틱각으로 못박는 이유이고, 그 목록은 [편 23 «커널이 아직 못 하는 것은 편파 분리·PTD·…»](23_kernel-open-items.ipynb) 에 있다.

## 이 절의 수치가 서 있는 자리

⚠ 위 상반성 표는 각도 모양 |Γ(θ)| 가 바이스태틱 경로에 들어오기 **전** 커널의 산출이다. 지금 그 경로는 각도 모양을 곱하고, 같은 씬에서 복소장과 σ 두 경로가 1e-14 dB [^22] 안에서 맞는다 [^23] — 상반성 표는 그 커널로 다시 낼 대상이다.

⚠ 그리고 **생산 검출 사슬은 이 커널을 부르지 않는다** — σ 를 모노스태틱 격자에서 이등분선 방향으로 조회한다. 그 근사의 크기가 β 15° 에서 rms 중앙 6.39 [^24] · p95 최대 18.9 dB [^25] 이므로, 검출 결과가 인용하는 β 가 몇 도인지가 그 근사의 유효성을 정한다 ([편 56 «TX·RX·표적 배치와 β·앙각·원거리장이 유…»](56_geometry.ipynb)).
