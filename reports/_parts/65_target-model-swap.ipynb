2. 자세 앙상블은 셀당 1080 [^4]자세 전수, (기체×밴드) 셀은 21 [^5]개이고 재현편차는 0.00 dB [^6] 다.
3. 낙차의 소유자는 정육면체다 — 자유공간에서 최대가 M2 인 셀이 21 [^7]개, 최소가 M1 인 셀이 21 [^8]개로 전수다.
4. 크기는 **추정량이 정한다** — 검출기가 읽는 p10 에서 맞추면 낙차가 2.30 dB [^9] 로 줄어든다. ⛔ 이때 M1 이 «가장 어려운 팔» 로 올라서는 것은 M1 의 자세분산이 0.0 dB [^10] 라 어느 분위수로 맞춰도 M1 이 같은 값인 산술이고, p10 순서 계수는 `M3<M2<M1` 10 [^11]셀 · `M2<M3<M1` 11 [^12]셀 둘뿐이다.
5. 각 다양성이 낙차를 줄인다 — 각 다양성이 0 인 자유공간에서 29.30 dB [^13], 가장 큰 앙상블에서 10.09 dB [^14] 다.

## 표적만 세 모형으로 갈아끼운다

같은 기하·같은 검출기·같은 동작점에서 표적만 셋으로 갈아끼운다 — 자세무관 평판 σ -12.81 dBsm [^1] (3GPP TR 38.901 RCS model 1 의 σ_M 상수, M1) · 정육면체(M2) · 우리 SBR+PO 격자(M3).

M3 팔이 선 판을 함께 찍는다 — ⚠ 판: 2026-08-04 [^2] 형상 정정 전 메쉬 · 2026-08-07 10:58:22 [^3] Γ(θ) 이전 커널. σ 격자의 신원은 「outputs/report13_sigma_grid.json (2026-07-29 생성, git 4aa668b) — 07-31 메쉬 개편(동체 셸·암 단면·블레이드) 이전 [^17]」 다. M1 은 3GPP 표값 상수이고 M2 는 현재 메쉬 bbox 로 잡은 모서리라, 이 판 표시는 M3 열에만 붙는다.

⚠ M1 의 «평평함» 은 절반만 3GPP 다 — TR 38.901 은 σ_M 과 함께 로그정규 σ_S 를 주고 그것을 경로마다 뽑는데, 헤드라인은 그것을 평균(=1)에 얼렸다. 원장이 규정대로 다시 뽑으니 M1 도 4.50 dB [^18] 를 물어 M3−M1 이 7.52 dB [^19] → 3.02 dB [^20] 로, 3모형 낙차가 29.30 dB [^21] → 24.80 dB [^22] 로 내려간다.

⛔ M1 열은 측정이 아니라 **눈금**이다 — 동작점 A_ref 를 «평판 -12.81 dBsm [^1] 이 앙상블평균 Pd = 0.9 에 정확히 앉도록» 정의했고, 원장이 그 자리에 «it is the ruler, not a result» 라고 적었다[^23].

자세 앙상블은 셀당 1080 [^4]자세 전수, (기체×밴드) 셀은 21 [^5]개이고, 자세평균을 맞춘 뒤 남는 **요구 추가이득**을 추정량별로 적는다(재현편차 0.00 dB [^6]).

| 무엇을 맞추나 | M1 평판 [dB] — 눈금(동작점 정의) | M2 정육면체 [dB] | M3 우리 SBR+PO 격자 [dB] (⚠ 판: 형상 정정 전 메쉬 · Γ(θ) 이전 커널) |
|---|---|---|---|
| 선형평균 — 이 실험의 규약 | +0.00 [^24] | +29.30 [^25] | +7.52 [^26] |
| 중앙값 | +0.00 [^27] | +16.01 [^28] | +4.78 [^29] |
| dB 평균 | +0.00 [^30] | +14.65 [^31] | +4.66 [^32] |
| p10 — 검출기가 읽는 분위수 | +0.00 [^33] | -2.20 [^34] | -2.08 [^35] |

⛔ M1 열의 네 개 +0.00 은 네 번의 독립 확인이 아니라 같은 항등식 하나다 — 원장이 재계산한 추정량 일곱 갈래(무정규화·선형평균·중앙값·dB 평균·p10·p95·최대) 전부에서 M1 이 부동소수 잡음까지 같은 값이다[^36].

## 낙차의 소유자는 정육면체다

자유공간에서 최대가 M2 인 셀이 21 [^7]개, 최소가 M1 인 셀이 21 [^8]개로 전수이고, M3 몫은 다섯 앙상블에서 20.2% [^37] ~ 33.0% [^38] 다.

각 다양성이 그 낙차를 줄인다 — 각 다양성이 0 인 자유공간(N_eff 1.0 [^39])에서 29.30 dB [^13], 각 다양성이 가장 큰 앙상블(N_eff 3.78 [^40])에서 10.09 dB [^14] 다.

## 크기는 추정량이 정한다

⚠ 낙차의 크기는 레벨을 무엇으로 맞추느냐가 정한다 — 선형평균 29.30 dB [^41] · 중앙값 16.01 dB [^42] · dB 평균 14.65 dB [^43] · p10 2.30 dB [^9] 다. 헤드라인은 선형평균 일치를 쓴다.

⛔ p10 에서 M1 이 «가장 어려운 팔» 로 올라서는 것은 분산 0 의 산술이다 — M1 의 자세분산이 0.0 dB [^10] 라 M1 은 p10 에서도 +0.00 dB [^33] 인데 M2 는 -2.20 dB [^34] · M3 는 -2.08 dB [^35] 로 내려간다. 이 정규화는 순환적이라 정본이 되지 못한다.

문턱은 잡음전력 기지 이상문턱이고 CA-CFAR 문턱은 세 팔에 같은 오프셋을 주므로, 교정표는 세 팔의 절대 소요이득만 옮긴다([^44]).

## 이 표에 구 대조군은 없다

구는 **부피를 맞게 고르면** σ 의 절대 레벨을 맞출 수 있는 단순 모형이면서 자세에 따른 변화를 0 으로 낸다. 레벨에서 우리 메쉬를 앞선 그 구는 논문이 적어 둔 상자 치수로 잡은 부피이고, 메쉬 부피로 잡으면 두 잣대 모두에서 우리 메쉬보다 나쁘다.

그래서 이 표의 낙차는 «자세 구조를 얼마나 담는가» 의 낙차로 읽는다. 구 팔을 넣는 일은 다음 단계에 있다.
