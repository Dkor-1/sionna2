# 리포트 40 — 지상 레이더는 기체를 아래에서 보므로 가림이 무는 자세가 우리 자세다

> ### 한 일
> **같은 기체를 코 쪽·배 쪽·배 옆 세 자세에서 돌려 가림과 변조가 자세에 따라 어떻게 움직이는지 쟀다.**

### 결과
1. 프로펠러는 동체 **위**에 있다 — 위에서 내려다보는 자세에서는 날개가 통째로 드러나고, 아래에서 보는 자세에서 가림이 문다.
2. 코 쪽(앙각 15 도 [^1])의 가림 효과는 변조 폭 +0.05 dB · 레벨 -0.04 dB [^2] 다 — 변조 폭은 두 팔 dB 열의 표준편차 차(`modulation_std_db`)다.
3. 배 쪽(앙각 -15 도 [^3])에서는 -0.62 dB · +1.31 dB [^4], 배 옆(방위 90 도 [^5])에서는 -0.36 dB · -0.85 dB [^6] 다 — 세 칸이 서로 다른 값을 낸다.
4. ⚠ 세 자세의 dB 는 판 한 장과 표본수 위에 있다 — 판을 반 칸 옮기면 레벨 차가 판 셋에서 4.16 dB [^7] p-p 로 흩어지고, 원장은 «레벨 차가 판 선택을 견디는가» 에 아니오 [^8] 라고 적는다. ⛔ 변조 깊이 p-p 는 판을 그대로 두고 표본만 150 개로 솎아도 배 옆에서 중앙값 -2.65 dB · 뽑기 p-p 56.31 dB 로 흔들린다. 이 편이 세우는 것은 **가림이 자세에 따라 다르게 문다**는 존재다.
5. 자유공간 바이스태틱의 이등분선 앙각은 전 구간 음수라 우리는 드론 배를 본다 — 자세 스윕이 그 문장에 독립적인 근거를 붙였다.

## 자세가 답을 바꾼다

프로펠러는 동체 **위**에 있다. 그래서 위에서 내려다보면 날개가 통째로 드러나고, 그 자세에서 가림은 0 에 가깝다.

지상 레이더는 비행 중인 기체를 **아래에서** 보므로 기체 좌표계로 앙각이 음수이고, 거기서 가림이 문다.

⛔ 이 편의 자세 축은 **아래 표(원장)** 로 읽는다. 자세·기체별 요약 그림 `../outputs/figures/report15b_f3.png` 의 dB 라벨은 지금 서 있는 원장 `outputs/report15b_microdoppler.json` 의 값과 다르다 — 그림 빌더(`benchmark/build_report15b_figs.py::fig3_summary`)를 다시 돌리면 그 라벨이 아래 표와 같은 세대가 되고, 그때 이 자리에 들어온다.

## 세 자세의 값

| 자세 | 방위 | 앙각 | 가림 · 변조 폭 | 가림 · 변조 깊이 p-p ⚠ | 가림 · 레벨 |
|---|---|---|---|---|---|
| 코 쪽 | 0° [^11] | 15° [^1] | +0.05 dB | +4.44 dB [^12] | -0.04 dB [^2] |
| 배 쪽 ⭐ | 0° [^13] | -15° [^3] | -0.62 dB | -4.79 dB [^14] | +1.31 dB [^4] |
| 배 옆 | 90° [^5] | -15° [^15] | -0.36 dB | +23.19 dB [^16] | -0.85 dB [^6] |

변조 폭 열은 두 팔 dB 열의 표준편차 차이고(`outputs/report15b_microdoppler.json : cells.*.arms.{F_blade_occ, G_blade_free}.modulation_std_db` 의 차), 세 열 모두 그 자세의 슬로타임 6,000 표본 [^9] 전부로 잰 값이다. 세 줄이 서로 다른 값을 낸다 — **가림이 자세에 따라 다르게 문다**는 것이 이 편의 내용이고, 그 dB 의 부호와 크기는 아래 절이 판 앙상블 평균 뒤로 미룬다.

⚠ **변조 깊이 p-p 열의 쓰임은 «그 자세에서 한 표본이 얼마나 깊이 떨어지나» 까지다** — 자세끼리의 비교는 변조 폭 열이 맡는다. 그 이유는 아래 «두 잣대를 표본수로 흔들었다» 절이 사다리로 인쇄한다.

마이크로도플러는 프레임 사이 위상차로 재는 양이라, 광선 격자가 프레임마다 움직이면 그 움직임이 표적 운동과 같은 자리에 실린다. 그래서 커널은 자세 전체에 한 격자를 고정하고(`grid_ref`), **이 표는 그 얼린 판으로 다시 난 원장에서 나왔다** [^17]. 옛 판의 값은 `outputs/prefreeze/` 에 사본으로 있다.

⚠ 메쉬 지문(`mesh_provenance`)은 도장 스크립트를 돌린 다음 세대부터 이 원장에 실린다. 지문이 찍히면 표가 어떤 메쉬 세대에서 났는지까지 자기가 말한다.

## 두 잣대를 표본수로 흔들었다

| 잣대 · 칸 | 150 표본 | 750 표본 | 3,000 표본 | 6,000 표본(원장) |
|---|---|---|---|---|
| 변조 깊이 p-p · 코 쪽 | -0.16 (p-p 24.63) | +0.17 (p-p 15.10) | +4.22 (p-p 0.45) | +4.44 |
| 변조 깊이 p-p · 배 쪽 | -3.16 (p-p 24.34) | -1.78 (p-p 24.00) | -4.71 (p-p 0.41) | -4.79 |
| 변조 깊이 p-p · 배 옆 | -2.65 (p-p 56.31) | +2.25 (p-p 54.87) | +18.00 (p-p 26.85) | +23.19 |
| 변조 폭 · 코 쪽 | +0.04 (p-p 1.18) | +0.06 (p-p 0.30) | +0.05 (p-p 0.02) | +0.05 |
| 변조 폭 · 배 쪽 | -0.67 (p-p 1.66) | -0.63 (p-p 0.58) | -0.62 (p-p 0.02) | -0.62 |
| 변조 폭 · 배 옆 | -0.44 (p-p 3.55) | -0.48 (p-p 1.30) | -0.36 (p-p 0.35) | -0.36 |

같은 기록을 stride 로만 솎았다 — 판도 자세도 로터도 그대로이고, 흔든 손잡이는 솎기 간격 하나다. 칸의 값은 «위상 stride 개 뽑기의 중앙값 (그 뽑기들의 p-p)» 이고 마지막 열이 원장이 쓴 6,000 표본 [^9] 전부다. 판 사이 차이는 아래 절이 따로 잰다.

⭐ 사다리 네 칸의 **중앙값**이 옮겨 간 폭은 깊이 p-p 가 코 쪽·배 쪽·배 옆 4.60 · 3.01 · 25.84 dB 이고(코 쪽과 배 옆은 부호까지 바뀐다), 같은 사다리에서 변조 폭은 0.02 · 0.05 · 0.12 dB 다. ⛔ 깊이 p-p 가 무엇을 읽는지는 배 옆 칸이 보여 준다 — F 팔의 변조 깊이 79.12 dB [^18] 가운데 18.47 dB 를 6,000 표본 [^9] 중 **가장 깊은 한 표본**이 낸다(그 한 표본을 빼면 60.65 dB 다).

## 판 한 장이 그 dB 의 부호를 정한다

판의 중심만 **반 칸** 옮긴 같은 크기의 판으로 같은 자세를 다시 태우면 가림 dB 가 판마다 이렇게 갈린다 — 세 판(P0 · 반 칸 둘)의 값이다. ⚠ 이 재실행은 생산 슬로타임 열을 stride 40 [^19] 로 솎은 150 표본 [^20] 판이라, 위 표(6,000 표본 [^9])와는 표본수가 다른 추정치다.

| 칸 | 가림 · 변조 깊이 p-p (판 셋) | 가림 · 레벨 (판 셋) |
|---|---|---|
| `matrice4e/belly` | -17.98 [^21] / -18.27 [^22] / +8.26 [^23] | +1.53 [^24] / +2.12 [^25] / -2.04 [^26] |
| `matrice4e/nose` | +14.47 [^27] / -9.95 [^28] / +0.09 [^29] | -0.19 [^30] / +0.21 [^31] / -0.19 [^32] |

⭐ 한 자세 안에서 부호가 갈린다. 레벨 차의 판 셋 흩어짐 4.16 dB [^7] p-p 가 위 세 자세의 레벨 크기를 삼키고, 원장은 «레벨 차가 판 선택을 견디는가» 에 아니오 [^8] 라고 적는다 [^33].

⛔ 깊이 p-p 열의 판 셋 흩어짐 26.52 dB [^10] p-p 를 «반 칸 옮기면 가림 효과가 그만큼 바뀐다» 로 읽지 않는다. 세 판이 같은 슬로타임 표본을 쓰므로(`benchmark/adv_freeze_plate_sensitivity.py` 의 `idx = np.arange(0, n_t, stride)` 한 줄을 세 판이 공유한다) 그 흩어짐이 판에서 온 것은 맞다. 같은 통계량은 판을 하나도 안 옮기고 어느 표본을 뽑느냐만 바꿔도 배 쪽에서 중앙값 -3.16 dB · 뽑기 p-p 24.34 dB 로 흔들리는데, 뽑기 수(판 셋 · 위상 40가지)가 달라 두 p-p 의 크기 비교는 이 원장 밖이다.

⇒ 이 편이 세우는 것은 **가림이 자세에 따라 다르게 문다**는 존재이고, 그 dB 의 부호와 크기는 판 앙상블 평균(오프셋 여러 판의 평균)이 낸다. 기전은 [편 18 «커널이 하는 일»](18_kernel-what.ipynb) 이 잰다.

## 이 결론은 우리 기하 명세와 같은 방향이다

자유공간 바이스태틱의 이등분선 앙각은 전 구간 음수라 우리는 드론 배를 본다. 자세 스윕이 그 문장에 **독립적인 근거**를 붙였다 — 기하 명세와 광선 추적이 서로 다른 길로 같은 자세를 가리킨다.

⚠ Mini 5 Pro 의 그림은 2026-08-10 **겹침 수리 이전** 원장으로 그린 것이라 헤드라인에서 뺐다. 같은 그림을 그 기체로도 그려 `../outputs/figures/report15b_f1b.png` · `../outputs/figures/report15b_f2b.png` 에 뒀다 — 옛 세대 값이라는 것을 알고 읽는다. ⭐현 메쉬의 프롭·벨 겹침은 스탠드오프 1.0 mm 로 **0.00 %** 다[^34].
