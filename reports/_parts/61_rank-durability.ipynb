# 리포트 61 — 그 순위는 자세평균이면 하나로 모이고, 자세평균 뒤집힘 문턱은 현실 봉투 안이다

> ### 한 일
> **σ 오차를 공통모드와 차분 두 종류로 나눠 넣고, 각각에서 밴드 순위가 어디까지 버티는지를 단일 자세·자세평균 두 인용 방식의 뒤집힘 문턱과 몬테카를로 보존확률로 재었다.**

### 결과
1. σ 는 SNR 에 선형이라 σ 오프셋 Δ dB 가 곧 SNR 오프셋 Δ dB 다 — 선형성 잔차 1.2e-05 [^1].
2. 공통모드 ±10 dB 에서 15칸 전부 순위가 유지되고, 절대거리만 -43.3% [^2] ~ +76.4% [^3] 움직인다.
3. 뒤집는 것은 차분 오차다 — 단일 자세 뒤집힘 문턱은 0.61 [^4] ~ 7.42 dB [^5], 자세평균 문턱은 0.09 [^6] ~ 2.95 dB [^7] 이고, 현실 봉투는 5.01 dB [^8] 다.
4. 작은 기체가 더 취약하다는 예상은 뒤집힌다 — 가장 작은 mini5pro [^9] 가 단일자세·자세평균 양쪽에서 가장 견고하다. 단일자세 뒤집힘 문턱과의 상관은 크기 쪽 -0.62 [^10], 밴드 간 σ 로브 산포 쪽 -0.31 [^11], 두 열 사이는 +0.09 [^12] 이고, 기체 5 대의 상관계수라 어느 열이 취약성을 정하는지는 이 표본으로 정할 수 없다.
5. σ 격자를 블레이드 형상 갱신본으로 바꾸는 것만으로 R90 이 최대 17.9% [^13] 움직이고 순위쌍 1 [^14]개가 뒤집힌다.

## 오차를 두 종류로 나눈다

**공통모드**는 세 밴드를 함께 옮기고, **차분**은 밴드마다 다르게 옮긴다. σ 는 SNR 에 선형이라 σ 오프셋 Δ dB 가 곧 SNR 오프셋 Δ dB 다(선형성 잔차 1.2e-05 [^1]).

| 오차 종류 | 무엇이 움직이나 | 크기 |
|---|---|---|
| 공통모드 ±10 dB | 순위 유지 · 절대거리만 이동 | 15칸 전부 순위 불변 · 거리 -43.3% [^2] ~ +76.4% [^3] |
| 차분(밴드별) — 단일 자세 ψ=0 | 순위가 뒤집힐 수 있는 축 | 뒤집힘 문턱 0.61 [^4] ~ 7.42 dB [^5] (현실 봉투 5.01 dB [^8]) · 봉투 안에서 3 [^17]/5 [^18]기체가 뒤집힌다 |
| 차분(밴드별) — 자세평균 σ | 합의 순위가 뒤집히는 축 | 뒤집힘 문턱 0.09 [^6] ~ 2.95 dB [^7] — 5기체가 모두 현실 봉투 안에 든다 |
| 밴드별 독립 오차 2 dB (몬테카를로) | 순위 보존 확률 | 0.28 [^19] ~ 0.82 [^20] |

## 작은 기체가 더 취약하다는 예상은 뒤집힌다

가장 작은 mini5pro [^9] (전장 0.378 m [^21], LTE 에서 D/λ = 2.32 [^22]) 가 단일자세·자세평균 양쪽에서 가장 견고하다.

단일자세 뒤집힘 문턱은 최대 치수 열과 밴드 간 σ 로브 산포 열 둘 다에 걸린다 — 상관은 크기 쪽 -0.62 [^10], 산포 쪽 -0.31 [^11] 로 크기 쪽이 더 강하고, 두 열 사이 상관은 +0.09 [^12] 다.

⚠ 기체가 5 대라 이 세 수는 서술용이다 — 같은 5행에서 순위상관을 내면 크기-문턱이 -0.90(p=0.04), 산포-문턱이 -0.10(p=0.87) 로 갈리고, 피어슨 p 는 각각 0.27 · 0.61 다(⛔ p 와 순위상관은 원장 밖의 재계산값이다). 원장 `size_vs_fragility.finding` 은 산포를 단독 원인으로 들지만, 이 편은 세 상관계수를 그대로 읽고 원인은 표본을 늘린 뒤로 미룬다.

## 통제되지 않은 σ 변화의 관측된 파급

σ 격자를 블레이드 형상 갱신본으로 바꾸는 것만으로 R90 이 최대 17.9% [^13] 움직이고 순위쌍 1 [^14]개가 뒤집힌다.

⚠ 이 사슬은 2026-08-04 [^23] 형상 정정 **전** 메쉬 위에 있다 — 다섯 기체 중 Matrice 4E 가 그 정정을 받았다 [^24].

⚠ 이 σ 원장은 Γ(θ) 각도 모양(기본 켬, 2026-08-07 10:58:22 [^25]) 이전 커널의 산출이다 — 방위평균 전체 드론 σ 이동은 +0.08 [^26] ~ +0.10 dB [^27], 프롭 채널 레벨 이동은 +5.16 [^28] ~ +6.48 dB [^29] 다. σ 격자를 Γ(θ) 켠 커널로 재생성한 뒤 이 편을 다시 짓는다.

하한 0.61 dB [^4] 를 내는 행이 그 정정을 받은 Matrice 4E 다. 이 하한은 **단일 자세** 기준이고, 같은 기체를 **자세평균**으로 읽으면 1.30 dB [^30] 다 — 실측 캠페인이 판정 문턱으로 드는 수가 그것이다([편 76 «기울기 판정의 문턱은 세션간 진폭 재현성이고»](76_session-drift.ipynb)).

자세평균 판에서 문턱이 가장 얇은 기체는 mavic4pro 이고 그 값은 0.09 dB [^31] 다 — 합의 순위의 아래 두 밴드가 그만큼 붙어 있다.
