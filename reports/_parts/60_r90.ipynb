# 리포트 60 — 앵커 σ 위의 R90 은 비교가능 12칸에서 3.69~7.44 km 이고, 밴드 순서는 기체마다 바뀐다

> ### 한 일
> **기울기 앵커의 Δσ 를 R90 근방 국소 지수로 옮겨 다섯 기체 × 세 밴드의 검출거리를 내고, 그 표에서 밴드 순서가 기체마다 어떻게 바뀌는지를 읽었다.**

### 결과
1. Δσ 를 R90 근방 국소 지수 3.99 [^1] 로 옮긴다 — `d` 축에서 국소적으로 $R \propto \sigma^{1/n}$ 이다.
2. 앵커 비교가능 기체 12칸의 R90 은 3.69 km [^2] ~ 7.44 km [^3] 다.
3. 밴드 순서는 기체마다 바뀐다 — 그 순서를 만드는 것은 자세별 로브 구조이고, 앵커는 밴드별 스칼라를 옮기면서 밴드 평균 레벨은 그대로 둔다.
4. 자세를 평균하면 다섯 기체가 한 순위 (LTE > 5G > WiFi [^4]) 로 모인다 — 단일 자세에서는 3 [^5]가지 순위가 나온다.
5. 아래 km 열은 **순위를 읽는 표**로 쓴다 — 공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% [^6] ~ +76.4% [^7] 옮긴다.

## 앵커 σ 위의 R90

Δσ 를 R90 근방 국소 지수 3.99 [^1] 로 옮긴다 — `d` 축에서 국소적으로 $R \propto \sigma^{1/n}$ 이다(`src/freespace_scene.py:56`).

아래 표의 R90 은 전부 공칭 헤딩 ψ=0 의 수이고, **σ 민감도 봉투와 함께 읽는다** — [편 61 «그 순위는 자세평균이면 σ 오차 아래에서 하나로 모인다»](61_rank-durability.ipynb) 가 그 봉투를 든다.

| mavic4pro | 5.81 km | 4.65 km | 6.61 km | scaled |
| matrice4e | 4.38 km | 6.34 km | 5.93 km | scaled |

## 왜 km 열을 순위로 읽는가

공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% [^6] ~ +76.4% [^7] 옮긴다.

그 봉투가 어디서 오는지도 함께 적는다 — 우리 세 밴드는 전부 PO 근사가 1 dB 안에 든다고 보장되는 부품 폭 문턱 아래에 부품을 남긴다([편 59 «레벨을 맞추려면 크기전이 법칙을 골라야 하므로…»](59_slope-anchor.ipynb)). σ 절대 레벨의 불확도는 선언된 ±10 dB 봉투와 별개로 크기가 아직 정해지지 않은 항을 하나 더 갖는다([^12]).

## 밴드 순서는 자세 인용 방식이 정한다

밴드 순서는 기체마다 바뀐다. 그 순서를 만드는 것은 자세별 로브 구조이고, 앵커는 밴드별 스칼라를 옮기면서 밴드 평균 레벨은 그대로 둔다. 자세를 평균하면 다섯 기체가 한 순위로 모인다.

| 인용 방식 | 서로 다른 순위 수 | 합의 순위 | 최악 뒤집힘 문턱 |
|---|---|---|---|
| 단일 자세 ψ=0 | 3 [^13] | 기체마다 다름 | 0.61 dB [^14] |
| 자세평균 σ | 1 [^15] | LTE > 5G > WiFi [^4] | 0.09 dB [^16] |
| 자세평균 + 기울기 앵커 | 1 [^17] | LTE > 5G > WiFi [^4] | 3.72 dB [^18] |
| 자세평균 + 앵커 + 듀티 | 2 [^19] | 기체마다 다름 | 0.82 dB [^20] |
