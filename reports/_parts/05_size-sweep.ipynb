# 리포트 05 — 면적을 1600배로 키워도 경로 진폭은 7.4e-07 dB 움직인다

> ### 한 일
> **빈 자유공간에 금속 평판 하나를 두고 변만 키우며 PO 단면적과 path solver 의 표적 경로 진폭을 같은 축에서 재고, 같은 실험을 드론 메쉬로 옮겼다.**

### 결과
1. 평판 면적을 1600 [^1]배 키우면 PO 단면적은 64.08 dB [^2] 커지고, path solver 의 표적 경로 진폭은 7.4e-07 dB [^3] 움직인다.
2. 경로 수는 전 구간 1 [^4]개이고, 그 진폭은 이미지-소스 해석해와 0.0017 dB [^5] 안에서 맞는다 — **엔진은 자기가 푸는 문제를 정확히 푼다**.
3. 같은 실험을 기체 메쉬로 옮겨 삼각형을 1.82 [^6] decade 깎으면, 실루엣이 살아 있는 단 전부에서 정반사 경로가 존재하는 자세는 36 [^7]자세 중 1 [^8]개(az 0° [^9] · el 0° [^10] · spp 256,000,000 [^11] · 깊이 1 [^12])이고, 그 자세의 정반사 경로 수는 2 [^13]→1 [^14] 로 준다.
4. ⭐ 그 구간의 **정반사 전용** 값은 -33.8709 dB [^15] → -39.8915 dB [^16] 이고, 1경로값에 20·log₁₀(2) 를 더하면 2경로값과 잔차 -5.5e-07 dB [^17] 로 같다 — 닫힌형 20·log₁₀(1/2) = -6.02 dB [^18] 와 같은 **항등식**이고, 적대검증 원장은 여기서 사라진 것을 산란체가 아니라 **중복 복사본**으로 읽는다. ⛔ 형상 판정에서 떨어지는 마지막 단(정반사 경로 0 [^19]개)의 계단은 이 편에서 폐기한다.
5. 반대 방향도 같은 뿌리다 — 같은 평판을 쪼개기만 해도 코히어런트 전력이 9.54 dB [^20] 부푼다. 답의 단위가 면적이 아니라 경로라서 면 수가 곧 답이 된다.

## 평판 하나를 키워 보면

빈 자유공간에 금속 평판 하나를 두고 변만 0.1 m [^26] → 4.0 m [^27] (면적 1600 [^1]배) 로 키운다. PO 단면적은 면적 dB 당 2.00 dB [^28]씩, 총 64.08 dB [^2] 오른다.

같은 구간에서 path solver 의 표적 경로 진폭은 7.4e-07 dB [^3] 움직이고 경로 수는 내내 1 [^4]개다 — 변 6단 × 예산 4단 = 24 [^21]셀 전부에서, 셀마다 시드 5 [^22]개를 평균해 깊이 1 [^23] 로 잰 값이다.

깊이 2 [^24] 는 격자 밖의 별도 확인이다 — 블록 `A_depth_check` 가 변 하나 · spp 16,000,000 [^25] 에서 깊이 1 [^29] 와 2 [^24] 두 항을 잰다.

## 드론 메쉬에서는 정반사 경로가 자세 하나에서만 살아남는다

같은 실험을 기체 메쉬로 옮긴다. 삼각형을 29,932 [^31]개 → 450 [^32]개(1.82 [^6] decade)로 여섯 단에 걸쳐 깎는다.

실루엣이 유지되는 것은 1,522 [^33]개까지다. 가장 성긴 450 [^32]개 판은 원장의 형상 판정에서 떨어진다(`shape_ok` = 아니오 [^34]) — 정면투영 편차 19.3% [^35] · 경계상자 편차 10.2% [^36] 다. 실루엣이 살아 있는 단 전부에서, 36 [^7]자세 중 정반사 경로가 존재하는 자세는 1 [^8]개다.

그 한 자세에서 스톡 솔버가 내놓는 **정반사 경로 수**는 2 [^13]→1 [^14] 로 줄고, 정반사만 남긴 코히어런트 값은 -33.8709 dB [^15] → -39.8915 dB [^16] 로 내려간다. 1경로값에 20·log₁₀(2) 를 더하면 2경로값과 잔차 -5.5e-07 dB [^17] 로 같다 — 닫힌형 20·log₁₀(1/2) = -6.02 dB [^18] 와 같은 **항등식**이다.

⚠ 적대검증 원장은 이 자리를 금지 문장 목록에 올려 뒀다 — '드론에서 면 하나가 없어지자 6 dB 가 날아갔다 = 산란체를 잃었다' — 잃은 것은 **중복 복사본**이다(1경로값+20log10(2) 와 5e-7 dB 일치). 물리 산란체가 아니다. [^37] (원장이 그 기전에 붙인 한정: Sionna 솔버 소스의 중복제거 코드를 읽지 않았다 — 기전은 출력에서의 추론 [^38])

⛔ 가장 성긴 450 [^32]개 판에서는 정반사 경로가 0 [^19]개이고 정반사 코히어런트 값이 `null` [^39] 이다. 그 단에 남는 coh_db 는 확산(diffuse_reflection=True)을 켠 실행의 값이고, 원장은 그런 값을 물리량으로 인용하는 것을 `must_not_say` 에 올린다 — 근거가 된 예산 스윕에서 같은 450 [^40]개 판이 spp 4,000,000 [^41] → 1,024,000,000 [^42] 에 29.89 dB [^43] 움직였다. 다만 그 스윕의 창은 el 15° [^44] 의 방위 두 개 · 시드 두 개 평균이라 여기 한 자세(az 0° [^9] · el 0° [^10])는 그 창 밖이고, 원장이 그 자세에 대해 들고 있는 행은 spp 256,000,000 [^45] 하나뿐이라 마지막 단의 계단 크기는 이 편에서 비워 둔다.

⭐ 면 수 축에서 이 편이 인쇄하는 것은 여기까지다. 형상 판정을 통과하는 구간에서 생산 설정 총에코는 19.7 [^46]배 데시메이션에 -0.098 dB [^47] ± 0.208 dB [^48] 움직이고(같은 방위·같은 시드끼리 짝지은 차), 그 이동은 실루엣 면적비 0.977 [^49] 로 예측한 -0.102 dB [^50] 와 잔차 0.004 dB [^51] 로 맞는다. 움직인 것은 면 수가 아니라 남은 형상이다.

⚠ 나머지 자세가 빈 이유는 따로 있다. 같은 자세에 확산을 켜면 표적경유 경로가 자세당 102 [^52]개 넘게 잡힌다. 비어 있는 것은 광선이 아니라 **거울 조건을 만족하는 삼각형**이다.

## 반대 방향도 같은 뿌리 — 쪼개기만 해도 답이 부푼다

같은 1 m [^53] 평판을 2 [^54]개 → 512 [^55]개 삼각형으로 **쪼개기만** 해도 코히어런트 전력이 9.54 dB [^20] 부푼다. 늘어난 경로들은 진폭 산포 0.0 dB [^56] · 지연 산포 0.0 ns [^57] · 위상 산포 0.0° [^58] 인 **완전한 복사본**이고, 합은 20·log₁₀(N) 을 0.0017 dB [^59] 안에서 따른다. 격자 위치를 옮겨도 중복은 남는다 (`offset_removes_duplication` = 아니오 [^60]).

메쉬를 깎으면 무너지고 쪼개면 부푼다. 두 방향이 같은 뿌리에서 나온다 — **답의 단위가 면적이 아니라 경로라서** 면 수가 곧 답이 된다.
