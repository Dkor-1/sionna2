# 리포트 26 — Phantom 3 를 문헌값을 보지 않고 내고 봉인을 풀었다

> ### 한 일
> **앵커 기체와 같은 기체를 문헌 상수를 한 번도 읽지 않은 경로로 돌리고 별도 스크립트가 봉인을 풀어 문헌과 맞댔다.**

### 결과
1. 같은 창 1.8 [^1]~18.2 GHz [^2] 에서 우리 커널로 돌렸다(`benchmark/p3_ours.py`, 14,542 s [^3]).
2. 우리 el=0 전대역 기울기는 0.420 [^4] ± 0.066 dB/GHz [^5] 이고, Das 공표 0.21 dB/GHz [^6] 대비 2.00 배 [^7] · 3.2 σ [^8] 다.
3. Yuan θ=90° 실측곡선 대비는 1.33 배 [^9] · 1.6 σ [^10] 이고 레벨은 -4.91 dB [^11] 다.
4. 우리 세 밴드가 전부 저주파 구간 안에 있고, 거기서 우리 μ(f) 는 1.56 [^12] (1.8–6.0 GHz)에서 0.20 dB/GHz [^13] (6.0–18.2 GHz)로 꺾인다.
5. 낙차의 방향은 얇은 판 참값이 정한다 — 전력을 만드는 쪽 기준이면 우리 PO 가 -4.02 [^14], 약한 쪽 기준이면 +7.53 dB [^15] 다.

## 무엇을 눈감고 냈나

앵커 기체와 같은 기체(DJI Phantom 3)를 같은 창 1.8 [^1]~18.2 GHz [^2] 에서 우리 커널로 돌렸다(`benchmark/p3_ours.py`, 14,542 s [^3]). 산출 과정은 문헌 상수를 한 번도 읽지 않았고 봉인은 별도 스크립트가 풀었다. 적합 창은 Das Table III · Yuan §IV 와 같다(일치 예 [^16]).

이 대조가 재는 것은 절대 레벨과 주파수 의존 두 스칼라이고, 각도패턴은 우리 기하에서 나온 그대로다. ⚠ 아래 표 네 행은 전부 **v1 메쉬** 산출이다 — 사진 실측으로 다시 지은 v2 메쉬의 같은 두 스칼라는 [편 27 «메쉬가 사는 축은 절대 크기가 아니라 각도 구…»](27_box-sphere-control.ipynb) 이 잇는다.

## 정본은 전대역 대 전대역이다

부분대역 행은 우리 μ(f) 가 1.56 [^12] (1.8–6.0 GHz)에서 0.20 dB/GHz [^13] (6.0–18.2 GHz)로 꺾인다는 것을 보이는 설명이다.

⚠ **우리 세 밴드(LTE 1.843 [^22] · 5G 3.5 [^23] · WiFi 5.21 GHz [^24])가 전부 그 저주파 구간 안에 있고, 거기서 우리 σ 는 실측곡선보다 낮다.**

## 낙차의 방향은 어디까지 말할 수 있나

이 낙차의 방향은 결과 밖에 둔다 — 우리 PO 적분은 편파를 가르는 대신 세기 하나만 내는 스칼라라서, 참값이 편파에 따라 갈리는 구간에서는 한쪽 편파 기준으로 낮게 다른 쪽 기준으로 높게 나온다 [^25].

⭐ 다만 두 쪽의 크기가 서로 다르다 — 얇은 판 참값에서 **전력을 만드는 쪽**(두 편파 중 σ 가 11.56 dB [^26] 큰 쪽) 기준이면 우리 PO 는 -4.02 dB [^14] (음수 = 우리가 낮다), 약한 쪽 기준이면 +7.53 dB [^15] 다.

그래서 **우리 σ 가 낮게 나와 있을 개연성 쪽이 크고**, 그 방향이라면 검출 산출물은 보수적인(비관적인) 쪽으로 틀린 것이다. 레벨을 만드는 굵은 부품일수록 이 어긋남이 작아서 이 항이 설명하는 몫은 관측된 낙차보다 작다 [^27] — 그래서 방향은 여기까지, 개연성으로만 적는다. 금칙 세 가지는 [편 22 «PO 유효 무릎을 부품 폭으로 옮기면 어느 부…»](22_po-knee.ipynb) 에 있다.
