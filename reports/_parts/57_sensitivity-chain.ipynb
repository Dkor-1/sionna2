# 리포트 57 — 세 밴드에서 값이 다른 항은 λ² 와 σ 둘뿐이다

> ### 한 일
> **1 km · Mavic 4 Pro · 수신소자 1개에서 출력 SNR 을 항별로 분해하고, 밴드 쌍의 격차를 그 항들로 쪼개 어느 항이 격차를 만드는지를 세었다.**

### 결과
1. 공통항(EIRP·수신이득·확산·1/N₀·CPI·듀티·손실)은 세 밴드가 같은 값을 쓴다 — 값이 다른 항은 $\lambda^2$ 와 σ 둘뿐이다.
2. WiFi−LTE 격차 +3.54 dB [^1] 는 $\lambda^2$ -9.03 dB [^2] 와 σ +12.57 dB [^3] 의 합이다.
3. 5기체 × 3쌍 15칸에서 σ 항이 더 큰 칸이 9 [^4]칸이다 — 그 셈은 `d` = 2000 m [^5] · 앵커 전 σ 판이고, σ-무관 축의 격차는 $\lambda^2$ 스프레드로 고정이다.
4. 유한 ECA 깊이에서는 직접파 잔차가 벽이 된다 — 기체 s1000plus [^6] · 모드 W1 [^7] 판에서 40 dB 의 R90 이 2786 [^8], 완전 억압이 9724 m [^9] 이고, solve 판 R90(기체 s1000plus [^6] · 모드 W1 [^7]) 은 완전 억압(ECA ∞) 행과 같은 값이다.
5. 듀티 항은 이 사슬에서 꺼져 있다 — 그 크기는 5G 가 LTE 대비 16.02 dB [^10] 다.

## σ 행이 서 있는 원장 세대

$\lambda^2$ 는 정의로 정해지지만 σ 행은 격자 원장에서 조회한 값이다 — 그 격자의 세대가 이렇다.

⚠ 이 사슬은 2026-08-04 [^19] 형상 정정 **전** 메쉬 위에 있다 — 다섯 기체 중 Matrice 4E 가 그 정정을 받았다 [^20].

⚠ 이 σ 원장은 Γ(θ) 각도 모양(기본 켬, 2026-08-07 10:58:22 [^21]) 이전 커널의 산출이다 — 방위평균 전체 드론 σ 이동은 +0.08 [^22] ~ +0.10 dB [^23], 프롭 채널 레벨 이동은 +5.16 [^24] ~ +6.48 dB [^25] 다. σ 격자를 Γ(θ) 켠 커널로 재생성한 뒤 이 편을 다시 짓는다.

## 밴드 쌍의 격차를 그 두 항으로 쪼갠다

$\lambda^2$ 는 정의로 정확하고, σ 는 자세 로브 구조(방위를 돌릴 때 σ 가 솟는 봉우리와 꺼지는 골)가 만든다.

| 쌍 | 출력 SNR 차 | $\lambda^2$ 항 | σ 항 |
|---|---|---|---|
| W1-L1 | +3.54 dB [^1] | -9.03 dB [^2] | +12.57 dB [^3] |
| W1-G1 | -2.04 dB [^26] | -3.46 dB [^27] | +1.41 dB [^28] |
| L1-G1 | -5.59 dB [^29] | +5.57 dB [^30] | -11.16 dB [^31] |

위 표의 σ 행은 `d` = 1000 m [^32] · 앵커 적용 판이다. 15칸을 세는 아래 문장은 다른 규약 위에 선다 — `d` = 2000 m [^5] · 앵커 전 σ 다.

그 판에서 5기체 × 3쌍 15칸 중 σ 항이 더 큰 칸은 9 [^4]칸이고, σ-무관 축의 격차는 $\lambda^2$ 스프레드 -9.03 [^33] / -3.46 [^34] / +5.57 dB [^35] 로 고정이다.

## 어느 벽이 거리를 정하나 — solve 판(기체 s1000plus · 모드 W1)

solve 판 R90(기체 s1000plus [^6] · 모드 W1 [^7]) 은 완전 억압(ECA ∞) 행과 같은 값이라, 그 동작점의 직접파 잔차 항은 0 이다. 유한 ECA 깊이에서는 직접파 잔차가 벽이 되고, 깊이에 따라 거리가 이렇게 움직인다. 이 감도는 solve 축 한 판의 값이고, 편 60 의 헤드라인 폭은 앵커 비교가능 기체 쪽에서 읽는다([편 60 «앵커 σ 위의 R90 은 비교가능 12칸에서…»](60_r90.ipynb)).

| ECA 깊이 | 40 dB | 60 dB | 90 dB | 완전 억압 |
|---|---|---|---|---|
| R90 (기체 s1000plus · 모드 W1) | 2786 m [^8] | 7618 m [^36] | 9720 m [^37] | 9724 m [^9] |

레이더 방정식 항등식 검사는 6 행 [^13] 에서 코드 경로와 dB 산술의 차이를 2.8e-14 dB [^38] 로 잡는다.

## 듀티 축의 크기

위 사슬은 기준신호가 CPI 전체를 채운다는 규약에서 풀린다. 실제 점유가 만드는 듀티 항은 밴드마다 다르고, 그 크기를 여기 적는다.

| 모드 | 기준신호 길이 T_ref | 프레임 M | 듀티 항 |
|---|---|---|---|
| WiFi | 5.20e-05 s [^39] | 100 [^40] | -12.84 dB [^41] |
| LTE | 1.00e-03 s [^42] | 100 [^43] | +0.00 dB [^44] |
| 5G | 5.00e-04 s [^45] | 5 [^46] | -16.02 dB [^47] |

이 항을 넣으면 5G 는 LTE 대비 16.02 dB [^10] 를 더 치른다. WiFi 의 -12.84 dB [^48] 는 조명원 원장의 패킷 듀티와 같은 값이다.
