# 리포트 36 — 두 엔진이 날개끝 주파수 아래에서 겹치고 그 위에서 갈린다

> ### 한 일
> **같은 로터 위상 스텝 절차를 스톡 Sionna 엔진과 우리 PO 커널 두 곳에 태워 같은 칸에서 나온 무늬를 맞댔다.**

### 결과
1. 운동학이 예측한 날개끝 주파수 **아래**에서 두 빗살이 겹친다 — 변조 깊이가 Sionna 10.91 dB [^1] · 순수 PO 17.70 dB [^2] 다(**nose · 3 m [^3] 한 칸**).
2. 그 위에서 갈린다 — 기하 절벽 너머 꼬리의 최대값이 Mini 2 에서 Sionna -19.56 dB [^4] 대 순수 PO -36.14 dB [^5] 다(**자세 5 × 거리 1·3·10 m = 15 칸 [^6] 의 중앙값**).
3. Sionna 의 빗이 **기하에서 온다** — 산란 물리를 전부 빼고 왕복 위상만 더한 기준과 겹치면 Mini 2 9 [^7]/10 [^8] 칸이 ±1 조화 안에서 일치한다(**근거리 1·3 m 칸**).
4. 빗 모양 코사인 중앙값은 Mini 2 0.9639 [^9] · Matrice 4E 0.8714 [^10] 다(**같은 15 칸의 중앙값**).
5. ⭐ 아래쪽에서 겹친다는 것이 «위상은 광선 엔진이 맞게 낸다» 의 근거이고, 위쪽에서 갈린다는 것이 «세기는 PO 커널이 맡는다» 의 근거다.

| 팔 셋 · 이 편의 낱말 | **S** = Sionna PathSolver · **B** = Ours (SBR+PO, 기본) · **P** = Ours, nothing blocked (control). 절마다 어느 팔이 재었는지를 이 세 이름으로 적는다 |

## 같은 로터를 두 엔진이 돌린다

![report07_f2](../outputs/figures/report07_f2.png)

**그림 1.** Sionna 자체 엔진과 우리 순수 PO 대조팔이 같은 로터에서 같은 무늬를 내는가?
로터 위상을 스텝하고 매번 다시 추적하는 같은 절차를 **두 팔**에 태웠다 — **S** = Sionna PathSolver 와 **P** = Ours, nothing blocked (control) 다. 이 그림은 **nose · 3 m 한 칸**이고 기선은 0.2 m 다(준-모노스태틱).

운동학이 예측한 날개끝 주파수 **아래**에서 두 빗살이 겹친다. 가림을 푸는 SBR 팔(B)은 이 그림 밖에 있고, 아래 «세 엔진을 같은 격자에» 절이 그 팔을 세운다.

## 그 위에서 갈린다

잰 팔은 **S** 와 **P** 둘이고, 셋째 열은 산란 물리를 뺀 기하 기준이다 — **자세 5 × 거리 1·3·10 m = 15 칸 [^6] 의 중앙값**이다.

| 기체 | S 꼬리 최대 | P 꼬리 | 기하 기준 꼬리 |
|---|---|---|---|
| Mini 2 | -19.56 dB [^4] | -36.14 dB [^5] | -26.52 dB [^11] |
| Matrice 4E | -24.60 dB [^12] | -35.83 dB [^13] | -29.85 dB [^14] |

Sionna 의 꼬리가 PO·기하 기준보다 뚜렷하게 높게 남는다. 그 꼬리는 블레이드가 만든 것이 아니므로, 가장자리를 자동으로 찾는 검출기는 물리적이지 않은 자리를 가장자리라고 보고한다. 그 꼬리의 원인은 [편 42 «광선예산»](42_md-ray-budget.ipynb) 이 잰다.

## 세 엔진을 같은 격자에 태우면

![report07_f5](../outputs/figures/report07_f5.png)

**그림 2.** Sionna·SBR·PO 를 같은 슬로타임 격자에 태우면 같은 무늬가 나오는가?
같은 기체·자세·주파수·PRF·로터별 회전수로 세 팔을 돌렸다 — 4096 표본 [^15] @ PRF 19700 Hz [^16] = 26 [^17] 블레이드 주기. 이 그림과 표는 **belly · 앙각 -15° [^18] · 3 m [^19] 한 칸**이다.

| 팔 | 무엇으로 쟀나 | 깊이 p-p ⛔ | 깊이 p5~p95 ⭐ | 비 | 전체 전력 중 날개끝 밖 | 무늬 |
|---|---|---|---|---|---|---|
| S | Sionna PathSolver | 32.6 dB [^20] | 7.5 dB [^21] | 4.3 배 [^22] | 5.06% [^23] | 얼룩이 대역을 채우고 날개끝 주파수 밖에도 남는다 |
| B | Ours (SBR+PO, 기본) | 5.3 dB [^24] | 2.5 dB [^25] | 2.2 배 [^26] | 0.18% [^27] | 빗살이 또렷하다 — ⚠순수 PO 보다는 여전히 35 dB [^28] 위다 |
| P | Ours, nothing blocked (control) | 2.8 dB [^29] | 1.2 dB [^30] | 2.3 배 [^31] | 0.00007% [^32] | 빗살이 가장 좁고 날개끝에서 절벽처럼 잘린다 |

⛔ **p-p 열을 물리량으로 인용하지 마라 — 특히 S 의 32.6 dB [^20] 는 표본 N 개의 max−min 이라 N 에 딸려 자란다:** 같은 4,096 표본 [^33] 열을 등간격으로 솎아 재면 S 의 p-p 중앙값이 N=64 13.2 → 512 18.5 → 4,096 32.6 dB 로 **전체 열까지 계속 자라고**, 같은 솎기의 p5~p95 는 7.3 → 7.5 dB 로 붙어 있다. 최솟값 자세 하나만 빼도 S 의 p-p 가 27.1 dB [^34] 로 내려간다. ⭐ 그래서 이 절이 깊이를 인용할 때 쓰는 자는 **p5~p95 열**이고, p-p 열과 «비» 열은 «이상치 몇 개가 폭을 정하는가» 를 보는 자로 읽는다.

⭐ 셋 다 (a) 0 도플러 동체 선, (b) 블레이드 통과율 간격 능선, (c) 날개끝 주파수 근처 감쇠를 낸다 — **구조가 일치한다**. 날개끝 주파수 안에서 스펙트럼 코사인이 0.745 [^35] (Sionna↔SBR) · 0.760 [^36] (Sionna↔PO) · 0.955 [^37] (SBR↔PO) 다.

## 그 코사인은 «닮았다» 까지를 말한다

위 코사인이 말하는 것은 «대역 안에서 무늬의 모양이 닮았다» 까지다. 세 엔진은 능선의 **세기와 밀도**에서 크게 갈리고, 갈리는 방향에 각각 이유가 있다.

| 팔 | 무엇이 그렇게 만드나 |
|---|---|
| P Ours, nothing blocked (control) | 가림을 빼고 계산하므로 모든 면이 항상 기여한다 → 변조가 씻긴다 |
| B Ours (SBR+PO, 기본) | 가림이 켜져 있어 능선이 굵다. 격자를 얼린 뒤 날개끝 밖에 남는 몫은 0.18% [^27] 이고, 그래도 순수 PO 보다 35 dB [^28] 위다 |
| S Sionna PathSolver | 경로가 열 개 남짓이라 표본이 성기고, 날개끝 밖에 5.06% [^23] 이 남는다 |

⭐ **날개끝 밖에 남던 능선의 원인은 광선 격자의 이산화가 아니라 자세마다 격자를 다시 정의하는 것이었다.** 격자를 촘촘히 해도 λ/12 → λ/32 에서 대역밖 절대 전력이 2.3 dB [^38] 만 내려간다 — 기울기 -0.56 [^39] 이고, d² 이산화 잡음이면 −2 여야 한다.

격자를 얼리면 같은 광선 수로 13.1 dB [^40] 내려가면서 기울기가 -2.19 [^41] 로 예측에 붙는다. **순수 PO 팔(P)을 잣대로 놓으면** 얼린 팔의 대역 안 일치도가 0.440 [^42] 에서 0.953 [^43] 으로 오른다. 기전과 대가는 [편 18 «커널이 하는 일»](18_kernel-what.ipynb) 이 잰다.

## 얼리기의 판정은 잣대로 삼은 팔이 정한다

⭐ **Sionna 팔(S)을 잣대로 놓으면 같은 원장이 반대로 움직인다** — 같은 격자(λ/12)에서 0.727 [^44] 에서 0.662 [^45] 로 내려가고, 사다리의 다섯 격자(λ/8 · λ/12 · λ/16 · λ/24 · λ/32) 전부에서 같은 방향이다.

| 격자 | P 잣대 (prod → froz) | S 잣대 (prod → froz) |
|---|---|---|
| λ/8 | 0.514 [^46] → 0.955 [^47] | 0.743 [^48] → 0.695 [^49] |
| λ/12 | 0.440 [^42] → 0.953 [^43] | 0.727 [^44] → 0.662 [^45] |
| λ/16 | 0.426 [^50] → 0.968 [^51] | 0.722 [^52] → 0.603 [^53] |
| λ/24 | 0.428 [^54] → 0.976 [^55] | 0.722 [^56] → 0.633 [^57] |
| λ/32 | 0.428 [^58] → 0.983 [^59] | 0.721 [^60] → 0.622 [^61] |

두 잣대가 재는 것이 다르다 — P 는 광선을 아예 안 쓰는 팔이라 격자를 얼리면 가까워지고, S 는 경로가 성긴 팔이라 그 성김이 얼리기와 함께 움직인다. 그래서 이 절의 판정은 **순수 PO 를 잣대로 놓았을 때** 성립한다고 적는다.

## 얼리기가 블레이드 대역에 한 일

⭐ 같은 재계산에서 **블레이드 대역 안**의 전력도 중앙값 -13.9 dB [^62] 내려갔다. 그것이 표적의 운동이었다면 얼리기가 물리를 지운 것이므로, 판정을 광선을 안 쓰는 엔진에게 맡겼다 — 블레이드 대역이 순수 PO 보다 +20.3 dB [^63] 위에 있던 것이 +4.0 dB [^64] 로 줄었고 PO 밑으로 내려간 열은 0 [^65] 개다. 같은 대역의 복소 일치는 0.34 [^66] → 0.75 [^67] 로 오르고, 플래시 대조비는 18 [^68]/27 [^69] 열에서 +3.0 dB [^70] 올라간다 — 자의 흔들림이 골짜기를 메우고 있었다는 뜻이다 [^71].

⚠ 대가는 비율 쪽에 있다 — 동체가 든 열에서 «대역밖 ÷ 블레이드 대역» 이 -8.8 dB [^72] 에서 -6.2 dB [^73] 로 올라간다(순수 PO 는 -29.6 dB [^74]). 잉여가 사라진 만큼 남은 잔차가 상대적으로 커 보이는 것이고, 그 잔차가 지금 이 맵의 동적범위 상한이다 [^75].

⭐ **이 편의 맵은 그 얼린 판으로 다시 났다.** 옛 판과의 전후 비교(레벨 · 변조 깊이 · 대역밖 전력 · 가산성)는 6 권의 «광선 격자를 어디에 매나» 절이 낸다. ⚠ 아직 옛 판인 것은 바이스태틱 스윕 원장과 PO 대조 원장이다.

## Sionna 의 빗은 기하에서 온다

잰 팔은 **S** 하나이고, 맞대는 상대는 산란 물리를 뺀 기하 기준이다.

| 기체 | ±1 조화 안 일치 (근거리 1·3 m 칸) | 빗 모양 코사인 중앙값 (15 칸) |
|---|---|---|
| Mini 2 | 9 [^7] / 10 [^8] | 0.9639 [^9] |
| Matrice 4E | 9 [^76] / 10 [^77] | 0.8714 [^10] |

같은 메쉬로 산란 물리를 전부 빼고 왕복 위상만 더한 빗과 Sionna 의 빗을 겹치면 근거리에서 거의 모든 칸이 ±1 조화 안에서 맞는다. 즉 스톡 PathSolver 는 **회전하는 기하의 왕복 위상을 제대로 따라간다** — 그 빗은 몬테카를로 잡음의 산물이 아니다.

## 프롭 정반사가 잡히느냐는 예산이 아니라 앙각이 정한다

⭐ **이 소절의 숫자는 전부 스톡 팔(S) 하나에서 나온다** — 광선 예산 사다리와 경로 인구조사, 그 둘뿐이다.

자세×로터위상 1,152 칸 [^78] 전수에서 프로펠러에 떨어진 정반사 경로는 0 칸 [^79] 이다(**이 인구조사의** 자세 격자 = 앙각 0~60° [^80]).

그 «0» 이 표적의 물리인지 격자의 성질인지 가르려고 축을 **두 개** 따로 열었다 — 광선을 사다리로 올리는 **예산 축**과, 프롭 법선이 향하는 가파른 쪽까지 여는 **앙각 축**이다.

| 기체 | 예산 꼭대기 | 그 예산의 프롭 정반사 | 앙각 85° [^81] 까지 열면 | 무엇이 갈랐나 |
|---|---|---|---|---|
| Matrice 4E | 4,096,000,000 발 [^82] | 0 칸 [^83] | 2 칸 [^84] / 960 칸 [^85] | 앙각 |
| Mini 2 | 4,096,000,000 발 [^86] | 0 칸 [^87] | 0 칸 [^88] / 960 칸 [^89] | 둘 다 아니다 |

⭐ **두 기체가 갈린다.** Matrice 4E 는 광선을 사다리 꼭대기까지 올려도 0 칸인데 가파른 앙각을 열면 프롭 정반사가 나온다 — 그 «0» 은 격자가 프롭 법선 근처를 안 본 결과다. Mini 2 는 예산 꼭대기에서도 앙각 끝에서도 0 칸이라, 이 기체에 대해서만 «스톡 RT 가 프롭 정반사를 못 낸다» 에 가까워진다.

⇒ 그래서 **이 인구조사의** 자세 격자(앙각 0~60° [^80])에서 스톡 팔이 내는 무늬는 전부 흩어져 되돌아오는 반사에서 나오고, 그 값은 광선을 다시 쏠 때마다 흔들린다.

⚠ **이 편의 맵은 다른 격자에 산다** — 앙각 -15° [^18] 한 칸이다. 두 격자는 따로 읽는다.

**위상은 광선 엔진이, 세기는 PO 커널이 맡는** 분업의 근거는 이 절 그림 1·2 의 두 팔 대조이고, 이 인구조사는 거기에 **스톡 팔 쪽 한계**를 보탠다.
