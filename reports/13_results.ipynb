> ### 한 일
> **σ = A(f)·B₁(φ,θ) 에서 A(f) 의 **기울기만** 측정에 맞추고, 레벨을 함께 맞추려면 무엇을 더 골라야 하는지와 그 대가의 크기를 원장에 적었다.**

### 결과
1. 재보정은 밴드별 스칼라 Δσ 하나씩이고, 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다.

4. 앵커가 통제한 것 밖의 항은 셋이다 — 규약 불확도 0.93 dB [^4], 크기전이 항, PO 유효성 항.

## 밴드 비교의 바닥 — 기울기 앵커 σ

σ = A(f)·B₁(φ,θ) 에서 **A(f) 의 기울기**를 Das 측정(IEEE WCL 2026 15:3731)에 맞춘다. **절대 레벨은 우리 PO 출력**이고, 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB [^2] 다.

레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB [^3] 갈리므로, 측정이 그 대가 없이 제약하는 기울기만 받는다.


## 재보정은 밴드별 스칼라 하나씩이다

그래서 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다 — 앵커가 옮기는 것은 레벨이지 모양이 아니다.

앵커가 통제한 것 밖의 항은 **셋**이다 — 규약 불확도 0.93 dB [^4], 크기전이 항, 그리고 **PO 유효성 항**이다([^8]).


## PO 유효성 항이 이번에 크기를 얻었다

PO 오차가 1 dB 아래로 내려가려면 부품의 폭이 파장의 0.729 [^5]배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

그 항의 **부호는 아직 정하지 못한다** — 우리 적분이 편파(전파의 전기장이 흔들리는 방향)를 가르지 않기 때문이다([^9]). 참값은 두 편파에서 서로 다른 값을 주는데 우리 커널은 그 둘을 구분 없이 하나로 내므로, 한쪽 편파를 기준으로 보면 낮고 다른 쪽을 기준으로 보면 높다.

이 사실은 원장에 「VV 측정 대 무편파 커널」 로 기록되어 있다. 이 부는 그 개연성을 결과에 넣지 않고 순위만 든다.

## 절 2. 앵커 σ 위의 R90 은 비교가능 12칸에서 3.69~7.44 km 이고, 밴드 순서는 기체마다 바뀐다



> ### 한 일
> **기울기 앵커의 Δσ 를 R90 근방 국소 지수로 옮겨 다섯 기체 × 세 밴드의 검출거리를 내고, 그 표에서 밴드 순서가 기체마다 어떻게 바뀌는지를 읽었다.**

## 왜 km 열을 순위로 읽는가

공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% [^23] ~ +76.4% [^24] 옮긴다.

그 봉투가 어디서 오는지도 함께 적는다 — 우리 세 밴드는 전부 PO 근사가 1 dB 안에 든다고 보장되는 부품 폭 문턱 아래에 부품을 남긴다(**절 1** «레벨을 맞추려면 크기전이 법칙을 골라야 하므로…»). σ 절대 레벨의 불확도는 선언된 ±10 dB 봉투와 별개로 크기가 아직 정해지지 않은 항을 하나 더 갖는다([^29]).


## 밴드 순서는 자세 인용 방식이 정한다

밴드 순서는 기체마다 바뀐다. 그 순서를 만드는 것은 자세별 로브 구조이고, 앵커는 밴드별 스칼라를 옮기면서 밴드 평균 레벨은 그대로 둔다. 자세를 평균하면 다섯 기체가 한 순위로 모인다.


| 인용 방식 | 서로 다른 순위 수 | 합의 순위 | 최악 뒤집힘 문턱 |
|---|---|---|---|

## 절 4. CPI 를 늘리면 세 파형 모두 블라인드율이 내려간다



> ### 한 일
> **도플러 가드가 헤딩 축을 얼마나 지우는지를 CPI 격자에서 재고, 세 파형의 블라인드 헤딩 비율과 5G 가 치르는 배수를 같은 표에 실었다.**

### 결과
1. 도플러 축을 지우는 기구는 둘이다 — **표본화**: `guard_hz = g*PRF/M >= PRF/2  <=>  M <= 2g  <=>  T_cpi <= 2g/PRF [^71]` 로 가드가 접힘 축 전체를 덮는다(반송파·속도·거리 무관). **진폭**: 짧은 CPI 에서 가드가 도플러 진폭을 덮는다(파형 공통).
2. 5G 의 눈먼 헤딩 비율은 CPI 0.1 s [^72] 에서 0.636 [^73], CPI 0.2 s [^74] 에서 0.303 [^75] 로 내려간다.
3. WiFi 대비 배수는 CPI 전 구간에서 12.1 [^76]~19.0 [^77]배 로 남는다 — 이것이 이 대가를 구조로 만드는 첫 번째 사실이다.
4. 1.5빈 규약에서 LTE 도 CPI ≤ 0.024 s [^78] 에서 전 헤딩 블라인드가 된다 — 5G 만의 성질이 아니라 CPI 가 짧을 때의 성질이다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 가드 규약 | 가드 반폭 = g빈 × PRF/M 이고 g 는 검출기 적용값 1.5빈과 선언값 2.5빈 **둘 다** 잰다 |
| 헤딩 격자 | 헤딩 720점 · 속도 5 m/s 로 스윕한다 — `benchmark/cpi_guard_sweep.py` |
| 두 기구를 가른다 | 표본화가 만드는 접힘 축과 진폭이 만드는 덮임을 분리해 각각의 CPI 의존을 적는다 |

## 도플러 축을 지우는 기구는 둘이다

5G 의 상시 기준신호(SSB)는 20 ms 주기라 PRF 50 Hz 를 준다. 그 축이 지워지는 경로가 둘이다.

**A. 표본화** — `guard_hz = g*PRF/M >= PRF/2  <=>  M <= 2g  <=>  T_cpi <= 2g/PRF [^71]` 로 가드가 접힘 축 전체를 덮는다(반송파·속도·거리 무관).

**B. 진폭** — 짧은 CPI 에서 가드가 도플러 진폭을 덮는다(파형 공통). 1.5빈 규약에서 LTE 도 CPI ≤ 0.024 s [^78] 에서 전 헤딩 블라인드가 된다.


## 기준 CPI 에서의 값

기준 CPI 는 0.1 s [^72] 다.

| 모드 | PRF | 기준 CPI 의 M | 접힘 축 ± | 블라인드(1.5빈) | 블라인드(2.5빈) |
|---|---|---|---|---|---|
| WiFi | 1000 Hz [^81] | 100 [^82] | 500.0 Hz [^83] | 0.028 [^84] | 0.083 [^85] |
| LTE | 1000 Hz [^86] | 100 [^87] | 500.0 Hz [^88] | 0.139 [^89] | 0.250 [^90] |
| 5G | 50 Hz [^91] | 5 [^92] | 25.0 Hz [^93] | 0.639 [^94] | 1.000 [^95] |

5G 의 커버리지 0 은 선언가드 2.5빈 · CPI ≤ 0.10 s [^96] 에서 성립한다. 검출기가 적용하는 1.5빈 규약의 경계는 0.06 s [^97] 다.
