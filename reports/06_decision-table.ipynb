## ⚠ 단서 하나 — 조명 방향이 둘이면 칸이 바뀐다

소거 논증은 표적이 **한 방향에서** 조명될 때의 것이다. 다중경로에서는 직접파와 바닥 반사가 서로 다른 방향에서 동시에 표적을 때리므로 표적 항이 방향마다 달라진다. 커널의 바이스태틱 일반형:  E ∝ Σ_hit |Γ| · e^{jk(û_i+û_s)·p} · d² [^4] 가 보여주듯 각 조명 방향마다 다른 E 가 나온다. 그때는 오른쪽 칸의 실험도 왼쪽으로 이사한다.

⭐ 이 표를 한 사례로 시험한 것이 [부 7 «마이크로도플러»](../README.md#부-7-마이크로도플러) 다 — 도는 로터는 비율만 필요하므로 오른쪽 칸에 앉는다.
