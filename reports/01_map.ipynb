# 리포트 1 — 이 연구가 묻는 것과 답한 방식

> 패시브 바이스태틱으로 드론을 **탐지하고 마이크로도플러로 분류**하는 것이 태스크이고, RCS 는 그 인프라다. 이 권은 나머지 열네 권의 지도다.

이 권은 아래 절로 이루어진다. 각 절은 **한 일 · 결과 · 방법 · 재현**을 자기 앞에 달고 있어, 필요한 절만 따로 읽어도 된다.

| 절 | 무엇을 말하나 |
|---|---|
| 1 | 읽는 목적이 셋이면 읽는 순서도 셋이다 |
| 2 | 네 관문을 동시에 통과한 게재본은 0편이고, 최근접 3편은 각각 다른 관문에서 걸린다 |
| 3 | 주장마다 판정 범위를 결판·사슬확인·캠페인 밖으로 적었다 |

⭐ 숫자는 전부 계산 결과 JSON(원장)에서 주입된다 — 각주가 그 출처다. 원장이 다시 계산되면 빌더를 돌리는 것만으로 본문 숫자가 따라 바뀐다.

## 절 1. 읽는 목적이 셋이면 읽는 순서도 셋이다



> ### 한 일
> **편 78개를 부 12개로 묶고, 읽는 목적마다 다른 진입 경로 셋을 깔았다.**

### 결과
1. 편은 78 개 [^1] 이고 부는 12 개 [^2] 다 — 한 편이 중심 메시지 하나를 들고, 편 제목이 곧 그 편의 결론 문장이다.
2. 재구성 전은 편 8개였고 그중 7개를 진단했다 — 화면 글자의 37.9% [^3] 가 본문이 아니라 출처 태그였고, 태그는 1,421 개 [^4] 였다.
3. 그래서 출처를 각주로 뺐다 — 태그를 지우면 문장이 183 [^5] → 76 자 [^6] 였고, 같은 (파일, 키) 재인용 454 회 [^7] 가 번호 하나로 모인다.
4. 경로는 셋이다 — **빨리 훑기**(부마다 결론 편 하나씩) · **왜 믿을 수 있나**(검증·대조·반증 편만) · **다시 돌리기**(리포트를 안 읽고 명령만 본다).
5. 논문 문장과 재현 절차는 리포트 밖에 산다 — [`docs/paper/`](../docs/paper/README.md) 와 [`docs/REPRODUCE.md`](../docs/REPRODUCE.md) 다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 편 하나 = 메시지 하나 | 제목이 «…다» 로 끝나는 평서문이고, 그 문장이 그 편이 판 결론이다. 제목만 읽어도 목차가 된다 |
| 부 = 물음 하나 | 부마다 답하는 질문이 하나다. 그 질문에 답하는 편들이 그 부에 모여 있다 |
| 색인 | 편마다 «어느 JSON 을 읽고 어느 명령으로 다시 나오나» 를 `outputs/reports_index.json` 에 기계용으로 적었다 — 각 편의 여는 블록에서 직접 읽어 만든다 |
| 경로 | 읽는 목적이 다르면 순서도 다르다. 세 경로가 같은 78편을 서로 다르게 가로지른다 |

## 이 편들을 읽을 때 알아 두면 되는 규약 다섯

| 규약 | 무슨 뜻인가 |
|---|---|
| 제목이 결론이다 | 편 제목은 «…다» 로 끝나는 평서문이고, 그 편이 판 결론 그 자체다. 물음표로 끝나는 제목은 규약이 막는다 |
| 숫자 옆의 `[^n]` | 그 숫자가 온 JSON 파일과 키다. 편 끝 «출처» 표에 파일·키·값이 있고, 그 값은 표를 만들 때 JSON 을 다시 열어 채운 것이다 |
| 여는 블록 넷 | 한 일 · 결과 · 방법 · 재현. **결과만 읽어도** 그 편의 답을 알 수 있게 쓴다 |
| 마지막 절은 «다음 단계» | 한계 목록이 아니라 **앞을 보는 행동**이다. 가운데 칸이 «그러면 결정되는 것» 이고, 그 칸이 빈 행은 규약이 막는다 |
| ⚠ 와 ⭐ | ⚠ 는 그 숫자가 서 있는 전제(사슬 세대·메쉬 판)를, ⭐ 는 그 절의 핵심 판단을 가리킨다 |

## 절 2. 네 관문을 동시에 통과한 게재본은 0편이고, 최근접 3편은 각각 다른 관문에서 걸린다



> ### 한 일
> **후보를 네 관문으로 하나씩 판정하고, 우리 파이프라인이 각 축에서 한 일을 값과 함께 같은 표에 올렸다.**

### 결과
1. 후보 12 편 [^8] 을 네 관문으로 판정해 동시 통과는 0 편 [^9] 이다.
2. 가장 가까운 셋은 각각 **다른** 관문에서 걸린다 — 그래서 «한 편만 넘으면 된다» 는 요약이 성립하지 않는다.
3. 우리 자세 패턴은 해석 PO 구 대비 0.201 dB [^10] (입사 48 방향 [^11])이고, 밴드 기울기는 Das 적합계수 0.21 dB/GHz [^12] 에 맞춘다.
4. 실측 전이까지 보는 축으로 좁히면 모집단이 달라진다 — 정독한 33편(아카이브 전문검색 313편 범위) 중 0편. 세 요소를 다 갖추고 '실측 RF 데이터에서 분류 성능'까지 보고한 드론 논문은 찾지 못함. [^13]
5. 레벨 규약은 출처 논문 안에서 갈린다 — 잔차 최소화 변환을 쓰면 2.51 dB [^14] 만큼 밝다.

## 판정 범위를 세 갈래로 적는다

네 번째 열이 **판정 범위**다 — `결판`(이 캠페인이 정한다) · `사슬 확인`(설계값과 대조한다) · `이 캠페인 밖`(다음 라운드나 통제 시뮬이 정한다). 세 번째 값을 그대로 적는 것이 이 표의 핵심이다.

표에 나오는 **편파**는 전파의 전기장이 흔들리는 방향을 말한다 — 세워서 보내고 세워서 받는 조합을 VV, 눕혀 보내고 눕혀 받는 조합을 HH, 한쪽만 눕힌 조합을 VH·HV 라 적는다. 우리 커널은 그 방향을 가르지 않고 세기 값 하나만 내므로 **무편파 스칼라**라고 부른다.
