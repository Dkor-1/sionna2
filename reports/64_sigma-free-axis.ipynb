> ### 한 일
> **같은 표적·같은 기하·같은 검출기에서 Pd = 0.5 에 필요한 출력 SNR 을 기준신호 대역과 점유 등급으로만 갈라, σ 와 무관하게 정해지는 순서를 냈다.**

### 결과
1. 상시 기준신호만 쓰는 제약의 대가는 5G 에서 +3.82 dB ⟨outputs/report05_derived.json : always_on_cost.G.cost_db⟩ 이고, WiFi 는 +0.21 dB ⟨outputs/report05_derived.json : always_on_cost.W.cost_db⟩ · LTE 는 +0.08 dB ⟨outputs/report05_derived.json : always_on_cost.L.cost_db⟩ 다.
2. 그 대가는 거리분해능으로도 읽힌다 — 5G 는 41.64 m ⟨outputs/report05_derived.json : always_on_cost.G.dr_g1_m⟩ 대 3.05 m ⟨outputs/report05_derived.json : always_on_cost.G.dr_g3_m⟩ 다.
3. 이 스윕은 단일 반송파 3.5 GHz ⟨outputs/report05_derived.json : bench.fc_ghz⟩ · 바이스태틱 거리 22.3 m ⟨outputs/report05_derived.json : bench.Rb_m⟩ · 고정 σ -27.58 dBsm ⟨outputs/report05_derived.json : bench.sigma_dbsm⟩ 라 표적·기하·σ 가 한 값에 묶여 있다.
4. 그래서 여기서 읽는 것은 파형 축 하나의 상대 비교다 — 절대 거리는 mavic4pro ⟨outputs/report05_derived.json : bench.drone⟩ 한 기체의 벤치 배치에서 나온다.

## σ 를 곱하기 전 축 — 기준신호 대역과 점유 등급

같은 표적·같은 기하·같은 검출기에서 Pd = 0.5 에 필요한 출력 SNR 은 기준신호 대역과 프레임 수가 정한다. 이 축은 σ 와 무관하게 세 파형의 순서를 정한다.

| 표준 | 상시 기준(등급 1) | 세션 기준(등급 3) | 상시 제약의 대가 | 거리분해능 대비 |
|---|---|---|---|---|
| WiFi | 11.87 dB ⟨outputs/report05_derived.json : always_on_cost.W.snr50_g1⟩ | 11.66 dB ⟨outputs/report05_derived.json : always_on_cost.W.snr50_g3⟩ | +0.21 dB ⟨outputs/report05_derived.json : always_on_cost.W.cost_db⟩ | 3.92 m ⟨outputs/report05_derived.json : always_on_cost.W.dr_g1_m⟩ ↔ 3.92 m ⟨outputs/report05_derived.json : always_on_cost.W.dr_g3_m⟩ |
| LTE | 13.76 dB ⟨outputs/report05_derived.json : always_on_cost.L.snr50_g1⟩ | 13.68 dB ⟨outputs/report05_derived.json : always_on_cost.L.snr50_g3⟩ | +0.08 dB ⟨outputs/report05_derived.json : always_on_cost.L.cost_db⟩ | 16.67 m ⟨outputs/report05_derived.json : always_on_cost.L.dr_g1_m⟩ ↔ 16.64 m ⟨outputs/report05_derived.json : always_on_cost.L.dr_g3_m⟩ |
| 5G NR | 15.03 dB ⟨outputs/report05_derived.json : always_on_cost.G.snr50_g1⟩ | 11.21 dB ⟨outputs/report05_derived.json : always_on_cost.G.snr50_g3⟩ | +3.82 dB ⟨outputs/report05_derived.json : always_on_cost.G.cost_db⟩ | 41.64 m ⟨outputs/report05_derived.json : always_on_cost.G.dr_g1_m⟩ ↔ 3.05 m ⟨outputs/report05_derived.json : always_on_cost.G.dr_g3_m⟩ |

이 스윕은 거리 표와 **다른 배치**에서 돈다 — X410 벤치(`src/experiment_x410.py:101`), 단일 반송파 3.5 GHz ⟨outputs/report05_derived.json : bench.fc_ghz⟩ 를 9모드 전부에 쓰고, 바이스태틱 거리 22.3 m ⟨outputs/report05_derived.json : bench.Rb_m⟩, 고정 σ -27.58 dBsm ⟨outputs/report05_derived.json : bench.sigma_dbsm⟩ 다. 표적·기하·σ 를 한 값으로 묶었으므로 여기서 읽는 것은 **파형 축 하나**의 상대 비교다.
