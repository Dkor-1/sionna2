# 리포트 30 — 사다리가 하나가 아니라 셋이었다 — 여섯 단이 서로 다른 운동학을 쓰고 있었다

> ### 한 일
> **표적을 점점 거칠게 바꾸는 사다리 여섯 단을 돌린 뒤 단마다 «무엇이 도는가» 를 세어 사다리를 세 벌로 갈랐다.**

### 결과
1. 여섯 단이 같은 운동학을 쓰지 않았다 — 구·정육면체·상자 단은 기체 전체를 덩어리로 바꿔 통째로 돌리고, 진짜 드론 단은 몸통이 서고 프로펠러만 돈다.
2. 그래서 가장 인용하기 좋은 «정육면체 대 메쉬» 차이 +26.05 dB [^1] 안에서 운동학 몫이 +23.42 dB [^2], 형상 몫이 -2.85 dB [^3] 다.
3. 운동학 비중은 Mini 2 74% [^4] · Matrice 4E 55% [^5] — 형상 몫보다 크다.
4. 세 몫의 합은 정의상 총 차이와 닫힌다 — 닫힘오차 0.00 dB [^6].
5. 그래서 사다리를 A(있는 그대로) · B(온몸 자전 고정) · C(진짜 비행 고정) 세 벌로 갈라 적는다. 답할 자격은 C 에만 있고 그 편이 [편 31 «교정 사다리»](31_ladder-calibrated.ipynb) 다.

## 무엇을 물었나

드론이 제자리에 떠 있어도 프로펠러가 돌기 때문에 되돌아오는 전파의 세기와 위상이 시간에 따라 흔들린다. 이 흔들림을 **마이크로도플러**라고 부른다 — 표적 전체가 움직여서 생기는 도플러가 아니라 표적의 *일부*(날개)가 움직여서 생기는 도플러라는 뜻이다.

지도교수의 지적은 «드론 RCS 정밀도는 연구 값어치가 없다» 이고, 절대 세기 축에서는 우리 데이터가 그 지적을 뒷받침한다. 이 라운드는 그 지적이 마이크로도플러에서도 성립하는지를 물었다.

방법은 **사다리**다 — 표적을 점점 거칠게 바꿔 가며 신호가 언제 무너지는지 본다.

## 사다리를 돌려 보니 여섯 단이 운동학을 섞어 쓰고 있었다

구·정육면체·상자 단은 기체 전체를 덩어리 하나로 바꿔 **통째로 돌린** 물체이고, 진짜 드론은 몸통이 서 있고 프로펠러만 돈다. 두 개를 나란히 놓고 «형상 차이» 라고 부르면 그 안에 형상 교체와 운동학 교체가 섞인다.

| 사다리 | 무엇이 도나 | 무엇이 바뀌나 | 질문에 답할 자격 |
|---|---|---|---|
| A. 있는 그대로 | 단마다 다름 | 형상 + 운동학 + 재질 | ❌ 섞여 있다 |
| B. 온몸 자전 고정 | 전부 온몸 자전 | 형상만 | △ 실제 드론이 아님 |
| C. ⭐ 진짜 비행 고정 | 전부 프로펠러만 | **프로펠러 형상만** | ✅ 이 축이 답이다 |

## 있는 그대로의 사다리 — 단마다 무엇이 도는가

| 표적 | 무엇으로 바꿨나 | 무엇이 도나 | 인용 자격 |
|---|---|---|---|
| sphere_eqvol | equal-volume sphere | whole body spins | 아니오 |
| cube_eqvol | equal-volume cube | whole body spins | 예 |
| box_bbox | bounding box | whole body spins | 예 |
| mesh_no_rotor | CAD body, rotors deleted | nothing moves | 아니오 |
| mesh_half_tri | CAD, triangles halved | rotors only | 예 |
| mesh_full | CAD, full resolution (reference) | rotors only | 예 |

출처 [^10]

D 단(`mesh_no_rotor`)은 도는 부품이 아예 없다. 변조 전력이 «작다» 가 아니라 정확히 0 이라([^11]) 네 지표 중 셋은 값 자체가 존재하지 않는다 — 아래 표에서 그 단을 뺐다.

**지표 넷이 무엇을 뜻하나** — 플래시 대조비: 날개가 시선에 수직으로 설 때의 번쩍임이 바닥보다 몇 dB 위인가. 풍부도 n_eff: 실질적으로 몇 개의 배음이 살아 있나. 폭: 도플러가 날개 끝 속도가 예측하는 만큼 넓게 퍼지나(1.0 이면 딱 맞음). 검출 단면적: 호버 표적이 실제로 쓸 수 있는 «가장 센 배음 한 줄의 RCS».

**그림 1.** 빨간 사다리와 초록 사다리는 같은 단에서 왜 다른 값을 내는가?
빨간 점선이 있는 그대로의 사다리(운동학이 단마다 바뀐다), 초록 실선이 프로펠러만 도는 교정 사다리다. 속 빈 표식은 «인용 자격 없음» — 변조 전력의 절반 이상이 운동학적으로 불가능한 자리에 있다는 뜻이다. E·F 단은 두 사다리에서 같은 팔이라 빨간 점이 초록 점 아래에 정확히 겹쳐 있다.

## 운동학 몫이 형상 몫보다 크다

| 기체 | 대리 형상 | 총 차이 | 운동학 몫 | 형상 몫 | 운동학 비중 |
|---|---|---|---|---|---|
| Mini 2 | 등가부피 정육면체 | +26.05 dB [^1] | +23.42 dB [^2] | -2.85 dB [^3] | 74% [^4] |
| Mini 2 | 외접 직육면체 | +34.41 dB [^12] | +23.42 dB [^13] | +5.50 dB [^14] | 68% [^15] |
| Matrice 4E | 등가부피 정육면체 | +15.11 dB [^16] | +23.30 dB [^17] | -13.71 dB [^18] | 55% [^5] |
| Matrice 4E | 외접 직육면체 | +17.89 dB [^19] | +23.30 dB [^20] | -10.93 dB [^21] | 59% [^22] |

즉 «정육면체가 틀린 이유» 는 대부분 «모양이 거칠어서» 가 아니라 **«무엇이 도는지를 틀리게 놓아서»** 다. 운동학이 같으면서 날개 디테일만 없앤 유일한 팔인 `slab` 의 차이는 -1.25 dB [^23] · -5.54 dB [^24] 로 훨씬 작다.

⚠ 같은 «정육면체 대 메쉬» 라도 무엇을 재느냐에 따라 값이 다르다. 위 표는 절대 AC 전력의 비이고, AC 를 DC(동체 반사)로 나눈 «상대 변조 깊이» 로 재면 +21.76 dB [^25] · +26.56 dB [^26] 가 된다. 프리미티브는 동체 반사도 함께 바꾸기 때문이다.
