# 리포트 59 — 레벨을 맞추려면 크기전이 법칙을 골라야 하므로 기울기만 받는다

> ### 한 일
> **σ = A(f)·B₁(φ,θ) 에서 A(f) 의 주파수 기울기만 측정에 맞추고, 레벨과 자세 패턴은 우리 PO 출력으로 둔 원장 위에 밴드 비교를 세웠다.**

### 결과
1. 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB ⟨outputs/report05_derived.json : anchor_scope.level_shift_abs_max_db⟩ 다 — 레벨은 우리 PO 출력 그대로다.
2. 레벨까지 맞추려면 크기전이 법칙을 하나 골라야 하고, L² 와 L⁴ 가 기체에 따라 최대 9.50 dB ⟨outputs/report05_derived.json : anchor_scope.size_law_spread_max_db⟩ 갈린다 — 측정이 그 대가 없이 제약하는 양이 기울기뿐이라 기울기만 받는다.
3. 재보정은 밴드별 스칼라 Δσ 하나씩이고, 정규화 각도 패턴은 9.6e-16 dB ⟨outputs/sigma_anchor.json : drones.phantom4.shape_invariance_max_abs_db⟩ 안에서 그대로 남는다.
4. 앵커가 통제한 것 밖의 항은 셋이다 — 규약 불확도 0.93 dB ⟨outputs/sigma_anchor.json : uncontrolled[1].size_db⟩, 크기전이 항, 그리고 PO 유효성 항이다.
5. PO 오차가 1 dB 아래로 내려가려면 부품 폭이 파장의 0.729 ⟨outputs/lowfreq_anchor.json : thin_plate.truth_2d_mom_fine_width_grid.knee_a_over_lam⟩ 배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

## 밴드 비교의 바닥 — 기울기 앵커 σ

σ = A(f)·B₁(φ,θ) 에서 **A(f) 의 기울기**를 Das 측정(IEEE WCL 2026 15:3731)에 맞춘다. **절대 레벨은 우리 PO 출력**이고, 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB ⟨outputs/report05_derived.json : anchor_scope.level_shift_abs_max_db⟩ 다. 레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB ⟨outputs/report05_derived.json : anchor_scope.size_law_spread_max_db⟩ 갈리므로, 측정이 그 대가 없이 제약하는 기울기만 받는다.

재보정은 밴드별 스칼라(방향 구분 없이 값 하나) Δσ 하나씩이고, 정규화 각도 패턴은 9.6e-16 dB ⟨outputs/sigma_anchor.json : drones.phantom4.shape_invariance_max_abs_db⟩ 안에서 그대로 남는다. 앵커가 통제한 것 밖의 항은 **셋**이다 — 규약 불확도 0.93 dB ⟨outputs/sigma_anchor.json : uncontrolled[1].size_db⟩, 크기전이 항, 그리고 **PO 유효성 항**이다(⟨outputs/sigma_anchor.json : uncontrolled⟩). 세 번째가 크기를 얻었다 — PO 오차가 1 dB 아래로 내려가려면 부품의 폭이 파장의 0.729 ⟨outputs/lowfreq_anchor.json : thin_plate.truth_2d_mom_fine_width_grid.knee_a_over_lam⟩ 배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다([편 22 «PO 유효 무릎»](22_po-knee.ipynb)).

그 항의 **부호는 이 부에서 정하지 않는다** — 우리 적분이 편파(전파의 전기장이 흔들리는 방향)를 가르지 않기 때문이다(⟨outputs/lowfreq_attack.json : q5_blast_radius.po_validity_blast_radius_the_real_one⟩). 참값은 두 편파에서 서로 다른 값을 주는데 우리 커널은 그 둘을 구분 없이 하나로 내므로, 한쪽 편파를 기준으로 보면 낮고 다른 쪽을 기준으로 보면 높다. 두 쪽 중 어느 쪽이 더 그럴듯한지는 [편 23 «커널의 열린 항목»](23_kernel-open-items.ipynb) 이 두 편파의 크기 차이로 적는다 — 이 편은 그 개연성을 결과에 넣지 않고 순위만 든다. 이 사실은 원장에 「VV 측정 대 무편파 커널」 로 기록되어 있다.

⚠ 이 절이 든 앵커 수(정규화 잔차 · 아래 Δσ 표)와 거리표는 `outputs/sigma_anchor.json` 에서 오고, 그 사슬은 2026-08-04 ⟨outputs/meshfix_applied.json : _meta.date⟩ 형상 정정 전 메쉬 위에 서 있다 — 다섯 기체 중 Matrice 4E 가 그 정정을 받았다([편 16 «형상 정정»](16_mesh-vs-real.ipynb)) ⟨outputs/meshfix_attack.json : Q6_invalidated_outputs.critical[3]⟩.
