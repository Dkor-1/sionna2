# 리포트 14 — 선행에서 빌린 절차와 아직 안 빌린 절차를 비용과 함께 센다

> ### 한 일
> **선행에서 이미 빌린 절차와 아직 안 빌린 절차를 순위·비용과 함께 세고, 의도적으로 안 빌린 셋에는 그 이유를 붙였다.**

### 결과
1. 이미 빌린 절차는 5가지이고, 항목마다 «어디서 빌렸나» 와 «그것이 사 준 것» 을 같은 행에 적었다.
2. 아직 안 빌린 절차는 7가지이고 순위와 비용이 붙어 있다 — 맨 위 두 줄이 이 편의 다음 단계와 같다.
3. 의도적으로 안 빌린 것은 3가지다. «Sionna stock diffraction=True 를 표적에 켜기 [^1]» 가 그중 하나다.
4. 밴드 기울기 앵커 0.21 dB/GHz [^2] 는 빌린 쪽의 대표 항목이고, 그 값이 우리 σ 의 주파수 축을 정한다.

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 빌린 것의 정의 | 절차·규약·값 중 선행 문헌에서 그대로 가져온 것만 센다 — 영감이나 문제의식은 세지 않는다 |
| 순위와 비용 | 안 빌린 항목마다 그것을 빌이는 데 드는 비용을 함께 적는다 — 순위가 곧 다음 단계의 순서다 |
| 안 빌린 이유 | 도구가 다르면 그 이유를 문장으로 적는다. «필요 없다» 로 끝내지 않는다 |

## 의도적으로 안 빌린 것

맨 위 두 줄이 이 편의 다음 단계와 같다 — 유효성 바닥을 기체 × 밴드 표로 인쇄하는 일과, 정준체를 캠페인 패드에 올려 절대 σ 에 처음으로 실측 검사를 붙이는 일이다.

의도적으로 안 빌린 것도 3가지다. «Sionna stock diffraction=True 를 표적에 켜기 [^1]» 가 그중 하나이고, 이유는 도구가 다르다는 것이다: D 는 σ 가 아니라 경로계수를 만들고, wedge 각을 인접 두 면의 법선에서 읽으므로 면분할된 셸에서 그 각은 기체의 성질이 아니라 **우리 메싱의 성질**이 된다.

⭐ 그래서 이 표는 «빚 목록» 이 아니라 **다음 라운드의 작업 순서**다. 같은 목록이 [편 23 «커널이 아직 못 하는 것은 편파·PTD·재테셀레이션 셋이고, 각각의 크기를 적었다»](23_kernel-open-items.ipynb) 의 열린 항목과 짝을 이룬다.
