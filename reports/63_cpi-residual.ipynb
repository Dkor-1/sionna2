# 리포트 63 — 모호속도는 표본화율의 성질이라 CPI 로 바뀌지 않는다

> ### 한 일
> **세 파형의 모호속도를 표본화율에서 직접 내고, 커버리지를 WiFi·LTE 수준으로 끌어올리는 데 드는 CPI 와 그 CPI 가 무는 재방문 시간을 함께 적었다.**

### 결과
1. 모호속도는 5G 1.07 m/s ⟨outputs/cpi_guard_sweep.json : unambiguous_speed.G1.v_unambiguous_ms⟩ · WiFi 14.40 m/s ⟨outputs/cpi_guard_sweep.json : unambiguous_speed.W1.v_unambiguous_ms⟩ · LTE 40.74 m/s ⟨outputs/cpi_guard_sweep.json : unambiguous_speed.L1.v_unambiguous_ms⟩ 다.
2. 커버리지를 WiFi 수준으로 올리는 CPI 는 1.00 s ⟨outputs/cpi_guard_sweep.json : cost_of_long_cpi.required_cpi_s.to_WiFi_parity⟩, LTE 수준은 0.38 s ⟨outputs/cpi_guard_sweep.json : cost_of_long_cpi.required_cpi_s.to_LTE_parity⟩ 이고 그 대가는 재방문 시간이다.
