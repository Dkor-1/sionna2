# 리포트 63 — 모호속도는 표본화율의 성질이라 CPI 와 무관한 상한이다

> ### 한 일
> **CPI 를 늘려 커버리지를 회복하는 길의 끝을 재고, 그 길이 무모호 속도와 코히어런스 한계에서 어디서 막히는지를 값으로 적었다.**

### 결과
1. 모호속도는 5G 1.07 m/s [^1] · WiFi 14.40 m/s [^2] · LTE 40.74 m/s [^3] 이고, CPI 를 늘려도 그대로다.
2. 커버리지를 WiFi 수준으로 올리는 CPI 는 1.00 s [^4], LTE 수준은 0.38 s [^5] 이고 그 대가는 재방문 시간이다.
3. 그 CPI 가 코히어런스 한계 안에 머무는 구간은 20 m/s [^6] 까지다 — 25 m/s [^7] 에서 WiFi 패리티 CPI 가 한계 (0.84 s [^8]) 를 넘어선다.
4. 거리·속도 격자 16 [^9]칸 중 8 [^10]칸이 WiFi 패리티를 허용한다.

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 모호속도 | PRF 와 λ 가 정한다. 적분시간을 늘려도 표본화율은 그대로이므로 이 상한은 안 움직인다 |
| 패리티 CPI | 5G 의 블라인드율을 다른 파형 수준으로 내리는 데 필요한 CPI 를 스윕에서 읽는다 |
| 코히어런스 한계 | 표적이 움직이는 동안 위상이 유지되는 시간이다. 패리티 CPI 가 그 한계를 넘으면 코히어런트 적분이 성립하지 않는다 |
| 대가 열 | 필요 CPI · SSB 버스트 수 · 헤드라인 대비 경과 · 거리워크 · 코히어런트 이득을 같은 표에 실어 대가의 형태를 갈라 놓는다 |

## 세 번째 사실이 결정적이다

모호속도는 표본화율의 성질이라 CPI 로 바뀌지 않는다 — 5G 1.07 m/s [^1] · WiFi 14.40 m/s [^2] · LTE 40.74 m/s [^3].

커버리지를 WiFi 수준으로 올리는 CPI 는 1.00 s [^4], LTE 수준은 0.38 s [^5] 이고 그 대가는 재방문 시간이다.

## 패리티의 대가

| 패리티 목표 (5 m/s) | 필요 CPI | SSB 버스트 | 헤드라인 대비 경과 | 거리워크 | 코히어런트 이득 |
|---|---|---|---|---|---|
| LTE 수준 | 0.38 s [^13] | 19 [^14] | 3.75 [^15]배 | 0.064 [^16]빈 | 5.75 dB [^17] |
| WiFi 수준 | 1.00 s [^18] | 50 [^19] | 10.00 [^20]배 | 0.170 [^21]빈 | 10.00 dB [^22] |

## 그 길이 막히는 자리

패리티 CPI 가 코히어런스 한계 안에 머무는 구간은 20 m/s [^6] 까지이고, 25 m/s [^7] 에서 WiFi 패리티 CPI 가 한계 (0.84 s [^8]) 를 넘어선다.

거리·속도 격자 16 [^9]칸 중 8 [^10]칸이 WiFi 패리티를 허용한다 — 그 밖의 칸에서는 CPI 로 메우는 길이 닫힌다.
