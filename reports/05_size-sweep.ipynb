# 리포트 05 — 면적을 1600배로 키워도 경로 진폭은 7.4e-07 dB 움직인다

> ### 한 일
> **빈 자유공간에 금속 평판 하나를 두고 변만 키우며 PO 단면적과 path solver 의 표적 경로 진폭을 같은 축에서 재고, 같은 실험을 드론 메쉬로 옮겼다.**

### 결과
1. 평판 면적을 1600 [^1]배 키우면 PO 단면적은 64.08 dB [^2] 커지고, path solver 의 표적 경로 진폭은 7.4e-07 dB [^3] 움직인다.
2. 경로 수는 전 구간 1 [^4]개이고, 그 진폭은 이미지-소스 해석해와 0.0017 dB [^5] 안에서 맞는다 — **엔진은 자기가 푸는 문제를 정확히 푼다**.
3. 같은 실험을 기체 메쉬로 옮기면 삼각형을 1.82 [^6] decade 깎는 동안 진폭이 총 49.02 dB [^7] 무너지고, 면 2→1 계단이 닫힌형 20·log₁₀(1/2) 에 붙는다.
4. 반대 방향도 같은 뿌리다 — 같은 평판을 쪼개기만 해도 코히어런트 전력이 9.54 dB [^8] 부푼다. 답의 단위가 면적이 아니라 경로라서 면 수가 곧 답이 된다.

## 평판 하나를 키워 보면

빈 자유공간에 금속 평판 하나를 두고 변만 0.1 m [^10] → 4.0 m [^11] (면적 1600 [^1]배) 로 키운다. PO 단면적은 면적 dB 당 2.00 dB [^12]씩, 총 64.08 dB [^2] 오른다.

같은 구간에서 path solver 의 표적 경로 진폭은 7.4e-07 dB [^3] 움직이고 경로 수는 내내 1 [^4]개다 — 예산 4단 · 시드 5개 · 깊이 2단 24 [^9]셀 전부에서.

## 그 진폭이 무엇인지도 같이 확인된다

![report00_f2](../outputs/figures/report00_f2.png)

**그림 1.** 표적을 키우면 무엇이 움직이고 무엇이 그대로인가?

⭐ RT 값은 이미지-소스 해석해와 0.0017 dB [^5] 안에서 맞는다. **엔진은 자기가 푸는 문제를 정확히 푼다.** 두 곡선이 만나는 자리는 변 0.93 m [^13] 한 점뿐이고, 그것은 우연이다.

## 드론 메쉬에서는 정반사 경로가 자세 하나에서만 살아남는다

같은 실험을 기체 메쉬로 옮긴다. 삼각형을 29,932 [^14]개 → 450 [^15]개(1.82 [^6] decade)로 깎아도 실루엣은 유지된다. 그런데 36 [^16]자세 중 정반사 경로가 존재하는 자세는 1 [^17]개다.

그 한 자세에서 진폭은 기여 면 개수를 따라 계단으로 떨어져 총 49.02 dB [^7] 무너진다 — 면 2→1 계단이 -6.05 dB [^18] 로 닫힌형 20·log₁₀(1/2) = -6.02 dB [^19] 에 붙는다.

⚠ 나머지 자세가 빈 이유는 따로 있다. 같은 자세에 확산을 켜면 표적경유 경로가 자세당 102 [^20]개 넘게 잡힌다. 비어 있는 것은 광선이 아니라 **거울 조건을 만족하는 삼각형**이다.

## 반대 방향도 같은 뿌리 — 쪼개기만 해도 답이 부푼다

같은 1 m [^21] 평판을 2 [^22]개 → 512 [^23]개 삼각형으로 **쪼개기만** 해도 코히어런트 전력이 9.54 dB [^8] 부푼다. 늘어난 경로들은 진폭 산포 0.0 dB [^24] · 지연 산포 0.0 ns [^25] · 위상 산포 0.0° [^26] 인 **완전한 복사본**이고, 합은 20·log₁₀(N) 을 0.0017 dB [^27] 안에서 따른다. 격자 위치를 옮겨도 중복은 남는다 (`offset_removes_duplication` = 아니오 [^28]).

메쉬를 깎으면 무너지고 쪼개면 부푼다. 두 방향이 같은 뿌리에서 나온다 — **답의 단위가 면적이 아니라 경로라서** 면 수가 곧 답이 된다.
