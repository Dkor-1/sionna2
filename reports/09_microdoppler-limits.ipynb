## 절 1. 지상 레이더는 기체를 아래에서 보므로 가림이 무는 자세가 우리 자세다



> ### 한 일
> **같은 기체를 코 쪽·배 쪽·배 옆 세 자세에서 돌려 가림과 변조가 자세에 따라 어떻게 움직이는지 쟀다.**

### 결과
1. 프로펠러는 동체 **위**에 있다 — 위에서 내려다보는 자세에서는 날개가 통째로 드러나고, 아래에서 보는 자세에서 가림이 문다.
2. 코 쪽(앙각 15 도 [^1])의 가림 효과는 변조 깊이 +0.43 dB [^2] · 레벨 -0.07 dB [^3] 다.
3. 배 쪽(앙각 -15 도 [^4])에서는 -25.42 dB [^5] · +1.04 dB [^6] 로 바뀐다.
4. 배 옆(방위 90 도 [^7])에서는 -2.29 dB [^8] · -0.92 dB [^9] 다.
5. 자유공간 바이스태틱의 이등분선 앙각은 전 구간 음수라 우리는 드론 배를 본다 — 자세 스윕이 그 문장에 독립적인 근거를 붙였다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 자세 세 칸 | 코 쪽(앙각 +15°) · 배 쪽(앙각 −15°) · 배 옆(방위 90° · 앙각 −15°). 기체 좌표계에서 앙각이 음수면 아래에서 보는 것이다 |
| 가림 단일축 | 칸마다 «동체가 막느냐» 만 다른 두 팔을 돌려 그 자세의 가림 효과를 낸다 |
| 왜 배 쪽이 헤드라인인가 | 지상 레이더는 비행 중인 기체를 아래에서 본다 — 자유공간 바이스태틱 기하가 이미 그렇게 적혀 있다 |
| Mini 5 Pro | 프로펠러와 모터 벨이 겹친 삼각형이 남아 있어 헤드라인에서 뺐다. 같은 그림을 그 기체로도 그려 따로 뒀다 |

## 절 2. 판정 잣대를 널 팔 15 칸과 이상 점산란자로 먼저 교정했다



> ### 한 일
> **변조가 나오면 안 되는 물체와 이상 점산란자를 같은 측정법에 넣어 판정 문턱이 거짓양성과 거짓음성을 내는지 먼저 쟀다.**

### 결과
1. 널 팔 15 칸 [^13] 에서 변조 판정이 0 번 [^14] 켜졌다 — 최댓값 3.83 dB [^15] 가 문턱 6.0 dB [^16] 아래에 든다.
2. 신호 팔은 6 칸 [^17] 중 5 칸 [^18] 이 켜진다 — 문턱이 양쪽에서 작동한다.
3. 이상 점산란자는 가장자리 시험을 9 [^19]/10 [^20] 칸에서 통과한다 — 그 시험이 교정됐는지는 아니오 [^21] 로 적혀 있다.
4. 널 팔 전용 재계산에서도 같은 결론이다 — 신호 대 인공물 여유 3.85 dB [^22], 비 77.4 배 [^23].
5. ⚠ 적대검증이 널 셈을 정정했다 — 실제로 시험된 널은 12 칸 [^24] 이고 나머지 3 칸 [^25] 은 AC 가 정확히 0 이라 시험 자체가 성립하지 않는 칸이다.


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 널 설계 다섯 갈래 | 메쉬 완전 동결 · 프로펠러 제거 · 회전대칭 원판 · 등가부피 구 z 회전 · 같은 구 텀블 — 각 널이 서로 다른 고장모드를 막는다 |
| 양성 대조 | 기체 전체를 돌린 팔을 함께 넣어 «켜져야 할 때 켜지는지» 를 확인했다 |
| 바닥 정규화 | 널의 재추적 잡음이 신호팔보다 조용하다는 반론에 대해, 바닥을 σ_signal/σ_null 로 환산해 보수적으로 다시 잡았다 |
| 이상 점산란자 | 같은 메쉬로 산란 물리를 빼고 왕복 위상만 더한 기준 — 이 모형이 같은 시험을 통과하지 못하면 시험이 틀린 것이다 |
| 해상도 사다리 | 삼각형 100/50/25/12.5 % 로 재고 ptp 뿐 아니라 곡선 상관까지 봤다 |

## 잣대를 먼저 교정하는 이유

«나왔다» 를 말하려면 «안 나와야 하는 것이 안 나온다» 를 먼저 보여야 한다. 회전대칭 물체는 돌려도 모양이 안 바뀌므로 변조를 만들 방법이 기하학적으로 막혀 있다 — 그런 물체에서 변조가 보이면 그것은 측정법이 만든 인공물이다.

반대 방향도 필요하다. 이상 점산란자처럼 «반드시 통과해야 하는» 모형이 같은 시험에서 떨어지면, 틀린 것은 대상이 아니라 시험이다.

## 적대검증이 정정한 두 가지

① 널 셈. 프로펠러를 제거한 팔 3 칸 [^25] 은 위상을 돌려도 기하가 글자 그대로 같아서 AC 가 **정확히 0** 이다 — 시험을 통과한 칸이 아니라 시험이 성립하지 않는 칸이다. 실제로 시험된 널은 12 칸 [^24] 이다.

② 분포 겹침. 널 최댓값 3.83 dB [^31] 가 신호 격자 최솟값 3.08 dB [^32] 보다 크다 — 문턱 6.0 dB [^33] 는 두 분포의 «빈 골짜기» 가 아니라 겹친 구간 위에 놓여 있다. 겹치는 칸들은 전부 무변조로 찍혔으므로 이 겹침이 거짓양성을 만들지는 않았다.

### 결과
