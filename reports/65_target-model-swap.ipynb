낙차의 소유자는 정육면체다 — 자유공간에서 최대가 M2 인 셀이 21 ⟨outputs/tm_attack.json : Q3_staleness.argmax_argmin_counts.E0_freespace.argmax_counts.M2⟩개, 최소가 M1 인 셀이 21 ⟨outputs/tm_attack.json : Q3_staleness.argmax_argmin_counts.E0_freespace.argmin_counts.M1⟩개로 전수이고, M3 몫은 다섯 앙상블에서 20.2% ⟨outputs/tm_attack.json : Q3_staleness.argmax_argmin_counts.E2_outdoor_canyon.m3_share_of_spread⟩ ~ 33.0% ⟨outputs/tm_attack.json : Q3_staleness.argmax_argmin_counts.E1_chamber_floor.m3_share_of_spread⟩ 다.

각 다양성이 그 낙차를 줄인다 — 각 다양성이 0 인 자유공간(N_eff 1.0 ⟨outputs/tm_result.json : verdicts.Q3_environment_dependence.predictor.E0_freespace.n_eff_pairs⟩)에서 29.30 dB ⟨outputs/tm_result.json : verdicts.Q3_environment_dependence.pure_pattern_spread_db_by_env.E0_freespace⟩, 각 다양성이 가장 큰 앙상블(N_eff 3.78 ⟨outputs/tm_result.json : verdicts.Q3_environment_dependence.predictor.E2b_outdoor_shadowed.n_eff_pairs⟩)에서 10.09 dB ⟨outputs/tm_result.json : verdicts.Q3_environment_dependence.pure_pattern_spread_db_by_env.E2b_outdoor_shadowed⟩ 다.

⚠ 크기는 **추정량이 정한다** — 검출기가 읽는 p10 에서 맞추면 낙차가 2.30 dB ⟨outputs/tm_attack.json : Q1_normalisation.recomputed_spread_db_by_matching_estimator.p10.spread_mean⟩ 로 줄고 세 팔의 순서가 뒤집혀 M1 이 가장 어려운 팔이 된다. 문턱은 잡음전력 기지 이상문턱이고 CA-CFAR 문턱은 세 팔에 같은 오프셋을 주므로, 교정표는 세 팔의 절대 소요이득만 옮긴다(⟨outputs/tm_result.json : protocol.pfa_convention⟩).

⚠ 이 표에 **구 대조군은 없다.** 구는 **부피를 맞게 고르면** σ 의 절대 레벨을 맞출 수 있는 단순 모형이면서 자세에 따른 변화를 0 으로 낸다 — 레벨에서 우리 메쉬를 앞선 그 구는 논문이 적어 둔 상자 치수로 잡은 부피이고, 메쉬 부피로 잡으면 두 잣대 모두에서 우리 메쉬보다 나쁘다([편 27 «상자·구 대조군»](27_box-sphere-control.ipynb)). 그래서 이 표의 낙차는 «자세 구조를 얼마나 담는가» 의 낙차로 읽는다.
