# 리포트 11 — 조달처 일곱 갈래 R1~R7 을 전수로 적었다

> ### 한 일
> **갈래마다 조달처와 그것이 사 주는 주장을 적고, 판정한 문서 전부를 갈래 태그와 게재상태와 함께 세 표로 실었다.**

### 결과
1. 갈래는 7개이고, 갈래마다 문서 수와 게재 편수를 함께 센다.
2. 카탈로그는 판정 문서 21 건 [^1] 전수를 담는다 — 표를 셋으로 나눈 이유는 화면 하나에 들어가게 하기 위해서다.
3. 각 행은 «논문 · 상태 · 조달처 · 그 편이 산 주장» 네 칸이다. 마지막 칸이 이 부의 중심 문장을 행마다 반복해서 보여준다.
4. 이 표가 논문 II 절 표의 원형이고, 그 문단은 [docs/paper/01_prior.md](../docs/paper/01_prior.md) 로 옮겨 두었다.

### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 갈래 태그 | 편마다 축자 인용으로 판정한 조달처를 `route_tag` 한 칸으로 압축한다 |
| 게재상태 열 | 논문 II 절 표의 원형이라 게재상태 열을 함께 싣는다 |
| 표 분할 | 행 순서는 근거 JSON 의 고정 순서 그대로이고, 인덱스로 잘랐다 |

## 갈래 일곱 — 조달처와 그것이 사 주는 주장

아래 표의 `CIR` 은 채널 임펄스 응답, 즉 한 번 때린 신호가 되돌아오는 모양이다.

| 갈래 | 조달처 | 그것이 사 주는 주장 | 문서 | 게재 |
|---|---|---|---|---|
| R1 | 그 기체를 직접 잰다 (무향실·CATR) | 그 기체 그 대역의 절대 σ 와 계수표 | 3 | 3 |
| R2 | full-wave 솔버로 그 기체를 푼다 (광선엔진 밖) | 바이스태틱 절대 σ 와 그것을 먹는 커버리지 예산 | 1 | 1 |
| R3 | 외부 σ 를 광선경로에 주입한다 | 엔진을 건드리지 않고 얻는 자세의존 진폭 | 1 | 0 |
| R4 | 해석 표적 모델 (thin-wire · 점산란체 연속체) | 마이크로도플러 서명의 모양과 상관계수 | 2 | 2 |
| R5 | 추상 계수 · 점산란체 · 기하 프리미티브 | 폐형식 신호모델과 검출·추정 통계 | 5 | 1 |
| R6 | 엔진 기본 상호작용에 맡긴다 (스톡 재질·확산계수) | 운동학 서명 · 지연-도플러 구조 · 데이터셋 | 6 | 3 |
| R7 | 엔진 안에서 메쉬 산란을 계산한다 | 기하와 일관된 서명, 멀티스태틱 확장 가능 | 3 | 1 |
