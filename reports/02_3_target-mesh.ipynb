## 표를 읽는 법, 그리고 형상 정정이 옮긴 양

맨 오른쪽 **kr** 은 표적을 파장으로 잰 크기다 — 외접반경 r 에 파수 k = 2π/λ 를 곱한 값이고, 클수록 파장에 비해 몸이 크다는 뜻이다. 외접반경 r 은 이 빌더가 메쉬에서 직접 다시 재고 갤러리 원장과 대조한다(최대 차이 0.00% [^16]).

위 그림과 표는 2026-08-04 [^17] 형상 정정 **후** 메쉬에서 재측정한 값이다. 그날 Matrice 4E · Mini 2 · X500 V2 의 형상 상수가 공식 CAD 실측으로 정정되면서 몸통을 공표 높이에 맞추던 z 축 배율이 Matrice 4E 0.950 [^19]→1.000 [^20] · Mini 2 0.906 [^21]→1.001 [^22] 로 움직였고 세 기체의 메쉬 지문이 셋 다 달라졌다. X500 V2 는 배율 대신 삼각형이 18,816 [^23]→19,030 [^24] 로 바뀐 쪽이다 [^25].

## 절 2. 메쉬 외형이 실제 기체와 얼마나 맞는지를 사진 IoU·CAD 치수·실물 스캔 세 자로 쟀다



> ### 한 일
> **사진 실루엣 IoU · 제조사 CAD 치수 · 실물 유래 메쉬 σ 세 가지 검사를 각각 자기 바닥과 함께 재서 한 표에 올렸다.**

### 결과
1. 사진 21 장 [^27] 을 정합해 실루엣 IoU 를 쟀다 — 기체마다 최고 1장을 골라 세운 7 개 [^28] 값이 0.478 [^29]~0.875 [^30] (자기복제 상한 대비 50 [^31]~91% [^32]) 이고, 사진 전수로는 0.363 [^33]~0.875 [^34] 다.
2. 그 상한 자체가 0.864 [^35]~0.957 [^36] 다 — 같은 메쉬로 만든 가짜 사진을 같은 파이프라인에 넣었을 때의 값이고, 암·블레이드가 몇 px 이라 1.0 아래에 있다.
3. 제조사 CAD 어셈블리 3 종 [^37] 중 우리 기체에 대응하는 것은 1종(maps_to 열의 `x500v2`) [^38] 이고, 맞댄 치수 20 개 [^39] 는 전부 그 기체다 — 최대 편차는 CAD 대비 0.56% [^40] (치수 19 개 [^41], 설계 의도 실현 검사) · 발행 제원 대비 0.56% [^42] (치수 10 개 [^43], 독립) 다.
4. 실물 유래 메쉬와의 Δ 방위평균 σ 는 -2.0 [^44] · -0.4 [^45] · +1.8 dB [^46] 이고, 자세별 RMS 는 세 대조 전체에서 5.2 [^47]~10.0 dB [^48] 다 — 양쪽 메쉬를 전부 PEC 로 덮고 껍데기 형상만 맞댄 값이고, 방위 48 [^49]·72 [^50]·48 [^51]칸을 하나씩 빼는 잭나이프 표준오차가 1.0·0.5·1.8 dB 다. 인용은 소수 첫째 자리까지로 정한다.
5. ⭐ 이 세 자가 σ 의 **인용 단위**를 정한다 — 인용 단위는 **Δ 방위평균 σ** 하나이고 소수 첫째 자리까지다. 로브 위치와 널 깊이는 인용 단위에서 뺐다: 방위 격자가 7.5·5.0·7.5° 이고, 두 메쉬의 방위 칸을 첨자 순서로 짝짓는다. 그 둘을 다시 재는 일은 다음 단계 표에 세웠다.

## 이 표가 σ 의 인용 단위를 정한다

인용 단위는 **Δ 방위평균 σ** 하나이고, 소수 첫째 자리까지다. **로브**(방위를 돌릴 때 σ 가 솟는 봉우리) **위치**와 **널 깊이**(그 봉우리 사이에서 σ 가 꺼지는 골의 깊이)는 인용 단위에서 뺐다 — 아래 두 줄이 그 근거를 숫자로 적는다.

⛔ **로브 위치** — 방위 격자는 7.5·5.0·7.5° 이고, 로브 폭의 눈금 λ/2D 는 7.0·4.7·1.5° 다(D = 실물 메쉬 bbox 의 수평 최대 변, 3.5 GHz [^54]). 이웃 칸 lag-1 자기상관은 여섯 패턴 전체에서 -0.19~+0.11 이고, 최대 σ 방위는 실물 120·50·127.5° 대 우리 180·70·180° 다.

⛔ **널 깊이** — 두 메쉬의 방위 칸을 첨자 순서로 짝짓는다. 우리 쪽 방위열을 한 칸씩 돌려 가며 자세별 RMS 를 다시 내면 Phantom 4 5.0~6.9 · Typhoon 4.7~8.2 · M600 8.2~10.3 dB 사이를 오가고, 회전 0 인 우리 짝짓기는 그 회전 널에서 41/48 · 2/72 · 37/48 번째다. 요각 정합은 다음 단계 표에 있다.

IoU 에는 눈금이 붙어 있다 — 표 오른쪽 끝의 **자기복제 상한**은 같은 메쉬로 만든 가짜 사진을 같은 파이프라인에 넣었을 때 나오는 값, 즉 이 검사가 낼 수 있는 최고점이다. 그 값이 0.864 [^35]~0.957 [^36] 다(암·블레이드가 몇 px 이라 최고점이 1.0 아래다). 같은 메쉬끼리도 자세가 1° 어긋나면 0.906 [^59], 2° 면 0.822 [^70] 로 내려간다.

⭐ 그래서 이 편이 파는 것은 «메쉬가 맞다» 가 아니라 **«어느 인용 단위까지 맞다»** 다. 그 단위 안에서 σ 를 쓰는 자리가 [리포트 3 «앵커와 검증»](03_anchor.ipynb) 이고, 단위 밖의 양을 인용하지 않는다.

⚠ 사진 IoU·CAD 표는 2026-08-04 [^57] 형상 정정 **전** 메쉬의 산출이고, σ 대조 세 값은 그 메쉬 위에서 2026-08-07 10:58:22 [^71] Γ(θ) 각도 모양(기본 켬) **이전** 커널로 돌린 것이다 — 재측정은 다음 단계 표에 있다.

## 절 3. 도전성 재질이 면적의 45.3% 로 Σ|Γ|A 의 73.5% 를 낸다



> ### 한 일
> **부품 그룹마다 재질을 붙여 7기체의 표면적을 재질로 나누고, 같은 면적을 반사계수 크기로 가중해 무엇이 반사 진폭을 내는지까지 셌다.**

### 결과
1. 재질은 6 가지 [^72] 이고, 7기체 합계 표면적은 3.236 m² [^73] 다.
2. 도전성 carbon, metal, camera_assembly, pcb [^74] 가 면적의 45.3% [^75] 를 차지하고 반사계수 가중 면적 Σ|Γ|A 의 73.5% [^76] 를 낸다.
3. 유전체 셸(body·canopy)은 반사계수 크기 0.28 [^77] 로 왕복 투과 τ = 1−|Γ|², 즉 0.71 dB [^78] (수직입사 기준) 를 곱해 통과시키고 그 뒤 금속(배터리·PCB)을 코히런트 합산한다(`src/rcs_sbr.py` `rcs_sbr(penetrate=True)`).
4. Sionna PathSolver(S) · 우리 SBR+PO 생산 커널(B) · 순수 PO 대조군(P)이 **같은 재질 표**를 읽는다(`src/materials.py` `MATERIALS`, `src/drones.py` `DRONE_GROUP_MAT`) — 전파 계산과 산란 계산이 같은 물성에서 나온다.
5. ⚠ 이 편의 면적·Σ|Γ|A 표와 그림은 2026-07-31 03:53:02 [^79] 산출이고, 형상 정정 2026-08-04 [^80] 보다 앞선 세대다 [^81] — 정정이 이 함대에서 건드린 Matrice 4E · X500 V2 [^82] 의 표면적이 그만큼 움직인다 (어느 기체에서 무엇이 얼마나 옮겼는지는 **절 1** «기체 7종을 제원과 제조사 CAD 치수에서 세우고 부품을 재질 그룹으로 유지했다»).


### 방법

| 무엇을 | 어떻게 얻었나 |
|---|---|
| 재질 배정 | 부품 그룹마다 재질을 붙인다 — 그룹은 메쉬를 지을 때 유지한 그 그룹이고, 재질 표는 전파 쪽과 공유한다 |

| 반사계수 가중 면적 | 같은 면적을 반사계수 크기로 가중한 장부다 — PO 적분에 들어가는 양이고, 위상이 없으므로 σ 자체는 아니다 |
| PO 실효 반사계수 | 수직입사 실효값이다 — SBR+PO 생산 커널(B, `src/rcs_sbr.py`)이 여기에 각도 모양 \|Γ(θ)\|/\|Γ(0)\| (TE·TM 전력평균)을 기본으로 곱한다 [^83] |

## 셸을 통과한 뒤 금속을 더한다

유전체 셸(body·canopy)은 |Γ| = 0.28 [^77] 로 왕복 투과 τ = 1−|Γ|², 즉 0.71 dB [^78] 를 곱해 통과시키고 그 뒤 금속(배터리·PCB)을 코히런트 합산한다 — τ 는 수직입사 기준이고, 각도 모양은 |Γ(θ)| 축이 따로 든다(`src/rcs_sbr.py` `rcs_sbr(penetrate=True)`).

⚠ Σ|Γ|A 는 **위상 없는 장부**다 — 어느 재질이 진폭에 얼마나 기여하는지의 크기 순서를 보는 데 쓰고, σ 자체는 위상을 가진 면적분에서 나온다. 그 적분이 [리포트 2 절 1 «가림 판정은 Sionna 광선엔진이 하고, 면적분은 우리 커널이 한다»](02_kernel.ipynb) 다.
