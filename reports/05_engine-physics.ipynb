## 두 엔진에 실제로 넘어간 스위치

엔진이 «할 수 있는 것»과 «이번에 한 것»은 다른 물음이다. 아래 표는 그 둘을 갈라 적는다 — 왼쪽 두 칸이 우리 팔(SBR + 물리광학), 오른쪽 두 칸이 스톡 PathSolver 다.

낱말부터 푼다. **가림**은 앞면이 뒷면을 막는 것, **투과**는 플라스틱 셸을 지나 안쪽 금속(배터리·PCB)까지 세는 것, **쐐기 회절**은 실루엣 모서리에서 빛이 꺾여 도는 것, **다중반사**는 표적 안에서 두 번 이상 튀는 것이다.

## 이 표가 서는 범위 — 어느 팔을 말하는가

위 표의 «이번에 켰나» 칸은 **기준 실행**(`--physics` 없이 돈 `sionna` 계열 네 팔)의 상태다. 같은 스크립트가 `--physics` 로 돌린 팔이 둘 더 있고, 그 팔은 같은 줄에서 `max_depth=3` 에 굴절·회절·모서리회절을 전부 켠다. 원장에도 그 팔의 행이 있다 — `engine="sionna_phys"` 7 행이고 그중 7 행이 `n_missing = 0` 인 완결 행이다[^9].

그래서 축마다 성립 범위가 다르다.

- **다중반사** — 기준 실행에서는 두 팔 다 끔이라 사과-대-사과다. `--physics` 판에서는 PathSolver 만 깊이 3 이 되고, 우리 팔은 스윕이 부르는 1 차 히트 커널에 다중반사 루프 자체가 없다(`src/rcs_sbr.py:1068`·투과 패스 `:1105`; 루프가 있는 `rcs_sbr(max_bounce≥2)` 는 스윕이 부르지 않는다).
- **투과** — 기준 실행에서는 우리 팔만 셸 안을 본다. 그 τ 는 수직입사 |Γ| 로 만든 값이고 굴절각·유전체 내 위상지연·두께는 들어가지 않는다.
- **조명** — 두 팔 다 실제 기하로 위상을 준다. 우리 커널의 평면파 갈래는 `range_m=None` 일 때만 켜지는 별개의 길이다.

⚠ 10 m 는 이 기체의 원거리장 경계 안쪽이다 — 두 팔 다 근거리장 판을 계산한 것이고, 원거리장 평면파 값과 나란히 놓는 비교는 이 절의 범위 밖이다.

**그림 1.** 네 스위치 중 어느 것이 나딧 레벨을 올리나?

회절만 켠 판 -64.23 dB [^3] 은 기준 판 -130.78 dB [^1] 보다 **66.55 dB** 위에 선다(두 칸의 차). 나머지 세 스위치는 기준 자리에 머물고, 네 스위치를 전부 켠 판의 레벨은 회절만 켠 판과 같다 — `--physics` 가 올린 것은 회절 하나다.


## 모서리 회절만 켠 판이 기준과 같은 이유

모서리회절만 켠 판은 레벨 -130.78 dB [^7] · 경로 중앙값 11 개 [^11] 로 기준 판과 같은 자리에 선다.

이유는 엔진 소스에 적혀 있다. `edge_diffraction` 은 `sionna/rt/path_solvers/sb_candidate_generator.py:600` 의 `if diffraction_enabled:` 블록 **안에서만** 읽힌다. 즉 `diffraction=False` 인 채로 `edge_diffraction=True` 를 주면 그 인자는 자유 모서리를 여는 대신 그냥 지나간다 — 이 행은 스위치의 크기가 아니라 **설정의 항등식**이고, 세 앙각 전부에서 네 값이 똑같이 겹치는 것이 그 표시다.

⭐그러니 `--physics` 의 네 스위치 중 실제로 레벨을 만든 것은 회절 하나이고, 모서리 회절의 크기는 `diffraction=True` 를 고정한 채 그 스위치만 뒤집는 판에서 잰다. 그 판은 다음 단계 표에 있다.

스톡 Sionna 2.0.1 이 여는 회절은 **1 차 한 번까지**다 — 소스가 «Only first order diffraction is supported» 라 적고(`sb_candidate_generator.py:394`), 회절을 뽑은 경로에서는 이후 회절과 확산반사를 끈다(같은 파일 `:405`~`:407`).


## 경로 수는 스위치의 지표이고, 세기는 레벨 칸이 잰다

이 절의 표에 경로 수를 함께 적은 것은 스위치가 무엇을 열었는지 보이기 위해서다. 같은 엔진이 반지름 0.3 m [^12] 짜리 금속 구에 광선 1,000,000 발 [^13] 을 쏘아 얻은 표적 경로는 0 개 [^14] 다 — 해석해가 있는 과녁인데도 표적 경로가 하나도 안 돌아온다.

⛔**여기 함께 적었던 «표적 근처 경로 12 개 중 진짜 에코 1 개» 는 내렸다**(2026-09-03). 그 수는 **실내 통제 기하**에서 낸 것이라 이 판의 «환경은 실외만» 규약 밖이다 — 낱말이 아니라 값이 그 기하의 함수다. 원장과 함께 `archive/chamber_0903/` 에 옮겨 뒀고, 실외 판에서 다시 내야 쓸 수 있다.

경로 수와 레벨이 갈라지는 자리도 이 표 안에 있다 — 전부 켠 판은 경로 중앙값 6 개 [^15] 로 회절만 켠 판의 14 개 [^16] 보다 적은데 레벨은 같다.

스톡 솔버의 경로 진폭이 표적 면적을 어떻게 따라가는지는 [리포트 2-2 절 5 «면적을 키운 스윕»](02_2_stock-engine.ipynb) 이 이미 재 놓았다. 여기서는 그 결론을 **경로 수를 읽는 주의**로만 쓴다.

## 절 2. AC/DC 가 +0.89 dB 에서 −68.13 dB 로 내려간 것은 분모가 66.55 dB 이상 커진 결과다



> ### 한 일

## 비가 내려가는 길은 둘이다

AC/DC 는 **변조 전력 ÷ 정적 전력**이다. 비가 내려가는 길은 둘뿐이다 — 분자인 변조 성분이 줄거나, 분모인 정적 성분이 커지거나.

둘을 가르는 잣대는 같은 표의 **레벨 열**이다. 레벨은 자세평균 크기라서 정적 성분이 커지면 함께 오르고, 변조만 사라지면 제자리에 머문다. 그래서 «비는 내려갔는데 레벨은 올랐다» 는 곧 분모 쪽 사건이라는 뜻이다.

## 경로를 무너뜨린 것은 굴절이다

굴절을 켜면 경로 중앙값이 11 개 [^21] 에서 2 개 [^22] 로, 평균이 11.50 개 [^31] 에서 1.65 개 [^32] 로 내려간다.

굴절 스위치는 표면에 닿은 광선이 **매질 속으로 들어가는 길**을 연다. Sionna 는 교점마다 반사·확산·투과 중 하나를 |R|²·|T|² 에 비례한 확률로 뽑으므로(`sionna/rt/radio_materials/radio_material.py:764`~`776`), 투과 계수가 큰 플라스틱 면에서는 대부분의 광선이 투과 가지로 간다. `max_depth=1` 이라 그 광선은 그 자리에서 예산을 다 쓴다.

⇒ 줄어드는 것은 **발견된 사슬의 수**다. 한 번이라도 발견된 사슬은 이미지법으로 정확히 다시 풀리므로 진폭은 표본 수와 무관하고, 그래서 경로가 5.5 배 줄어도 레벨은 -132.48 dB [^23] 로 기준판 -130.78 dB [^17] 에서 1.70 dB 안에 머문다.

그 1.70 dB 는 나딧의 값이다 — 같은 스위치가 el −45° 에서는 레벨을 -125.96 [^33] 에서 -130.30 dB [^34] 로 내리고, el 0° 에서는 -59.65 [^35] 에서 -59.66 dB [^36] 다. 그 두 판은 자세 16 개 [^37] 로 돌린 표본이라 나딧 판(자세 20 개 [^38])과 표본 수가 다르다.

나딧 판의 AC/DC 는 +4.60 dB [^39] 로 올라가는데, 남은 경로 2 개 [^22] 가 자세마다 크게 흔들린 결과다.

굴절 하나는 `--physics` 의 경로 감소를 분해하지 않는다 — 넷을 다 켠 판의 중앙값은 6 개 [^40] 이고, 회절만 켠 판은 14 개 [^41] 로 오히려 늘어난다.


## 레벨을 올린 것은 회절이고, 그 크기는 판 대 판의 차다

회절만 켠 판의 레벨은 -64.23 dB [^18] 이고 넷을 다 켠 판도 -64.23 dB [^42] 다 — 귀속은 **절 1** «스위치 단일축» 가 세웠다.

⭐ 그 66.55 dB 는 «회절 항 대 정반사 항» 의 비가 아니라 **판 대 판의 레벨 차**다. 여섯 판 모두 정반사와 확산 반사를 함께 켜고 굴절을 끈 깊이 1 로 돌았으므로(`benchmark/diag_physics_paths.py:61`), 기준판의 이름은 «반사만» 이다.

이 차에는 등급이 있다 — el −45° 에서 -125.96 [^33] → -97.02 dB [^43] 로 28.94 dB, el 0° 에서 -59.65 [^35] → -59.65 dB [^44] 로 0.00 dB 다 — 두 판 다 자세 16 개 [^37] 표본이다.

⭐ 나딧은 표적 바닥이 파면과 정면으로 마주 보는 자리다. 회절을 켠 판은 el 0° 에서 나딧까지 4.58 dB 내려가는 동안 반사만 켠 판은 71.13 dB 내려간다. 그래서 이 레벨이 묻는 것은 «회절이 그만큼 세다» 가 아니라 «스톡 솔버의 반사 경로가 나딧에서 무엇을 잃고 회절 경로가 그 자리를 채우는가» 다.


## 그래서 −68 dB 는 분모 쪽 사건이다

회절 판의 AC/DC 는 -68.13 dB [^20] 라 그 판의 간극 δ 는 0.00 dB 이고, 레벨이 곧 정적 성분이다. 기준판은 AC/DC 가 +0.89 dB [^19] 라 δ 의 상한이 3.48 dB 다.

그래서 정적 성분의 상승은 66.55 ~ 70.03 dB 구간이고, 비가 -69.02 dB 내려간 것과 합치면 변조 성분의 이동은 -2.47 ~ +1.01 dB 구간이다 — 분자는 몇 dB 안에 갇혀 있고, 수십 dB 움직인 것은 분모다.

그러니 「물리를 켜니 나딧 도플러가 이론대로 0 이 됐다」는 문장은 이 분모 상승의 다른 이름이다. 회절이 정적 성분을 수십 dB 올렸고, 변조는 그 자리에 남았다.

⚠ 레벨차 66.55 dB 와 합산값 -2.47 dB 는 각각 구간의 **끝**으로 읽는다. 점값을 내려면 자세별 복소 합을 저장해 |mean E|² 를 직접 재야 하고, 그것이 다음 단계 표의 둘째 줄이다.


## 모서리회절 행은 항등식이고, 깊이 축은 둘째 자리다

모서리회절만 켠 판은 경로 11 개 [^24] · 레벨 -130.78 dB [^25] · AC/DC +0.89 dB [^26] 로 기준판과 **네 값이 모두 같다.** 그 스위치가 `diffraction` 의 하위 옵션이라 `diffraction=False` 인 판에서는 계산에 들어가지 않기 때문이다 — 세 앙각(0°·−45°·−90°) 전부에서 네 값이 똑같이 겹치는 것이 그 표시다. 모서리회절의 크기는 다음 단계 표의 판이 잰다.

다중반사(최대 깊이 3 [^45])는 이 앙각에서 레벨 -130.76 dB [^46] · AC/DC +0.88 dB [^47] 로 소수점 둘째 자리에서 갈리고, 경로 중앙값만 11 개 [^21] → 12 개 [^48] 로 하나 는다. 세 앙각을 다 보면 깊이 축이 레벨을 0.29 dB(el −45°), AC/DC 를 0.56 dB(el 0°) 안에서 움직인다[^49].

선행 스위치 연구는 깊이 축에서 갈린다 — 도시 규모 경로손실을 관측량으로 삼은 판은 `max_depth` 무감을 보고하고, 방위 추정과 라디오맵을 관측량으로 삼은 두 판은 깊이가 결과를 옮긴다고 적는다(`prior_work/sionna_solver_settings_survey.md` 의 도시 판·방위 판·라디오맵 판). 우리 소형 표적 후방산란은 앞쪽과 같은 방향이고, 갈리는 것은 불리언 스위치 쪽이다.


## 자세 수에 걸리는 칸과 걸리지 않는 칸

이 절이 여는 el −90 진단판은 자세 20 개 [^38] 로 돌린 표본이고, el −45·el 0 판은 자세 16 개 [^37] 다. 같은 스위치를 켜고 자세 4,096 개 [^27] 로 돌린 스윕 본판의 나딧 행은 레벨 -64.23 dB [^28] · 경로 6 개 [^29] 로 진단판의 «전부 켬» 칸과 **같은 값**이다. 그 판은 AC/DC 가 −68 dB 라 자세축 변동이 평균의 0.04 % 이고, 자세를 몇 개 뽑아도 자세평균 크기가 같은 자리에 온다.

물리를 끈 같은 엔진의 나딧 행은 **자세 수에 걸린다** — 본판은 레벨 -130.29 dB [^50] · 경로 12 개 [^51] 이고, 진단판의 기준 칸은 -130.78 dB [^17] · 11 개 [^21] 다. 그 판의 AC/DC 가 +0.89 dB [^19] 라 자세마다 크게 흔들리기 때문이다. 두 행 모두 자세가 하나도 비지 않은 완결 행이다[^52].

⇒ 이 절의 레벨 상승 66.55 dB 는 **el −90 진단판의 자세 20 개 기준**으로 닫아 계산한 값이다. 본판 행을 피감수로 쓰면 같은 뺄셈이 66.06 dB 가 되므로, 두 판의 행은 따로 읽는다.

⚠ 이 대조는 **PathSolver 계열 안에서만** 성립한다. 우리 커널은 정규화가 달라 레벨 축을 따로 쓴다.


## 이 판이 서는 범위

여섯 줄은 el -90° [^53] 한 점에서, 표적 `matrice4e` 하나로, 거리 10 m [^54] · 반송주파수 3.5 GHz[^55] · 광선 11,111,111 개 [^56] · 자세 20 개 [^38] 로 잰 값이다.

분자·분모 가르기는 부등식이 주는 **구간**이다. 회절 판은 AC/DC 가 -68.13 dB [^20] 라 간극이 0.00 dB 이고, 기준판은 +0.89 dB [^19] 라 간극의 상한이 3.48 dB 다. 그 폭이 -2.47 ~ +1.01 dB 라는 구간을 만든다.

같은 여섯 판을 돌린 원장은 앙각 3 점에 있다 — el −90° · el −45° · el 0° 이고, el −45·el 0 판은 자세 16 개 [^37] 표본이다. 남은 앙각으로 넓히는 일이 아래 표의 첫 줄이다.

## 상한 위 몫이 무엇을 재는가

날개끝 속도가 정하는 도플러 상한은 f_tip = 1272.9 Hz [^66] × cos(앙각) 이다. 시선 방향으로 가장 빠른 점이 날개끝이므로 그보다 높은 주파수는 블레이드가 닿는 자리 밖이고, 거기 담긴 전력은 전부 엔진이 만든 인공물이다.

관찰 상한은 나이퀴스트 9850 Hz [^67] 라, 앙각 0° 에서는 f_tip 위로 7.7 배(= 9850 ÷ 1272.9) 폭의 자리를 들여다본다. 그 자리에 무엇이 얼마나 앉는지가 이 절의 잣대다.

몫은 팔마다 자기 전체 전력으로 나눈 값이라 눈금과 무관하다. 팔 사이 레벨 비교는 같은 엔진 안에서만 하고, 그 자리는 [리포트 4 절 5 «예산 사다리»](04_elevation-coverage.ipynb) 다.


## 앙각 0° 에서 스위치 축과 예산 축을 따로 움직인다

| 팔 | 광선 예산 | 물리 스위치 | 경로/자세(중앙값) | 상한 위 몫 |
|---|---|---|---|---|
| 우리 커널 (SBR+PO) | 얼린 격자(자세 합집합 bbox), λ/12 [^68] | PO · 투과 켬 · PTD 끔 | — | 0.02285 [^62] |
| PathSolver | 11.1M | 끔 | 9 개 [^69] | 0.29027 [^57] |
| PathSolver | 11.1M | 켬 | 5 개 [^70] | 0.83078 [^58] |
| PathSolver | 250M | 끔 | 127 개 [^60] | 0.87023 [^59] |
| PathSolver | 250M | 켬 | 33 개 [^61] | 0.87107 [^71] |


규칙 예산에서 스위치를 켜면 몫이 0.290 [^57] 에서 0.831 [^58] 로 올라간다. 그 상승은 예산 축 혼자서도 나온다 — 물리를 끈 채 광선만 22.5 배 올린 팔이 0.870 [^59] 다.

두 축을 가르는 자리는 광선을 맞춘 짝이다. 250M 에서 물리 끔 0.87023 대 물리 켬 0.87107 이고, 뺄셈이 0.00084 다.

앙각 0° 가 그렇게 되는 것은 그 자리가 이 잣대의 **포화점**이기 때문이다 — 상한 위 대역이 관찰 대역의 87.08% 를 덮으므로, 시간 구조가 남지 않은 시계열이 받는 점수가 그 값이다. 두 팔 다 이미 그 포화값에 앉아 있어 스위치가 더할 몫은 소수점 넷째 자리에 남는다.

| -15° | 0.30871 [^73] | 0.85437 [^74] | +0.54566 |

## 같은 광선 수는 팔마다 다른 경로 수로 갈린다

물리를 켜면 광선 하나가 굴절 · 회절 · 모서리회절 · 3 회 반사로 갈라진다. 앙각 0° 에서 자세당 경로 중앙값은 11.1M 에서 9 [^69] → 5 [^70], 250M 에서 127 [^60] → 33 [^61] 로 줄어든다.

부호는 앙각이 정한다 — 같은 11.1M 짝이 앙각 −30° 에서는 6 [^79] → 7 [^80] 로 오히려 는다. 굶기는 스위치는 굴절 하나이고, 회절만 켜면 나딧에서 경로가 늘어난다[^81].

이 열이 세는 것은 **자세당 되돌아온 경로의 개수**이고 표면 표본 밀도는 이 원장 밖이다. 그리고 굶음은 같은 예산 짝 안의 상대량이다 — 250M 물리 팔의 33 [^61] 개는 규칙 예산 물리 끔 팔의 9 [^69] 개보다 3.67 배 많다.

그래서 11.1M 짝의 0.290 [^57] 대 0.831 [^58] 는 스위치와 경로 수가 함께 움직인 값이고, 같은 예산에서 스위치만 남긴 크기는 위 표의 뺄셈 열이다.


## 비용 — 같은 예산 짝에서 벽시계가 2 배다

| 짝 | 물리 끔 | 물리 켬 | 배율 |
|---|---|---|---|
| 앙각 0° · 11.1M | 1549.9 s [^82] | 3361.5 s [^83] | 2.17 배 |
| 앙각 −90° · 11.1M | 1818.0 s [^84] | 3621.6 s [^85] | 1.99 배 |
| 앙각 0° · 250M | 1221.3 s [^86] | 1530.0 s [^87] | 1.25 배 |

«배율» 은 켬 ÷ 끔 이다. 초를 4096 자세 [^65] 로 나누면 자세당 0.38 → 0.82 초(앙각 0°), 0.44 → 0.88 초(앙각 −90°) 다.


이 초는 여러 실행이 같은 GPU 를 나눠 쓴 조건에서 잰 벽시계다. 광선을 22.5 배 더 쏜 팔의 벽시계가 오히려 짧은 칸이 있고(1221.3 ÷ 1549.9 = 0.79 배), 같은 스위치의 배율도 2.17 · 1.25 로 갈린다. 비용 축은 카드 한 장을 비워 놓고 다시 재야 확정된다 — 이 절이 비용에 대해 세우는 주장은 «같은 예산 짝에서 켜면 더 걸린다» 까지다.

**그림 5.** 우리 팔의 el −15° 무늬는 이미 발표한 덱의 판과 같은 물건인가?

7 판의 빗살 간격과 예측 대비 오차, 그리고 정규화와 무관한 양인 AC/DC(0 Hz 정지 성분 대비 변조 성분의 비)를 한 표에 놓는다.

|---|---|---|---|

| deck:R15/E | 125.05 | -1.62 | -19.67 |
| deck:R40/E | 125.05 | -1.62 | -19.72 |
| deck:R3/E | 125.05 | -1.62 | -19.54 |
| sionna/el-15 | 125.05 | -1.62 | -19.99 |

| sionna_p250000000_phys/el-15 | 125.05 | -1.62 | -30.27 |

출처 [^115]


간격 칸이 모든 판에서 한 값인 것은 **눈금이 한 칸뿐이기 때문**이다. 첨두는 주파수 격자 위에만 놓이고, PRF 19,700 Hz [^110] 를 4,096 자세로 잰 격자의 한 칸은 4.81 Hz 다. 125.05 Hz 는 그 26 칸이고, 입력한 f_flash 126.67 Hz [^112] 는 26.34 칸 자리라 26 칸으로 내려앉는다 — 예측 대비 오차 −1.62 Hz 는 커널이 낸 값이라기보다 그 내림 나머지다.

⇒ 이 칸이 가르는 것은 **반 칸(±2.4 Hz)보다 큰 박자 차이**뿐이고, 모든 판이 그 안에 들어 있다. 박자 자체는 우리가 입력한 회전수가 정하므로 이 칸은 산란 커널을 재지 않는다 — 팔을 가르는 잣대는 [리포트 4 절 3 «상한 위 누설»](04_elevation-coverage.ipynb) 가 앙각 여섯 점에서 다룬다. AC/DC 는 우리 팔·덱·물리 끔 팔이 -19.54 dB [^117] 에서 -19.99 dB [^118] 사이에 모이고, 물리를 켠 두 팔은 -26.38 dB [^119] · -30.27 dB [^120] 로 그 아래에 앉는다 — 이 값은 판 안에서의 비라 정규화가 달라도 나란히 놓을 수 있는 칸이다.

우리 팔과 덱 판은 같은 모양의 사다리를 그린다 — 3 차에서 골이 지고 6~7 차에서 마루가 선다. 우리 팔의 3 차는 24.30 dB [^121], 7 차는 42.90 dB [^122] 이고, 덱 15 m 판의 같은 두 칸은 24.25 dB [^123] 과 42.42 dB [^124] 다.

물리를 끈 PathSolver 판에도 3 차 골은 있다 — 29.55 dB [^125] 로 그 행 여덟 칸 중 가장 낮다. 갈리는 것은 **골의 깊이와 마루의 자리**다. 그 판의 마루는 7 차(31.88 dB [^126])가 아니라 4 차(37.70 dB [^127])에 서고, 최고−최저가 8.15 dB 로 우리 팔 18.60 dB 의 절반보다 작다. 상관 0.6961 [^100] 의 정체가 이 **덜 굴곡짐**이다.


물리를 켠 두 팔은 사다리 자체가 낮다 — 여덟 칸이 8.64 [^128]~13.26 dB [^129] (물리 켬 11.1M) 과 7.23 [^130]~12.99 dB [^131] (물리 켬 250M) 안에 다 들어간다. 그 폭은 우리 팔 18.60 dB 의 3 분의 1 아래이고, 상관 0.6903 [^101] · 0.6291 [^102] 이 그 평평함의 점수다.

`line_snr_db` 는 팔마다 **자기 대역외 바닥**에 대한 비다. 바닥이 다르면 여덟 칸이 통째로 평행이동하므로, 비교되는 것은 한 팔 안의 **모양**이다.


## 이 대조에서 빠진 팔은 하나다

el −15° 의 채점표에 자리가 빈 팔은 모서리 회절을 켠 우리 커널 하나다[^132] — 그 팔의 시계열이 아직 계산되지 않았고, 그 계산이 다음 단계 표에 있다.

물리를 켠 두 팔은 el −15° 에서 자세가 다 차 있어 채점됐다 — `n_missing` 이 0 [^106] · 0 [^107] 다. 광선 250M 에 물리를 끈 팔(`sionna_p250000000/el-15`)도 완결 행이고[^133] 이 대조표에는 그 팔의 시계열이 아직 들어가 있지 않다.

완결성 문턱은 대조표를 만드는 스크립트 밖에 있다[^134] — 스크립트는 npz 에 키가 있으면 패널로 넣고, 자세 완결성은 `outputs/elevation_sweep_md.json` 의 `n_missing` 이 따로 들고 있다.

## 날개는 λ 의 절반보다 좁다 — 회절 근사가 서는 자리

3.5 GHz [^178] 에서 λ = 0.0857 m [^155] 다. matrice4e 날개는 반경 1.6 [^179] λ · 최대 코드 0.40 [^156] λ · 최대 두께 2.99 mm [^180] = lambda/28.7 [^157] 다.

그 두께는 블레이드 법칙의 **두께비**(무차원 비율) 상수와 코드 분포에서 스팬을 따라 계산한 최대값이다 [^181] — 파장 대비는 위의 mm 값이 정한다.

반경만 기체 관측(공표 프로펠러 지름)에 묶여 있고, 코드와 두께는 **CAD 블레이드 법칙에서 계산한 모델값**이다. 그 상수는 참조 프롭에서 상속한 것이라 이 기체를 잰 값이 아니다 — 어느 쪽 숫자를 쓰든 코드가 λ 의 절반보다 좁다는 결론은 선다.

1차 UTD 와 1차 PTD 는 «모서리가 고립돼 있고 인접면이 여러 λ 에 걸쳐 평평하다» 를 전제로 선다. Ziganshin 외(`arXiv:2604.05991`, IEEE OJAP 투고본)가 그 전제를 facet 변길이로 못 박았다 — 축자 «the facet size should be of several wavelengths in order to satisfy the UTD assumptions, typically E > 1.5λ».


우리 메쉬의 `prop` 그룹 facet 변은 중앙 0.078 [^182] λ 이고, 굽은 여섯 그룹은 그 전제를 15.4 [^183] ~ 85.1 [^184] 배 벗어난다.

Sionna 의 쐐기 회절은 그 면들 위에 붙고 1차까지 계산한다. 두 엔진 다 이 날개에서는 점근 영역 밖에서 회절을 다룬다.

코드가 λ 의 절반보다 좁은 유전체 날개에서 1차 회절 근사의 오차 크기를 잰 선행은 우리 조사 두 편(`prior_work/po_refinement_survey.md` · `prior_work/sionna_sensing_survey.md`)에 없다. 그 크기는 다음 단계 표의 마지막 행이 재는 양이다.
