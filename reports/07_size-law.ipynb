## 절 1. 사다리가 하나가 아니라 셋이었다 — 여섯 단이 서로 다른 운동학을 쓰고 있었다



> ### 한 일
> **표적을 점점 거칠게 바꾸는 사다리 여섯 단을 돌린 뒤 단마다 «무엇이 도는가» 를 세어 사다리를 세 벌로 갈랐다.**

### 결과
1. 여섯 단이 같은 운동학을 쓰지 않았다 — 구·정육면체·상자 단은 기체 전체를 덩어리로 바꿔 통째로 돌리고, 진짜 드론 단은 몸통이 서고 프로펠러만 돈다.
2. 그래서 가장 인용하기 좋은 «정육면체 대 메쉬» 차이 +26.05 dB [^1] 안에서 운동학 몫이 +23.42 dB [^2], 형상 몫이 -2.85 dB [^3] 다.
3. 운동학 비중은 Mini 2 74% [^4] · Matrice 4E 55% [^5] — 형상 몫보다 크다.
4. 세 몫의 합은 정의상 총 차이와 닫힌다 — 닫힘오차 0.00 dB [^6].
5. 그래서 사다리를 A(있는 그대로) · B(온몸 자전 고정) · C(진짜 비행 고정) 세 벌로 갈라 적는다. 답할 자격은 C 에만 있고 그 편이 **절 2** «교정 사다리» 다.

## 무엇을 물었나

드론이 제자리에 떠 있어도 프로펠러가 돌기 때문에 되돌아오는 전파의 세기와 위상이 시간에 따라 흔들린다. 이 흔들림을 **마이크로도플러**라고 부른다 — 표적 전체가 움직여서 생기는 도플러가 아니라 표적의 *일부*(날개)가 움직여서 생기는 도플러라는 뜻이다.

지도교수의 지적은 «드론 RCS 정밀도는 연구 값어치가 없다» 이고, 절대 세기 축에서는 우리 데이터가 그 지적을 뒷받침한다. 이 라운드는 그 지적이 마이크로도플러에서도 성립하는지를 물었다.

방법은 **사다리**다 — 표적을 점점 거칠게 바꿔 가며 신호가 언제 무너지는지 본다.


## 사다리를 돌려 보니 여섯 단이 운동학을 섞어 쓰고 있었다

구·정육면체·상자 단은 기체 전체를 덩어리 하나로 바꿔 **통째로 돌린** 물체이고, 진짜 드론은 몸통이 서 있고 프로펠러만 돈다. 두 개를 나란히 놓고 «형상 차이» 라고 부르면 그 안에 형상 교체와 운동학 교체가 섞인다.

| 사다리 | 무엇이 도나 | 무엇이 바뀌나 | 질문에 답할 자격 |
|---|---|---|---|
| A. 있는 그대로 | 단마다 다름 | 형상 + 운동학 + 재질 | ❌ 섞여 있다 |
| B. 온몸 자전 고정 | 전부 온몸 자전 | 형상만 | △ 실제 드론이 아님 |
| C. ⭐ 진짜 비행 고정 | 전부 프로펠러만 | **프로펠러 형상만** | ✅ 이 축이 답이다 |

## 운동학 몫이 형상 몫보다 크다

| 기체 | 대리 형상 | 총 차이 | 운동학 몫 | 재질 몫 | 형상 몫 | 운동학 비중 |
|---|---|---|---|---|---|---|
| Mini 2 | 등가부피 정육면체 | +26.05 dB [^1] | +23.42 dB [^2] | +5.49 dB [^13] | -2.85 dB [^3] | 74% [^4] |
| Mini 2 | 외접 직육면체 | +34.41 dB [^14] | +23.42 dB [^15] | +5.49 dB [^16] | +5.50 dB [^17] | 68% [^18] |
| Matrice 4E | 등가부피 정육면체 | +15.11 dB [^19] | +23.30 dB [^20] | +5.52 dB [^21] | -13.71 dB [^22] | 55% [^5] |
| Matrice 4E | 외접 직육면체 | +17.89 dB [^23] | +23.30 dB [^24] | +5.52 dB [^25] | -10.93 dB [^26] | 59% [^27] |

즉 «정육면체가 틀린 이유» 는 대부분 «모양이 거칠어서» 가 아니라 **«무엇이 도는지를 틀리게 놓아서»** 다. 운동학이 같으면서 날개 디테일만 없앤 유일한 팔인 `slab` 의 차이는 -1.25 dB [^28] · -5.54 dB [^29] 로 훨씬 작다.

⚠ 같은 «정육면체 대 메쉬» 라도 무엇을 재느냐에 따라 값이 다르다. 위 표는 절대 AC 전력의 비이고, AC 를 DC(동체 반사)로 나눈 «상대 변조 깊이» 로 재면 +21.76 dB [^30] · +26.56 dB [^31] 가 된다. 프리미티브는 동체 반사도 함께 바꾸기 때문이다.

## 절 2. 몸통은 진짜 CAD, 프로펠러만 갈아 끼운 교정 사다리가 답할 자격이 있는 유일한 축이다



> ### 한 일
> **여섯 단 전부를 «몸통은 진짜 CAD · 프로펠러만 돈다» 는 같은 운동학에 고정하고 프로펠러 형상 하나만 바꿔 사다리를 다시 세웠다.**

## 읽는 법 — 네 구간

- **공짜 — 해상도.** 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 -0.01 dB [^45], 템플릿 손실이 0.00 dB [^46] 움직인다. 방위에 따른 흔들림 7.99 dB [^79] 에 묻히는 크기라 실측이 가르는 범위 밖이다.
- **싼 구간 — 날개를 평판으로.** 스팬·두께·부피가 같은 평판 두 장이면 검출 단면적이 한 자릿수 dB 안에 든다. «변조가 있나 없나» 만 보는 검출기라면 충분하고, 파형을 본뜨는 검출기라면 이미 비싸다.
- **비싼 구간 — 날개를 덩어리로.** 감싸는 상자는 속을 꽉 채웠으니 검출 단면적을 과대하게 낸다. 회전반경 위의 공은 Matrice 4E 에서 레벨만 보면 «잘 맞는» 것처럼 보이지만 파형 상관이 0.224 [^61] 뿐이라 템플릿 손실이 13.00 dB [^60] 다 — **레벨이 맞았다고 신호가 맞은 것이 아니다**.
- **낭떠러지 — 회전대칭.** 원판으로 바꾸면 검출 단면적이 -54.44 dB [^43] 무너진다. 이것만은 어떤 커널 결함에도 안 흔들리는 기하학이다.


## 사다리의 순서는 삼각형 수가 아니라 보존량이 정한다

감싸는 상자(C)가 평판(D)보다 오차가 큰 것은 «형상 정보가 더 적어서» 가 아니라 부피를 안 지켜서다. 즉 «단순화의 값» 을 정하는 것은 삼각형 수가 아니라 **무엇을 보존했는가**(회전대칭 깨짐 → 스팬 → 부피 → 두께 순)다.

⭐ 한 줄로: **«모양이 있느냐 없느냐» 는 50 dB 대를 가르지만, «모양이 얼마나 정밀하냐» 는 한 자릿수 dB 안에서 논다.** 지도교수의 지적은 앞쪽이 아니라 뒤쪽을 겨눈 것이고, 뒤쪽에서는 지적이 맞는다.


## 지도교수 지적에 답한다 — 네 축

| 축 | 무엇을 쟀나 | 결과 |
|---|---|---|
| ① 절대 세기 | 실측 앵커 대비 rms | 자유 모수 0 개 [^80] 짜리 등가부피 구가 우리 메쉬를 rms 1.73 dB [^81] 이긴다 |
| ② 방위 산포 | 실측 대비 오차 | 우리 메쉬 +1.42 dB [^82] · 구 -5.46 dB [^83] · 상자 +4.48 dB [^84] |
| ③ 시간 변조 | 메쉬 − 구 간격 | 62.0 [^50]~100.8 dB [^51] 중 CAD 정밀도 단독 몫은 0.51 [^85]~4.97 dB [^86] |
| ④ 단순화의 값 | 교정 사다리 | 위 두 표가 그 답이다 |

① 에서는 지적이 맞는다. ② 는 구가 원리적으로 못 내는 축이다 — 구의 방위 산포는 «작다» 가 아니라 정확히 0 이고, 회전대칭이라 만들 방법이 막혀 있다. ③ 의 큰 간격은 «메쉬가 정밀해서» 가 아니라 «회전대칭이 아니라서» 번 것이다.

⚠ ① 의 «자유 모수 0 개» 는 **구의 부피를 논문표기 상자로 잡았을 때**의 값이다 — 그 구는 레벨오차 +0.96 dB [^87] 이고, 같은 기체에서 **메쉬 부피**로 잡은 구는 -4.47 dB [^88] 로 우리 메쉬 -3.30 dB [^89] 보다 멀다. 즉 이 축의 자유 매개변수는 구의 모수가 아니라 **어느 부피에 맞출지의 선택**이고, 그 대조군 표는 [리포트 6 절 4 «상자·구 대조군»](06_anchor.ipynb) 에 있다.

⭐ 그래서 다시 잡아야 할 방향은 «정밀도» 가 아니라 «무엇이 도는가» 다(**절 1** «사다리가 셋이었다»).


## 자기반증 — 이 «공짜» 가 어디까지 참인가

이 커널에서 동체는 AC 에 정확히 0 을 기여한다. 동체를 아무리 깎아도 마이크로도플러 지표가 눈금 하나 움직이지 않는 것은 형상의 성질이 아니라 **가림 없는 커널의 성질**이다.

| 검사 | Mini 2 | Matrice 4E |
|---|---|---|
| 동체까지 깎았을 때 AC 의 상대 변화 | 9.7e-15 [^54] | 5.1e-15 [^90] |
| 같은 팔에서 동체:날개 비의 이동 | -1.88 dB [^91] | +0.34 dB [^92] |

첫 줄이 기계 정밀도라는 것은 AC 가 동체를 아예 안 본다는 뜻이다. 둘째 줄이 0 에서 떨어져 있는 것은 동체 간략화가 «몸통 대 날개 비» 는 움직인다는 뜻이다 — 면적·부피를 잃기 때문이다. 가림을 켜면 날개가 동체 뒤로 사라졌다 나타나면서 동체 형상이 AC 에 들어오기 시작한다.

## 치명적인 둘

**P2 — 커널이 가림(그늘)을 안 본다.** 커널 렌즈가 깊이버퍼 그늘을 넣어 보니, 이 라운드에서 가장 인용될 대조(«평판이 CAD 보다 배음이 풍부하다», n_eff 간격)가 두 대역 × 두 기체 네 구석 전부에서 **부호가 뒤집혔다**. 가림 없는 결과는 방향조차 못 정한다.

**P4 — 사다리 여섯 단이 운동학을 섞어 썼다.** «정육면체 대 메쉬» 로 인용되는 차이의 55% [^98]~74% [^99] 가 형상이 아니라 운동학이다(**절 1** «사다리가 셋이었다»). 교정 사다리가 그 몫을 갈랐으므로 P4 는 **절 2** «교정 사다리» 안에서 닫힌다.


## 프로펠러는 우리 커널의 사각지대다

우리 PO(물리광학) 커널이 1 dB 안으로 맞으려면 부품의 특징 폭이 파장의 0.729 [^103] 배 이상이어야 한다. 프로펠러 날개 폭 13.78 mm [^100] 는 생산 대역 3.50 GHz [^101] 에서 파장의 0.161 [^102] 배에 그쳐 문턱에 4.53 [^104] 배 모자란다.

동체는 2.68 GHz [^106] 부터 유효하므로 통과한다. ⭐ 즉 **마이크로도플러를 만드는 바로 그 부품이 우리 커널이 가장 약한 부품**이다. 이 부의 모든 마이크로도플러 숫자는 그 사실을 안고 읽는다.

PO 유효 무릎 자체의 근거는 [리포트 5 절 5 «PO 무릎을 부품 폭으로»](05_kernel.ipynb) 에 있다.

## 절 5. 두 기체를 함께 재면 크기전이 법칙이 차등신호의 부호 하나로 갈린다



> ### 한 일
> **앵커보다 큰 기체와 작은 기체를 한 캠페인에 넣고 두 크기전이 법칙이 예측하는 μ 를 밴드마다 계산해 차등신호를 냈다.**

### 결과
1. Matrice 4E 는 앵커보다 크고(크기비 1.254 [^111]), Mini 5 Pro 는 작다(0.786 [^112]).
2. 두 법칙의 예측이 **반대 방향**으로 갈리므로 두 기체의 μ 차이가 법칙을 직접 고른다 — 차등신호는 4.06 dB [^113] 다.
3. 기체별 두 법칙의 벌어짐은 Matrice 4E +1.96 dB [^114] · Mini 5 Pro -2.09 dB [^115] 다.
4. 앵커 원장은 크기전이 법칙을 미해소 항목으로 두고 그 크기를 9.50 dB [^116] 로 적어 뒀다 — 이 캠페인이 그 항목을 닫으러 간다.
5. 판정은 값이 아니라 **부호**다 — 두 기체 μ 차이의 부호 하나가 법칙을 고른다.


### 방법

| 무엇을 | 어떻게 얻었나 |

| 크기전이 법칙 | 앵커 기체의 σ 를 우리 기체로 옮길 때 길이의 몇 제곱으로 잡느냐 — L²(면적 비례)와 L⁴(공진 영역) 두 가설 |
| 예상 σ | Das 앵커 레벨에 두 법칙을 각각 적용 — `src/sigma_anchor.py:255`. 설계 계산에만 쓰고 생산 σ 원장은 기울기만 앵커한다 |
| 차등신호 | (L⁴−L²)_matrice4e − (L⁴−L²)_mini5pro [dB]. 두 기체가 앵커보다 각각 크고 작으므로 두 법칙의 예측이 반대 방향으로 갈린다 |

## 앵커 원장이 이 항목을 열어 뒀다

앵커 원장은 크기전이 법칙을 미해소 항목으로 두고 그 크기를 9.50 dB [^116] 로 적어 뒀다. 이 캠페인이 그 항목을 닫으러 간다.

정의는 하나로 고정한다 — `(L4-L2)_matrice4e - (L4-L2)_mini5pro [dB]. 두 기체가 앵커보다 각각 크고 작으므로 두 법칙의 예측이 **반대 방향**으로 갈린다.`

판정은 값이 아니라 **부호**다. 두 기체 μ 차이의 부호 하나가 법칙을 고르고, 그 뒤에 σ 원장이 괄호를 지운다.
