# 리포트 60 — 앵커 σ 위의 R90 은 3.69~11.10 km 이고, 밴드 순서는 기체마다 바뀐다

> ### 한 일
> **기울기 앵커의 Δσ 를 R90 근방 국소 지수로 옮겨 다섯 기체 × 세 밴드의 검출거리를 내고, 그 순서가 인용 방식에 따라 어떻게 갈리는지를 한 표에 놓았다.**

### 결과
1. 다섯 기체 × 세 밴드 15칸의 R90 은 3.69 km ⟨outputs/report05_derived.json : r90.span_comparable_min_km⟩ ~ 11.10 km ⟨outputs/report05_derived.json : r90.excluded_max_km⟩ 이고, 위 끝은 앵커 비교가능성이 `s1000plus ⟨outputs/report05_derived.json : r90.excluded[0]⟩` 인 기체가 갖는다 — 앵커 비교가능 12 ⟨outputs/report05_derived.json : r90.n_cells_comparable⟩칸만 보면 3.69 km ⟨outputs/report05_derived.json : r90.span_comparable_min_km⟩ ~ 7.44 km ⟨outputs/report05_derived.json : r90.span_comparable_max_km⟩ 다.
2. 국소 지수는 3.99 ⟨outputs/report13_freespace.json : ranges.*.*.equal_psd.full_waveform_capture.by_N.1.n_local_at_R90 → 15칸 평균⟩ 이고, `d` 축에서 국소적으로 R ∝ σ^(1/n) 이다.
3. 단일 자세 ψ=0 에서는 다섯 기체가 3 ⟨outputs/sigma_sensitivity.json : configurations.by_config.as_published.n_distinct_orders⟩가지 순위를 내고, 자세를 평균하면 LTE > 5G > WiFi ⟨outputs/report05_derived.json : ranking.consensus_order_aspect_avg⟩ 한 순위로 모인다.
4. 공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.minus10⟩ ~ +76.4% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.plus10⟩ 옮긴다 — 그래서 km 열은 순위를 읽는 표로 쓴다.

## 기울기 앵커 σ 위의 R90

Δσ 를 R90 근방 국소 지수 3.99 ⟨outputs/report13_freespace.json : ranges.*.*.equal_psd.full_waveform_capture.by_N.1.n_local_at_R90 → 15칸 평균⟩ 로 옮긴다 — `d` 축에서 국소적으로 $R \propto \sigma^{1/n}$ 이다(`src/freespace_scene.py:56`). 아래 표의 R90 은 전부 공칭 헤딩 ψ=0 의 수이고, **민감도 봉투와 함께 읽는다** — 공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.minus10⟩ ~ +76.4% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.plus10⟩ 옮긴다.

⚠ 그 봉투가 어디서 오는지도 함께 적는다 — 우리 세 밴드는 전부 PO 근사가 1 dB 안에 든다고 보장되는 부품 폭 문턱 아래에 부품을 남긴다([편 22 «PO 유효 무릎»](22_po-knee.ipynb)). 즉 σ 절대 레벨의 불확도는 **선언된 ±10 dB 봉투와 별개로 크기가 아직 정해지지 않은 항**을 하나 더 갖는다(⟨outputs/lowfreq_attack.json : q5_blast_radius.po_validity_blast_radius_the_real_one⟩). 그래서 아래 km 열은 **순위를 읽는 표**로 쓴다.

| 기체 | WiFi | LTE | 5G | 앵커 비교가능성 |
|---|---|---|---|---|
| mini5pro | 3.69 km | 7.44 km | 5.10 km | scaled |
| mavic4pro | 5.81 km | 4.65 km | 6.61 km | scaled |
| matrice4e | 4.38 km | 6.34 km | 5.93 km | scaled |
| phantom4 | 5.93 km | 5.76 km | 5.65 km | direct |
| s1000plus | 8.46 km | 11.10 km | 10.11 km | not_comparable |
