# 리포트 12 — 외부 σ 주입은 게재돼 있고, 그 값의 대가는 검증이 치른다

> ### 한 일
> **주입을 아카이브 고유문서 전수 위에서 좁은 정의와 넓은 정의로 갈라 다시 세고, 게재된 사례가 무엇을 인쇄했는지 판독했다.**

### 결과
1. 모집단은 아카이브 고유문서 327 편 [^1] (텍스트 해시 중복 제거 후)이고, 판정 문서 21 건 [^2] 과 다르다.
2. 넓은 정의(외부 서명표 → 시뮬 경로·링크)로 16 건 [^3] 이고 그중 게재는 10 편 [^4] 이다.
3. 인구조사를 PDF 로 다시 열어 판정하면 확정 게재 주입은 3 편 [^5] 이다(인구조사 주장 5 편 [^6]).
4. ⭐ 게재된 주입 사례가 인쇄한 것은 아키텍처가 아니라 **검증**이다 — 금속구와 held-out 밴드, 자동차 레이다 end-to-end, 정규화 서명 상관이 그것이다.

## 주입을 두 정의로 갈라 다시 세면

주입(엔진 밖에서 만든 σ 표를 자세로 조회해 시뮬 경로에 곱하는 구조)을 아카이브 고유문서 327 편 [^1] 위에서 다시 셌다. 정의를 둘로 갈라 적는다 — 좁은 정의는 `외부 상용 솔버 → 광선추적 경로계수`, 넓은 정의는 `외부 서명표 → 시뮬 경로·링크` 다.

| 갈래 · 정의 | 이 조사가 센 것 | 게재 | 인쇄된 검증 |
|---|---|---|---|
| 좁은 정의 — 외부 상용 솔버 → 광선추적 경로계수 | LAMBDA, arXiv:2607.03826, preprint [^7] | 0 편 [^8] | NO [^9] |
| 넓은 정의 — 외부 서명표 → 시뮬 경로·링크 | 16 건 [^3] | 10 편 [^4] (프리프린트 6 편 [^10]) | YES, 단 조건부 [^11] |
| 재판독 감사 — 인구조사를 PDF 로 다시 열어 판정 | 인구조사 10 건 [^12] 재판독 | 3 편 [^5] | 확정 주입 · 예산수준 · 주입 아님 셋으로 갈린다 |

## 통계 추출 — 표적 진폭을 분포에서 뽑는 쪽

엔진 밖 조달의 다른 갈래는 표적 진폭을 식 (108) CN(0, σ²) 에서 뽑는다. 그 클러터 서베이(Proc. IEEE)의 41쪽 전문을 세면 `physical optics` 0 회 [^13] · `dBsm` 0 회 [^14] · `max_depth` 0 회 [^15] · `mesh` 2 회 [^16] · `Blender` 1 회 [^17] 다.

클러터 서베이가 표적 축을 비운 것은 그 편의 설계다 — 표적 진폭은 통계 추출이고 광선추적은 건물 클러터에 쓴다. 그 빈 칸이 우리 자리이고, 메쉬 해상도·광선예산·기종을 인쇄하는 것이 우리 재현성 규약이다.

## 입장료는 검증이다

⭐ **게재된 주입 사례가 인쇄한 것은 아키텍처가 아니라 검증이다** — 0.5 m 금속구와 28 GHz held-out 밴드(Zhang, IEEE JSAC), 77 GHz 자동차 레이다 end-to-end(Deep, IET RSN), 정규화 서명 상관(Costa, IEEE J-STEAP).

입장료가 거기 있으므로 우리도 같은 값을 치른다. 그 값을 치르는 자리가 [부 5 «앵커와 검증»](../README.md#부-5-앵커와-검증) 와 [부 11 «실측 설계»](../README.md#부-11-실측-설계) 이고, 검증량이 σ 자체에서 서명 형태까지 갈린다는 사실이 우리 검증 층 설계의 근거다.
