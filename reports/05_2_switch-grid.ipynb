# 리포트 5-2 — 물리 스위치 격자: 회절은 리듬을 지우지 않고 리듬 없는 에코로 덮는다

이 편은 [리포트 5 «엔진의 물리 스위치»](05_engine-physics.ipynb) 의 **별편**이다.

**판**: matrice4e · 3.5 GHz · 15 m · 자세 8192 · 광선 4e9 · 깊이 2 · 확산 켬 · 정본 메쉬(mfixbatteryi5_blperairframe) · 앙각 −30° 한 자리. 갈리는 축은 굴절 R · 회절 D · 모서리회절 E 세 스위치뿐이다.

### 한 일
> 세 스위치의 의미 있는 조합 7 개를 같은 자리·같은 광선 예산(4×10⁹ 발)에서 **몫이 아니라 절대 dB** 로 재어, 회절이 리듬을 지우는지 덮는지를 갈랐다. 일곱 칸 모두 확산은 켠 채다 — 이 표적은 확산을 끄면 비스듬한 각에서 에코가 아예 없어 견줄 것이 남지 않는다(절 «영 에코»). STFT 맵과 블레이드 대역 에너지는 그 중 **다섯 팔**(다 끔·굴절·회절·굴절+회절, 그리고 우리 커널)에 대해 그렸다. 우리 커널은 광선 예산 축 밖의 별개 엔진이라 기준 팔로만 함께 둔다.

### 결과
1. ⭐**회절 스위치 하나가 경계선이다** — 회절을 켜면 상한 위 **바닥이 +35.4 dB**(-161.7 → -126.3 dB) 오르고, 빗살 선의 바닥 위 솟음은 +14.6 dB → +0.1 dB 로 무너진다. 리듬 몫이 12.9~13.2 %(백색 = 12.7)로 떨어진 것은 분자가 준 것이 아니라 **분모가 커진 것**이다. 회절이 없는 칸은 58.7~80.6 %.
2. ⭐**회절이 얹는 것은 리듬 없는 에코다** — 회절을 켠 시계열은 끈 시계열을 계수 1.04·위상 0.6° 로 그대로 품고 있고, 뺀 나머지에는 날개 박자가 없다(몫 11.2~12.4 % = 백색). 원래 빗살은 새 바닥보다 21.5 dB 아래에 잠겨 있을 뿐 사라지지 않았다 — 아래 «덮개 시험» 절.
3. ⭐**바꾸는 축은 하나, 얹는 축이 셋이다** — 굴절 R 만 원래 시계열 자체를 다른 것으로 만들고(담김계수 0.44~0.69), 회절 D · 모서리 E · 확산 F 는 원래 것을 남긴 채 위에 새 항을 더한다(≈0.98~1.10).
4. **모서리회절 스위치는 혼자서는 아무것도 안 한다** — ⚠이 한 줄만 **다른 판**(깊이 1 · 메쉬 전 세대)에서 읽는다. 모서리만 켠 팔이 그 판에만 있기 때문이다. 같은 판의 «다 끔» 과 견주면 리듬 80.50 %, 바닥 -162.4 dB 로 같다. 두 칸이 함께 가진 물리 수치 24 개 가운데 23 개가 비트단위로 같고, 남은 1 개도 배정밀도 끝자리 차이(Δ = 1.4e-14)다 — 갈리는 것은 원장 줄번호와 벽시계뿐. 모서리회절 후보를 만드는 자리가 회절 스위치 안에 있기 때문이다.
5. **굴절은 깎지만 죽이지 않는다** — 혼자 켜면 리듬 80.6 → 58.7 %, 빗살 솟음 +14.6 → +9.9 dB. 선의 자리는 126.1 Hz 로 그대로다.
6. **봉우리 위치는 다섯 팔 전부 126.1 Hz** — 예측 126.7 Hz 의 자리다. 위치가 아니라 **선명도**가 갈린다.
7. ⚠**사전등록 문턱 밖에 남은 것은 회절 켠 조합의 절대 레벨 하나다** — 깊이 1↔3 을 견준 13 쌍 중 문턱 밖은 판 위(−30°) R1D1** 4 쌍(깊이 3 에서 세 열 전부 +2.2~+2.4 dB)이고, 그 +2 dB 의 기전은 아직 안 세워졌다. ⛔전 판이 여기 함께 적었던 «판 밖 −60° 의 리듬 붕괴» 는 그 칸 자세 8,192 개 중 하나가 만든 값이라 **철회했다**(2026-08-16) — 아래 «깊이» 절.

### 잣대 — 무엇을 재나 (쉬운 말)
- **상한 위** — 날개 끝이 낼 수 있는 최고 도플러(f_tip) **위쪽** 자리. 몸통은 여기까지 못 온다. 즉 날개만 만들 수 있는 구역이다.
- **바닥** — 상한 위에서 빗살 자리를 뺀 나머지 에너지. 리듬이 없는 부분이다.
- **빗살** — 예측 박자 126.7 Hz 의 정수배 ±8 Hz 자리. 날개 플래시가 앉는 곳이다.
- **빗살 솟음** — 빗살 자리의 빈 하나당 에너지를 바닥 빈 하나당 에너지로 나눈 것. 0 dB 면 «선이 없다 = 백색» 이라는 뜻이다.
- **리듬 몫** — 상한 위 에너지 중 f_flash 정수배 ±8 Hz 에 붙은 몫[%] — 백색 12.7, 이상 로터 100. 몫이라서 바닥이 커져도 내려간다 — 그래서 절대 dB 를 함께 본다.
- **1차 선** — 블레이드 대역 전력 변조 스펙트럼의 1 차 선 — 국소 바닥 위 dB 와 봉우리 위치

⭐세 열은 전부 **절대 dB** — 원장 시계열 진폭의 제곱 단위(행의 level_db 와 같은 눈금). 비율이 아니다.

⚠**바닥 dB 열은 세로로만 읽는다.** 우리 커널 행과 Sionna 행은 애초에 단위가 다르다 — 우리 커널이 내는 것은 산란장(면적 차원)이고 Sionna 가 내는 것은 무차원 채널계수다. 그래서 두 엔진의 절대 dB 를 서로 빼는 것은 뜻이 없고, **같은 엔진 안에서 스위치를 바꿨을 때의 차이**만 이 열의 쓰임이다. 빗살 솟음 열은 같은 시계열 안의 비(선 ÷ 바닥)라 단위가 상쇄되므로 엔진 사이에서도 읽힌다.

표를 가로로 읽으면 결과 1 이 그대로 보인다. 회절이 켜진 2 행은 **바닥 열이 35 dB 넘게 위로 올라와 있고**, 그래서 빗살 솟음이 0 dB 근처(=선 없음)로 주저앉고, 그 결과로 리듬 몫이 백색 자리에 앉는다. 리듬 몫 하나만 보면 «선이 사라졌다» 로 읽히지만, 바닥 열을 같이 보면 «바닥이 올라왔다» 다.

⭐두 원장이 서로를 재현한다 — 5 행 전부 자세 8,192 개가 다 찼고, 리듬 몫이 0.03 %p 안에서 맞는다. ⛔전 판이 «굴절+회절» 행에 달았던 «1 차 선 갱신 대기» 꼬리표는 뗐다 — 그림 원장을 이 팔의 자세가 다 찬 뒤 다시 구웠고, 이제 그 행이 정확히 맞는다.

## 덮개 시험 — 회절은 지우나, 덮나

![switch factorial absolute dB](../outputs/figures/switch_factorial_abs_db.png)

**그림 5.** 같은 격자를 절대 dB 로 다시 읽은 판. 왼쪽 열이 깊이 1, 오른쪽 열이 깊이 3 이다. 위 두 패널은 세 열(움직이는 성분의 세기 AC · 상한 위 바닥 · 상한 위 빗살)의 절대 세기, 아래 두 패널은 «빗살 빈이 바닥 빈보다 얼마나 솟았나» 다 — 0 dB 면 선이 없다는 뜻이다. 회색 음영 줄이 회절을 끈 칸이다.

회색 줄(회절 끔)에서는 바닥 막대가 짧고 아래 패널 막대가 오른쪽(초록)으로 나가 있다. 흰 줄(회절 켬)에서는 바닥 막대가 길게 자라 있고 아래 패널 막대는 0 에 붙는다. **선이 사라진 것이 아니라 바닥이 선 높이까지 올라온 그림**이다.

### 세 가지로 갈라 물었다

판정은 설계서가 정한 판 — 앙각 −30° · matrice4e · 15 m · 광선 4×10⁹ — 의 5 쌍(회절만 켜고 끈 짝)에서만 내린다.

1. ⛔**«빗살 총합이 그대로냐»로 물으면 아니다.** 올라온 바닥이 빗살 자리 안에도 똑같이 들어차서 빗살 빈 총합도 함께 올라간다. 이 정의로는 «선» 을 잴 수 없다.
2. ✅**«선이 바닥 위로 솟았느냐»로 물으면 맞다.** 바닥은 27.0~37.2 dB 오르고, 빗살 솟음은 0 dB(=백색)로 주저앉는다.
3. ✅**덮개 시험 — 켠 판이 끈 판을 품고 있느냐.** 켠 시계열을 «계수 a × 끈 시계열 + 나머지» 로 갈라 보면 a 가 1.04~1.10 이고 위상 틀어짐은 많아야 4.7° 다. 5 쌍 중 5 쌍이 «a = 1» 을 3σ 안에서 담는다. 나머지의 전력은 두 판 전력의 차와 0.03 dB 안에서 같다 — 원래 신호와 겹치지 않는 **새 항이 더해졌다**는 뜻이다.

그리고 그 «나머지» 만 따로 재면 리듬 몫이 11.2~12.4 % 로 백색 밴드(10~16 %) 안이다. **얹힌 것에는 날개 박자가 없다.**

### 얼마나 깊이 잠겼나

원래 빗살(회절 끈 판)과 그 자리를 채운 새 바닥(회절 켠 판)의 높이 차 — 판 위 5 쌍에서 16.5~22.6 dB, 기본 칸에서 21.5 dB 다. 리듬은 사라진 것이 아니라 그만큼 아래에 있다.

⇒ 기전이 다르면 다음 실험이 달라진다. **지운다면** 회절 모형 자체가 틀린 것이고, **덮는다면** 회절 항의 절대 크기가 문제다. 남은 질문은 하나로 좁혀진다 — 상한 위로 새는 그 회절 항이 물리인가, 아니면 광선을 유한하게 쏴서 생기는 계산 흔들림인가.

## 깊이 — 판 밖 −60° 반례는 철회했다, 남은 것은 판 위 +2 dB 하나다

사전등록한 문턱은 «깊이 1↔3 쌍 전부에서 세기 차 < 2 dB 이고 리듬 차 < 3 %p 면 깊이 축 종결» 이었다. 실측은 견줄 수 있는 쌍이 13 쌍(판 위 10 쌍) + 두 판 모두 경로가 0 이라 견줄 것이 없는 죽은 쌍 2 쌍이다. 이 원장(R13)은 5 쌍을 문턱 밖에 적었는데 그중 −60° 한 쌍은 뒤 원장이 **철회**했으므로(아래 2 번), 남는 것은 **4 쌍**이다. 아래 1·2 번 값은 15 m · 자세 8,192 개 생값 · 광선 4×10⁹ 발 · PRF 19,700 Hz · 빗살 반폭 ±8 Hz 한 설정에서 잰 것이다(«자세를 뺀» 값은 가장 튄 자세 1 개 또는 8 개를 뺀 판이다).

1. **판 위(−30°)** — R1D1** 칸 4 개가 깊이 3 에서 세 열 전부 +2.2~+2.4 dB 다. 2 dB 밴드 바로 밖이다. 가장 튄 자세 1 개·8 개를 빼도 그대로이고, 그 팔의 재실행 문턱(같은 물리를 두 이름으로 독립 재실행한 폭) 0.07 dB 의 30~33 배라 **차이 자체는 실재한다.** ⚠다만 **그 +2 dB 의 기전은 모른다** — «회절 항이 깊이를 타고 한 번 더 얹힌 것» 은 후보일 뿐이고, 깊이 3 에서 광선 사다리도 시드 복제도 아직 안 돌렸다. 물리인지 경로 표집의 부산물인지 이 원장으로는 못 가른다.
2. ⛔**판 밖(−60°) 은 반례가 아니었다 — 자세 하나다(2026-08-16 철회).** «세기는 +0.09 dB 로 같은데 상한 위 바닥만 +12.7 dB 오르고 리듬 몫이 86.6 → 32.4 %(낙차 54.25 %p)로 무너진다» 는 그 칸 자세 8,192 개 중 **#3399 하나**가 만든 값이다. 그 자세의 움직이는 성분 |AC| 는 중앙값의 20.1 배인데 둘째 자세는 2.89 배뿐이고(고립도 = 최대÷둘째 = 6.97 — 이 원장 43 칸의 고립도 중앙값은 1.018 다), 깊이 1 판의 같은 칸은 고립도 1.005 로 평범하다. 경로 수도 정상이고(2,308 ≈ 중앙값 2,260), 로터 4 회 대칭 짝 셋도 0.69 · 1.47 · 1.21 배로 안 튄다. **그 자세 하나만 빼면** 두 판이 리듬 85.52 ↔ 85.24 %(차 -0.28 %p) · 세기 차 -0.059 dB 로 붙는다. 원장은 이 판정을 세 방법(이웃 평균으로 갈아 끼우기 · 죄 없는 자세를 빼는 대조검사 · 그 자세가 다른 어느 칸에서도 안 튄다는 전 칸 훑음)으로 각각 확인해 뒀다. ⚠**왜 그 자세가 튀는지는 모른다** — 이웃도 대칭 짝도 경로 수도 정상이라 구조적 정반사 플래시로는 안 읽힌다.

⛔**여기서 «깊이는 상한 위 바닥을 올리는 축» 이라고는 못 쓴다** — 전 판이 그렇게 쓴 근거는 위 2 번 하나였다. 표준 프레임이 싣는 두 팔(PS 다 끔 · PS 굴절만)의 깊이 짝 7 개는, 가장 튄 자세 1 개를 뺀 값으로 리듬 몫 차가 최대 0.40 %p 이고, 빗살 대비 차는 값이 잡히는 빗각·거리 5 칸에서 최대 0.21 dB 다(앙각 4 개 0° · −30° · −60° · −90°, 거리 15·30·120 m). ⚠**빗살 대비의 이 범위에 정면 0° 는 안 든다** — 그 칸은 +2.27 dB 로 범위 밖이고, 판 밖 −90° 는 빗살 대비 값 자체가 없다. 그리고 «작다» 라고 쓰는 근거는 격자 밴드가 아니라 관측값과 ≈0 인 재실행 문턱이다 — 밴드는 «판정 불가» 만 말한다(원장 scorecard.wording_rule_ko).

밴드로도 채점해 뒀다. **정본(전역)** 밴드로는 깊이 쌍 22 개 중 판독을 바꾼 쌍이 0 개다. ⚠전역 밴드는 앙각별 밴드보다 넓어 «안 바뀐다» 쪽에 유리하다고 원장이 스스로 적어 뒀다(scorecard.band_rule_caveat_ko) — **앙각별 밴드로 채점하면 2 개**다. 둘 다 정면 0° 의 빗살 대비 칸이고, 양쪽 판이 다 백색 널 자리라 «빗살 없음» 이라는 읽기 자체는 안 바뀐다.

⚠**리듬 몫 자체가 자유 파라미터 하나를 탄다.** 빗살 반폭을 ±2 → ±32 Hz 로 흔들면 이 잣대의 기하학적 바닥(= 2·hw/f_flash, 백색 대조군이 앉는 자리)이 3.2 → 50.5 % 로 옮겨 간다. 위 값들은 ±8 Hz 한 값에서 잰 **두 판의 차**로만 쓴다 — 리듬 몫의 절대값을 다른 문서로 들고 나가지 않는다.

⇒ 이 원장으로 확실한 것은 둘이다. ①**표준 프레임의 두 팔에서는 깊이 1 과 3 의 판독이 같다** — 큐의 표준 팔에 `--max-depth 3` 을 다시 태울 이유가 없다. ②**회절을 켠 조합의 절대 레벨을 인용할 때만 «깊이 1 한정» 꼬리표를 단다.** 「깊이 축이 통째로 살아 있다」도 「죽었다」도 이 원장으로는 못 쓴다.

## 곁가지 — 순정 공장 기본값은 빗각에서 에코가 정확히 0
