# 리포트 5-2 — 물리 스위치 격자: 회절은 리듬을 지우지 않고 리듬 없는 에코로 덮는다

이 편은 [리포트 5 «엔진의 물리 스위치»](05_engine-physics.ipynb) 의 **별편**이다.

**판**: matrice4e · 3.5 GHz · 15 m · 자세 8192 · 광선 4e9 · 깊이 1 · 확산 켬 · 앙각 −30° 한 자리 (⭐**센서가 드론보다 30° 아래에서 올려다본다** — 땅에 둔 레이다가 하늘의 드론을 보는 자리다. 저장소의 음수 앙각은 전부 «올려다봄» 이고, 코드로 확인했다: el −30 → 센서 z = −7.5 m). 갈리는 축은 굴절 R · 회절 D · 모서리회절 E 세 스위치뿐이다.

### 한 일
> 세 스위치의 의미 있는 조합 7 개를 같은 자리·같은 광선 예산(4×10⁹ 발)에서 재고, STFT 맵과 블레이드 대역 에너지로 나란히 놓았다. 그 다음 같은 칸들을 **몫이 아니라 절대 dB** 로 다시 읽어, 회절이 리듬을 지우는지 덮는지를 갈랐다. 우리 커널은 광선 예산 축 밖의 별개 엔진이라 기준 팔로만 함께 둔다.
> ⭐여기에 둘을 더했다 — **정지 클러터를 띠로 지운 판**(ECA 계열 부분공간 소거)에서 다시 그렸고, **회절이 얹는 항을 꺼내 그 성질을 쟀다**(이웃 자세 상관 · 도플러 분포 · 광선 예산 사다리). 그리고 이 편의 문장들을 **반증하려는 시험 여덟 개**를 따로 돌렸다.

### 결과
1. ⭐**회절 스위치 하나가 경계선이다** — 회절을 켜면 상한 위 **바닥이 +36.0 dB**(-162.4 → -126.3 dB) 오르고, 빗살 선의 바닥 위 솟음은 +14.6 dB → -0.4 dB 로 무너진다. 리듬 몫이 11.6~12.5 %(백색 = 12.7)로 떨어진 것은 분자가 준 것이 아니라 **분모가 커진 것**이다. 회절이 없는 칸은 62.1~80.5 %.
2. ⭐**회절이 얹는 것은 리듬 없는 에코다** — 회절을 켠 시계열은 끈 시계열을 계수 1.04·위상 0.6° 로 그대로 품고 있고, 뺀 나머지에는 날개 박자가 없다(몫 11.2~12.4 % = 백색). 원래 빗살은 새 바닥보다 21.5 dB 아래에 잠겨 있을 뿐 사라지지 않았다 — 아래 «덮개 시험» 절.
3. ⭐**바꾸는 축은 하나, 얹는 축이 셋이다** — 굴절 R 만 원래 시계열 자체를 다른 것으로 만들고(담김계수 0.44~0.69), 회절 D · 모서리 E · 확산 F 는 원래 것을 남긴 채 위에 새 항을 더한다(≈0.98~1.10).
4. **모서리회절은 무동작이다** — «모서리만» 은 «다 끔» 과 리듬 80.50 %, 바닥 -162.4 dB 로 같다. 두 시계열의 차이는 −309 dB(배정밀도 반올림 한계)이고 상관은 1.00000000 이다. 모서리회절 후보를 만드는 자리가 회절 스위치 안에 있기 때문이다.
5. **굴절은 깎지만 죽이지 않는다** — 혼자 켜면 리듬 80.5 → 62.1 %, 빗살 솟음 +14.6 → +10.5 dB. 선의 자리는 126.1 Hz 로 그대로다.
6. **봉우리 위치는 여덟 팔 전부 126.1 Hz** — 예측 126.7 Hz 의 자리다. 위치가 아니라 **선명도**가 갈린다.
7. ⚠**깊이 축은 아직 살아 있다** — 깊이 1↔3 을 견준 13 쌍 중 5 쌍이 문턱 밖이다. 깊이 축을 큐에서 뺄 수 없다 — 아래 «깊이» 절.
8. ⭐⭐**회절이 얹는 항은 슬로우타임 백색이다** — 담김계수 1.043∠0.6° 로 원래 신호를 그대로 두고 위에 얹는데, 그 항이 이웃 자세 상관 **0.0225**(정반사는 0.9907) · f_tip 위 **89.06 %** · 리듬 **11.56 %** 다. 완전 백색의 닫힌 식(88.82 % · 12.63 %)과 백색잡음 대조군에 그대로 앉는다. ⇒ **선이 낮아진 것이 아니라 바닥이 올라온 것**이다. 아래 «회절이 얹는 항의 정체» 절.
9. ⭐**그 항은 광선을 늘려도 안 줄어든다 — 오히려 는다**(깊이 1 에서 광선 ×4 에 요동 전체 **+5.25 dB**, 표본 잡음이면 −6 dB 여야 한다). 4×10⁹ 광선에서도 미수렴이다. ⛔기전은 아직 모른다 — «경로가 늘어서» 는 반증됐다(경로 178 배에 +0.43 dB). ⇒ 회절 켠 팔의 f_tip 위 수치는 인용하지 않는다.
10. **클러터를 제대로 지워도 안 달라진다** — 이 장면의 정지 클러터는 정확히 DC 라 평균 빼기가 이미 다 지운다(띠로 지우면 0.0~1.0 % 만 더 나온다). ⭐회절이 얹는 항은 0 Hz 근처가 아니라 **전 대역에 깔린다**(⚠노치 대역과 잣대 대역이 안 겹치니 «못 걷는다» 를 발견으로 쓰지 말 것).
11. **적대적 검증 — 합격 7 + 귀무 판정 1** — 잣대를 정의만 보고 다시 구현해 대조(최악 0.044 %p) · 원장 두 개 대조(0.04 %p) · 백색 귀무분포로 «이 잣대로는 안 보인다» 를 판정으로. ⚠그 과정에서 «모서리만 = 다 끔, 비트단위로 같다» 가 과장임이 드러나 결과 4 를 고쳤다.

### 잣대 — 무엇을 재나 (쉬운 말)
- **상한 위** — 날개 끝이 낼 수 있는 최고 도플러(f_tip) **위쪽** 자리. 몸통은 여기까지 못 온다. 즉 날개만 만들 수 있는 구역이다.
- **바닥** — 상한 위에서 빗살 자리를 뺀 나머지 에너지. 리듬이 없는 부분이다.
- **빗살** — 예측 박자 126.7 Hz 의 정수배 ±8 Hz 자리. 날개 플래시가 앉는 곳이다.
- **빗살 솟음** — 빗살 자리의 빈 하나당 에너지를 바닥 빈 하나당 에너지로 나눈 것. 0 dB 면 «선이 없다 = 백색» 이라는 뜻이다.
- **리듬 몫** — 상한 위 에너지 중 f_flash 정수배 ±8 Hz 에 붙은 몫[%] — 백색 12.7, 이상 로터 100. 몫이라서 바닥이 커져도 내려간다 — 그래서 절대 dB 를 함께 본다.
- **1차 선** — 블레이드 대역 전력 변조 스펙트럼의 1 차 선 — 국소 바닥 위 dB 와 봉우리 위치

⭐세 열은 전부 **절대 dB** — 원장 시계열 진폭의 제곱 단위(행의 level_db 와 같은 눈금). 비율이 아니다.

## 클러터를 제대로 지우면 달라지나 — 안 달라진다

위 판의 «정지 성분 제거» 는 **평균 한 값**만 뺀 것이다(도플러 0 Hz 한 칸). 실제 레이다는 정지
클러터를 **띠로** 지운다. 그래서 표준 방식으로 다시 지우고 같은 자리에서 다시 그렸다
(`outputs/switch_grid_clutter_0818.json` · 생성기 `benchmark/switch_clutter_stft_0818.py`).

**⭐ECA 계열 부분공간 소거** — 느린시간에서 $|f|\le f_{\rm cut}$ 인 도플러 부분공간을 **투영으로**
들어낸다. DFT 격자 위에서 그 부분공간의 기저는 서로 직교하므로, 직교투영은 «그 칸들을 0 으로» 와
정확히 같다. $f_{\rm cut}=100$ Hz — 날개 박자 첫 선(126.7 Hz)보다 **아래**라 신호는 안 건드린다.

### ⭐그리고 클러터 소거로는 회절 팔을 못 살린다

회절 켠 팔의 바닥은 0 Hz 근처에 쌓여 있는 것이 아니라 **도플러 전 대역에 깔려 있다**(앞 절의
백색 판정). 노치는 1.0 % 만 들어내고 빗살 대비는 −0.12 dB 그대로다.

⚠**이것을 «발견» 으로 쓰면 안 된다.** 노치는 \|f\| ≤ 100 Hz 를 지우고 빗살 대비는
\|f\| ≥ f_tip(1101.6 Hz) 에서 재므로 **두 자리가 애초에 안 겹친다** — 안 변하는 것이 당연하다.
여기서 말할 수 있는 것은 딱 하나다: **회절이 얹는 항은 0 Hz 근처가 아니라 전 대역에 깔린다.**

### ⚠3-펄스 MTI 는 이 잣대와 같이 쓰면 안 된다

교과서 대조군으로 (1, −2, 1) 캔슬러도 걸어 봤다. 이것은 고역통과라 응답이

$$|H(f)|^{2}=16\sin^{4}\!\left(\frac{\pi f}{\rm PRF}\right)
\;\;\Rightarrow\;\;
f_{\rm tip}\text{ 에서 }-18.3\ {\rm dB},\;\;{\rm Nyquist}\text{ 에서 }+12.0\ {\rm dB}$$

로 **30 dB 기울어 있다**. 그러면 상한 위로 에너지를 **만들어 올린다** — 정반사 팔의 f_tip 위 몫이
1.06 % → **94.1 %** 가 되고 리듬 몫이 80.5 % → **22.5 %** 로 떨어진다. 표적이 변한 것이 아니라
**필터가 분모를 바꾼 것**이다. (적대 대조 A1 은 MTI 가 없는 빗살을 **만들지는** 않음을 보인다 —
백색 입력에서 12.13 % 로 백색값에 머문다. 문제는 비교 가능성이지 위조가 아니다.)

## 왜 그렇게 되나 — 회절이 얹는 항의 정체

회절은 «물체 뒤로 파가 돌아 들어가는» 현상이다. 켜면 에코가 조금 늘 것 같은데 날개 무늬가 통째로
사라진다. 그 항을 꺼내서 성질을 쟀다 (`outputs/diffraction_mechanism_0818.json`).

![](../outputs/figures/diffraction_mechanism.png)

### ① 회절은 원래 신호를 지우지 않는다

회절 켠 시계열 $x_D$ 를 끈 시계열 $x_S$ 에 투영한다. ⭐정지 성분을 **먼저 빼고** 재야 한다:

$$a=\frac{\langle x_S,\;x_D\rangle}{\langle x_S,\;x_S\rangle}=1.043\;\angle\!+\!0.6^{\circ}$$

$a\approx 1$ 은 «원래 에코가 손상 없이 그대로 들어 있다» 는 뜻이다. 남은 것
$r=x_D-a\,x_S$ 가 회절이 **얹은** 항이고, 움직이는 성분 전력의 **91.2 %** 를 차지한다
(원래 신호는 8.8 %).

### ② 얹힌 항은 자세 하나 만에 완전히 잊는다

| | 이웃 자세 상관 (lag-1) | f_tip 위 에너지 | 빗살에 붙은 몫 |
|---|---|---|---|
| 정반사만 | **0.9907** | 1.06 % | 80.50 % |
| ⭐회절이 얹은 항 | **0.0225** | 89.06 % | 11.56 % |
| (대조) 백색잡음 20 회 | 0.0096 | 88.85 ± 0.40 % | 12.79 ± 0.44 % |

정반사는 스무 자세 남짓 걸쳐 천천히 잊는다 — 그 길이가 $1/f_{\rm tip}$(17.9 자세)와 맞는다.
얹힌 항은 **한 자세** 만에 잊는다.

그리고 완전히 백색이면 두 수가 **닫힌 식**으로 정해진다:

$$\text{f\_tip 위 몫}=\frac{{\rm PRF}/2-f_{\rm tip}}{{\rm PRF}/2}
=\frac{9850-1101.6}{9850}=88.82\,\%,\qquad
\text{리듬 몫}=\frac{2\,hw}{f_{\rm flash}}=\frac{16}{126.67}=12.63\,\%$$

실측이 **89.06 %** 와 **11.56 %** 다. 얹힌 항은 통계적으로 백색과 구별되지 않는다.

### ③ 그래서 왜 무늬가 사라지나 — 분모가 커진 것이다

$x[n]=s[n]+d[n]$ 로 쓰면 도플러 스펙트럼은

$$S(f)=|S_s(f)|^{2}+|S_d(f)|^{2}+2\,{\rm Re}\{S_s S_d^{*}\}$$

$s$ 는 날개 플래시라 $f_{\rm flash}$ 정수배에 **선**으로 서고 $|f|\le f_{\rm tip}$ 안에 갇힌다
(강체 회전의 최대 시선속도가 날개끝이므로). $d$ 는 백색이라 $|S_d|^2$ 는 **평평한 바닥**이다.
두 잣대가 이렇게 움직인다:

$$\text{빗살 대비}=\frac{\text{선}+\text{바닥}}{\text{바닥}}\;\longrightarrow\;1\;(0\ {\rm dB}),
\qquad
\text{리듬 몫}\;\longrightarrow\;\frac{2\,hw}{f_{\rm flash}}=12.63\,\%$$

⭐**선이 낮아진 것이 아니라 바닥이 올라온 것이다.** 원래 신호가 $a\approx1$ 로 살아 있다는 것이
그 증거다. 실측 빗살 대비 −0.4 dB, 리듬 몫 11.7 % 가 위 두 극한값이다.

### ④ 이 항이 표적의 운동일 수 있나 — 없다

슬로우타임에서 백색이라는 것은 **한 자세(50.8 µs) 만에 완전히 새 값이 된다**는 뜻이다.
그 사이 로터는 1.16° 돌고 날개끝은 2.4 mm 움직인다. 강체가 만드는 에코는 도플러가
$f_{\rm tip}$ 에서 **막혀 있어** 그렇게 빨리 못 변한다 — 정반사 팔이 실제로 그렇다
(천장 위 1.06 %). 얹힌 항은 **89 %** 가 그 천장 위에 있다.

### ⑤ 광선을 늘리면 줄어드나 — 오히려 는다

«그냥 표본 잡음 아니냐» 를 광선 예산 사다리로 가른다. 표본 잡음이면 광선 ×4 에 **−6 dB** 여야 한다.

| 깊이 | 광선 | 움직이는 성분 [dB] | 앞 칸 대비 | 표본잡음 예측 |
|---|---|---|---|---|
| 1 | 1.0e9 | −131.85 | — | — |
| 1 | 4.0e9 | −126.60 | **+5.25 dB** | −6.0 dB |
| 3 | 1.1e7 | −136.58 | — | — |
| 3 | 2.5e8 | −127.28 | **+9.30 dB** | −13.5 dB |
| 3 | 4.0e9 | −124.42 | **+2.86 dB** | −12.0 dB |

⇒ «표본 잡음» 은 **반증**됐다 — 방향이 반대다. 늘어나는 폭이 점점 작아지니 언젠가 포화하겠지만,
**4×10⁹ 광선에서도 아직 수렴하지 않았다.**

⛔**기전은 아직 모른다.** «경로가 늘어 무작위 위상으로 더해진다(전력 ∝ 경로 수)» 는 그럴듯하지만
**반증됐다** — 경로 수를 **178 배**로 늘려도 요동은 **+0.43 dB** 밖에 안 는다(자유 수확 사다리).
그러니 여기서는 **관측만** 적는다: «광선을 늘리면 는다». 왜 그런지는 열린 물음이다.

### ⚠ 이 절이 말하지 않는 것

«Sionna 회절이 틀렸다» 가 **아니다**. 말할 수 있는 것은 «우리 조건(드론 삼각망 · 4×10⁹ 광선 ·
자세 8,192 · 15 m · 앙각 −30°)에서 회절 항이 수렴하지 않았고, 그 미수렴분이 슬로우타임에서
백색으로 나타난다» 까지다. 수렴시켰을 때 무엇이 남는지는 이 원장으로 답할 수 없다.

⇒ **인용 규칙**: 회절을 켠 팔의 f_tip 위 수치는 인용하지 않는다. 팔 비교에서 «에코가 더 크다» 가
아니라 «수렴 안 된 항이 얹혀 있다» 로 읽는다.
