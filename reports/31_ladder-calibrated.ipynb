# 리포트 31 — 몸통은 진짜 CAD, 프로펠러만 갈아 끼운 교정 사다리가 답할 자격이 있는 유일한 축이다

> ### 한 일
> **여섯 단 전부를 «몸통은 진짜 CAD · 프로펠러만 돈다» 는 같은 운동학에 고정하고 프로펠러 형상 하나만 바꿔 사다리를 다시 세웠다.**

### 결과
1. 여섯 단이 같은 운동학을 쓴다 — rotors only (真 flight, matched) [^1] 하나로 전부 통일했다.
2. 바뀌는 것은 프로펠러 모양 하나다 — 회전대칭 원판 · 회전반경 위의 공 · 감싸는 상자 · 평판 두 장 · 삼각형 절반 · 진짜 CAD 여섯 단이다.
3. 기준선(진짜 CAD)의 검출 단면적은 -48.15 dBsm [^2], 플래시 대조비 16.84 dB [^3], 풍부도 7.94 [^4] 다.
4. 회전대칭 원판 단만 인용 자격이 아니오 [^5] 다 — 변조 전력의 0% [^6] 만 운동학이 허용하는 자리에 든다.
5. 이 축의 오차는 전부 **같은 방위끼리** 뺀 값이다 — 방위 24 점 [^7] 을 짝지어 뺐다.

## 왜 이 축만 답할 자격이 있는가

«표적을 얼마나 단순화해도 되나» 라는 물음에 답하려면 바뀌는 것이 형상 하나여야 한다. 있는 그대로의 사다리는 형상과 함께 운동학·재질도 바꾸므로 그 답을 못 준다([편 30 «앞 편»](30_ladder-three.ipynb)).

교정 사다리는 그 셋 중 둘을 묶는다 — 몸통은 언제나 진짜 CAD 이고, 도는 것은 언제나 프로펠러다. 그래서 단 사이의 차이는 **프로펠러 모양의 차이**다.

## 원판 단이 인용 자격에서 떨어지는 이유

회전대칭 원판은 돌려도 모양이 안 바뀐다. 그래서 변조를 **만들 방법**이 기하학적으로 막혀 있고, 남는 것은 격자 잔재다 — 변조 전력의 0% [^6] 만 운동학이 허용하는 자리에 든다.

Mini 2 쪽도 같다 — 0% [^10] 다. 이 단은 «형상을 얼마나 거칠게 그려도 되나» 의 답이 아니라 **«형상을 아예 지우면 어떻게 되나» 의 답**이므로, 사다리의 낭떠러지로 따로 읽는다.
