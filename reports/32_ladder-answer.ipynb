**그림 1.** 여섯 표적 모델의 마이크로도플러는 눈으로 얼마나 갈리는가?
열두 칸 모두 같은 색눈금과 같은 축이다. 세로축은 도플러(Hz), 가로축은 느린 시간(ms), 색은 그 도플러 칸에서 보이는 RCS 다. 점선은 운동학이 예측하는 날개 끝 도플러다. 윗줄이 있는 그대로의 사다리, 아랫줄이 교정 사다리다.

## 눈으로 읽히는 것 셋

1. 어느 칸에서나 가장 밝은 것은 0 Hz 가로줄 — 몸통이다. 날개 신호는 그보다 한참 어둡다. 이 낙차(호버 벌금)가 10 기체 평균 25.4 ⟨outputs/report16_synthesis.json : professor_answer.hover_penalty_db.mean_min⟩~38.7 dB ⟨outputs/report16_synthesis.json : professor_answer.hover_penalty_db.mean_max⟩ 다.
2. 회전대칭 표적(구·원판)은 0 Hz 한 줄만 있고 위아래가 비어 있다. 돌려도 모양이 안 바뀌니 변조를 만들 방법이 막힌다 — 이것은 계산 결과가 아니라 기하학이다.
3. 아랫줄 A → F 로 갈수록 배음 사다리가 촘촘해지지만, D(평판)와 F(진짜 CAD) 사이의 차이는 눈으로 잘 안 보인다. 그 «잘 안 보임» 이 이 편의 핵심 숫자다.

## 읽는 법 — 네 구간

- **공짜 — 해상도.** 도는 부품의 삼각형을 절반으로 줄이면 검출 단면적이 -0.01 dB ⟨outputs/report16_synthesis.json : professor_answer.axis4_what_simplification_actually_costs.rows.matrice4e.mesh_half_tri.sigma_ac_peak_err_db⟩, 템플릿 손실이 0.00 dB ⟨outputs/report16_synthesis.json : professor_answer.axis4_what_simplification_actually_costs.rows.matrice4e.mesh_half_tri.template_loss_db⟩ 움직인다. 방위에 따른 흔들림 7.99 dB ⟨outputs/report16_synthesis.json : professor_answer.axis3_time_modulation.azimuth_sd_db.mesh_max⟩ 에 묻히는 크기라 실측이 가르는 범위 밖이다.
- **싼 구간 — 날개를 평판으로.** 스팬·두께·부피가 같은 평판 두 장이면 검출 단면적이 한 자릿수 dB 안에 든다. «변조가 있나 없나» 만 보는 검출기라면 충분하고, 파형을 본뜨는 검출기라면 이미 비싸다.
- **비싼 구간 — 날개를 덩어리로.** 감싸는 상자는 속을 꽉 채웠으니 검출 단면적을 과대하게 낸다. 회전반경 위의 공은 Matrice 4E 에서 레벨만 보면 «잘 맞는» 것처럼 보이지만 파형 상관이 0.224 ⟨outputs/report16_synthesis.json : professor_answer.axis4_what_simplification_actually_costs.rows.matrice4e.sph_blade_rg.waveform_corr⟩ 뿐이라 템플릿 손실이 13.00 dB ⟨outputs/report16_synthesis.json : professor_answer.axis4_what_simplification_actually_costs.rows.matrice4e.sph_blade_rg.template_loss_db⟩ 다 — **레벨이 맞았다고 신호가 맞은 것이 아니다**.
- **낭떠러지 — 회전대칭.** 원판으로 바꾸면 검출 단면적이 -54.44 dB ⟨outputs/report16_synthesis.json : professor_answer.axis4_what_simplification_actually_costs.rows.matrice4e.disc.sigma_ac_peak_err_db⟩ 무너진다. 이것만은 어떤 커널 결함에도 안 흔들리는 기하학이다.

## 사다리의 순서는 삼각형 수가 아니라 보존량이 정한다

감싸는 상자(C)가 평판(D)보다 오차가 큰 것은 «형상 정보가 더 적어서» 가 아니라 부피를 안 지켜서다. 즉 «단순화의 값» 을 정하는 것은 삼각형 수가 아니라 **무엇을 보존했는가**(회전대칭 깨짐 → 스팬 → 부피 → 두께 순)다.

⭐ 한 줄로: **«모양이 있느냐 없느냐» 는 50 dB 대를 가르지만, «모양이 얼마나 정밀하냐» 는 한 자릿수 dB 안에서 논다.** 지도교수의 지적은 앞쪽이 아니라 뒤쪽을 겨눈 것이고, 뒤쪽에서는 지적이 맞는다.

## 자기반증 — 이 «공짜» 가 어디까지 참인가

이 커널에서 동체는 AC 에 정확히 0 을 기여한다. 동체를 아무리 깎아도 마이크로도플러 지표가 눈금 하나 움직이지 않는 것은 형상의 성질이 아니라 **가림 없는 커널의 성질**이다.

| 검사 | Mini 2 | Matrice 4E |
|---|---|---|
| 동체까지 깎았을 때 AC 의 상대 변화 | 9.7e-15 ⟨outputs/report16_synthesis.json : frame_blindness_audit.rows.mini2.ac_relative_difference_frame_decimation.max⟩ | 5.1e-15 ⟨outputs/report16_synthesis.json : frame_blindness_audit.rows.matrice4e.ac_relative_difference_frame_decimation.max⟩ |
| 같은 팔에서 동체:날개 비의 이동 | -1.88 dB ⟨outputs/report16_synthesis.json : frame_blindness_audit.rows.mini2.dc_ac_shift_from_frame_decimation_db.mean⟩ | +0.34 dB ⟨outputs/report16_synthesis.json : frame_blindness_audit.rows.matrice4e.dc_ac_shift_from_frame_decimation_db.mean⟩ |

첫 줄이 기계 정밀도라는 것은 AC 가 동체를 아예 안 본다는 뜻이다. 둘째 줄이 0 에서 떨어져 있는 것은 동체 간략화가 «몸통 대 날개 비» 는 움직인다는 뜻이다 — 면적·부피를 잃기 때문이다. 가림을 켜면 날개가 동체 뒤로 사라졌다 나타나면서 동체 형상이 AC 에 들어오기 시작한다.
