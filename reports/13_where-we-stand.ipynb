# 리포트 13 — 네 관문을 동시에 통과한 게재본은 0편이고, 최근접 3편은 각각 다른 관문에서 걸린다

> ### 한 일
> **후보를 네 관문으로 하나씩 판정하고, 우리 파이프라인이 각 축에서 한 일을 값과 함께 같은 표에 올렸다.**

### 결과
1. 후보 12 편 [^1] 을 네 관문으로 판정해 동시 통과는 0 편 [^2] 이다.
2. 가장 가까운 셋은 각각 **다른** 관문에서 걸린다 — 그래서 «한 편만 넘으면 된다» 는 요약이 성립하지 않는다.
3. 우리 자세 패턴은 해석 PO 구 대비 0.201 dB [^3] (입사 48 방향 [^4])이고, 밴드 기울기는 Das 적합계수 0.21 dB/GHz [^5] 에 맞춘다.
4. 실측 전이까지 보는 축으로 좁히면 모집단이 달라진다 — 정독한 33편(아카이브 전문검색 313편 범위) 중 0편. 세 요소를 다 갖추고 '실측 RF 데이터에서 분류 성능'까지 보고한 드론 논문은 찾지 못함. [^6]
5. 레벨 규약은 출처 논문 안에서 갈린다 — 잔차 최소화 변환을 쓰면 2.51 dB [^7] 만큼 밝다.

## 우리 파이프라인이 각 축에서 한 일

우리는 Sionna 의 Mitsuba/OptiX 광선엔진으로 first-hit(광선이 처음 부딪히는 면만 조명면으로 치는 판정) 가림을 풀고, 조명면 위에 부품별 재질 PO 를 적분한다. 레벨과 주파수 의존성은 측정 적합계수에서 받는다.

| 축 | 우리가 한 일 | 값 | 어디서 자세히 |
|---|---|---|---|
| 자세 패턴 | 광선엔진 first-hit 가림 + 조명면 PO 적분, 부품별 재질 | 해석 PO 구 대비 0.201 dB [^3], 입사 48 방향 [^4] | [편 21 «해석 PO 구 대비 구현오차는 kr 전 구간에…»](21_kernel-vs-reference.ipynb) |
| 절대 레벨 · 밴드 기울기 | A(f)·B1(φ)·B2 분해로 Das 적합계수에 정렬 | 0.21 dB/GHz [^5] | [편 24 «σ = A(f)·B₁·B₂ 에서 A(f) 의…»](24_anchor-mode.ipynb) |
| 바이스태틱 | 수신기 방향 그림자 광선으로 출사쪽 가시성 판정 | β ≤ 45° | [편 20 «수신 방향 그림자 광선을 켜면 상반성 위반이…»](20_bistatic-exit.ipynb) |
| 검출 | 경험 Pfa(실제로 울린 오경보율)로 CFAR 문턱 교정 | GPU 몬테카를로 2717 s [^10] | [편 53 «운용 형상에서 경험 Pfa 를 재니 명목값의…»](53_cfar-calib.ipynb) |
| 파형 비교 | 한 표적 · 한 검출기로 LTE · 5G · WiFi | 점유 · 대역폭 · PRF(펄스 반복 주파수) | [부 8 «조명원»](../README.md#부-8-조명원) |

## 앵커의 세 문헌, 그리고 규약이 갈리는 자리

Das 는 평균을 **측정한 방위 구간의 선형평균**으로 정의하고(Phantom 3 는 반평면), Zhang 식 (15)는 전 각도의 선형평균이다. Yuan 은 정의를 적지 않으므로 같은 축으로 읽는 것은 우리 추론이다 [^12].

| 문헌 | 기체 · 대역 | 기울기 | 우리가 쓰는 자리 |
|---|---|---|---|
| Das (IEEE WCL 2026) | Phantom 3 · 1.8 [^13]–18.2 GHz [^14] | 0.21 dB/GHz [^5] | 기울기 앵커 |
| Yuan (EuCAP 2025) | Phantom 3 · 같은 실험실 · VV | 0.315 dB/GHz [^15] | 고도 3측면 대조 |
| Zhang (IEEE JSAC 2026) | M350 · 10–36 GHz | 0.31 dB/GHz [^16] | 분해 σ=A(f)·B1(φ)·B2 형식 |

Das 의 Phantom 3 자료는 Das 자신의 참고문헌 [7](Wang 외, China Communications)에서 왔고 [^17], 우리가 쓰는 것은 그 기울기 하나다. 레벨 규약은 출처 논문 안에서 갈린다 — 정의를 그대로 쓰면 인쇄값이고, 잔차 최소화 변환을 쓰면 2.51 dB [^7] 만큼 밝다. 규약을 하나로 맞추는 자리가 [편 24 «σ = A(f)·B₁·B₂ 에서 A(f) 의 기울기만 측정에서 받고, 레벨과 각패턴은 우리 PO 출력이다»](24_anchor-mode.ipynb) 다.
