### 절 11.1 요약 수치 — 팔 × 앙각의 리듬 몫 [%]

날개끝 상한 **위**에 남은 에너지 가운데 예측 박자의 정수배에 앉은 몫이다. ⚠**상한 위만 재는 잣대**라 낮은 값이 «날개가 없다» 는 뜻이 아니다(절 13.7 · 팔마다의 표에 상한 **아래**를 보는 «빗살 대비[dB]» 를 함께 실었다). 백색잡음 값은 칸마다 다르다 — DJI Matrice 4E ≈ 12.6 · DJI Mavic 4 Pro ≈ 13.5 · DJI Mini 5 Pro ≈ 8.7 · DJI S1000+ ≈ 10.8.
