### 절 11.1 요약 수치 — 팔 × 앙각의 리듬 몫 [%]

날개끝 상한 **위**에 남은 에너지 가운데 예측 박자의 정수배에 앉은 몫이다. ⚠**상한 위만 재는 잣대**라 낮은 값이 «날개가 없다» 는 뜻이 아니다(절 13.7 · 팔마다의 표에 상한 **아래**를 보는 «빗살 대비[dB]» 를 함께 실었다). 백색잡음 값은 칸마다 다르다 — DJI Matrice 350 RTK ≈ 20.1 · DJI Matrice 4E ≈ 12.6 · DJI Mavic 4 Pro ≈ 13.4 · DJI Mini 5 Pro ≈ 8.7 · DJI Phantom 4 ≈ 8.7 · DJI S1000+ ≈ 10.8 · Holybro X500 V2 ≈ 8.8.

| 팔 | 표적 · 박자 | 거리 | 0° | −30° | −60° | −90° |
|---|---|---|---|---|---|---|
| `ours_r15_n8192_pw` | DJI Matrice 4E · 126.667 Hz | 15 m | — | — | — | 59\* |
| `ours_r15_n8192_pw_mfixbatteryi5_blperairframe` | DJI Matrice 4E · 126.667 Hz | 15 m | 62 | 87 | 53 | 56\* |

표기 — `—` 원장에 없는 칸 · `∅` 아무것도 안 돌아옴 · `◐` 자세가 덜 참 · `○` 움직이는 것이 없음 · `*` −90° 는 잣대 퇴화.
