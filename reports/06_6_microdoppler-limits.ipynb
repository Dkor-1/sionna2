## 절 1. 지상 레이더는 기체를 아래에서 보므로 가림이 무는 자세가 우리 자세다



> ### 한 일
> **같은 기체를 코 쪽·배 쪽·배 옆 세 자세에서 돌려 가림과 변조가 자세에 따라 어떻게 움직이는지 쟀다.**

### 결과
1. 프로펠러는 동체 **위**에 있다 — 위에서 내려다보는 자세에서는 날개가 통째로 드러나고, 아래에서 보는 자세에서 가림이 문다.
2. 코 쪽(앙각 15 도 [^1])의 가림 효과는 변조 폭 +0.05 dB · 레벨 -0.04 dB [^2] 다 — 변조 폭은 두 팔 dB 열의 표준편차 차(`modulation_std_db`)다.
3. 배 쪽(앙각 -15 도 [^3])에서는 -0.62 dB · +1.31 dB [^4], 배 옆(방위 90 도 [^5])에서는 -0.36 dB · -0.85 dB [^6] 다 — 세 칸이 서로 다른 값을 낸다.
4. ⚠ 세 자세의 dB 는 판 한 장과 표본수 위에 있다 — 판을 반 칸 옮기면 레벨 차가 판 셋에서 4.16 dB [^7] p-p 로 흩어지고, 원장은 «레벨 차가 판 선택을 견디는가» 에 아니오 [^8] 라고 적는다. ⛔ 변조 깊이 p-p 는 판을 그대로 두고 표본만 150 개로 솎아도 배 옆에서 중앙값 -2.65 dB · 뽑기 p-p 56.31 dB 로 흔들린다. 이 편이 세우는 것은 **가림이 자세에 따라 다르게 문다**는 존재다.
5. 자유공간 바이스태틱의 이등분선 앙각은 전 구간 음수라 우리는 드론 배를 본다 — 자세 스윕이 그 문장에 독립적인 근거를 붙였다.

## 자세가 답을 바꾼다

프로펠러는 동체 **위**에 있다. 그래서 위에서 내려다보면 날개가 통째로 드러나고, 그 자세에서 가림은 0 에 가깝다.

지상 레이더는 비행 중인 기체를 **아래에서** 보므로 기체 좌표계로 앙각이 음수이고, 거기서 가림이 문다.

⛔ 이 편의 자세 축은 **아래 표(원장)** 로 읽는다. 자세·기체별 요약 그림 `../outputs/figures/report15b_f3.png` 의 dB 라벨은 지금 서 있는 원장 `outputs/report15b_microdoppler.json` 의 값과 다르다 — 그림 빌더(`benchmark/build_report15b_figs.py::fig3_summary`)를 다시 돌리면 그 라벨이 아래 표와 같은 세대가 되고, 그때 이 자리에 들어온다.


## 세 자세의 값

| 자세 | 방위 | 앙각 | 가림 · 변조 폭 | 가림 · 변조 깊이 p-p ⚠ | 가림 · 레벨 |
|---|---|---|---|---|---|
| 코 쪽 | 0° [^11] | 15° [^1] | +0.05 dB | +4.44 dB [^12] | -0.04 dB [^2] |
| 배 쪽 ⭐ | 0° [^13] | -15° [^3] | -0.62 dB | -4.79 dB [^14] | +1.31 dB [^4] |
| 배 옆 | 90° [^5] | -15° [^15] | -0.36 dB | +23.19 dB [^16] | -0.85 dB [^6] |

변조 폭 열은 두 팔 dB 열의 표준편차 차이고(`outputs/report15b_microdoppler.json : cells.*.arms.{F_blade_occ, G_blade_free}.modulation_std_db` 의 차), 세 열 모두 그 자세의 슬로타임 6,000 표본 [^9] 전부로 잰 값이다. 세 줄이 서로 다른 값을 낸다 — **가림이 자세에 따라 다르게 문다**는 것이 이 편의 내용이고, 그 dB 의 부호와 크기는 아래 절이 판 앙상블 평균 뒤로 미룬다.

⚠ **변조 깊이 p-p 열의 쓰임은 «그 자세에서 한 표본이 얼마나 깊이 떨어지나» 까지다** — 자세끼리의 비교는 변조 폭 열이 맡는다. 그 이유는 아래 «두 잣대를 표본수로 흔들었다» 절이 사다리로 인쇄한다.

마이크로도플러는 프레임 사이 위상차로 재는 양이라, 광선 격자가 프레임마다 움직이면 그 움직임이 표적 운동과 같은 자리에 실린다. 그래서 커널은 자세 전체에 한 격자를 고정하고(`grid_ref`), **이 표는 그 얼린 판으로 다시 난 원장에서 나왔다** [^17]. 옛 판의 값은 `outputs/prefreeze/` 에 사본으로 있다.

⚠ 메쉬 지문(`mesh_provenance`)은 도장 스크립트를 돌린 다음 세대부터 이 원장에 실린다. 지문이 찍히면 표가 어떤 메쉬 세대에서 났는지까지 자기가 말한다.


## 두 잣대를 표본수로 흔들었다

| 잣대 · 칸 | 150 표본 | 750 표본 | 3,000 표본 | 6,000 표본(원장) |
|---|---|---|---|---|
| 변조 깊이 p-p · 코 쪽 | -0.16 (p-p 24.63) | +0.17 (p-p 15.10) | +4.22 (p-p 0.45) | +4.44 |
| 변조 깊이 p-p · 배 쪽 | -3.16 (p-p 24.34) | -1.78 (p-p 24.00) | -4.71 (p-p 0.41) | -4.79 |
| 변조 깊이 p-p · 배 옆 | -2.65 (p-p 56.31) | +2.25 (p-p 54.87) | +18.00 (p-p 26.85) | +23.19 |
| 변조 폭 · 코 쪽 | +0.04 (p-p 1.18) | +0.06 (p-p 0.30) | +0.05 (p-p 0.02) | +0.05 |
| 변조 폭 · 배 쪽 | -0.67 (p-p 1.66) | -0.63 (p-p 0.58) | -0.62 (p-p 0.02) | -0.62 |
| 변조 폭 · 배 옆 | -0.44 (p-p 3.55) | -0.48 (p-p 1.30) | -0.36 (p-p 0.35) | -0.36 |

같은 기록을 stride 로만 솎았다 — 판도 자세도 로터도 그대로이고, 흔든 손잡이는 솎기 간격 하나다. 칸의 값은 «위상 stride 개 뽑기의 중앙값 (그 뽑기들의 p-p)» 이고 마지막 열이 원장이 쓴 6,000 표본 [^9] 전부다. 판 사이 차이는 아래 절이 따로 잰다.

⭐ 사다리 네 칸의 **중앙값**이 옮겨 간 폭은 깊이 p-p 가 코 쪽·배 쪽·배 옆 4.60 · 3.01 · 25.84 dB 이고(코 쪽과 배 옆은 부호까지 바뀐다), 같은 사다리에서 변조 폭은 0.02 · 0.05 · 0.12 dB 다. ⛔ 깊이 p-p 가 무엇을 읽는지는 배 옆 칸이 보여 준다 — F 팔의 변조 깊이 79.12 dB [^18] 가운데 18.47 dB 를 6,000 표본 [^9] 중 **가장 깊은 한 표본**이 낸다(그 한 표본을 빼면 60.65 dB 다).


## 판 한 장이 그 dB 의 부호를 정한다

판의 중심만 **반 칸** 옮긴 같은 크기의 판으로 같은 자세를 다시 태우면 가림 dB 가 판마다 이렇게 갈린다 — 세 판(P0 · 반 칸 둘)의 값이다. ⚠ 이 재실행은 생산 슬로타임 열을 stride 40 [^19] 로 솎은 150 표본 [^20] 판이라, 위 표(6,000 표본 [^9])와는 표본수가 다른 추정치다.

| 칸 | 가림 · 변조 깊이 p-p (판 셋) | 가림 · 레벨 (판 셋) |
|---|---|---|
| `matrice4e/belly` | -17.98 [^21] / -18.27 [^22] / +8.26 [^23] | +1.53 [^24] / +2.12 [^25] / -2.04 [^26] |
| `matrice4e/nose` | +14.47 [^27] / -9.95 [^28] / +0.09 [^29] | -0.19 [^30] / +0.21 [^31] / -0.19 [^32] |

⭐ 한 자세 안에서 부호가 갈린다. 레벨 차의 판 셋 흩어짐 4.16 dB [^7] p-p 가 위 세 자세의 레벨 크기를 삼키고, 원장은 «레벨 차가 판 선택을 견디는가» 에 아니오 [^8] 라고 적는다 [^33].

⛔ 깊이 p-p 열의 판 셋 흩어짐 26.52 dB [^10] p-p 를 «반 칸 옮기면 가림 효과가 그만큼 바뀐다» 로 읽지 않는다. 세 판이 같은 슬로타임 표본을 쓰므로(`benchmark/adv_freeze_plate_sensitivity.py` 의 `idx = np.arange(0, n_t, stride)` 한 줄을 세 판이 공유한다) 그 흩어짐이 판에서 온 것은 맞다. 같은 통계량은 판을 하나도 안 옮기고 어느 표본을 뽑느냐만 바꿔도 배 쪽에서 중앙값 -3.16 dB · 뽑기 p-p 24.34 dB 로 흔들리는데, 뽑기 수(판 셋 · 위상 40가지)가 달라 두 p-p 의 크기 비교는 이 원장 밖이다.

⇒ 이 편이 세우는 것은 **가림이 자세에 따라 다르게 문다**는 존재이고, 그 dB 의 부호와 크기는 판 앙상블 평균(오프셋 여러 판의 평균)이 낸다. 기전은 [리포트 2 절 1 «커널이 하는 일»](02_kernel.ipynb) 이 잰다.


## 이 결론은 우리 기하 명세와 같은 방향이다

자유공간 바이스태틱의 이등분선 앙각은 전 구간 음수라 우리는 드론 배를 본다. 자세 스윕이 그 문장에 **독립적인 근거**를 붙였다 — 기하 명세와 광선 추적이 서로 다른 길로 같은 자세를 가리킨다.

⚠ Mini 5 Pro 의 그림은 2026-08-10 **겹침 수리 이전** 원장으로 그린 것이라 헤드라인에서 뺐다. 같은 그림을 그 기체로도 그려 `../outputs/figures/report15b_f1b.png` · `../outputs/figures/report15b_f2b.png` 에 뒀다 — 옛 세대 값이라는 것을 알고 읽는다. ⭐현 메쉬의 프롭·벨 겹침은 스탠드오프 1.0 mm 로 **0.00 %** 다[^34].

## 절 2. 문턱은 널 팔이 교정했고, 가장자리 시험은 아직 교정되지 않았다



> ### 한 일
> **변조가 나오면 안 되는 물체와 이상 점산란자를 같은 측정법에 넣어 판정 문턱이 거짓양성과 거짓음성을 내는지 먼저 쟀다.**

### 결과
1. 널 팔 15 칸 [^35] 에서 변조 판정이 0 번 [^36] 켜졌다 — 최댓값 3.83 dB [^37] 가 문턱 6.0 dB [^38] 아래에 든다.
2. 신호 팔은 6 칸 [^39] 중 5 칸 [^40] 이 켜진다 — 문턱이 양쪽에서 작동한다. 그 신호 칸은 **S** 팔(Sionna PathSolver · 스톡 그대로)로 낸 **Mini 2** 한 기체이고, 거리 1·3·10 m × 채널 둘이다 — 칸 이름이 원장에 그대로 있다(`criterion_a_calibration.rows`).
3. ⚠ 두 번째 잣대인 **가장자리 시험은 아직 안 섰다** — 반드시 통과해야 하는 이상 점산란자가 10 [^41] 칸 중 9 [^42] 칸만 통과하고, 산출물도 그 시험이 교정됐느냐에 아니오 [^43] 라고 적는다. 이 편이 세운 것은 문턱(기준 a)이고, 가장자리(기준 b)는 아니다.
4. 널 팔 전용 재계산에서도 같은 결론이다 — 신호 대 인공물 여유 3.85 dB [^44] 이고, 신호 변조 깊이를 인공물 바닥으로 **dB 값끼리 나눈 비**(분자·분모가 둘 다 dB 값이다)가 77.4 [^45] 다.
5. ⚠ 적대검증이 널 셈을 정정했다 — 실제로 시험된 널은 12 칸 [^46] 이고 나머지 3 칸 [^47] 은 AC 가 정확히 0 이라 시험 자체가 성립하지 않는 칸이다.

## 잣대를 먼저 교정하는 이유

«나왔다» 를 말하려면 «안 나와야 하는 것이 안 나온다» 를 먼저 보여야 한다. 회전대칭 물체는 돌려도 모양이 안 바뀌므로 변조를 만들 방법이 기하학적으로 막혀 있다 — 그런 물체에서 변조가 보이면 그것은 측정법이 만든 인공물이다.

반대 방향도 필요하다. 이상 점산란자처럼 «반드시 통과해야 하는» 모형이 같은 시험에서 떨어지면, 틀린 것은 대상이 아니라 시험이다.


## 널과 신호가 같은 문턱 아래에서 갈린다

![report15_f1](../outputs/figures/report15_f1.png)

**그림 1.** 변조가 나오면 안 되는 물체에서도 무늬가 보이는가?
위 네 줄이 신호 팔이다 — 두 기체(Mini 2 · Matrice 4E) 각각을 **S** = Sionna PathSolver(그림의 `Sionna` 줄 · 스톡 그대로) 와 **P** = 가림 없는 순수 PO 커널(그림의 `PO kernel` 줄) 로 낸 프로펠러 채널 네 줄이다. 맨 아래 한 줄이 **널 팔 넷 중 가장 시끄러운 것**이다 — 대표는 자료가 고른다(세 거리 최대 측대역의 최댓값).

⚠ 이 그림의 네 줄과 위 교정 격자는 범위가 다르다 — 교정 격자의 신호 칸 6 개 [^39] 는 **S** 팔의 Mini 2 뿐이다.

가장 시끄러운 널이 조용하면 나머지도 조용하므로, 나머지 세 줄은 그리지 않고 수치로 접었다. 접는 규칙과 접힌 값은 그림 빌더가 `null_rows_folded` 로 함께 낸다(`benchmark/report15_verdict_build.py::fig1_spectrograms`). ⚠ 지금 원장(`outputs/report15_verdict.json : figures.f1`)에는 그 키가 없다 — 그림을 다시 낼 때 채워진다.

색눈금은 각 패널 자신의 정적 반사를 0 dB 로 잡은 값이다. ⭐**신호 줄은 블레이드 박자로 띠가 뛰고, 널 줄은 시간에 전혀 안 변한다** — 표시 규약이 0 도플러를 지우지 않으므로 널 줄에도 정지 반사 그 자체는 0 dB 로 남는다.

즉 «검다» 가 아니라 «안 움직인다» 가 널의 표시다. 판마다 붙은 뱃지는 정적 성분을 지운 뒤의 최대 측대역을 그 패널 자신의 정적 반사 대비로 적은 값이고, 그 뱃지로 읽으면 신호 줄은 몇 dB 안쪽 · 널 줄은 수십 dB 아래다(뱃지 값 자체는 그림에만 있고 아직 원장 키로는 안 실린다).

## 절 3. 두 기체가 갈리는 축은 메쉬 품질이 아니라 표적 크기 대비 광선예산이다 — 예산을 맞춰 확인하는 시험은 이 하네스에서 아직 못 한다



> ### 한 일
> **기하를 고정하고 광선예산만 사다리로 흔들어 다시 추적해 판정 통계가 경로수를 어떻게 따라가는지 쟀다.**

### 결과
1. 판정 통계는 경로수 1 dB 당 1.01 dB [^57] 로 따라 올라간다 — 칸 6 개 [^58] 의 중앙값이다.
2. 두 기체의 프롭 경로수 비는 5.41 배 [^59] (Matrice 4E / Mini 2) 다. 예산 보정을 하면 자세 짝 15 [^60] 중 12 [^61] 에서 Mini 2 가 앞선다.
3. 예산 보정 차이의 중앙값은 -5.73 dB [^62] 이고, 경로수를 맞추면 Mini 2 가 +7.40 dB [^63] 올라간다.
4. ⚠ 예산을 맞추려면 5.41 배 [^64] 가 필요한데 표본 인자가 uint32 라 올릴 수 있는 여유는 2.0 배 [^65] 다 — 이 하네스로 기체 비교를 닫으려면 표본 규칙을 먼저 바꿔야 한다.
5. 그래서 Mini 2 의 무변조 칸 10/oblique/prod/prop [^66] 계열은 표적의 성질이 아니라 예산의 성질이다.


### 방법

**그림 2.** 경로수가 위상마다 껐다 켜지는가, 그리고 두 엔진은 거리에 따라 어떻게 갈리는가?
위 두 칸이 확산 채널의 경로수이고, 가운데가 정반사 채널, 아래가 거리별 두 엔진 일치도다.

경로수 인구조사의 판정은 «확산 채널: 껐다켜짐 없음(연속) / 정반사 채널: 50/60 칸이 통째로 비어 있고, 4 칸이 껐다켜진다 [^69]» 이다 — 확산 채널은 위상에 따라 매끄럽게 변하고, 정반사 채널은 대부분의 칸이 통째로 비어 있다.

오른쪽 아래에서 Matrice 4E 의 «hot» 자세만 거리와 함께 무너진다 — 그 자세의 경로수가 가장 적기 때문이다.

⚠ **이 편의 «예산» 은 확산 채널의 이야기다 — 정반사 채널의 «0 칸» 까지 같은 축으로 읽지 마라.** 그 «0» 은 예산 축과 앙각 축을 따로 열어 별도로 시험했고 답이 기체마다 갈렸다. Matrice 4E: ⭐앙각이었다 — 예산으로는 안 나오는데 가파른 앙각을 열면 나온다. «0» 은 격자가 프롭 법선 근처를 안 본 결과다. [^70] / Mini 2: 예산도 앙각도 아니다 — 사다리 꼭대기·앙각 85° 까지 프롭 정반사가 0 칸이다. 이제서야 «스톡 RT 가 프롭 정반사를 못 낸다» 에 가까워진다. [^71] 그 시험은 [리포트 6 절 3 «두 엔진»](06_3_pattern.ipynb) 에 있다.

## 남이 쏘는 신호를 쓰면 표본율을 못 고른다

패시브 레이더는 남이 쏘는 신호를 쓴다. 그 신호가 얼마나 자주 반복되는지가 우리가 볼 수 있는 도플러의 상한을 정한다.

날개끝 도플러를 접힘 없이 보려면 표본율이 그 두 배는 돼야 한다. 그 요구치와 각 모드가 실제로 주는 반복률을 같은 표에 놓으면 어디까지 보이는지가 바로 읽힌다.


## 네 모드의 반복률과 요구 표본율

| 모드 | 반복률 | 요구 표본율 | 접힘 없이 보이나 |
|---|---|---|---|
| WiFi VHT-LTF | 1000 Hz | 1980 Hz | 아니오 |
| LTE CRS | 1000 Hz | 1980 Hz | 아니오 |
| 5G SSB | 50 Hz | 1980 Hz | 아니오 |
| 5G NR-PRS | 200 Hz | 1980 Hz | 아니오 |

출처 [^90]

표적은 mini5pro [^91] · 반송파 3.5e+09 Hz [^92] · 호버 5500 rpm [^93] 이고, 이때 날개끝 주파수가 989.8 Hz [^94] · 블레이드 통과율이 183.3 Hz [^89] 다.


## 그래서 무엇이 남는가

⚠ 표의 네 줄 중 셋만 상시다 — 5G NR-PRS 는 측위 세션이 설정될 때 켜지는 옵션이라, 그 줄은 «상시로 기대하는 값» 이 아니라 낙관적 상한이다. WiFi 줄의 반복률도 표준이 정한 값이 아니라 트래픽 가정에서 온 선언값이다.

네 모드 전부 요구 표본율 아래에 있으므로 날개끝 확산은 접힌다. 살아남는 것은 **블레이드 통과율** 이다 — 헤드라인 기체(Matrice 4E)에서 126.7 Hz [^88] · 이 표의 기체(Mini 5 Pro)에서 183.3 Hz [^89] 이고, 둘 다 LTE·WiFi 의 나이퀴스트 안에 든다.

5G 쪽 두 모드는 반복률이 낮아 그 통과율마저 접는다. 5G 가 치르는 두 배의 대가가 마이크로도플러 축에서도 그대로 나타난다.

⭐ 이 구분이 실험 설계를 바꾼다. 날개끝 확산을 보려면 기준 안테나가 **풀 파형을 받아야** 하고, 그건 «상시 신호만 쓴다» 와 다른 조건이다.

그 기준 안테나가 «완벽하다» 는 가정을 푸는 편이 따로 있다 — ⛔«기준채널이 현실이면 얼마를 잃는가» 별편(2026-09-03 내림 — 동작점이 실내 통제 기하다, `archive/chamber_0903/`) 가 기준채널의 잡음·다중경로를 단일축으로 흔들어 검출이 무는 대가를 재고, 그 편의 마지막 절이 «검출이 죽은 칸에서 빗살은 어디까지 남는가» 를 같은 사슬에서 본다.
