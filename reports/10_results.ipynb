## 절 1. 레벨을 맞추려면 크기전이 법칙을 골라야 하므로 기울기만 받는다



> ### 한 일
> **σ = A(f)·B₁(φ,θ) 에서 A(f) 의 **기울기만** 측정에 맞추고, 레벨을 함께 맞추려면 무엇을 더 골라야 하는지와 그 대가의 크기를 원장에 적었다.**

### 결과
1. 재보정은 밴드별 스칼라 Δσ 하나씩이고, 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다.
2. 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB [^2] 다 — 절대 레벨은 우리 SBR+PO 커널(팔 B — sbr (rcs_sbr_batch, 1-bounce, penetrate=True) [^3]) 이 낸 값이다.
3. 레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB [^4] 갈린다 — 측정이 그 대가 없이 제약하는 것은 기울기뿐이다.
4. 앵커가 통제한 것 밖의 항은 4 개 [^5] 다 — 편파(UNRESOLVED) · 크기전이(UNRESOLVED) · 단일 실험실(UNRESOLVED) · 앙각 정합(PARTIAL). 규약 불확도 0.93 dB [^6] 는 원장에서 RESOLVED_EMPIRICALLY 이고, 채택 변환과 탈락 변환의 차로 앵커에 딸린다.
5. PO 오차가 1 dB 아래로 내려가려면 부품 폭이 파장의 0.729 [^7]배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

## 밴드 비교의 바닥 — 기울기 앵커 σ

σ = A(f)·B₁(φ,θ) 에서 **A(f) 의 기울기**를 Das 측정(IEEE WCL 2026 15:3731)에 맞춘다. **절대 레벨은 우리 SBR+PO 커널(팔 B — sbr (rcs_sbr_batch, 1-bounce, penetrate=True) [^3]) 이 낸 값**이고, 생산 모드 `slope_only` 의 세 밴드 평균 레벨이동은 0.00 dB [^2] 다.

레벨까지 맞추려면 크기전이 법칙을 골라야 하고 L² 와 L⁴ 가 최대 9.50 dB [^4] 갈리므로, 측정이 그 대가 없이 제약하는 기울기만 받는다.


## 재보정은 밴드별 스칼라 하나씩이다

그래서 정규화 각도 패턴은 9.6e-16 dB [^1] 안에서 그대로 남는다 — 앵커가 옮기는 것은 레벨이지 모양이 아니다.

앵커가 통제한 것 밖의 항은 **4 개 [^5]** 다 — 편파(UNRESOLVED) · 크기전이(UNRESOLVED) · 단일 실험실(UNRESOLVED) · 앙각 정합(PARTIAL). 개수는 원장 `uncontrolled`[^10] 의 status 로 센다.

규약 불확도 0.93 dB [^6] 는 같은 원장에서 RESOLVED_EMPIRICALLY 이고, 채택 변환과 탈락 변환의 차로 앵커에 딸려 다닌다. 앙각 정합 항의 크기는 0.73 dB [^11], 크기전이 항은 9.50 dB [^12] 다.


## PO 유효성 항이 이번에 크기를 얻었다

PO 오차가 1 dB 아래로 내려가려면 부품의 폭이 파장의 0.729 [^7]배 이상이어야 하는데, 우리 세 밴드는 전부 그 문턱 아래에 부품을 남긴다.

그 항의 **부호는 아직 정하지 못한다** — 우리 적분이 편파(전파의 전기장이 흔들리는 방향)를 가르지 않기 때문이다([^13]). 참값은 두 편파에서 서로 다른 값을 주는데 우리 커널은 그 둘을 구분 없이 하나로 내므로, 한쪽 편파를 기준으로 보면 낮고 다른 쪽을 기준으로 보면 높다.

이 사실은 원장에 「VV 측정 대 무편파 커널」 로 기록되어 있다. 이 부는 그 개연성을 결과에 넣지 않고 순위만 든다.

## 절 2. 앵커 σ 위의 R90 은 비교가능 12칸에서 3.69~7.44 km 이고, 밴드 순서는 기체마다 바뀐다



> ### 한 일
> **기울기 앵커의 Δσ 를 R90 근방 국소 지수로 옮겨 다섯 기체 × 세 밴드의 검출거리를 내고, 그 표에서 밴드 순서가 기체마다 어떻게 바뀌는지를 읽었다.**

### 결과
1. Δσ 를 R90 근방 국소 지수 3.99 [^22] 로 옮긴다 — `d` 축에서 국소적으로 $R \propto \sigma^{1/n}$ 이다.
2. 앵커 비교가능 기체 12칸의 R90 은 3.69 [^23] ~ 7.44 km [^24] 다 — 원키 `R90_C50_m`[^25] 에 앵커 Δσ[^26] 를 R90 근방 국소 지수로 옮긴 값이고, **공칭 헤딩 ψ=0 한 점**에서 푼 거리다[^27].
3. 밴드 순서는 기체마다 바뀐다 — 그 순서를 만드는 것은 자세별 로브 구조이고, 앵커는 밴드별 스칼라를 옮기면서 밴드 평균 레벨은 그대로 둔다.
4. 자세를 평균하면 다섯 기체가 한 순위 (LTE > 5G > WiFi [^28]) 로 모인다 — 단일 자세에서는 3 [^29]가지 순위가 나온다.
5. 아래 km 열은 **순위를 읽는 표**로 쓴다 — 공통모드 σ 오차 ±10 dB 가 이 열 전체를 -43.3% [^30] ~ +76.4% [^31] 옮긴다. ⚠ 같은 리프의 형제 키 `E_psi_Pd_at_R90` 는 그 거리에서의 헤딩 평균 검출확률을 15칸에서 0.00 ~ 0.65(5G 다섯 칸은 전부 0.0) 로 적으므로, 이 km 열은 **한 헤딩의 도달거리**로 읽는다.


### 방법

| 무엇을 | 어떻게 얻었나 |

| R90 정의 | **공칭 헤딩 ψ=0 한 점**의 σ 로 만든 SNR(d) 가 교정 문턱과 **최외곽 하강교차**하는 수평거리 `d` 다 — 거리축은 송·수신 기선의 중점에서 표적까지의 수평거리이고, 해를 찾는 칸을 고르는 유효 게이트는 β ≤ 90° 와 원거리장 둘이다[^27]. 헤딩 축은 같은 리프의 형제 키 `coverage_ceiling` · `E_psi_Pd_at_R90` 가 따로 든다 |
| 앵커 전이 | Δσ 를 R90 근방 국소 지수로 1차 전이한다 — `src/freespace_scene.py:56` |

## R90 이 무엇인가

**공칭 헤딩 ψ=0 한 점**의 σ 로 만든 SNR(d) 가 교정 문턱을 마지막으로 아래로 뚫는 수평거리다. 거리축은 송신국과 수신국을 잇는 기선의 중점에서 표적까지의 수평거리이고, 해를 찾는 칸을 고르는 유효 게이트는 β ≤ 90° 와 원거리장 둘이다[^27].

⛔ 키 이름의 `C50` 은 스펙 §2.3 의 커버리지 백분위(P_ψ[Pd≥0.9] ≥ 0.50)를 가리키지만, 이 실행이 그 자리에 적은 값은 단일 헤딩 solve 의 거리다 — 원장의 `key_note` 가 그렇게 적고 있다[^34]. 헤딩 축은 형제 키가 따로 든다: `E_psi_Pd_at_R90` 는 15칸에서 0.00 ~ 0.65 이고, 5G 다섯 칸은 전부 0.0 다.

⚠ 그 ψ=0 은 이 권의 헤드라인 방위 φ=90° 에서 도플러가 0 인 헤딩이다 — 2026-09-01 재계산으로 |f_d(ψ=0)| 는 15칸 전부 1 mHz 아래이고, CPI 0.1 s 의 가드 반폭은 선언 2.5빈 25 Hz [^35] · 검출기 적용 1.5빈 15 Hz 다. 이 표는 그 가드를 끄고 푼 거리이고, 가드를 켜고 세는 몫은 위의 형제 키 쪽에 있다.

표의 오른쪽 끝 «앵커 비교가능성» 은 그 기체의 σ 를 앵커 기체(Phantom 3)에서 얼마나 곧장 옮길 수 있나를 세 낱말로 적은 것이다.

## 앵커 σ 위의 R90

Δσ 를 R90 근방 국소 지수 3.99 [^22] 로 옮긴다 — `d` 축에서 국소적으로 $R \propto \sigma^{1/n}$ 이다(`src/freespace_scene.py:56`).

아래 표의 R90 은 전부 공칭 헤딩 ψ=0 의 수이고, **σ 민감도 봉투와 함께 읽는다** — **절 3** «그 순위는 자세평균이면 하나로 모이고, 자세평균 뒤집힘 문턱은 현실 봉투 안이다» 가 그 봉투를 든다.

## 도플러 축을 지우는 기구는 둘이다

5G 의 상시 기준신호(SSB)는 20 ms 주기라 PRF 50 Hz 를 준다. 그 축이 지워지는 경로가 둘이다.

**A. 표본화** — `guard_hz = g*PRF/M >= PRF/2  <=>  M <= 2g  <=>  T_cpi <= 2g/PRF [^91]` 로 가드가 접힘 축 전체를 덮는다(반송파·속도·거리 무관).

**B. 진폭** — 짧은 CPI 에서 가드가 도플러 진폭을 덮는다(파형 공통). 1.5빈 규약에서 LTE 도 CPI ≤ 0.024 s [^100] 에서 전 헤딩 블라인드가 된다.


## 기준 CPI 에서의 값 — 앵커 재현 격자

기준 CPI 는 0.1 s [^92] 다. 아래 표는 발표된 앵커를 그대로 재현하는 헤딩 격자 72 점 [^101] 위의 값이다.

| 모드 | PRF | 기준 CPI 의 M | 접힘 축 ± | 블라인드(1.5빈) | 블라인드(2.5빈) |
|---|---|---|---|---|---|
| WiFi | 1000 Hz [^105] | 100 [^106] | 500.0 Hz [^107] | 0.028 [^108] | 0.083 [^109] |
| LTE | 1000 Hz [^110] | 100 [^111] | 500.0 Hz [^112] | 0.139 [^113] | 0.250 [^114] |
| 5G | 50 Hz [^115] | 5 [^116] | 25.0 Hz [^117] | 0.639 [^118] | 1.000 [^119] |

5G 의 커버리지 0 은 선언가드 2.5빈 · CPI ≤ 0.10 s [^120] 에서 성립한다. 검출기가 적용하는 1.5빈 규약의 경계는 0.06 s [^121] 다.

두 수는 **표본화(A) 하나로** 정한 경계다. 진폭(B)까지 함께 관측한 경계는 각각 0.109 s [^122] · 0.070 s [^123] 이고, 아래 LTE 의 0.024 s [^100] 가 그 관측 잣대의 수다.


## CPI 를 늘리면 — 촘촘한 격자

세 파형 모두 블라인드율이 내려간다. 5G 가 치르는 **배수**는 5칸 전부에서 11.0 [^98] ~ 19.0 [^99]배 로 남는다 — 이것이 이 대가를 구조로 만드는 첫 번째 사실이다.

아래 표는 헤딩 격자 720 점 [^102] 위의 값이라, 같은 CPI 라도 앞의 앵커 재현 표와 값이 갈린다 — 격자가 촘촘하면 가드에 걸리는 헤딩 구간의 경계가 더 곱게 세어진다. 본문이 드는 수는 이 촘촘한 격자 쪽이다.

|---|---|---|---|---|---|
| 0.1 s [^92] | 0.053 [^124] | 0.158 [^125] | 0.636 [^126] | 12.1 [^127]배 | 4.0 [^128]배 |
| 0.2 s [^94] | 0.025 [^129] | 0.081 [^130] | 0.303 [^131] | 12.1 [^132]배 | 3.8 [^133]배 |
| 0.5 s [^134] | 0.008 [^135] | 0.031 [^136] | 0.119 [^137] | 14.3 [^138]배 | 3.9 [^139]배 |

⛔ 두 블라인드 열은 헤딩 격자 720점 위의 **칸 수**다 — WiFi 열이 38·18·6·2·2 칸, 5G 열이 458·218·86·38·22 칸이다. 두 열 다 CPI 와 함께 칸 수가 내려가고, WiFi 열은 마지막 두 칸에서 같은 2칸에 머문다. 5G/WiFi 열의 분모는 가장 작은 칸에서 2칸이라 한 칸(=0.0014)만 달라져도 그 칸의 배수가 2/3배~2배로 갈린다 — 이 격자가 정하는 것은 칸 수이고, 배수의 폭과 그 아래 WiFi 열의 움직임은 격자를 올려 다시 세어야 정해진다.


## 두 번째 사실 — 접힘 비율은 CPI 와 무관하다

5G 의 alias 비율 0.861 [^152] 는 적분시간이 아니라 표본화율의 성질이라 CPI 와 무관한 상수이고, WiFi·LTE 는 0.000 [^153] 다.

## 절 5. 모호속도는 표본화율의 성질이라 CPI 와 무관한 상한이다



> ### 한 일
> **CPI 를 늘려 커버리지를 회복하는 길의 끝을 재고, 그 길이 무모호 속도와 코히어런스 한계에서 어디서 막히는지를 값으로 적었다.**

### 결과
1. 모호속도는 5G 1.07 m/s [^154] · WiFi 14.40 m/s [^155] · LTE 40.74 m/s [^156] 이고, CPI 를 늘려도 그대로다.
2. 커버리지를 WiFi 수준으로 올리는 CPI 는 1.00 s [^157], LTE 수준은 0.38 s [^158] 이고 그 대가는 재방문 시간이다.
3. 그 CPI 가 코히어런스 한계 안에 머무는 구간은 20 m/s [^159] 까지다 — 25 m/s [^160] 에서 WiFi 패리티 CPI 가 한계 (0.84 s [^161]) 를 넘어선다.
4. 거리·속도 격자 16 [^162]칸 중 8 [^163]칸이 WiFi 패리티를 허용한다 — 그중 여유가 가장 얇은 칸은 6.3 km · 20 m/s 로, 코히어런스 한계 1.05 s 가 필요 CPI 1.00 s [^157] 보다 5 % 높다.
5. ⚠ 그 필요 CPI 는 헤딩 격자 720점의 눈먼 **칸 수**를 견줘 읽은 값이다 — 5G 의 눈먼 칸 수(38칸)와 목표인 CPI 0.1 s 의 WiFi 칸 수(38칸)가 **같은 정수**라 `≤` 판정이 등호로 통과한 자리다(5G 쪽이 한 칸 = 0.0014 만 더 컸으면 이 CPI 는 통과 목록에서 빠진다). 여유 5 % 인 칸이 통과 쪽에 남는지는 이 원장으로 정할 수 없다.

## 세 번째 사실이 결정적이다

모호속도는 표본화율의 성질이라 CPI 로 바뀌지 않는다 — 5G 1.07 m/s [^154] · WiFi 14.40 m/s [^155] · LTE 40.74 m/s [^156].

커버리지를 WiFi 수준으로 올리는 CPI 는 1.00 s [^157], LTE 수준은 0.38 s [^158] 이고 그 대가는 재방문 시간이다.

⛔ WiFi 패리티 CPI 는 `parity_cpi()` 가 헤딩 격자 720점의 눈먼 **칸 수**를 `≤` 로 견줘 읽은 값이다 — 5G 의 눈먼 칸 수(38칸)와 목표인 CPI 0.1 s 의 WiFi 칸 수(38칸)가 **같은 정수**라 `≤` 판정이 등호로 통과한 자리다(5G 쪽이 한 칸 = 0.0014 만 더 컸으면 이 CPI 는 통과 목록에서 빠진다). 그래서 이 원장이 정하는 것은 «칸 수가 같아지는 자리» 다.


## 패리티의 대가

| 패리티 목표 (5 m/s) | 필요 CPI | SSB 버스트 | 헤드라인 CPI 대비 경과 | 거리워크 | 코히어런트 이득 |
|---|---|---|---|---|---|
| LTE 수준 | 0.38 s [^166] | 19 [^167] | 3.75 [^168]배 | 0.064 [^169]빈 | 5.75 dB [^170] |
| WiFi 수준 | 1.00 s [^171] | 50 [^172] | 10.00 [^173]배 | 0.170 [^174]빈 | 10.00 dB [^175] |

⛔ WiFi 행의 경과·이득이 딱 떨어지는 것은 그 행의 필요 CPI 가 헤드라인 CPI 0.1 s 의 10 [^173]배로 읽혔기 때문이다 — 5G 의 눈먼 칸 수(38칸)와 목표인 CPI 0.1 s 의 WiFi 칸 수(38칸)가 **같은 정수**라 `≤` 판정이 등호로 통과한 자리다(5G 쪽이 한 칸 = 0.0014 만 더 컸으면 이 CPI 는 통과 목록에서 빠진다). 두 열은 그 CPI 에서 따라 나오는 산수다(경과 = T / 헤드라인 CPI, 코히어런트 이득 = 10log₁₀ 그 배수).


## 그 길이 막히는 자리

패리티 CPI 가 코히어런스 한계 안에 머무는 구간은 20 m/s [^159] 까지이고, 25 m/s [^160] 에서 WiFi 패리티 CPI 가 한계 (0.84 s [^161]) 를 넘어선다.

거리·속도 격자 16 [^162]칸 중 8 [^163]칸이 WiFi 패리티를 허용한다 — 그 밖의 칸에서는 CPI 로 메우는 길이 닫힌다.

⛔ 통과하는 칸 중 여유가 가장 얇은 것은 6.3 km · 20 m/s 로, 코히어런스 한계 1.0517 s 가 필요 CPI 1.00 s [^157] 보다 5.2 % 높을 뿐이다. 그 필요 CPI 는 5G 의 눈먼 칸 수(38칸)와 목표인 CPI 0.1 s 의 WiFi 칸 수(38칸)가 **같은 정수**라 `≤` 판정이 등호로 통과한 자리다(5G 쪽이 한 칸 = 0.0014 만 더 컸으면 이 CPI 는 통과 목록에서 빠진다) — 그래서 이 한 칸이 통과 쪽에 남는지는 격자를 올려 다시 세어야 정해진다.
