## 그 순위는 σ 오차 아래에서 어디까지 버티나

σ 오차를 두 종류로 나눠 넣는다. **공통모드**는 세 밴드를 함께 옮기고, **차분**은 밴드마다 다르게 옮긴다. σ 는 SNR 에 선형이라 σ 오프셋 Δ dB 가 곧 SNR 오프셋 Δ dB 다(선형성 잔차 1.2e-05 ⟨outputs/sigma_sensitivity.json : _meta.sigma_linearity_check⟩).

| 오차 종류 | 무엇이 움직이나 | 크기 |
|---|---|---|
| 공통모드 ±10 dB | 순위 유지 · 절대거리만 이동 | 15칸 전부 순위 불변 · 거리 -43.3% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.minus10⟩ ~ +76.4% ⟨outputs/sigma_sensitivity.json : common_mode.abs_range_shift_at_10db_pct.plus10⟩ |
| 차분(밴드별) | 순위가 뒤집힐 수 있는 축 | 뒤집힘 문턱 0.61 ⟨outputs/sigma_sensitivity.json : differential.smallest_flip_span_db_overall⟩ ~ 7.42 dB ⟨outputs/sigma_sensitivity.json : differential.largest_flip_span_db_overall⟩ (현실 봉투 5.01 dB ⟨outputs/sigma_sensitivity.json : differential.realistic_span_db⟩) |
| 밴드별 독립 오차 2 dB (몬테카를로) | 순위 보존 확률 | 0.28 ⟨outputs/sigma_sensitivity.json : monte_carlo_per_band_error → 5기체 최소⟩ ~ 0.82 ⟨outputs/sigma_sensitivity.json : monte_carlo_per_band_error → 5기체 최대⟩ |

취약성을 정하는 것은 기체 크기가 아니라 **밴드 간 σ 로브 산포**다 — 가장 작은 mini5pro ⟨outputs/sigma_sensitivity.json : size_vs_fragility.smallest_airframe⟩(전장 0.378 m ⟨outputs/sigma_sensitivity.json : size_vs_fragility.by_drone.mini5pro.extent_m⟩, LTE 에서 D/λ = 2.32 ⟨outputs/sigma_sensitivity.json : size_vs_fragility.by_drone.mini5pro.D_over_lambda_lte⟩) 가 가장 견고하고, 크기와 뒤집힘 문턱의 상관은 -0.62 ⟨outputs/sigma_sensitivity.json : size_vs_fragility.corr_extent_vs_flip_single⟩ 다.

또 하나의 실측 사실을 같은 자리에 적는다: σ 격자를 블레이드 형상 갱신본으로 바꾸는 것만으로 R90 이 최대 17.9% ⟨outputs/sigma_sensitivity.json : staleness_and_mesh_update.by_drone → 5기체 max_range_change_pct 최대⟩ 움직이고 순위쌍 1 ⟨outputs/sigma_sensitivity.json : staleness_and_mesh_update.n_orders_changed⟩개가 바뀌었다 — 통제되지 않은 σ 변화의 파급을 관측한 값이다.

⚠ 위 표의 뒤집힘 문턱은 2026-08-04 ⟨outputs/meshfix_applied.json : _meta.date⟩ 형상 정정 전 메쉬 위에 있다 — 하한 0.61 dB ⟨outputs/sigma_sensitivity.json : differential.smallest_flip_span_db_overall⟩ 를 내는 행이 그 정정을 받은 Matrice 4E 다([편 16 «형상 정정»](16_mesh-vs-real.ipynb)). 이 하한은 **단일 자세** 기준이고, 같은 기체를 **자세평균**으로 읽으면 1.30 dB ⟨outputs/sigma_sensitivity.json : aspect_averaged.by_drone.matrice4e.smallest_flip_span_db⟩ 다 — 캠페인 판정 문턱으로 드는 수가 그것이다([편 74 «캠페인이 결판내는 양»](74_sim-vs-meas.ipynb)).
