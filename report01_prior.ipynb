# 이 편은 갈라졌습니다 — 선행 연구가 표적 서명을 어디서 조달했나

> ⭐ **새 편은 `reports/` 에 있습니다.** 이 파일은 안내만 남깁니다.

재구성으로 리포트가 **8편 → 78편**이 됐습니다. 한 편에 중심 메시지 하나만 담고, 편당 **10분** 안에 읽히도록 나눴습니다(전에는 34~88분이었습니다).

이 편의 내용은 **부 2 「선행 연구」** 로 갔습니다.

| 새 편 | 제목 |
|---|---|
| [`08_census-published`](reports/08_census-published.ipynb) | 게재본 중 드론 메쉬에서 산란을 계산한 것은 0편이다 |
| [`09_census-preprint`](reports/09_census-preprint.ipynb) | 프리프린트는 따로 세고, Ziganshin 은 두 판을 구별해 인용한다 |
| [`10_procurement`](reports/10_procurement.ipynb) | 표적 서명을 어디서 조달했는지가 그 논문이 낼 수 있는 주장의 크기를 정한다 |
| [`11_procurement-catalog`](reports/11_procurement-catalog.ipynb) | 조달처 일곱 갈래 R1~R7 을 전수로 적었다 |
