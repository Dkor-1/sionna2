## §1. 표적 서명이 쓰는 양, Sionna RT 가 계산하는 양

Sionna RT 기술보고서 Version 1.2 (2025-11-24) ⟨outputs/prior_work_survey.json : engine.technical_report.version⟩ 를 열어 스톡 솔버가 무엇을 계산하는지 1차 사료로 확인했다. 경로는 SBR 로 후보를 만들고 이미지법으로 확정하며, 면과의 상호작용은 네 가지다(p.9).

| 스톡 솔버가 계산하는 것 | 표적 서명이 쓰는 것 |
|---|---|
| 정반사 — 프레넬 계수 곱 (식 127–130, p.46) | 조명면 위의 위상 코히런트 면적분 |
| 확산반사 — 산란계수 S 와 정규화 산란패턴 (p.50–52) | 자세에 따른 진폭 σ(θ, φ) |
| 굴절 · 1차 회절 (p.16) | 에지·정점 회절 항 |
| 경로 지연 τ · 도플러 · 가림 기하 | 그 기하를 σ 로 환산하는 단계 |

59쪽 전문에서 `physical optics` 0 ⟨outputs/prior_work_survey.json : engine.technical_report.term_counts.physical_optics⟩회, `radar cross section`/`RCS` 0 ⟨outputs/prior_work_survey.json : engine.technical_report.term_counts.radar_cross_section⟩회, `dBsm` 0 ⟨outputs/prior_work_survey.json : engine.technical_report.term_counts.dbsm⟩회다. `SBR` 은 44 ⟨outputs/prior_work_survey.json : engine.technical_report.term_counts.sbr⟩회 나온다 — 광선을 쏘고 튕기는 층은 있고, **면적분과 RCS 출력이 그 위에 얹히는 층**이다.

## §3. ⭐ 조달 카탈로그 — 표적 서명을 어디서 가져왔고, 그것이 무슨 주장을 사 주었나

편마다 표적 서명의 조달처를 골랐고, 고른 조달처가 그 편이 낼 수 있는 주장의 크기를 정한다. 갈래는 7개이고, 갈래마다 게재본 수를 함께 센다.

| 갈래 | 조달처 | 그것이 사 주는 주장 | 문서 | 게재 |
|---|---|---|---|---|
| R1 | 그 기체를 직접 잰다 (무향실·CATR) | 그 기체 그 대역의 절대 σ 와 계수표 | 3 | 3 |
| R2 | full-wave 솔버로 그 기체를 푼다 (광선엔진 밖) | 바이스태틱 절대 σ 와 그것을 먹는 커버리지 예산 | 1 | 1 |
| R3 | 외부 σ 를 광선경로에 주입한다 | 엔진을 건드리지 않고 얻는 자세의존 진폭 | 1 | 0 |
| R4 | 해석 표적 모델 (thin-wire · 점산란체 연속체) | 마이크로도플러 서명의 모양과 상관계수 | 2 | 2 |

Publication status was fixed per document from the PDF metadata subject field written by IEEE typesetting, from the page footer stamp, and from the arXiv stamp; an acceptance sentence inside the body of an arXiv manuscript scores partial and the document remains a preprint, which is the rule under which every count in this report was computed. Engine facts were read first-hand from the Sionna RT technical report (document Version 1.2, 59 pages) and from the installed Sionna 2.0.1 package, whose sionna.rt namespace was enumerated programmatically.
