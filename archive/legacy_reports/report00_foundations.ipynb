# 이 편은 갈라졌습니다 — 기초: Sionna 로 되는 것과, 표적 산란이 시작되는 자리

> ⭐ **새 편은 `reports/` 에 있습니다.** 이 파일은 안내만 남깁니다.

재구성으로 리포트가 **8편 → 78편**이 됐습니다. 한 편에 중심 메시지 하나만 담고, 편당 **10분** 안에 읽히도록 나눴습니다(전에는 34~88분이었습니다).

이 편의 내용은 **부 1 「스톡 엔진이 하는 일과 안 하는 일」 · 부 4 「우리 커널」** 로 갔습니다.

| 새 편 | 제목 |
|---|---|
| [`01_stock-says`](reports/01_stock-says.ipynb) | Sionna 기술보고서 59쪽에 physical optics 는 0회, SBR 은 44회 나온다 |
| [`02_engine-paths`](reports/02_engine-paths.ipynb) | 경로는 SBR 과 이미지법이 정한다 |
| [`03_engine-amplitude`](reports/03_engine-amplitude.ipynb) | 진폭은 국소 평면파–평면경계 해 하나로 만들어진다 |
| [`04_eight-factors`](reports/04_eight-factors.ipynb) | 필드 갱신 인자 여덟 개에 면적·곡률·치수·λ 가 없다 |
| [`05_size-sweep`](reports/05_size-sweep.ipynb) | 면적을 1600배로 키워도 경로 진폭은 7.4e-07 dB 움직인다 |
| [`06_decision-table`](reports/06_decision-table.ipynb) | 표적 항이 비에서 소거되는가와 절대값이 필요한가, 두 물음이 실험을 네 칸으로 가른다 |
| [`07_why-po`](reports/07_why-po.ipynb) | 완전파는 정확도의 과녁이고, SBR+PO 는 표를 만들 수 있는 유일한 비용대다 |
