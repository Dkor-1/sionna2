# 이 편은 갈라졌습니다 — X410 로 교정된 σ 를 얻는 세션 설계

> ⭐ **새 편은 `reports/` 에 있습니다.** 이 파일은 안내만 남깁니다.

재구성으로 리포트가 **8편 → 78편**이 됐습니다. 한 편에 중심 메시지 하나만 담고, 편당 **10분** 안에 읽히도록 나눴습니다(전에는 34~88분이었습니다).

이 편의 내용은 **부 11 「측정 설계」** 로 갔습니다.

| 새 편 | 제목 |
|---|---|
| [`70_calibration-sphere`](reports/70_calibration-sphere.ipynb) | 구가 σ 를 절대량으로 만들고, 반경 17.8 cm 를 고른다 |
| [`71_subband`](reports/71_subband.ipynb) | 표적을 한 거리빈에 넣는 최대 대역은 200 MHz 다 |
